## The Information Game

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Before writing a formula for entropy, play a game. One player thinks of
an object drawn from a known finite list. The other may ask only yes/no
questions. How many questions do you need in the worst case? That count
is already a measure of information: it is how much uncertainty you must
remove to name the object.

With a list of size $N$ and equal prior probability on each item, an
ideal strategy halves the search space at every step. After $k$ answers
you can distinguish at most $2^k$ hypotheses, so you need $$
k \ge \lceil\log_2 N\rceil
$$ questions in the worst case. Twenty questions reach a million
possibilities: $2^{20} = 1\,048\,576$. A *fair* yes/no answer therefore
carries one *bit* of information. That is Shannon’s unit before
Shannon’s formula: information as the number of balanced binary
distinctions required to resolve uncertainty.

If the prior is not uniform, the best question is no longer a literal
half-split of the *count* of remaining objects. It is a half-split of
their *probability*. The Shannon entropy $H = -\sum_i p_i\log_2 p_i$ is
the information budget, not the question count. The optimal expected
number of yes/no questions $L$ satisfies $H\le L<H+1$; equality $L=H$
holds only when every $p_i$ is a power of $1/2$. The lecture’s own die
is the exhibit: $H=\log_2 6\approx 2.585$ bits, but a Huffman strategy
needs $8/3\approx 2.667$ questions on average. We will meet the formula
next; the game already shows what the number is counting.

## MacKay’s Balance and Balls

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

David MacKay opens the compression half of *Information Theory,
Inference, and Learning Algorithms* with a weighing puzzle, Exercise 4.1
(MacKay, 2003). State the puzzle first. Do not yet say that the
instrument is ternary. The information-game habit is to ask how we
should proceed — what we must name — and then to try the instrument as
if it had two readings, like a yes/no question. Keep the *mechanism*
(what we put on the pans) on a different slide from the *maths* (how
many bits that use can return). Only after that sequence do we notice
that the balance has a third reading.

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/mackay-balance-balls.svg" class="" width="80%" style="vertical-align:middle;">

Figure: <i>A two-pan balance and twelve identical balls.</i>

You are given twelve balls, identical in appearance. Exactly one differs
in weight from the others; you do not know whether that odd ball is
heavier or lighter. A two-pan balance compares two equal-sized groups.
Design a strategy that finds the odd ball and the direction of the
defect.

Resist jumping to a particular grouping. Count first: how many states of
the world. Then try the instrument as a two-outcome device, because that
is the game we already know. Keep what we put on the pans separate from
the information budget those pans can return.

Under a uniform prior over those twenty-four possibilities, the Shannon
uncertainty we must remove is $$
H_{\mathrm{start}} = \log_2 24 \approx 4.58 \text{ bits}.
$$ At the end we need a single named hypothesis: $H_{\mathrm{end}} = 0$.
The *deficit* to close is therefore about $4.58$ bits. That difference —
start minus end — is the information the experiment must supply. We have
not yet said what the balance returns.

## Try Two Outcomes

A two-outcome instrument, used optimally, returns at most $\log_2 2 = 1$
bit per measurement. The information budget then says $$
2^4 = 16 < 24 \le 32 = 2^5,
$$ so five uses are not ruled out on paper and four cannot suffice.
Whether a physical balance *achieves* that one bit is a different
question: it depends on what we put on the pans.

$$
\log_2 24 \approx 4.58,\qquad 2^4 = 16 < 24 \le 32 = 2^5
$$

Write two columns if you can: left, the information budget; right, the
pans. The left column does not know about equal group sizes, leftover
balls, or borrowing a ball of known mass. Those constraints live on the
right. The rest of this block is the right-hand column, scored
afterwards by the left.

The move 6 vs 6 is the binary-search instinct. If we already knew the
odd ball was heavier, it would be a perfect split of twelve balls. We do
not know that. A down-left reading means “one of the left six is heavy,
*or* one of the right six is light”: twelve of the original twenty-four
hypotheses remain. One bit toward the 24, and *no* ball has been
eliminated. Mechanism and maths have already come apart: the pans moved,
the pile of balls did not shrink.

The information gain can be perfect for the two-outcome model — an even
split of 24 — while the mechanism fails to name a set of six suspect
*balls*. What we carry forward is six maybe-heavy and six maybe-light.
If we now *pretend* it was heavier and keep only the down pan, we have
thrown away half of the remaining hypotheses by assumption, not by
measurement.

This is the sequence people often initially sketch on the board: 6 vs 6,
then 3 vs 3. It is the right mechanism for a different puzzle — twelve
balls, one *heavier*. Hold that sequence, because this is where the
mechanism starts departing from our assumptions.

Firstly, what are the possible outcomes for this 3 vs 3. What if we
choose the pile that doesn’t contain the odd ball?

OK, so that’s revealed a flaw in our reasoning a balance actually has 3
outcomes.

## Three Readings

A two-pan balance has a ternary alphabet. The reading “level” is not an
annoyance to be designed out; it is a symbol that can carry information.
The first move of the two-outcome story, 6 vs 6, sets the probability of
that symbol to zero: with exactly one odd ball among the twelve, the
pans cannot balance. We spent a mechanism choice to turn a three-outcome
device into a two-outcome device, and then paid further mechanism taxes
(the known ball) to keep the third symbol from coming back.

$$
\log_2 3 \approx 1.58,\qquad 3^2 = 9 < 24 \le 27 = 3^3
$$

A binary question yields at most one bit. If the three readings are
equiprobable, one weighing contributes $$
\log_2 3 \approx 1.585 \text{ bits}.
$$ After $k$ weighings the strategy tree has at most $3^k$ leaves, so it
can name at most $3^k$ hypotheses. Compare budgets: $$
\begin{align*}
3^2 &= 9 < 24,\\
3^3 &= 27 \ge 24.
\end{align*}
$$ Two weighings cannot suffice no matter how clever the groupings.
Three weighings *can*: after naming the 24 hypotheses, three paths are
left over ($27-24=3$). Those leftover paths are not a thirteenth ball —
$13\times 2=26$ still sits under $27$, but three weighings cannot
identify an odd ball among thirteen without a known genuine. In bits, $$
3\log_2 3 \approx 4.75 > \log_2 24 \approx 4.58.
$$ The gap is small. That is why the puzzle feels tight. It is the same
comparison we already made with $\log_2 2$, with a larger per-use
budget.

The first weighing that realises the ternary budget puts four against
four and leaves four off. Level then means the odd ball is in the four
left off; a tip couples a group of four with a direction, as 6 vs 6 did,
but now the third symbol is alive. We do not give the rest of the tree.
The design principle is the one already present in the information game:
choose each measurement so that its outcomes are as nearly equiprobable
as possible under the current posterior. That maximises expected
information gain per weighing and is what makes the $3^3 \ge 24$ budget
achievable in practice. Adding a known-normal ball, or arranging a
weighing that can never balance, is the same kind of mechanism tax we
already computed for $h_2(2/3)$.

We stop at the existence argument. Constructing the adaptive tree —
which balls on which pan at each step — is MacKay’s exercise, and
solving it is the best way to feel what “informative experiment” means.
What the lecture owns is the split between ledgers: $\log_2 24$ versus
$\log_2 2$ or $\log_2 3$ is the information account; 6 vs 6, 4 vs 4, and
padding with a known ball are the mechanical account. The two are not
the same, and the known-ball tax is the exhibit.

Take a fair six-sided die, $X\in\{1,\ldots,6\}$ with $p(x)=1/6$, so Ask
questions in two stages: first learn which *group* the face falls in,
then resolve the face inside the group.

$$
H(G_{\mathrm{parity}})=1 \text{ bit},
\qquad
H(X\mid G_{\mathrm{parity}})=\log_2 3 \approx 1.585 \text{ bits}
$$

Let $G_{\mathrm{parity}}=0$ on even faces and $1$ on odd. Then
$w=(1/2,1/2)$, so $H(G_{\mathrm{parity}})=1$. Conditional on the group,
three faces remain, each with probability $1/3$, so
$H(X\mid G_{\mathrm{parity}})=\log_2 3$. The stages add:

$$
w=\Bigl(\tfrac16,\tfrac56\Bigr),
\qquad
H(G_{\mathrm{six}})=h_2\Bigl(\tfrac16\Bigr)\approx 0.65 \text{ bits}
$$

Let $G_{\mathrm{six}}=\mathbf{1}_{\{X=6\}}$. The group weights are
unbalanced, $w=(1/6,5/6)$, so the binary entropy of the first answer is
That is not one bit. After a “yes” the face is known ($H=0$). After a
“no” five faces remain ($H=\log_2 5$). Averaging, Again
$H(G_{\mathrm{six}})+H(X\mid G_{\mathrm{six}})=H(X)$. Note that
“$\ge 4$?” on a fair die is *also* a balanced cut ($\{1,2,3\}$ vs
$\{4,5,6\}$), so it *is* one bit — the unbalanced example has to be
chosen carefully.

Nothing about the cube stops you asking “odd?” after “low?” The question
can be asked; it cannot be *balanced* ($h_2(1/3)\approx 0.918$ bits). A
Huffman strategy on the six faces needs $8/3\approx 2.667$ questions on
average against $H=\log_2 6\approx 2.585$. The physical point about the
*balance* is an instrument with a fixed alphabet. The die point is atoms
that are not a power of two.

## Grouping Composition

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-grouping-composition.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-grouping-composition.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

$$
H(X)=H(w)+\sum_i w_i\, H(p_i)
$$

Partition the outcomes of $X$ into groups labelled by $i$, with group
probabilities $w_i$ and within-group distributions $p_i$. Then The die
examples are the case check: $H(w)$ is the entropy of the coarse
question; $\sum_i w_i H(p_i)$ is what remains once the group is known.
If $G=g(X)$ is the group label, this identity is exactly the chain rule
because $G$ is a function of $X$ (so $H(G\mid X)=0$) and
$H(X\mid G)=\sum_i w_i H(p_i)$.

In the reverse direction, observing $X$ fixes $G$ with no leftover
uncertainty. The mutual information $I(X;G)=H(G)-H(G\mid X)=H(G)$ is
exactly the price of the coarse question. After you know the group, the
remaining work is $H(X\mid G)$ — which of the fine outcomes inside that
group. Coarse-graining *keeps* $H(G)$ as your description and *discards*
$H(X\mid G)$:

The die paid $\log_2 6$ whichever cut came first. That invariance under
regrouping is not an accident of base-two logarithms; it is part of what
characterises Shannon entropy among candidate uncertainty measures
(Shannon’s third requirement in 1948, (Shannon, 1948); (Khinchin,
1957)). Categorical accounts are also explicit about this (Baez et al.,
2011).

## Grouping Is a Choice of Abstraction

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/grouping-as-macrostate.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/grouping-as-macrostate.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The idealised information game fails when we turn to the mechanism. But
we do have some control. We can coarse-grain, and reduce the search
space. For example, the low/high grouping is a choice of abstraction. We
keep a two-valued group label and throw away which of the three faces
occurred. Coarse graining in thermodynamics can be seen similarly. We
cannot track Laplace’s curves for every molecule, instead we keep
macroscopic numbers. Those macroscopic numbers split into *extensive*
quantities that scale with the system (e.g. $N$, $V$, energy, entropy)
and *intensive* quantities that do not (e.g. temperature, pressure). The
grouping property *constrains the form* of $H$; it is not that the form
of $H$ creates the grouping. Accounting for information works through
the groupings: the macrostate is the group label $G$, its entropy is
$H(w)$, and the unresolved motion inside it is $\sum_i w_i H(p_i)$.

## Gaussian, Maxwell–Boltzmann, and the Missing Measure

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/gaussian-maxwell-missing-measure.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/gaussian-maxwell-missing-measure.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Entropy depends on which degrees of freedom you count. Jaynes, following
Wigner, called that the anthropomorphic character of entropy: steam
tables do not include electron spin. Two observers who neglect different
parts of the uncertainty can still share a consistent account of what
they *do* measure, because the same functional form of $H$ sits
underneath. That is the bridge into Harry’s question. Each Cartesian
velocity component peaks at zero; the speed histogram in three
dimensions does not. The disagreement is not a paradox about folding. It
is a change of the undeclared measure — first made visible by drawing a
grid and concentric shells in velocity space, then named as Jaynes’ $m$.

```
import numpy as np

rng = np.random.default_rng(0)
n_velocities = 10000
velocity_sigma = 1.0
n_histogram_bins = 40
velocities = rng.normal(0.0, velocity_sigma, size=(n_velocities, 3))
speed = np.linalg.norm(velocities, axis=1)
velocity_x = velocities[:, 0]
speed_1d = np.abs(velocity_x)


def discrete_entropy(counts):
    n = counts.sum()
    p = counts[counts > 0] / float(n)
    return -np.sum(p * np.log(p))


def discrete_entropy_from_masses(p):
    p = np.asarray(p, dtype=float)
    p = p[p > 0]
    return -np.sum(p * np.log(p))


def bin_midpoints(edges):
    return 0.5 * (edges[:-1] + edges[1:])


def riemann_masses(pdf_at_mids, widths):
    # Midpoint masses p_i ≈ f(x_i) Δ_i, renormalised on the truncated support.
    mass = np.asarray(pdf_at_mids, dtype=float) * np.asarray(widths, dtype=float)
    total = mass.sum()
    if total <= 0.0:
        return mass
    return mass / total


def gaussian_pdf(x, sigma):
    return np.exp(-0.5 * x**2 / sigma**2) / np.sqrt(2.0 * np.pi * sigma**2)


def gaussian_differential_entropy(sigma):
    return 0.5 * np.log(2.0 * np.pi * np.e * sigma**2)


def half_normal_pdf(s, sigma):
    # Folded |N(0, sigma^2)| on s >= 0; mode remains at zero.
    return (
        np.sqrt(2.0 / (np.pi * sigma**2))
        * np.exp(-s**2 / (2.0 * sigma**2))
    )


def maxwell_speed_pdf(s, sigma):
    return (
        np.sqrt(2.0 / np.pi)
        * (s**2 / sigma**3)
        * np.exp(-s**2 / (2.0 * sigma**2))
    )


def maxwell_speed_differential_entropy(sigma):
    return np.log(sigma * np.sqrt(2.0 * np.pi)) + np.euler_gamma - 0.5


def speed_range_max(s, p_hi=0.995):
    return float(np.quantile(s, p_hi))


def equal_delta_speed_edges(s_max, n_bins):
    return np.linspace(0.0, s_max, n_bins + 1)


def equal_volume_speed_edges(s_max, n_bins):
    u_edges = np.linspace(0.0, s_max**3, n_bins + 1)
    return u_edges ** (1.0 / 3.0)
```

## Velocity Space: Grid and Speed Rings

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/velocity-space-shell-2d.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/velocity-space-shell-2d.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Harry’s question — why does the speed histogram peak away from zero when
each Cartesian component peaks at zero? — is answered by a change of
variables in velocity space, not by folding alone. Draw the plane of two
components $(v_x,v_y)$. An isotropic Gaussian is a circular cloud
centred at the origin. Partition that plane with an equal-$\Delta v$
grid: every square has the same area. Now draw concentric circles of
radius $s=|\mathbf{v}|$. The ring between $s$ and $s+\Delta s$ contains
more and more grid cells as $s$ grows, because its circumference is
$2\pi s$. Counting samples by which ring they fall in therefore weights
larger speeds more heavily than the Cartesian grid does. That
circumferential factor is the two-dimensional shell volume. In three
dimensions the same geometry becomes a spherical shell of area
$4\pi s^2$, which is the $s^2$ in the Maxwell speed law. Absolute value
alone (the one-dimensional fold) never moves the mode; the shell volume
does.

## The Continuous Limit Drops $\log\Delta$

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/log-delta-divergence.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/log-delta-divergence.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Refine the equal-width partition of a Gaussian component. For each bin
width $\Delta$, Plotting $H$ against $-\log\Delta$ gives a line of slope
one, offset by the constant $h=\tfrac12\log(2\pi e\sigma^2)$. The
discrete entropy diverges as the bins shrink because each mass
$p_i\approx f\Delta$ contributes an extra $-\log\Delta$. Differential
entropy is defined to be the $\Delta$-independent remainder. The
continuous functional therefore *ignores a term that is driving to
$+\infty$*. Differential entropy $h$ can be negative because densities
can exceed $1$ (mass concentrated below the unit scale). Dropping
$-\log\Delta$ *permits* that; it is not the cause. And $h$ is not a
code-length bound.

```
import matplotlib.pyplot as plt
import mlai.plot as plot
import mlai
```

```
vx_lim = 4.0
n_list = np.array([5, 8, 10, 16, 20, 25, 32, 40, 50, 64, 80, 100])
Hs = []
deltas = []
for n_bins in n_list:
    edges = np.linspace(-vx_lim, vx_lim, n_bins + 1)
    delta = edges[1] - edges[0]
    counts, _ = np.histogram(velocity_x, bins=edges)
    Hs.append(discrete_entropy(counts))
    deltas.append(delta)
Hs = np.array(Hs)
deltas = np.array(deltas)
h = gaussian_differential_entropy(velocity_sigma)
minus_log_delta = -np.log(deltas)

fig, ax = plt.subplots(figsize=plot.big_wide_figsize)
ax.plot(minus_log_delta, Hs, 'o', color='#4C72B0', ms=8, label='$H$')
xline = np.linspace(minus_log_delta.min() - 0.1,
                    minus_log_delta.max() + 0.1, 50)
ax.plot(xline, h + xline, color='#C44E52', lw=2, label='$h-\\log\\Delta$')
ax.axhline(h, color='#55A868', ls='--', lw=2, label='$h$')
ax.set_xlabel('$-\\log\\Delta$')
ax.set_ylabel('nats')
ax.legend(fontsize=10)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
mlai.write_figure('log-delta-divergence.svg',
                  directory='./information')
```

## Bin Widths Are a Radon-Nikodym Derivative

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/discrete-jacobian-radon-nikodym.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/discrete-jacobian-radon-nikodym.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

For a partition that is not equal-width, the bin-width correction is no
longer a single $\log\Delta$. Probability conservation in cell $i$ still
reads $p_i \approx f\,\Delta_i$, so Take the equal-volume cells of the
previous snippets, written once in $u=s^3$ and once in $s$. The masses
$\{p_i\}$ (hence $H$) are the same. The two differential entropies
differ by the average log-ratio of widths, That ratio
$\Delta s_i/\Delta u_i$ is called the discrete Radon-Nikodym derivative
of Lebesgue-on-speed with respect to Lebesgue-on-volume. In the
continuum limit it becomes the Jacobian $1/(3s^2)$. Probability
conservation $f_s(s)\,{\mathrm{d}}s = f_u(u)\,{\mathrm{d}}u$ is the same
statement.

The Maxwell speed density relative to three-dimensional Lebesgue is the
shell factor $4\pi s^2$. Equivalently, the pushforward of Lebesgue
measure on $\mathbb{R}^3$ has density $4\pi s^2$ in $s$: The Gaussian
$h(v_x)$ and the Maxwell $h(s)$ disagree because they are different
coordinates *and* because each $h$ is taken against a different
*undeclared* Lebesgue measure.

```
import matplotlib.pyplot as plt
import mlai.plot as plot
import mlai
```

```
s_max = speed_range_max(speed)
edges_s = equal_volume_speed_edges(s_max, n_histogram_bins)
edges_u = edges_s**3
delta_s = np.diff(edges_s)
delta_u = np.diff(edges_u)
counts, _ = np.histogram(speed, bins=edges_s)
p = counts / float(counts.sum())
H = discrete_entropy(counts)
occupied = counts > 0
s_mid = 0.5 * (edges_s[:-1] + edges_s[1:])
rn_discrete = np.divide(delta_s, delta_u, out=np.zeros_like(delta_s),
                        where=delta_u > 0)
mean_log_j = np.sum(p[occupied] * np.log(rn_discrete[occupied]))
h_s_est = H + np.sum(p[occupied] * np.log(delta_s[occupied]))
h_u_est = H + np.sum(p[occupied] * np.log(delta_u[occupied]))
h_s = maxwell_speed_differential_entropy(velocity_sigma)
ss = np.linspace(max(s_mid[occupied].min(), 1e-3), s_max, 400)

fig, ax = plt.subplots(figsize=plot.big_wide_figsize)
ax.plot(s_mid[occupied], rn_discrete[occupied], 'o', color='#4C72B0',
        ms=6, label='$\\Delta s_i/\\Delta u_i$')
ax.plot(ss, 1.0 / (3.0 * ss**2), color='#C44E52', lw=2,
        label='$1/(3s^2)$')
ax.set_xlabel('$s$')
ax.set_ylabel('width ratio')
ax.legend(fontsize=10)
ax.text(0.98, 0.95,
        f'$\\sum p_i\\log(\\Delta s_i/\\Delta u_i)={mean_log_j:.3f}$\n'
        f'$H+\\sum p_i\\log\\Delta s_i={h_s_est:.3f}$\n'
        f'$H+\\sum p_i\\log\\Delta u_i={h_u_est:.3f}$\n'
        f'$h(s)={h_s:.3f}$',
        transform=ax.transAxes, va='top', ha='right', fontsize=12,
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.9))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
mlai.write_figure('discrete-jacobian-radon-nikodym.svg',
                  directory='./information')
```

## Jaynes: Promote the Bins to a Measure

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/jaynes-limiting-density.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/jaynes-limiting-density.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

$$
h_m(f)=-\int f\log\frac{f}{m}\,{\mathrm{d}}x=-\text{KL}\left( f\,\|\,m \right)
$$

Differential entropy $h=-\int f\log f\,{\mathrm{d}}x$ pretends the
coordinate measure is part of the data. The histogram arithmetic showed
that it is not: $H$ is a functional of masses, and the passage to a
density subtracts $\sum_i p_i\log\Delta_i$, whose continuum limit is the
log of a reference density $m$. Jaynes called $m$ the *limiting density
of discrete points* — the binning measure, promoted to a density
(Jaynes, 1963).

Once $m$ is written down, Taking $m$ uniform (Lebesgue) in the Cartesian
component $v_x$ recovers the ordinary Gaussian remainder $h(v_x)$.
Taking $m$ uniform in three-dimensional velocity, and writing it in the
speed coordinate, gives $m(s)\propto s^2$ — the shell volume that the
concentric rings made geometric and that had been hiding in the
Jacobian. The two Maxwell–Boltzmann entropies disagreed because each $h$
used a different undeclared $m$. Relative entropy makes the choice
explicit. The Jacobian cancels inside $\text{KL}\left( f\,\|\,m \right)$
under a common invertible map; the discrete KL has no divergent
$-\log\Delta$ term. The map $\mathbf{v}\mapsto s=|\mathbf{v}|$ is *not*
invertible. Equality of KL in the speed coordinate with KL in 3D
velocity holds *here* because both $f$ and Lebesgue on $\mathbb{R}^3$
are isotropic, so both lose the same direction information. In general
the data-processing inequality gives $\le$, not $=$.

Going continuous buys analytic machinery; it does not remove the need to
declare $m$. Which binning — which degrees of freedom — you treat as
\`\`equally likely a priori’’ remains an anthropomorphic choice in
Jaynes’ sense. This names $m$. Maximising $h_m$ at fixed mean energy —
Jaynes’ maximum entropy principle — is the next lecture.

## Maxwell’s Demon

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwells-demon.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwells-demon.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Maxwell’s demon is a thought experiment described by James Clerk Maxwell
in his book, *Theory of Heat* (Maxwell, 1871) on page 308. He first
described the being in an 1867 letter to Tait; Thomson named it a
“demon” in 1874.

> But if we conceive a being whose faculties are so sharpened that he
> can follow every molecule in its course, such a being, whose
> attributes are still as essentially finite as our own, would be able
> to do what is at present impossible to us. For we have seen that the
> molecules in a vessel full of air at uniform temperature are moving
> with velocities by no means uniform, though the mean velocity of any
> great number of them, arbitrarily selected, is almost exactly uniform.
> Now let us suppose that such a vessel is divided into two portions, A
> and B, by a division in which there is a small hole, and that a being,
> who can see the individual molecules, opens and closes this hole, so
> as to allow only the swifter molecules to pass from A to B, and only
> the slower ones to pass from B to A. He will thus, without expenditure
> of work, raise the temperature of B and lower that of A, in
> contradiction to the second law of thermodynamics.
>
> James Clerk Maxwell in *Theory of Heat* (Maxwell, 1871) page 308

He goes on to say:

> This is only one of the instances in which conclusions which we have
> drawn from our experience of bodies consisting of an immense number of
> molecules may be found not to be applicable to the more delicate
> observations and experiments which we may suppose made by one who can
> perceive and handle the individual molecules which we deal with only
> in large masses

## A Compartment Model of the Demon

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-demon-compartment-model.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-demon-compartment-model.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Maxwell’s sentence is a selective membrane. The useful first cut for
Worksheet 3 is to *compartmentalise* that picture. Treat each bulk —
left of the gate, and right of the gate — as a region that is close to
local equilibrium at its own temperature, provided one stays a mean free
path or more away from the membrane. Between those bulks sits a kinetic
layer where the velocity rule continually removes hot particles one way
and cold particles the other. That layer is not a Maxwellian. Counting
the membrane itself as a fourth compartment is optional: it is the place
where the microscopic gate rule lives. Compartmentalisation here is
coarse-graining — assigning a legal description to each region. The
transition across the gate is the microscopic detailed balance one then
tries to write. The two ledgers that follow ask whether particle flux
and energy flux can both close under those descriptions.

## Detailed Balance Across the Membrane

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-demon-detailed-balance.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-demon-detailed-balance.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

In the simulation the membrane does not move. A ball on the left is
allowed through only when its speed exceeds $v_\star$; a ball on the
right is allowed through only when its speed is at most $v_\star$.
Between crossings, elastic collisions redistribute energy, so a ball
that is cold now can become hot later and then cross. The question is
not whether sorting begins — it does — but where the *net* particle and
energy flows settle. The compartment schematic above is the bookkeeping:
each bulk hopes to be locally Maxwellian; the kinetic layer at the gate
is where that hope fails.

Take equal chamber areas and a 2D ideal gas (the billiard world). On
each side the velocity density is Maxwellian, $$
f(\mathbf{v}\mid\beta)=\frac{\beta}{2\pi}\exp\!\left(-\tfrac{1}{2}\beta v^2\right),\qquad \beta=\frac{1}{k_B T},
$$ with mean kinetic energy $k_B T$ per ball (two quadratic degrees of
freedom). The one-way effusion flux through a unit aperture is
$\Phi_{\mathrm{tot}}=n/\sqrt{2\pi\beta}$. Those formulae assume a small
aperture relative to the mean free path (Knudsen). In the billiard
simulation with large balls that may not hold. Restricting to the hot or
cold set gives Particle-number detailed balance across the membrane is
$$
\Phi^{\mathrm{hot}}(n_L,\beta_L)=\Phi^{\mathrm{cold}}(n_R,\beta_R).
$$ Total ball number $N=N_L+N_R$ and total energy
$E=N_L k_B T_L+N_R k_B T_R$ are conserved. For the simulation parameters
($v_\star=3$, mean kinetic energy $12.5$ in code units, equal
populations $N_L=N_R=N/2$ with $N=39$) this fixes the temperatures
uniquely. The equal-$N$ slice is a cut, not the whole story: a search
over $N_L/N$ (with $T_L<T_R$) finds no Maxwellian pair that balances
both number and energy flux.

```
import numpy as np
from scipy.special import erfc
from scipy.optimize import brentq
import matplotlib.pyplot as plt
import mlai

def flux_total(n, beta):
    return n / np.sqrt(2 * np.pi * beta)

def flux_hot(n, beta, v_star):
    tail = np.sqrt(np.pi / (2 * beta)) * erfc(v_star * np.sqrt(beta / 2.0))
    return (n / np.pi) * (v_star * np.exp(-0.5 * beta * v_star**2) + tail)

def flux_cold(n, beta, v_star):
    return flux_total(n, beta) - flux_hot(n, beta, v_star)

def side_entropy(N, beta, area=1.0):
    # Position (Stirling) + 2D Maxwellian differential entropy; k_B = 1
    return N * (2.0 + np.log(area / N) - np.log(beta / (2 * np.pi)))

N = 39.0
E_tot = N * 12.5
v_star = 3.0
N_L = N_R = N / 2.0
beta0 = N / E_tot
S0 = 2 * side_entropy(N / 2.0, beta0)

def particle_balance(log_beta_L):
    beta_L = np.exp(log_beta_L)
    E_L = N_L / beta_L
    E_R = E_tot - E_L
    if E_R <= 0:
        return 1.0
    beta_R = N_R / E_R
    return flux_hot(N_L, beta_L, v_star) - flux_cold(N_R, beta_R, v_star)

log_beta_L = brentq(particle_balance, -4.0, 1.0)
beta_L = np.exp(log_beta_L)
beta_R = N_R / (E_tot - N_L / beta_L)
T_L, T_R = 1.0 / beta_L, 1.0 / beta_R
S_ss = side_entropy(N_L, beta_L) + side_entropy(N_R, beta_R)
delta_S = S_ss - S0

# Sweep threshold at fixed equal populations / total energy
v_grid = np.linspace(0.5, 6.0, 80)
T_L_grid, T_R_grid, dS_grid = [], [], []
for vs in v_grid:
    def bal(log_bL, vs=vs):
        bL = np.exp(log_bL)
        E_L = N_L / bL
        E_R = E_tot - E_L
        if E_R <= 0:
            return 1.0
        bR = N_R / E_R
        return flux_hot(N_L, bL, vs) - flux_cold(N_R, bR, vs)
    try:
        lb = brentq(bal, -4.5, 1.5)
    except ValueError:
        T_L_grid.append(np.nan); T_R_grid.append(np.nan); dS_grid.append(np.nan)
        continue
    bL = np.exp(lb)
    bR = N_R / (E_tot - N_L / bL)
    T_L_grid.append(1.0 / bL)
    T_R_grid.append(1.0 / bR)
    dS_grid.append(side_entropy(N_L, bL) + side_entropy(N_R, bR) - S0)

fig, axes = plt.subplots(1, 2, figsize=(9.5, 3.8))
axes[0].plot(v_grid, T_L_grid, color='#4488cc', label='$T_L$ (cold side)')
axes[0].plot(v_grid, T_R_grid, color='#cc4444', label='$T_R$ (hot side)')
axes[0].axvline(v_star, color='gray', ls='--', alpha=0.7)
axes[0].axhline(1.0 / beta0, color='gray', ls=':', alpha=0.7)
axes[0].set_xlabel(r'threshold $v_\star$')
axes[0].set_ylabel(r'temperature ($k_B=1$)')
axes[0].set_title('Particle-flux steady state')
axes[0].legend(frameon=False, fontsize=9)
axes[1].plot(v_grid, dS_grid, color='#333333')
axes[1].axvline(v_star, color='gray', ls='--', alpha=0.7)
axes[1].axhline(0.0, color='gray', ls=':', alpha=0.7)
axes[1].scatter([v_star], [delta_S], color='red', zorder=3)
axes[1].set_xlabel(r'threshold $v_\star$')
axes[1].set_ylabel(r'$\Delta S = S_{\mathrm{ss}}-S_0$ (nats)')
axes[1].set_title('Entropy relative to one bath')
fig.tight_layout()
mlai.write_figure('maxwell-membrane-detailed-balance.svg', directory='./physics')
print(f'T_L={T_L:.3f}, T_R={T_R:.3f}, S0={S0:.2f}, S_ss={S_ss:.2f}, dS={delta_S:.2f}')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/physics/maxwell-membrane-detailed-balance.svg" class="" width="85%" style="vertical-align:middle;">

Figure: <i>Left: temperatures on each side of the selective membrane
from particle-flux detailed balance at equal populations (simulation
energy budget). Right: entropy of the two-Maxwellian state minus the
single-temperature equilibrium. The red point is the simulation
threshold $v_\star=3$.</i>

With $k_B=1$, equal populations, $N=39$, $v_\star=3$ and total energy
matching the simulation’s mean kinetic energy $12.5$, particle-flux
detailed balance gives $T_L\approx 1.94$ and $T_R\approx 23.1$. Those
temperatures are the equal-$N$ slice. The ideal-gas Shannon entropy of a
2D Maxwellian chamber (position Stirling term plus velocity differential
entropy) $$
S(N,\beta)=N\left[2+\log\frac{A}{N}-\log\frac{\beta}{2\pi}\right]
$$ falls by about $24$ nats relative to the single-temperature state at
the same $N$ and $E$ — about $-0.63$ nats per ball. That drop is the
thermodynamic content of the sorting you see on the canvas: hot
accumulates on the right, cold on the left, and the joint distribution
over $(\mathrm{side},\mathbf{v})$ is more concentrated than the
unconstrained Maxwellian.

The compartments, kinetic layers, and two-ledger closure are this
course’s construction, not a result in Leff and Rex. Worksheet 3 builds
on them.

A Maxwellian is symmetric in a way the membrane is not. Conditioning on
$v>v_\star$ selects a high-energy set; conditioning on $v\le v_\star$
selects a low-energy set. When the two *number* fluxes match, the
*energy* fluxes generally do not. Equal number of crossings is not equal
energy of crossings: the hot direction carries more energy per ball.
That is the flaw in treating number balance alone as a stationary state.
The true non-equilibrium steady state therefore cannot be a pair of
perfect Maxwellians: the left hot tail is depleted by leakage, the right
cold body is depleted by leakage the other way, and collisions
continually refill those sets. The local-temperature calculation remains
the right first cut — it predicts a cold left, a hot right, and a lower
gas entropy — but the precise $(T_L,T_R)$ pair is an approximation to a
non-Maxwellian steady state.

The canvas can *look* settled for a second reason that is an artefact,
not thermodynamics: a discrete-time billiard with finite resolution can
stall, so fewer balls appear to move even when the continuous
idealisation would keep exchanging. Treat apparent stationarity in the
simulation as a prompt for the two-ledger question, not as proof that
both ledgers have closed.

The selective membrane is Maxwell’s demon written as a dynamical rule.
The detailed-balance calculation shows that the rule really does drive
the gas toward a lower-entropy steady state. Restoring Clausius requires
putting the membrane — or the memory that controls an equivalent trap
door — on the thermodynamic ledger. Szilard (1929) put the cost in
*measurement* (Szilard, 1929). Landauer (1961) priced *erasure* but did
not apply it to the demon (Landauer, 1961). Bennett (1982) made the
erasure resolution and argued that measurement can be reversible — not
one continuous story (Bennett, 1982). Earman and Norton (1998–99) is one
of the dissenting books. The simulation’s “velocity-bin entropy” is a
coarse proxy for the same story: as sorting proceeds, the displayed
histogram entropy typically drops relative to the unsorted gas.

## The Demon Sorts Groups

Maxwell’s sorting rule partitions velocity space into coarse sets —
`hot enough to cross'' versus`cold enough to cross’’ — and applies
different gates to each set. That is exactly this lecture’s grouping: a
deterministic label $G=g(\mathbf{v})$ plus residual uncertainty inside
each label. The membrane calculation assumed each chamber stays close to
a Maxwellian at its own $T$ — local equilibrium, timescale separation
between collisions and crossings. When that fails, the true steady state
is no longer a pair of perfect Maxwellians (tails deplete). Information
engines and demons play on the tension between moving probability among
groups ($H(w)$) and the within-group bath term $\sum w_i H(p_i)$ that
thermodynamics prices.

Compartmentalisation is the coarse-graining move: assign a legal
description to each bulk and to the kinetic layer. The transition at the
gate is then a microscopic detailed-balance question — number flux and
energy flux — not a single fluent story about \`\`the demon.’’ The
compartments, kinetic layers, and two-ledger closure are this course’s
construction, not a result in Leff and Rex. Worksheet 3 builds on them.
I will not mark you wrong for disagreeing with the bookkeeping; I will
mark you wrong for treating it as a textbook theorem.

Kemeny–Snell lumpability is the precise condition under which a
partition of a Markov chain induces another Markov chain on the groups.
When it fails, the macro description needs memory of how the system
entered the group. Name the issue; do not develop projection-operator
formalism here.

Worksheet 3 is the coarse-graining of Maxwell: which description is
legal for each compartment, and where the missing entropy is allowed to
sit. Szilard, Landauer, Ellis, and the information-engine scale argument
wait until that worksheet needs them.
