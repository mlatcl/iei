## The Information Game

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The information game is the following idea. One player thinks of an
object drawn from a known finite list. For example, “I’m thinking of an
animal.” The other may ask only yes/no questions. For example “Does your
animal have four legs?” If we assume that the choice is uniform across
that list, and if we can ask optimal questions, what number of questions
do you need to guarantee the answer? How does information theory help us
with this?

```
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
import mlai.plot as plot
import mlai
```

```
def draw_yesno_tree(ax, node, x, y, dx, dy):
    """Draw a yes/no question tree. Leaves are named objects; internal nodes are questions.
    Each child dict may set leaf=True. Left is no, right is yes.
    """
    is_leaf = bool(node.get('leaf', False))
    label = node['label']
    width = max(0.16, 0.012 * len(label) + 0.08)
    height = 0.09
    facecolor = '#F4E2C9' if is_leaf else '#D6EAF8'
    box = FancyBboxPatch((x - width / 2, y - height / 2), width, height,
                         boxstyle='round,pad=0.012,rounding_size=0.02',
                         facecolor=facecolor, edgecolor='black', linewidth=1.2)
    ax.add_patch(box)
    ax.text(x, y, label, ha='center', va='center', fontsize=11)
    if is_leaf:
        return
    y_child = y - dy
    for key, x_child, name, ha, dx_text in (
            ('no', x - dx, 'no', 'right', -0.02),
            ('yes', x + dx, 'yes', 'left', 0.02)):
        child = node.get(key)
        if child is None:
            continue
        ax.plot([x, x_child], [y - height / 2, y_child + height / 2],
                color='black', lw=1.2, zorder=0)
        ax.text((x + x_child) / 2 + dx_text,
                (y + y_child) / 2,
                name, ha=ha, va='bottom', fontsize=9, color='#1B4F72')
        draw_yesno_tree(ax, child, x_child, y_child, dx / 2, dy)

animal_tree = {
    'label': 'Four legs?',
    'no': {
        'label': 'Wings?',
        'no': {'label': 'Snake', 'leaf': True},
        'yes': {'label': 'Owl', 'leaf': True},
    },
    'yes': {
        'label': 'Storybook?',
        'no': {'label': 'Fox', 'leaf': True},
        'yes': {'label': 'Gruffalo', 'leaf': True},
    },
}

fig, ax = plt.subplots(figsize=plot.big_wide_figsize)
draw_yesno_tree(ax, animal_tree, x=0.5, y=0.88, dx=0.22, dy=0.36)
ax.set_xlim(0.0, 1.0)
ax.set_ylim(0.02, 1.0)
ax.axis('off')
mlai.write_figure('information-game-animal-tree.svg',
                  directory='./information')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/information-game-animal-tree.svg" class="" width="70%" style="vertical-align:middle;">

Figure: <i>A balanced yes/no tree for four animals. Each question halves
the remaining list, so two answers name one of $2^2=4$ hypotheses.</i>

With a list of size $N$ and equal prior probability on each item, an
ideal strategy halves the search space at every step. After $k$ answers
you can distinguish at most $2^k$ hypotheses, so you need $$
k \ge \log_2 N
$$ questions in the worst case. Sometimes the game is called “twenty
questions.” Twenty questions would reach a million possibilities:
$2^{10}=1024\approx 10^3$, so $2^{20}=2^{10}\times 2^{10}\approx 10^6$.
Exactly $2^{20}=1\,048\,576$. A problem that drives an equiprobable
yes/no answer delivers to you one *bit* of information. That is the most
information you can gain at ech turn. This gives us another
interpretation of Shannon’s bit, it is the amount of information you
gain if you manage to answer a balanced yes/no or true/false question.

If the prior is not uniform, the best question is no longer a literal
half-split of the *count* of remaining objects. It is a half-split of
their *probability*. The average number of questions under optimal play
becomes the Shannon entropy $H = -\sum_i p_i\log_2 p_i$.

## MacKay’s Balance and Balls

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

David MacKay opens the compression half of *Information Theory,
Inference, and Learning Algorithms* with a weighing puzzle (MacKay,
2003, p. Exercise\~4.1).

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/mackay-balance-balls.svg" class="" width="80%" style="vertical-align:middle;">

Figure: <i>A two-pan balance and twelve identical balls.</i>

You are given twelve balls, identical in appearance. Exactly one differs
in weight from the others; you do not know whether that odd ball is
heavier or lighter. You are given a two-pan balance. Design a strategy
that finds the odd ball and the direction of the defect.

How many outcomes does that give us? How much information can we get per
measurement? And how many measurements do we (therefore) require?

Under a uniform prior over those twenty-four possibilities, the Shannon
uncertainty we must remove is $$
H_{\text{start}} = \log_2 24 \approx 4.58 \text{ bits}.
$$ At the end we would like a single hypothesis: $H_{\mathrm{end}} = 0$.
This means that the information *deficit* is around $4.58$ bits. We need
to design a set of experiments that will supply this information.

## Try Two Outcomes

A two-outcome instrument, used optimally, returns at most $\log_2 2 = 1$
bit per measurement. The information budget then says $$
2^4 = 16 < 24 \le 32 = 2^5,
$$ so five weighings would work, but four cannot. Whether we can
organise the physical balance to achieve that one bit per measurement is
a separate question and associated with the *mechanism*.

$$
\log_2 24 \approx 4.58,\qquad 2^4 = 16 < 24 \le 32 = 2^5
$$

Write two columns if you can: left, the information budget; right, the
pans. The left column does not know about equal group sizes, leftover
balls, or borrowing a ball of known mass. Those constraints live on the
right. The rest of this block is the right-hand column, scored
afterwards by the left.

The move 6 vs 6 is a binary-search instinct that comes to us from the
information game. If we already knew the odd ball was heavier, it would
be a perfect split of twelve balls. We do not know that. A down-left
reading means "one of the left six is heavy, *or* one of the right six
is light’’: twelve of the original twenty-four hypotheses remain. One
bit toward the 24, and *no* ball has been eliminated.

## Three Readings

A two-pan balance has a ternary alphabet, the level reading can also
give us information. This is an instrument that can give us $\log_2 3$
bits of information per measurement. By starting with measuring 6 vs 6
we set the probability of a level answer to zero: with exactly one odd
ball among the twelve, the pans cannot balance.

$$
\log_2 3 \approx 1.58,\qquad 3^2 = 9 < 24 \le 27 = 3^3
$$

A binary question yields at most one bit. If the three readings are
equiprobable, one weighing contributes $$
\log_2 3 \approx 1.585 \text{ bits}.
$$ After $k$ weighings the strategy tree has at most $3^k$ leaves, so it
can name at most $3^k$ hypotheses. Compare budgets: $$
\begin{align*}
3^2 &= 9 < 24,\\
3^3 &= 27 \ge 24.
\end{align*}
$$ Two weighings cannot suffice no matter how clever the groupings.
Three weighings *can*: there is spare capacity of three unused leaves
($27-24=3$). In bits, $$
3\log_2 3 \approx 4.75 > \log_2 24 \approx 4.58.
$$ The gap is small. That is why the puzzle feels tight. It is the same
comparison we already made with $\log_2 2$, with a larger per-use
budget.

## Coarse-Graining a Die

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/die-coarse-graining.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/die-coarse-graining.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Take a fair six-sided die, $X\in\{1,\ldots,6\}$ with $p(x)=1/6$, so In
the information game we may ask any yes/no we can name. The pans already
showed that a physical instrument will not always realise the split the
budget wants. The cube is the same lesson on a smaller object. Its
sample space has six atoms, not $2^k$. A grouping is a partition the
*mechanism* affords — looking at the face through some physical cut. The
lead cut combines $\{1,2,3\}$ against $\{4,5,6\}$. Odd versus even is
not the next question inside that grouping: it is a different cut of the
original six. After a $3$–$3$ split, three faces remain. Three cannot be
split by a fair yes/no, just as three remaining balls could not sit
evenly on the pans. Nested 1-bit questions would need a power of two —
an eight-sided die, which would still live in the game’s world. This
cube does not.

$$
H(G_{\text{low}})=1 \text{ bit},
\qquad
H(X\mid G_{\text{low}})=\log_2 3 \approx 1.585 \text{ bits}
$$

Let $G_{\mathrm{low}}=0$ on $\{1,2,3\}$ and $1$ on $\{4,5,6\}$. Then
$w=(1/2,1/2)$, so $H(G_{\mathrm{low}})=1$. Conditional on the group,
three faces remain, each with probability $1/3$, so
$H(X\mid G_{\mathrm{low}})=\log_2 3$. The stages add: The first answer
names the group. The residual $\log_2 3$ is the information the
mechanism threw away by combining three faces into one physical cut.
That residual is *not* a second 1-bit question the cube still owes us.

$$
h_2\Bigl(\tfrac13\Bigr)\approx 0.92 < 1 \text{ bit}
$$

Suppose we have learned $G_{\mathrm{low}}=0$, so the face is in
$\{1,2,3\}$. Asking odd or even now splits $\{1,3\}$ against $\{2\}$.
The answers are not equiprobable: $w=(2/3,1/3)$, so the question is
worth $h_2(1/3)\approx 0.92$ bits, not one bit. That is the same tax as
three remaining balls padded with a known normal. Successive balanced
1-bit questions form a binary tree of equal splits. That tree has $2^k$
leaves. A six-sided die cannot sit on it, because the *instrument* has
six faces. An eight-sided die can: first $\{1,2,3,4\}$ versus
$\{5,6,7,8\}$, then a $2$–$2$ split inside the named half, then a
$1$–$1$ split — three fair coins, $H=\log_2 8=3$ bits exactly. That
object would still live in the information game. This cube does not. So
we do not invent a nested question the mechanism cannot ask. We go back
to all six faces and name a *different* grouping the cube *does* afford.

$$
H(G_{\mathrm{parity}})=1 \text{ bit},
\qquad
H(X\mid G_{\mathrm{parity}})=\log_2 3 \approx 1.585 \text{ bits}
$$

Let $G_{\mathrm{parity}}=0$ on even faces and $1$ on odd, as a partition
of the original six, not of a group we already named. Then
$w=(1/2,1/2)$, so $H(G_{\mathrm{parity}})=1$, and
$H(X\mid G_{\mathrm{parity}})=\log_2 3$ as before. The stages add to
$\log_2 6$ again. Low/high and odd/even are two equally good 1-bit
groupings the *cube* affords. They keep different distinctions. They do
not nest, because the instrument is not a free questioner. Asking both
would give four cells $\{1,3\}$, $\{2\}$, $\{5\}$, $\{4,6\}$ — not four
equiprobable atoms, and not two bits. Which grouping you choose is which
physical cut you look through. The next snippet will name that cut as an
abstraction. The point today is that mechanism, not the information
budget, chose the menu.

$$
w=\Bigl(\tfrac16,\tfrac56\Bigr),
\qquad
H(G_{\mathrm{six}})=h_2\Bigl(\tfrac16\Bigr)\approx 0.65 \text{ bits}
$$

Let $G_{\mathrm{six}}=\mathbf{1}_{\{X=6\}}$. The group weights are
unbalanced, $w=(1/6,5/6)$, so the binary entropy of the first answer is
one bit. After a `yes'' the face is known ($H=0$). After a`no’’ five
faces remain ($H=\log_2 5$). Averaging, Again
$H(G_{\mathrm{six}})+H(X\mid G_{\mathrm{six}})=H(X)$. The low/high cut
*is* one bit because the two groups are equiprobable. The six-or-not cut
is not.

## Grouping Composition

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-grouping-composition.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-grouping-composition.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Often we do not need to know the exact outcome. If all I care about is
whether the die shows five or above, I can stop once that is settled. If
I need a fair 50/50 decision and only have a die, I can call $\{1,2,3\}$
“low” and $\{4,5,6\}$ “high”; the die, coarse-grained, *is* a fair coin.
In both cases we have replaced the fine variable $X$ (the face) with a
coarse variable $G=g(X)$ (the group), and $G$ has its own entropy:
$H(G)=\log_2 2 = 1$ bit for the coin, and
$H(G)=H(\tfrac{2}{3},\tfrac{1}{3})\approx 0.918$ bits for “five or
above.” That number is the uncertainty in the coarse-grained state, and
it is all we pay if we stop there.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
import mlai.plot as plot
```

```
def entropy(p):
    p = np.asarray(p, dtype=float)
    p = p[p > 0]
    return -np.sum(p * np.log2(p))

def draw_die_tree(ax, groups, names, title):
    faces = [f for g in groups for f in g]
    xface = np.linspace(0.08, 0.92, len(faces))
    xpos = dict(zip(faces, xface))
    w = np.array([len(g) for g in groups]) / 6.0
    yroot, ygroup, yface = 0.92, 0.55, 0.15
    ax.plot(0.5, yroot, 'o', color='k', markersize=10)
    ax.text(0.5, yroot + 0.05, 'die', ha='center', va='bottom', fontsize=12)
    cols = ['tab:blue', 'tab:orange', 'tab:green']
    for i, g in enumerate(groups):
        xg = np.mean([xpos[f] for f in g])
        ax.plot([0.5, xg], [yroot, ygroup], '-', color=cols[i], lw=1 + 6 * w[i])
        ax.text((0.5 + xg) / 2 + (0.03 if xg > 0.5 else -0.03), (yroot + ygroup) / 2,
                'w=%d/6' % len(g), ha='left' if xg > 0.5 else 'right',
                fontsize=11, color=cols[i])
        ax.plot(xg, ygroup, 's', color=cols[i], markersize=12)
        ax.text(xg, ygroup + 0.04, names[i], ha='center', va='bottom', fontsize=11)
        for f in g:
            ax.plot([xg, xpos[f]], [ygroup, yface], '-', color=cols[i], lw=1.2)
            ax.plot(xpos[f], yface, 'o', color=cols[i], markersize=8)
            ax.text(xpos[f], yface - 0.06, str(f), ha='center', va='top', fontsize=12)
    Hw = entropy(w)
    Hres = np.sum([wi * np.log2(len(g)) for wi, g in zip(w, groups)])
    ax.text(1.02, (yroot + ygroup) / 2, 'H(G) = %.3f' % Hw, fontsize=11, va='center')
    ax.text(1.02, (ygroup + yface) / 2, 'H(X|G) = %.3f' % Hres, fontsize=11, va='center')
    ax.text(1.02, 0.0, 'total = %.3f bits' % (Hw + Hres), fontsize=11, va='center',
            fontweight='bold')
    ax.set_title(title, fontsize=13)
    ax.set_xlim(-0.05, 1.3)
    ax.set_ylim(-0.12, 1.05)
    ax.axis('off')

fig, axs = plt.subplots(1, 2, figsize=plot.big_wide_figsize)
draw_die_tree(axs[0], [[1, 2, 3, 4], [5, 6]], ['below 5', '5 or above'],
              'Is it five or above?')
draw_die_tree(axs[1], [[1, 2, 3], [4, 5, 6]], ['low', 'high'],
              'Low or high? (a fair coin)')
mlai.write_figure(filename='die-coarse-graining-trees.svg',
                  directory='./information')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/die-coarse-graining-trees.svg" class="" width="90%" style="vertical-align:middle;">

Figure: <i>Two coarse-grainings of a fair die. The top level is the
coarse question, with branch widths set by the group probabilities
$w_i$; the bottom level is the residual detail. Stopping at the squares
costs $H(G)$. Going all the way down costs $H(G)+H(X\mid G)$, which is
$\log_2 6$ for either cut.</i>

$$
H(X)=\underbrace{H(w)}_{\text{coarse state}}+\underbrace{\sum_i w_i\, H(p_i)}_{\text{residual detail}}
$$

Partition the outcomes of $X$ into groups labelled by $i$, with group
probabilities $w_i$ and within-group distributions $p_i$. Then The first
term is the entropy of the coarse-grained state; the second is the
average uncertainty left once the group is known. For “five or above,”
$H(w)=H(\tfrac{2}{3},\tfrac{1}{3})\approx 0.918$ and the residual is
$\tfrac{2}{3}\log_2 4+\tfrac{1}{3}\log_2 2\approx 1.667$; for the coin,
$1+\log_2 3$. Both sum to $\log_2 6$. Different cuts split the bill
differently, but the bill is the same.

If $G=g(X)$ is the group label, the identity is exactly the chain rule
because $G$ is a function of $X$ (so $H(G\mid X)=0$) and
$H(X\mid G)=\sum_i w_i H(p_i)$.

```
stages = [
    ['five or above?', [entropy([2 / 3., 1 / 3.]), (2 / 3.) * 2 + (1 / 3.) * 1],
     ['H(G)', 'H(X|G)']],
    ['low or high?', [1.0, np.log2(3)], ['H(G)', 'H(X|G)']],
    ['five+, pair, face',
     [entropy([2 / 3., 1 / 3.]), (2 / 3.) * 1, 1.0],
     ['H(G)', "H(G'|G)", "H(X|G')"]],
]
cols = ['tab:blue', 'tab:orange', 'tab:green']
fig, ax = plt.subplots(figsize=plot.big_wide_figsize)
for j, (name, parts, labels) in enumerate(stages):
    left = 0.0
    for k, (val, lab) in enumerate(zip(parts, labels)):
        ax.barh(j, val, left=left, color=cols[k], edgecolor='k', height=0.6)
        ax.text(left + val / 2, j, '%s\n%.3f' % (lab, val), ha='center', va='center',
                fontsize=9, color='black', fontweight='bold')
        left += val
ax.axvline(np.log2(6), color='k', ls='--')
ax.text(np.log2(6), -0.5, r'$\log_2 6 = %.3f$ bits' % np.log2(6),
        ha='center', va='bottom', fontsize=11)
ax.set_yticks(range(len(stages)))
ax.set_yticklabels([s[0] for s in stages], fontsize=11)
ax.invert_yaxis()
ax.set_xlabel('bits', fontsize=11)
ax.set_xlim(0, 3.0)
for s in ['top', 'right']:
    ax.spines[s].set_visible(False)
mlai.write_figure(filename='die-coarse-graining-budget.svg',
                  directory='./information')
```

Coarse-grainings can themselves be coarse-grained. Refine “five or
above” ($G$) into pairs $G'\in\{\{1,2\},\{3,4\},\{5,6\}\}$ and then into
faces. Each stage is a grouping of the next, so the chain rule applies
at every level: Numerically $0.918+0.667+1=\log_2 6$. The middle term is
$\tfrac{2}{3}$ because “below five” still has two pairs to choose
between (1 bit) while “five or above” is already a single pair (0 bits).
The practical reading: the entropy of whatever coarse-grained state you
stop at is the running total up to that point, $H(G')=H(G)+H(G'\mid G)$,
and refining further simply adds the next term. Nothing is
double-counted and nothing is lost.

In the reverse direction, observing $X$ fixes $G$ with no leftover
uncertainty. The mutual information $I(X;G)=H(G)-H(G\mid X)=H(G)$ is
exactly the price of the coarse question. After you know the group, the
remaining work is $H(X\mid G)$ — which of the fine outcomes inside that
group. Coarse-graining *keeps* $H(G)$ as your description and *discards*
$H(X\mid G)$: Using the die as a coin keeps one bit and throws away
$\log_2 3$.

Every bar in the budget figure ends at $\log_2 6$, whichever cut came
first and however many stages we used. That invariance under regrouping
is not an accident of base-two logarithms; it is part of what
characterises Shannon entropy among candidate uncertainty measures
(Khinchin, 1957). Categorical accounts make the same demand precise
(Baez et al., 2011). For this lecture it is enough that staged questions
*add*, so a coarse-grained state has a well-defined entropy and
coarse-graining has a well-defined price.

## Grouping Is a Choice of Abstraction

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/grouping-as-macrostate.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/grouping-as-macrostate.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The idealised information game fails when we turn to the mechanism. But
we do have some control. We can coarse-grain, and reduce the search
space. For example, the low/high grouping is a choice of abstraction. We
keep a two-valued group label and throw away which of the three faces
occurred. Coarse graining in thermodynamics can be seen similarly. We
cannot track Laplace’s curves for every molecule, instead we keep
macroscopic numbers. Those macroscopic numbers split into *extensive*
quantities that scale with the system (e.g. $N$, $V$, energy, entropy)
and *intensive* quantities that do not (e.g. temperature, pressure).
What we’ve seen is that these entropies can compose, so that the
coarse-graining is coming from the form of entropy (as a function). This
means that accounting for information works through the groupings, if
the macrostate is $H(w)$; the unresolved motion inside it is
$\sum_i w_i H(p_i)$.

## Gaussian, Maxwell–Boltzmann, and the Missing Measure

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/gaussian-maxwell-missing-measure.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/gaussian-maxwell-missing-measure.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

```
import numpy as np

rng = np.random.default_rng(0)
n_velocities = 10000
velocity_sigma = 1.0
n_histogram_bins = 40
velocities = rng.normal(0.0, velocity_sigma, size=(n_velocities, 3))
speed = np.linalg.norm(velocities, axis=1)
velocity_x = velocities[:, 0]


def discrete_entropy(counts):
    n = counts.sum()
    p = counts[counts > 0] / float(n)
    return -np.sum(p * np.log(p))


def gaussian_differential_entropy(sigma):
    return 0.5 * np.log(2.0 * np.pi * np.e * sigma**2)


def maxwell_speed_pdf(s, sigma):
    return (
        np.sqrt(2.0 / np.pi)
        * (s**2 / sigma**3)
        * np.exp(-s**2 / (2.0 * sigma**2))
    )


def maxwell_speed_differential_entropy(sigma):
    return np.log(sigma * np.sqrt(2.0 * np.pi)) + np.euler_gamma - 0.5


def speed_range_max(s, p_hi=0.995):
    return float(np.quantile(s, p_hi))


def equal_delta_speed_edges(s_max, n_bins):
    return np.linspace(0.0, s_max, n_bins + 1)


def equal_volume_speed_edges(s_max, n_bins):
    u_edges = np.linspace(0.0, s_max**3, n_bins + 1)
    return u_edges ** (1.0 / 3.0)
```

## Histograms as Riemann Sums

Approximating a density by its value at the centre of each bin is a
piecewise-constant (midpoint-rule) approximation to the integral. We
call the corresponding sums Riemann sums. Writing
$p_i\approx f(x_i)\Delta$ inside the entropy gives with both sums
converging to integrals as $\Delta\to 0$.

If we sample isotropic three-dimensional velocities and histogram one
Cartesian component with equal-width bins. The discrete Shannon entropy
of the occupancy is Note that empty bins don’t contribute to the
entropy. For a smooth density and a fine equal-width partition,
$p_i\approx f(x_i)\Delta$ and The figure below reports both numbers.
Note that they agree up to sampling and tail truncation. From this we
can see that the differential entropy, $h$, is the $\Delta$-independent

```
vx_lim = 4.0
edges = np.linspace(-vx_lim, vx_lim, n_histogram_bins + 1)
delta = edges[1] - edges[0]
counts, _ = np.histogram(velocity_x, bins=edges)
H = discrete_entropy(counts)
h = gaussian_differential_entropy(velocity_sigma)
xs = np.linspace(-vx_lim, vx_lim, 400)
fig, ax = plt.subplots(figsize=plot.big_wide_figsize)
ax.hist(velocity_x, bins=edges, density=True, color='#4C72B0', alpha=0.6,
        label='equal-$\\Delta v$ histogram')
ax.plot(xs, np.exp(-0.5 * xs**2 / velocity_sigma**2)
        / np.sqrt(2.0 * np.pi * velocity_sigma**2),
        color='#C44E52', lw=2, label='$\\mathcal{N}(0,\\sigma^2)$')
ax.set_xlim(-vx_lim, vx_lim)
ax.set_xlabel('$v_x$')
ax.set_ylabel('density')
ax.text(0.02, 0.95,
        f'$H={H:.3f}$\n$h-\\log\\Delta={h - np.log(delta):.3f}$\n$h={h:.3f}$',
        transform=ax.transAxes, va='top', fontsize=12,
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.9))
ax.legend(loc='upper right', fontsize=9)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
mlai.write_figure('gaussian-histogram-discrete-entropy.svg',
                  directory='./physics')
```

## The Continuous Limit Drops $\log\Delta$

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/log-delta-divergence.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/log-delta-divergence.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Refine the equal-width partition of a Gaussian component. For each bin
width $\Delta$, Plotting $H$ against $-\log\Delta$ gives a line of slope
one, offset by the constant $h=\tfrac12\log(2\pi e\sigma^2)$. The
discrete entropy diverges as the bins shrink because each mass
$p_i\approx f\Delta$ contributes an extra $-\log\Delta$. Differential
entropy is defined to be the $\Delta$-independent remainder. The
continuous functional therefore *ignores a term that is driving to
$+\infty$*. That is why $h$ can be negative, and why it is not a
code-length bound.

```
import matplotlib.pyplot as plt
import mlai.plot as plot
import mlai
```

```
vx_lim = 4.0
n_list = np.array([5, 8, 10, 16, 20, 25, 32, 40, 50, 64, 80, 100])
Hs = []
deltas = []
for n_bins in n_list:
    edges = np.linspace(-vx_lim, vx_lim, n_bins + 1)
    delta = edges[1] - edges[0]
    counts, _ = np.histogram(velocity_x, bins=edges)
    Hs.append(discrete_entropy(counts))
    deltas.append(delta)
Hs = np.array(Hs)
deltas = np.array(deltas)
h = gaussian_differential_entropy(velocity_sigma)
minus_log_delta = -np.log(deltas)

fig, ax = plt.subplots(figsize=plot.big_wide_figsize)
ax.plot(minus_log_delta, Hs, 'o', color='#4C72B0', ms=8, label='$H$')
xline = np.linspace(minus_log_delta.min() - 0.1,
                    minus_log_delta.max() + 0.1, 50)
ax.plot(xline, h + xline, color='#C44E52', lw=2, label='$h-\\log\\Delta$')
ax.axhline(h, color='#55A868', ls='--', lw=2, label='$h$')
ax.set_xlabel('$-\\log\\Delta$')
ax.set_ylabel('nats')
ax.legend(fontsize=10)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
mlai.write_figure('log-delta-divergence.svg',
                  directory='./information')
```

## Bin Widths Are a Radon-Nikodym Derivative

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/discrete-jacobian-radon-nikodym.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/discrete-jacobian-radon-nikodym.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

For a partition that is not equal-width, the bin-width correction is no
longer a single $\log\Delta$. Probability conservation in cell $i$ still
reads $p_i \approx f\,\Delta_i$, so Take the equal-volume cells of the
previous snippets, written once in $u=s^3$ and once in $s$. The masses
$\{p_i\}$ (hence $H$) are the same. The two differential entropies
differ by the average log-ratio of widths, That ratio
$\Delta s_i/\Delta u_i$ is called the discrete Radon-Nikodym derivative
of Lebesgue-on-speed with respect to Lebesgue-on-volume. In the
continuum limit it becomes the Jacobian $1/(3s^2)$. Probability
conservation $f_s(s)\,{\mathrm{d}}s = f_u(u)\,{\mathrm{d}}u$ is the same
statement.

The Maxwell speed density relative to three-dimensional Lebesgue is the
shell factor $4\pi s^2$: The Gaussian $h(v_x)$ and the Maxwell $h(s)$
disagree because they are different coordinates *and* because each $h$
is taken against a different *undeclared* Lebesgue measure.

```
import matplotlib.pyplot as plt
import mlai.plot as plot
import mlai
```

## Jaynes: Promote the Bins to a Measure

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/jaynes-limiting-density.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/jaynes-limiting-density.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

$$
h_m(f)=-\int f\log\frac{f}{m}\,{\mathrm{d}}x=-\text{KL}\left( f\,\|\,m \right)
$$

Differential entropy $h=-\int f\log f\,{\mathrm{d}}x$ pretends the
coordinate measure is part of the data. The histogram arithmetic showed
that it is not: $H$ is a functional of masses, and the passage to a
density subtracts $\sum_i p_i\log\Delta_i$, whose continuum limit is the
log of a reference density $m$. Jaynes called $m$ the *limiting density
of discrete points* — the binning measure, promoted to a density
(Jaynes, 1963).

Once $m$ is written down, Taking $m$ uniform (Lebesgue) in the Cartesian
component $v_x$ recovers the ordinary Gaussian remainder $h(v_x)$.
Taking $m$ uniform in three-dimensional velocity, and writing it in the
speed coordinate, gives $m(s)\propto s^2$ — the shell volume that had
been hiding in the Jacobian. The two Maxwell–Boltzmann entropies
disagreed because each $h$ used a different undeclared $m$. Relative
entropy makes the choice explicit. The Jacobian cancels inside
$\text{KL}\left( f\,\|\,m \right)$ under a common invertible map; the
discrete KL has no divergent $-\log\Delta$ term.

This names $m$. Maximising $h_m$ at fixed mean energy — Jaynes’ maximum
entropy principle — is a later step.

## Compartments Around the Membrane

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/demon-spatial-compartments.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/demon-spatial-compartments.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The canvas already implements Maxwell’s sorting as a dynamical rule. A
ball on the left crosses only when its speed exceeds $v_\star$; a ball
on the right crosses only when its speed is at most $v_\star$. Hot
therefore accumulates on the right and cold on the left. That is the JS
convention, and we keep it.

The new move is to refuse a single temperature on each side. Insert two
extra partitions, distant from the membrane. Between the left wall and
the left partition sits a *bulk*: many collisions, few crossings.
Between the right partition and the right wall sits another bulk.
Against the membrane, on each side, sits a kinetic layer of thickness a
mean free path or so. In the bulks we may argue for local equilibrium —
the same timescale split as the grouping identity, $p_i$ relaxing
quickly to $\pi_i$. In the layers we may not. What hits the gate is a
selected tail or body; what just crossed has not remixed. The
neighbourhood of the membrane is not at equilibrium.

Each distant bulk is a group in the sense we already have. We keep a few
intensive and extensive numbers and refuse the molecular faces. Local
equilibrium is the claim that collisions inside the group are fast
compared with exchange across its boundary, so the within-group law has
settled to $\pi_i$. That claim is plausible far from the membrane. It is
not plausible in the kinetic layer. The gate is a non-reciprocal filter:
the left-going stream and the right-going stream are different velocity
sets. Arrivals are conditioned on that filter, and a ball that has just
crossed has not yet collided enough times to forget how it entered.
Extending a Maxwellian up to the membrane is the same mistake as
treating $\{1,2,3\}$ as already mixed when the cube has only just shown
the group label.

## Two Ledgers at the Membrane

In the simulation the membrane does not move. A ball on the left is
allowed through only when its speed exceeds $v_\star$; a ball on the
right is allowed through only when its speed is at most $v_\star$.
Between crossings, elastic collisions in each bulk redistribute energy,
so a ball that is cold now can become hot later and then cross. The
question is not whether sorting begins — it does — but which currents
through the gate can vanish together.

For the closed vessel to sit in a non-equilibrium *steady* state, every
conserved current through the membrane must vanish. Particle number is
conserved, so the number of balls crossing left to right must equal the
number crossing right to left, otherwise the bulk densities keep
changing. Energy is conserved, so the energy those streams carry must
also balance, otherwise one bulk keeps heating and the other cooling.

That pair of flux balances is not detailed balance. Detailed balance
would require each velocity class that crosses to be matched by its
reverse. The gate refuses: a hot left-to-right crossing has no matching
hot right-to-left crossing, because hot balls on the right are
reflected. The allowed streams are a high-energy set one way and a
low-energy set the other. Equal *counts* are then not equal *energies*.

```
import numpy as np
from scipy.special import erfc
from scipy.optimize import brentq
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import mlai
import mlai.plot as plot
```

Figure: <i>Two ledgers at the same gate. For a closed box, a steady bulk
*number* requires equal counts both ways. The hot stream carries more
energy per ball than the cold stream, so those equal counts are not an
energy balance. A steady bulk *energy* would need the lower pair of

Suppose, illegally, that each kinetic layer is the same Maxwellian as
its distant bulk, at temperatures $T_L$ and $T_R$. Conditioning on
$v>v_\star$ selects a high-energy set; conditioning on $v\le v_\star$
selects a low-energy set. Matching the two *number* fluxes then leaves a
net *energy* flux into the right-hand side. Matching the energy fluxes
instead leaves a net number flux. The only Maxwellian pair that closes
both ledgers is $T_L=T_R$ with no sort — and at equal temperatures the
filter itself drives a current, so that pair is not stationary either.

A closed box can therefore sit still only if the velocity distributions
*at the gate* are not the bulk Maxwellians. The extra partitions were
not decoration: they mark the contour where the local-equilibrium coarse
graining stops being legal. Collisions rebuild the tails in the bulks;
the membrane continually sculpts the layers.

Take equal chamber areas and a 2D ideal gas (the billiard world). On
each side the velocity density is Maxwellian, $$
f(\mathbf{v}\mid\beta)=\frac{\beta}{2\pi}\exp\!\left(-\tfrac{1}{2}\beta v^2\right),\qquad \beta=\frac{1}{k_B T},
$$ with mean kinetic energy $k_B T$ per ball (two quadratic degrees of
freedom). The one-way effusion flux through a unit aperture is
$\Phi_{\mathrm{tot}}=n/\sqrt{2\pi\beta}$. Restricting to the hot or cold
set gives Particle-number balance across the membrane is $$
\Phi^{\mathrm{hot}}(n_L,\beta_L)=\Phi^{\mathrm{cold}}(n_R,\beta_R).
$$ Total ball number $N=N_L+N_R$ and total energy
$E=N_L k_B T_L+N_R k_B T_R$ are conserved. For the simulation parameters
($v_\star=3$, mean kinetic energy $12.5$ in code units, equal
populations $N_L=N_R$) this fixes the temperatures uniquely *under the
Maxwellian-at-the-gate assumption*. It does not close the energy ledger.

```
def flux_total(n, beta):
    return n / np.sqrt(2 * np.pi * beta)

def flux_hot(n, beta, v_star):
    tail = np.sqrt(np.pi / (2 * beta)) * erfc(v_star * np.sqrt(beta / 2.0))
    return (n / np.pi) * (v_star * np.exp(-0.5 * beta * v_star**2) + tail)

def flux_cold(n, beta, v_star):
    return flux_total(n, beta) - flux_hot(n, beta, v_star)

def side_entropy(N, beta, area=1.0):
    # Position (Stirling) + 2D Maxwellian differential entropy; k_B = 1
    return N * (2.0 + np.log(area / N) - np.log(beta / (2 * np.pi)))

N = 39.0
E_tot = N * 12.5
v_star = 3.0
N_L = N_R = N / 2.0
beta0 = N / E_tot
S0 = 2 * side_entropy(N / 2.0, beta0)

def particle_balance(log_beta_L):
    beta_L = np.exp(log_beta_L)
    E_L = N_L / beta_L
    E_R = E_tot - E_L
    if E_R <= 0:
        return 1.0
    beta_R = N_R / E_R
    return flux_hot(N_L, beta_L, v_star) - flux_cold(N_R, beta_R, v_star)

log_beta_L = brentq(particle_balance, -4.0, 1.0)
beta_L = np.exp(log_beta_L)
beta_R = N_R / (E_tot - N_L / beta_L)
T_L, T_R = 1.0 / beta_L, 1.0 / beta_R
S_ss = side_entropy(N_L, beta_L) + side_entropy(N_R, beta_R)
delta_S = S_ss - S0

# Sweep threshold at fixed equal populations / total energy
v_grid = np.linspace(0.5, 6.0, 80)
T_L_grid, T_R_grid, dS_grid = [], [], []
for vs in v_grid:
    def bal(log_bL, vs=vs):
        bL = np.exp(log_bL)
        E_L = N_L / bL
        E_R = E_tot - E_L
        if E_R <= 0:
            return 1.0
        bR = N_R / E_R
        return flux_hot(N_L, bL, vs) - flux_cold(N_R, bR, vs)
    try:
        lb = brentq(bal, -4.5, 1.5)
    except ValueError:
        T_L_grid.append(np.nan); T_R_grid.append(np.nan); dS_grid.append(np.nan)
        continue
    bL = np.exp(lb)
    bR = N_R / (E_tot - N_L / bL)
    T_L_grid.append(1.0 / bL)
    T_R_grid.append(1.0 / bR)
    dS_grid.append(side_entropy(N_L, bL) + side_entropy(N_R, bR) - S0)

fig, axes = plt.subplots(1, 2, figsize=(9.5, 3.8))
axes[0].plot(v_grid, T_L_grid, color='#4488cc', label='$T_L$ (cold side)')
axes[0].plot(v_grid, T_R_grid, color='#cc4444', label='$T_R$ (hot side)')
axes[0].axvline(v_star, color='gray', ls='--', alpha=0.7)
axes[0].axhline(1.0 / beta0, color='gray', ls=':', alpha=0.7)
axes[0].set_xlabel(r'threshold $v_\star$')
axes[0].set_ylabel(r'temperature ($k_B=1$)')
axes[0].set_title('Number-flux balance only')
axes[0].legend(frameon=False, fontsize=9)
axes[1].plot(v_grid, dS_grid, color='#333333')
axes[1].axvline(v_star, color='gray', ls='--', alpha=0.7)
axes[1].axhline(0.0, color='gray', ls=':', alpha=0.7)
axes[1].scatter([v_star], [delta_S], color='red', zorder=3)
axes[1].set_xlabel(r'threshold $v_\star$')
axes[1].set_ylabel(r'$\Delta S = S_{\mathrm{ss}}-S_0$ (nats)')
axes[1].set_title('Entropy relative to one bath')
fig.tight_layout()
mlai.write_figure('maxwell-membrane-detailed-balance.svg', directory=r'./physics')
print(f'T_L={T_L:.3f}, T_R={T_R:.3f}, S0={S0:.2f}, S_ss={S_ss:.2f}, dS={delta_S:.2f}')
```

Figure: <i>Left: temperatures from balancing *number* flux at equal
populations, forcing Maxwellian bulks all the way to the gate. Right:
entropy of that two-Maxwellian state minus the single-temperature
equilibrium. The red point is the simulation threshold $v_\star=3$.

With $k_B=1$, equal populations, $v_\star=3$ and total energy matching
the simulation’s mean kinetic energy $12.5$, number-flux balance gives
$T_L\approx 1.94$ and $T_R\approx 23.1$. The ideal-gas Shannon entropy
of a 2D Maxwellian chamber (position Stirling term plus velocity
differential entropy) $$
S(N,\beta)=N\left[2+\log\frac{A}{N}-\log\frac{\beta}{2\pi}\right]
$$ falls by about $24$ nats relative to the single-temperature state at
the same $N$ and $E$. Read that drop as the cost of the illegal coarse
graining, not as a machine. Hot accumulates on the right, cold on the
left, and the joint distribution over $(\mathrm{side},\mathbf{v})$ looks
more concentrated than the unconstrained Maxwellian — *provided* we
pretend the kinetic layers are Maxwellian too. They are not. The true
non-equilibrium state depletes the left hot tail and the right cold
body; collisions refill those sets in the bulks. The local-temperature
calculation remains the right first cut on *direction*. It is not a
closed book.

## Where Can the Entropy Sit?

Once the compartments are drawn, the demon is no longer a single puzzle.
It is a choice of which coarse graining is legal, and therefore of where
the missing entropy is allowed to sit.

If we keep Maxwellians all the way to the gate, particle-flux balance
and energy-flux balance cannot hold together. The calculation that
produced $T_L\approx 1.9$, $T_R\approx 23$ already used that illegal
extension: it closed only the number ledger. The first resolution is to
refuse the extension. The kinetic layers are not groups of the same kind
as the bulks; their entropy is not $H(\pi_i)$ at a local $T$.

If the gate is a physical trapdoor, the door is a degree of freedom in
the same bath. It jitters, it fails to stay open for the tail and shut
for the body, and the rectification dies (Smoluchowski, 1914). Feynman’s
pawl is the same object.

If the gate is a policy that remembers who may pass, the record is on
the ledger. Szilard isolates one bit of that record; Landauer prices
erasing it (Bennett, 1982; Landauer, 1961; Szilard, 1929).

If we refuse compartments altogether and write one equilibrium ensemble
for the closed box, the sorting we see on the canvas is a fluctuation.
Maxwell already said the second law need not apply to a being who
handles individual molecules (Maxwell, 1871).

Those are different books, not competing slogans. The next worksheet is
to coarse-grain Maxwell: which of those descriptions you are allowed to
write for each compartment, and what entropy each description may claim.

## The Demon Sorts Groups

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/demon-groups-hot-cold.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/demon-groups-hot-cold.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Maxwell’s sorting rule partitions velocity space into coarse sets —
`hot enough to cross'' versus`cold enough to cross’’ — and applies
different gates to each set. That is week 2’s grouping again, now
stacked on the spatial compartments: a deterministic label
$G=g(\mathbf{v})$ plus residual uncertainty inside each label. Local
equilibrium — timescale separation between collisions and crossings — is
the claim we already restricted to the *distant* bulks. Extending it
through the kinetic layers is what made number balance and energy
balance incompatible. When that extension fails, the true steady state
is no longer a pair of perfect Maxwellians (tails deplete). Information
engines and demons play on the tension between moving probability among
groups ($H(w)$) and the within-group bath term $\sum w_i H(p_i)$ that
thermodynamics prices.

Kemeny–Snell lumpability is the precise condition under which a
partition of a Markov chain induces another Markov chain on the groups.
When it fails, the macro description needs memory of how the system
entered the group. Name the issue; do not develop projection-operator
formalism here.

Worksheet 3 is the coarse-graining of Maxwell: which description is
legal for each compartment, and where the missing entropy is allowed to
sit. Szilard, Landauer, Ellis, and the information-engine scale argument
are parked until that worksheet needs them.
