## Exponential Family

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/mlatcl/iei/edit/gh-pages/_lamd/channels-dpi-good-regulator.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/mlatcl/iei/edit/gh-pages/_lamd/channels-dpi-good-regulator.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

On a finite state space the constrained MaxEnt distribution is an
exponential family. With sufficient statistics $T_k(x)=f_k(x)$ and base
density $h_0$ of the counting measure $\#$ on the allowed outcomes,
where $h_0(x)=\mathbf{1}[x\in\mathcal{X}]$ (one on allowed labels, zero
elsewhere). The Lagrange multipliers from the Week 5 proof are the
negative natural parameters, $\theta_k=-\lambda_k$. The minus sign keeps
high-energy states down-weighted when $\theta_1=-\beta<0$.

A probability is always a density with respect to some reference measure
$\mu$: the Radon–Nikodym derivative ${\mathrm{d}}P/{\mathrm{d}}\mu$.
Every entropy-like quantity in the course is the same expression, with a
different $\mu$ each time. Lecture 4 already said that choosing a
discretisation is choosing a measure; the counting measure $\#$ is the
case so natural that nobody notices the choice has been made.

Separate two uses of the word Lebesgue. The *Lebesgue integral*
$\int f\,{\mathrm{d}}\mu$ is a single integration machine that works for
*any* measure $\mu$: a sum is a Lebesgue integral against $\#$, and the
ordinary calculus integral is a Lebesgue integral against Lebesgue
measure $\lambda$. *Lebesgue measure* $\lambda$ itself is one particular
choice of $\mu$: length, area, or volume, so $\lambda([a,b])=b-a$.
Integrating against $\#$ gives a sum,
$\int f\,{\mathrm{d}}\#=\sum_x f(x)$, and the density of $P$ with
respect to $\#$ is the probability mass function $p(x)=P(\{x\})$.
Shannon entropy is exactly $H_\#(P)=-\sum_x p(x)\log p(x)$.

Counting measure is a prior: every outcome gets weight 1. On a finite
set of $N$ outcomes write $u=\#/N$ for the uniform distribution. Then So
Shannon entropy *is* a negative KL divergence against the uniform that
$\#$ defines. Three immediate consequences: (i) $H\ge 0$, because each
atom has counting measure 1 so $p(x)\le 1$; (ii) $H$ is unchanged by any
relabelling of the outcomes ($\#$ is invariant under bijections); (iii)
on a countably infinite set $\#$ has infinite total mass, there is no
uniform $u$, and entropy need not be finite.

The same template covers the continuous and relative cases. Lebesgue
measure $\lambda$ is the continuum counterpart: differential entropy is
$H_\lambda$, can be negative, and changes under nonlinear
reparametrisation because $\lambda$ does. Two distinct effects meet here
and should stay separate. Under a change of variable the Jacobian
$|{\mathrm{d}}y/{\mathrm{d}}x|$ appears — that is Lecture 4’s
density-of-states factor (for example $v^2$ in the speed coordinate).
Separately, putting counting measure on a grid of spacing $\Delta$ and
rescaling by $\Delta$ converges to $\lambda$ as $\Delta\to 0$; a binned
Shannon entropy then carries an additive $-\log\Delta$, which is the log
of that rescaling factor between $\#$ and $\lambda$, not a Jacobian. Why
not just use counting measure on $\mathbb{R}$? Because $\#$ and
$\lambda$ are mutually singular: $\lambda(\{x\})=0$ for every point,
while a discrete law puts positive mass on points, so neither measure
admits a density with respect to the other. A mixed law — daily rainfall
exactly zero with probability $0.3$, otherwise continuous — has a
density only with respect to something like $\delta_0+\lambda$. Keep
that aside in the notes; do not board it live.

KL divergence uses a probability $Q$ as reference, provided $P$ is
absolutely continuous with respect to $Q$ ($P\ll Q$). If $P$ puts mass
anywhere $Q$ does not, the Radon–Nikodym derivative fails and
$\mathrm{KL}(P\|Q)=+\infty$. When both densities are written against a
common base $\mu$, the ratio of densities cancels $\mu$, which is why KL
(and mutual information) has no $-\log\Delta$ divergence and is
unchanged when the same invertible map is applied to both arguments.

Maximum entropy relative to $\mu$ gives
${\mathrm{d}}P/{\mathrm{d}}\mu=\exp(\theta\cdot T-\psi)$. The factor
$h_0$ in the exponential family is the density of that base measure when
sums or integrals are written against a default chart. Today $\nu=\#$ on
the allowed set and
$h_0={\mathrm{d}}\nu/{\mathrm{d}}\#=\mathbf{1}[x\in\mathcal{X}]$, so
$\psi$ is a log-sum. Even in discrete cases the base need not be plain
counting: Poisson uses $h_0(x)=1/x!$ on $\mathbb{N}$, binomial uses
$\binom{n}{x}$ — combinatorial multiplicity as the discrete counterpart
of the $v^2$ density of states. We write $h_0$ rather than $h$ to avoid
colliding with differential entropy and with the regulator map $A=h(S)$
later in the module.

Each of these is MaxEnt for its moment constraints, so each belongs to
the exponential family. The canonical ensemble is the mean-energy case
already recovered in Week 5. A Gaussian is MaxEnt for a fixed mean and
variance. A two-level system with energies $0$ and $\varepsilon$ is a
Bernoulli with natural parameter $\theta=-\beta\varepsilon$. The
distribution bank that follows is where both measure morals land:
Bernoulli / categorical use plain counting ($h_0=1$); binomial / Poisson
/ multinomial use combinatorial $h_0$ as discrete density of states.

## Common Distributions as Exponential Families

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_statistics/includes/common-distributions-exponential-family.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_statistics/includes/common-distributions-exponential-family.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Week 1 named five families without writing them in natural coordinates:
Bernoulli, binomial, Poisson, multinomial/categorical, and Gaussian.
Each is MaxEnt for its moment constraints relative to an appropriate
base measure $h_0$, so each belongs to the exponential family $$
p(x\mid\boldsymbol{\theta})=\exp\bigl(\boldsymbol{\theta}\cdot T(x)-\psi(\boldsymbol{\theta})\bigr)\,h_0(x).
$$ The slides below record, for each family, the natural parameter(s)
$\boldsymbol{\theta}$, the sufficient statistic $T$, the cumulant
generating function $\psi$, and — the point of this block — which $h_0$
is in force. Two emphases run through the tour. First, when $h_0=1$
(Bernoulli; categorical $n=1$), the base is counting measure: every
allowed label has prior weight one, the same unnoticed prior as
$H=\log N-\mathrm{KL}(P\|u)$. Second, when $h_0$ is non-constant
(binomial $\binom{n}{k}$, Poisson $1/k!$, multinomial
$n!/\prod_j x_j!$), that factor is the discrete density of states: same
moment constraints relative to plain $\#$ would give a *different*
family. Gaussian uses Lebesgue — the continuum counterpart of $\#$.

## Bernoulli as Exponential Family

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_statistics/includes/bernoulli-as-exponential-family.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_statistics/includes/bernoulli-as-exponential-family.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

A Bernoulli($\pi$) trial with $P(X=1)=\pi$ is already in
exponential-family form once we take the logit as the natural parameter,
$$
\theta=\log\frac{\pi}{1-\pi},\qquad T(x)=x,\qquad \psi(\theta)=\log(1+e^{\theta}).
$$ The base density is $h_0(x)=1$: counting measure on the two labels.
That is the unnoticed prior from the counting-measure slide — each of
$\{0,1\}$ has weight one, so Shannon $H$ on a Bernoulli is
$\log 2-\mathrm{KL}(P\|u)$ with $u$ uniform on those two outcomes.
Recovering the mean parameter is the logistic map:
$\pi=\sigma(\theta)=e^{\theta}/(1+e^{\theta})$, and $\psi'(\theta)=\pi$.
The two-level thermal system with energies $0$ and $\varepsilon$ is the
same family with $\theta=-\beta\varepsilon$.

## Binomial as Exponential Family

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_statistics/includes/binomial-as-exponential-family.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_statistics/includes/binomial-as-exponential-family.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Binomial$(n,\pi)$ is $n$ i.i.d. Bernoulli trials counted. The natural
parameter is the same logit as the Bernoulli, $$
\theta=\log\frac{\pi}{1-\pi},\qquad T(k)=k,
$$ but the cumulant generating function scales with the number of
trials, $$
\psi(\theta)=n\log(1+e^{\theta}),
$$ and the base measure carries the combinatorial multiplicity
$h_0(k)=\binom{n}{k}$. That multiplicity is the discrete density of
states — the same role $v^2$ played for Maxwell–Boltzmann. MaxEnt under
a mean-count constraint *relative to plain counting measure* does *not*
recover the binomial; MaxEnt relative to $\binom{n}{k}\,{\mathrm{d}}\#$
does. Same $\theta$ as Bernoulli; different prior weight on the count
$k$. Mean and variance follow by differentiation: $\psi'(\theta)=n\pi$,
$\psi''(\theta)=n\pi(1-\pi)$.

## Poisson as Exponential Family

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_statistics/includes/poisson-as-exponential-family.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_statistics/includes/poisson-as-exponential-family.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Poisson($\lambda$) has probability mass $$
P(k)=\frac{\lambda^k e^{-\lambda}}{k!}.
$$ In exponential-family coordinates the rate becomes a natural
parameter through $\theta=\log\lambda$, the count is the sufficient
statistic $T(k)=k$, and the cumulant generating function is simply $$
\psi(\theta)=e^{\theta}=\lambda.
$$ The factorial in the denominator is the base measure $h_0(k)=1/k!$
with respect to counting measure on $\mathbb{N}_0$. This is the sharpest
discrete illustration of the Weeks 4–5 story: the exponential tilt
$\exp(\theta\,T-\psi)$ is only half the model; the base $h_0$ is the
other half. MaxEnt under a fixed mean count *relative to plain counting
measure* ($h_0=1$) yields a geometric distribution. MaxEnt relative to
$h_0=1/k!$ yields Poisson. Same constraint, different measure, different
family — the discrete counterpart of Lebesgue versus $v^2\mathrm{d}v$.
Differentiating recovers the familiar moments: $\psi'(\theta)=\lambda$
and $\psi''(\theta)=\lambda$ (mean equals variance).

## Multinomial as Exponential Family

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_statistics/includes/multinomial-as-exponential-family.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_statistics/includes/multinomial-as-exponential-family.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

A categorical distribution on $K$ exclusive labels is a multinomial with
a single trial ($n=1$). Write the outcome as a one-hot vector
$\mathbf{x}$ with $\sum_{j=1}^K x_j=1$. Because of the simplex
constraint we use $K-1$ free natural parameters, $$
\theta_j=\log\frac{\pi_j}{\pi_K},\qquad j=1,\ldots,K-1,
$$ with sufficient statistics $T(\mathbf{x})=(x_1,\ldots,x_{K-1})$ and
cumulant generating function $$
\psi(\boldsymbol{\theta})=\log\Bigl(1+\sum_{j=1}^{K-1}e^{\theta_j}\Bigr)
$$ (the log-sum-exp that recovers
$\pi_K=1/\bigl(1+\sum_j e^{\theta_j}\bigr)$). For $n$ independent trials
the counts $\mathbf{x}$ are multinomial: the same $\boldsymbol{\theta}$
and $T$, but $$
\psi(\boldsymbol{\theta})=n\log\Bigl(1+\sum_{j=1}^{K-1}e^{\theta_j}\Bigr)
$$ and base measure $h_0(\mathbf{x})=n!/(x_1!\cdots x_K!)$. Softmax
regression is this family with features in place of the raw one-hot.

## Gaussian as Exponential Family

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_statistics/includes/gaussian-as-exponential-family.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_statistics/includes/gaussian-as-exponential-family.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The univariate Gaussian $\mathcal{N}(\mu,\sigma^2)$ is MaxEnt under
fixed mean and variance, so it belongs to a two-parameter exponential
family. Expanding the quadratic energy gives natural parameters $$
\theta_1=\frac{\mu}{\sigma^2},\qquad \theta_2=-\frac{1}{2\sigma^2},
$$ sufficient statistics $T(y)=(y,y^2)$, and cumulant generating
function $$
\psi(\theta_1,\theta_2)=-\frac{\theta_1^2}{4\theta_2}-\frac12\log(-2\theta_2)-\frac12\log(2\pi).
$$ The base measure is Lebesgue — the continuous counterpart of counting
measure $\#$ — and the $(2\pi)$ factor sits in $\psi$ (equivalently in a
constant piece of $h_0$). Differentiating recovers the moments:
$\nabla\psi=(\mu,\mu^2+\sigma^2)$. With *known* variance the family
collapses to one parameter, $\theta=\mu/\sigma^2$, $T(y)=y$, and
$\psi(\theta)=\tfrac12\sigma^2\theta^2$ (up to an additive constant
independent of $\theta$). The multivariate case replaces $\theta_1$ by
$\Sigma^{-1}\boldsymbol{\mu}$ and $\theta_2$ by $-\tfrac12\Sigma^{-1}$
acting on $yy^\top$.

## $\psi(\theta)$ as Cumulant Generating Function

The Quiz 2 debrief already read $\log Z$ as available energy,
$F=-k_BT\log Z$. In exponential-family coordinates the same function is
$\psi(\boldsymbol{\theta})=\log Z(-\boldsymbol{\theta})$. Its first
derivative is the mean of $T$. Differentiating again will give
$\mathrm{Cov}[T]$, but that identity is earned after the entropy form,
not asserted here.

## Entropy of an Exponential Family

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/exponential-family-entropy-fisher.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/exponential-family-entropy-fisher.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Once $p$ is written as an exponential family, Shannon entropy collapses
to an algebraic identity in $\psi$ and $\boldsymbol{\eta}$. That
identity is the bridge to Fisher information: the same $\log p$ whose
expectation is $-H$ has a score whose second moment is $\nabla^2\psi$.

Write
$\log p(x\mid\boldsymbol{\theta})=\boldsymbol{\theta}\cdot T(x)-\psi(\boldsymbol{\theta})+\log h_0(x)$.
When the base is counting measure on the allowed set ($h_0=1$), Shannon
entropy collapses to with $\boldsymbol{\eta}=\nabla\psi$. In general the
base survives as so the short form $H=\psi-\theta\cdot\eta$ is the
$h_0=1$ special case (Bernoulli, categorical, Jaynes’ die), not a
licence to drop $1/k!$ from a Poisson. Equivalently, when $h_0=1$,
$H=-\psi^*(\boldsymbol{\eta})$ where $\psi^*$ is the Legendre conjugate
of $\psi$. That is the form that makes the next differentiation step
cheap.

Differentiate $\log p=\boldsymbol{\theta}\cdot T-\psi$ under the
expectation: $$
\nabla_{\boldsymbol{\theta}}\log p(x\mid\boldsymbol{\theta})=T(x)-\nabla\psi(\boldsymbol{\theta})=T(x)-\boldsymbol{\eta}.
$$ The score has mean zero. Fisher’s information matrix is the second
moment of the score, $$
G_{ij}(\boldsymbol{\theta})=\mathbb{E}_{\boldsymbol{\theta}}[\partial_i\log p\,\partial_j\log p]=\mathrm{Cov}_{\boldsymbol{\theta}}[T]_{ij}.
$$ Differentiating $\boldsymbol{\eta}=\nabla\psi$ once more yields the
same matrix as a Hessian: $G=\nabla^2\psi$. So the cumulant-generating
property *is* the Fisher metric on an exponential family — earned from
the entropy calculation, not postulated.

## Fisher Metric from the Exponential Family

*Class test timer (Fisher geometry): 10:35*

Fisher (1925) called the expected squared score *information* about a
parameter. For a scalar parameter, $$
I(\theta)=\mathbb{E}_\theta\!\left[\left(\frac{\partial}{\partial\theta}\log p(x\mid\theta)\right)^2\right].
$$ On an exponential family that is exactly
$\nabla^2\psi=\mathrm{Cov}[T]$, already boarded. Shannon’s $H$ measures
uncertainty in a random variable; Fisher’s $I$ measures how much a
sample tells you about $\theta$. Conflating the two words is the first
mistake to block — and the exponential family is where both appear from
one $\log p$.

Three readings of one matrix: covariance of the sufficient statistics,
Fisher’s expected squared score, and the Hessian of the cumulant
generating function. Fréchet–Rao–Cramér identifiability
($\mathrm{cov}(\hat{\boldsymbol{\theta}})\succeq G^{-1}$) is the
estimation reading. That matrix is the Fisher information metric on the
statistical manifold.

The Fisher matrix defines infinitesimal information distance. Larger $G$
means a given parameter step is more distinguishable. Integrating
$\mathrm{d}s$ along a path of equilibria would give Crooks’
thermodynamic length; that development is dropped from the module. Today
the metric itself is the object.

Amari’s dual flatness: exponential families admit affine coordinates in
$\boldsymbol{\theta}$ and in $\boldsymbol{\eta}=\nabla\psi$, with dual
potentials whose Hessians are inverse Fisher metrics. The Legendre dual
is what they already computed for entropy:
$\psi^*(\eta)=\theta\cdot\eta-\psi=-H$ when $h_0=1$ (otherwise subtract
$\mathbb{E}[\log h_0]$ from $H$). The Pythagorean theorem for KL on dual
flats stays thin today so Shannon and the Good Regulator keep their
clock.

## Shannon’s Communication Theorem

*Class test timer (Shannon): 11:05*

Shannon’s communication theorem (noisy-channel coding) is the
information-theoretic twin of Carnot: it forbids reliable rates above
capacity and does not construct a code. Capacity $C=\max_{p(x)}I(X;Y)$
is the no-go number; the capacity-achieving input distribution is the
prescription. Week 5 already made mutual information first-class via
soft partitions; today that $I$ is the rate currency on a channel.

## Scaffolding, Not Outcomes

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/channel-capacity-chain-rule.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/channel-capacity-chain-rule.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The chain rule $H(X,Y)=H(X)+H(Y|X)$ is the algebraic source of
multi-information. Mutual information $I(X;Y)=H(X)-H(X|Y)$ is the
pairwise case; we do not yet treat $n>2$. Channel capacity $C$ is a
no-go on rate; the capacity-achieving input distribution is the
prescription. The data-processing inequality is the no-go on $I$:
processing cannot create information. Cover and Thomas Theorem 2.8.1 is
the reading; the proof waits for week 8, when $I$ is first-class.

```
eps = 0.1
p = np.linspace(0.01, 0.99, 200)
H = lambda q: -q*np.log2(q)-(1-q)*np.log2(1-q)
I = H(p) - H(eps)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(p, I, linewidth=2)
ax.scatter([0.5], [H(0.5) - H(eps)], s=80, color='red', zorder=3)
ax.set_xlabel('input bias $p(x=1)$')
ax.set_ylabel('$I(X;Y)$ (bits)')
ax.set_title('Binary symmetric channel')
mlai.write_figure('bsc-capacity.svg', directory='./ml')
```

## Good Regulator Overview

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/good-regulator-overview.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/good-regulator-overview.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Overview only. Full Conant–Ashby development — concavity argument,
existence of a deterministic optimum, and the two caveats in depth — is
the Week 8 review. Today’s job is the setup, the notation, and the
slogan with fences. Diagrams follow Joery’s oneshot figure in the
good-regulator notes: disturbance $D$, system state $S$, action $A$,
reguland $Z$.

Conant and Ashby (Conant and Ashby, 1970) often label the regulator $R$.
We write $A$ for the action so the policy $\pi(a\mid s)$ is first-class.
The reguland — the variable whose entropy successful regulation
minimises — is $Z$. The system state is $S$; the disturbance is $D$. The
cumulant generating function earlier today was
$\psi(\boldsymbol{\theta})=\log Z(-\boldsymbol{\theta})$; that $Z$ was
the partition function, not the reguland — different objects, same
letter in the physics tradition.

A disturbance $D$ causes events in a system $S$; the regulator counters
with an action $A$; the pair produces a reguland $Z=\phi(S,A)$ (Conant
and Ashby, 1970). Cause-controlled regulation acts on $D$ (or on an
inferred model of $D$). Error-controlled regulation acts on the state
$S$ caused by $D$, so by the data-processing inequality
$I(A;D)\le I(S;D)$. Both diagrams are on today’s board; the existence
proof that an $H(Z)$-optimal policy may be taken deterministic waits for
Week 8.

Successful regulation, for Conant and Ashby, minimises the entropy of
the reguland: $\min_\pi H(Z)$. Among optima there exists a deterministic
policy $A=h(S)$, equivalently $H(A\mid S)=0$. That is the sense in which
the regulator is a
`model'' --- deliberately weak. Do not collapse the slogan into`$A$
contains a rich model of $S$’’: a constant action also has
$H(A\mid S)=0$. Do not confuse low $H(Z)$ with a desirable outcome.
Requisite variety in Shannon form will use the same letters: residual
uncertainty satisfies $H(Z)\ge H(D)-I(D;A)$ when $A$ is the regulatory
response. Week 8 reviews the theorem properly and applies it to
capability claims.

Ashby’s law of requisite variety in Shannon form (Ashby, 1956), written
in today’s letters: residual uncertainty in the reguland is at least the
disturbance entropy minus the information the action carries about the
disturbance. Requisite variety asks how much relevant $I$ is required;
the Good Regulator Theorem asks how that $I$ may be organised. Full
development of both, with the concavity argument, is Week 8.
