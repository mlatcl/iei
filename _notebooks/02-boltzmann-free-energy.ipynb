## Laplace’s Molecule: an Anachronism?

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/laplace-molecule-anachronism.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/laplace-molecule-anachronism.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

In lecture 1 we read Laplace’s line that the curve of a simple molecule
of air or vapour is regulated as certainly as a planetary orbit, the
difference being only our ignorance (Laplace, 1814). To a modern ear
that is anachronistic. Ludwig Boltzmann spent his career defending a
molecular picture of matter; energeticists such as Ostwald and
positivists such as Mach treated atoms as optional or metaphysical.
Boltzmann died in 1906; Einstein’s Brownian-motion theory and Perrin’s
experiments (1908–1909) are what made the molecular hypothesis
unavoidable for the mainstream. So why is Laplace, writing in 1814,
already talking about molecules?

In early modern and Enlightenment usage, *molecule* often meant a minute
body or corpuscle — the Newtonian residue of a particulate theory of
matter — rather than the stoichiometrically defined chemical molecule of
the late nineteenth century. Laplace’s point in the *Essai* is
epistemological: even if every air particle followed a determined
trajectory, we lack the data, the laws in usable form, and the
computation to track it, so we turn to probability. The ontological
fight over whether those particles *exist* is a later, sharper
controversy. The person who gave Laplace a concrete kinetic picture of
air was Daniel Bernoulli, three-quarters of a century earlier.

## What Bernoulli Assumes

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/bernoulli-kinetic-assumptions.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/bernoulli-kinetic-assumptions.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Daniel Bernoulli’s kinetic chapter in *Hydrodynamica* (Mikhailov, n.d.)
asks what pressure *is* if air is a swarm of particles. The modelling
assumptions are severe and productive:

Conservation of mechanical effect (energy in later language) lets him
relate microscopic motion to macroscopic push. That is an extraordinary
claim for 1738. It is also an *idealisation*: real molecules attract and
vibrate; Bernoulli’s gas is a coarse mechanical cartoon chosen so that
Newton’s laws have something to act on.

Even at the birth of kinetic theory, the useful object is not the full
list of paths. Bernoulli averages. The statistical physicists who follow
— Maxwell, Boltzmann, Gibbs — will turn that averaging into an explicit
probability over microstates. The billiard simulation is Bernoulli’s
assumption made visible; the Shannon entropy of the velocity histogram
is already a coarse-grained proxy for how much velocity detail we are
keeping.

## From Paths to an Ensemble

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/ensemble-as-coarse-graining.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/ensemble-as-coarse-graining.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Laplace’s demon would integrate every molecular trajectory. Laplace’s
gremlin says we lack the information and the compute. Statistical
mechanics answers by changing the object of study. A *microstate* is a
point in the high-dimensional space of all positions and momenta — the
collection of curves. A *macrostate* is what we can actually hold fixed
or measure: particle number, volume, energy, or temperature and
pressure. An *ensemble* is a probability distribution over microstates
consistent with that macrostate. We replace ignorance of the path with a
measure on the set of paths. That replacement *is* a coarse-graining:
many micro-histories are declared equivalent for the purpose of
prediction.

The billiard canvas already coarse-grains: it reports a velocity-bin
Shannon entropy, not the full thermodynamic entropy of the gas. Maxwell,
Boltzmann and Gibbs elevate the same move. Once the macroscopic
constraints are fixed, the remaining uncertainty over microstates is
what their $S$ counts. The fight over atoms was partly a fight over
whether that microscopic sample space was real. The mathematical move —
probability over an ensemble — survives either reading, and it is the
move this course will keep using when energy becomes a score and
$p_i\propto e^{-\beta E_i}$ becomes an occupation.

What follows in this lecture is not a new ontology so much as the
bookkeeping that makes the ensemble usable: Gibbs–Boltzmann occupation,
free energy $F=U-TS$, and the bath that justifies fixing $T$. The
molecular hypothesis will win historically; the course needs the
*abstraction* today.

The important point about the uncertainty being represented here is that
it is not genuine stochasticity, it is a lack of knowledge about the
system. The techniques proposed by Maxwell, Boltzmann and Gibbs allow us
to exactly represent the state of the system through a set of parameters
that represent the sufficient statistics of the physical system. We know
these values as the volume, temperature, and pressure. The challenge for
us, when approximating the physical world with the techniques we will
use is that we will have to sit somewhere between the deterministic and
purely stochastic worlds that these different scientists described.

One ongoing characteristic of people who study probability and
uncertainty is the confidence with which they hold opinions about it.
Another leader of the Cavendish laboratory expressed his support of the
second law of thermodynamics (which can be proven through the work of
Gibbs/Boltzmann) with an emphatic statement at the beginning of his

## The Information Game

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Before writing a formula for entropy, play a game. One player thinks of
an object drawn from a known finite list. The other may ask only yes/no
questions. How many questions do you need in the worst case? That count
is already a measure of information: it is how much uncertainty you must
remove to name the object.

With a list of size $N$ and equal prior probability on each item, an
ideal strategy halves the search space at every step. After $k$ answers
you can distinguish at most $2^k$ hypotheses, so you need $$
k \ge \log_2 N
$$ questions in the worst case. Twenty questions reach a million
possibilities: $2^{20} = 1\,048\,576$. A *fair* yes/no answer therefore
carries one *bit* of information. That is Shannon’s unit before
Shannon’s formula: information as the number of balanced binary
distinctions required to resolve uncertainty.

If the prior is not uniform, the best question is no longer a literal
half-split of the *count* of remaining objects. It is a half-split of
their *probability*. The average number of questions under optimal play
becomes the Shannon entropy $H = -\sum_i p_i\log_2 p_i$. We will meet
that formula next; the game already shows what the number is counting.

## MacKay’s Balance and Balls

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

David MacKay opens the compression half of *Information Theory,
Inference, and Learning Algorithms* with a weighing puzzle (MacKay,
2003, p. Exercise\~4.1). It is the information game with a ternary
instrument instead of a binary question. We use it here only as a
*counting* argument: how much uncertainty we start with, how much each
use of the balance can remove, and why three weighings are enough in
principle — without exhibiting a particular sequence of measurements.

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/mackay-balance-balls.svg" class="" width="80%" style="vertical-align:middle;">

Figure: <i>A two-pan balance and twelve identical balls. Each use of the
balance returns one of three readings: left heavy, balance, or right
heavy.</i>

You are given twelve balls, identical in appearance. Exactly one differs
in weight from the others; you do not know whether that odd ball is
heavier or lighter. A two-pan balance compares two equal-sized groups
and returns one of three readings: left side heavier, right side
heavier, or equal. Design a strategy that finds the odd ball and the
direction of the defect.

Under a uniform prior over those twenty-four possibilities, the Shannon
uncertainty we must remove is $$
H_{\mathrm{start}} = \log_2 24 \approx 4.58 \text{ bits}.
$$ At the end we need a single named hypothesis: $H_{\mathrm{end}} = 0$.
The *deficit* to close is therefore about $4.58$ bits. That difference —
start minus end — is the information the experiment must supply.

A binary question yields at most one bit. The balance is richer: its
alphabet size is three. If the three readings are equiprobable, one
weighing contributes $$
\log_2 3 \approx 1.585 \text{ bits}.
$$ After $k$ weighings the strategy tree has at most $3^k$ leaves, so it
can name at most $3^k$ hypotheses. Compare budgets: $$
\begin{align*}
3^2 &= 9 < 24,\\
3^3 &= 27 \ge 24.
\end{align*}
$$ Two weighings cannot suffice no matter how clever the groupings.
Three weighings *can*: there is spare capacity of three unused leaves
($27-24=3$). In bits, $$
3\log_2 3 \approx 4.75 > \log_2 24 \approx 4.58.
$$ The gap is small. That is why the puzzle feels tight, and why a first
weighing that wastes an outcome (for example putting six against six,
which can never balance if exactly one ball is odd) fails: you have
thrown away ternary resolution and fallen back toward a binary split.

We stop at the existence argument. Constructing the adaptive tree —
which balls on which pan at each step — is MacKay’s exercise, and
solving it is the best way to feel what \`\`informative experiment’’
means. The only design principle we need today is the one already
present in the information game: choose each measurement so that its
outcomes are as nearly equiprobable as possible under the current
posterior. That maximises expected information gain per weighing and is
what makes the $3^3 \ge 24$ budget achievable in practice.

That last number is Shannon entropy in nats:
$H(p)=\mathbb{E}_{p}[-\log p(y)]=-\sum_y p(y)\log p(y)$. Same arithmetic
under another name.

Under Shannon’s definition, entropy is dimensionless. It is the
expectation of the negative log surprisal,
$-\left\langle\log P(y)\right\rangle$. The information game and MacKay’s
balance give the operational reading: $H$ is how much uncertainty you
must remove; the log alphabet size of your instrument bounds how much
each query can buy.

## Bernoulli Entropy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/bernoulli-entropy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/bernoulli-entropy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

A Bernoulli($p$) random variable takes values $\{0,1\}$ with
probabilities $1-p$ and $p$. Its Shannon entropy is the binary entropy
$$
H(X) = -p\log_2 p - (1-p)\log_2(1-p)
$$ bits. The panels below show, for a fixed $p$, the distribution, the
pointwise surprisal $-\log_2 P(x)$, and the summands $-P(x)\log_2 P(x)$
whose total is $H$.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
import mlai.plot as plot
```

```
p_success = 0.3
xs = np.array([0, 1])
probs = np.array([1 - p_success, p_success])
surprisal = -np.log2(probs)
contrib = probs * surprisal
H = contrib.sum()

fig, axes = plt.subplots(1, 3, figsize=plot.big_wide_figsize)
colours = ['#4C72B0', '#DD8452', '#55A868']
axes[0].bar(xs, probs, color=colours[0], width=0.5)
axes[0].set_xticks(xs)
axes[0].set_xlabel('$x$')
axes[0].set_ylabel('$P(x)$')
axes[0].set_title(f'Bernoulli($p={p_success}$)')
axes[0].set_ylim(0, 1)

axes[1].bar(xs, surprisal, color=colours[1], width=0.5)
axes[1].set_xticks(xs)
axes[1].set_xlabel('$x$')
axes[1].set_ylabel('$-\\log_2 P(x)$')
axes[1].set_title('surprisal')

axes[2].bar(xs, contrib, color=colours[2], width=0.5)
axes[2].set_xticks(xs)
axes[2].set_xlabel('$x$')
axes[2].set_ylabel('$-P(x)\\log_2 P(x)$')
axes[2].set_title(f'summands ($H={H:.3f}$ bits)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('bernoulli-entropy-panels.svg', directory='./information')
```

## Binomial Entropy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/binomial-entropy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/binomial-entropy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

A Binomial($n,p$) count is the sum of $n$ i.i.d. Bernoulli trials. There
is no elementary closed form for $H$, but the same three-panel reading
applies: plot $P(k)$, plot $-\log_2 P(k)$, and sum $-P(k)\log_2 P(k)$.
For large $n$ a Gaussian approximation gives
$H\approx\tfrac12\log_2\bigl(2\pi e\, np(1-p)\bigr)$ bits.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
import mlai
import mlai.plot as plot
```

```
n, p = 12, 0.4
ks = np.arange(0, n + 1)
probs = binom.pmf(ks, n, p)
# Guard zeros for log (binom pmf is positive on 0..n)
surprisal = -np.log2(np.clip(probs, 1e-300, None))
contrib = probs * surprisal
H = contrib.sum()

fig, axes = plt.subplots(1, 3, figsize=plot.big_wide_figsize)
colours = ['#4C72B0', '#DD8452', '#55A868']
axes[0].bar(ks, probs, color=colours[0], width=0.8)
axes[0].set_xlabel('$k$')
axes[0].set_ylabel('$P(k)$')
axes[0].set_title(f'Binomial($n={n}, p={p}$)')

axes[1].bar(ks, surprisal, color=colours[1], width=0.8)
axes[1].set_xlabel('$k$')
axes[1].set_ylabel('$-\\log_2 P(k)$')
axes[1].set_title('surprisal')

axes[2].bar(ks, contrib, color=colours[2], width=0.8)
axes[2].set_xlabel('$k$')
axes[2].set_ylabel('$-P(k)\\log_2 P(k)$')
axes[2].set_title(f'summands ($H={H:.3f}$ bits)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('binomial-entropy-panels.svg', directory='./information')
```

## Poisson Entropy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/poisson-entropy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/poisson-entropy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

A Poisson($\lambda$) distribution on the non-negative integers has $$
P(k)=\frac{\lambda^k e^{-\lambda}}{k!}.
$$ Entropy again has no elementary closed form; numerically
$H=\sum_k -P(k)\log_2 P(k)$. For large $\lambda$,
$H\approx\tfrac12\log_2(2\pi e\lambda)$ bits.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
import mlai
import mlai.plot as plot
```

```
lam = 4.0
ks = np.arange(0, 16)
probs = poisson.pmf(ks, lam)
surprisal = -np.log2(np.clip(probs, 1e-300, None))
contrib = probs * surprisal
H = contrib.sum()

fig, axes = plt.subplots(1, 3, figsize=plot.big_wide_figsize)
colours = ['#4C72B0', '#DD8452', '#55A868']
axes[0].bar(ks, probs, color=colours[0], width=0.8)
axes[0].set_xlabel('$k$')
axes[0].set_ylabel('$P(k)$')
axes[0].set_title(f'Poisson($\\lambda={lam:g}$)')

axes[1].bar(ks, surprisal, color=colours[1], width=0.8)
axes[1].set_xlabel('$k$')
axes[1].set_ylabel('$-\\log_2 P(k)$')
axes[1].set_title('surprisal')

axes[2].bar(ks, contrib, color=colours[2], width=0.8)
axes[2].set_xlabel('$k$')
axes[2].set_ylabel('$-P(k)\\log_2 P(k)$')
axes[2].set_title(f'summands ($H={H:.3f}$ bits)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('poisson-entropy-panels.svg', directory='./information')
```

## Multinomial / Categorical Entropy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/multinomial-entropy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/multinomial-entropy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

One draw from a multinomial (a categorical distribution) on $K$
categories with probabilities $\mathbf{p}=(p_1,\ldots,p_K)$ has entropy
$$
H = -\sum_{k=1}^K p_k\log_2 p_k
$$ bits — exactly Shannon’s formula. The four-outcome table from the
expectation example is already a categorical; we reuse those
probabilities here.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
import mlai.plot as plot
```

```
# Same four-outcome distribution as the expectation / entropy example
cats = np.array([1, 2, 3, 4])
probs = np.array([0.3, 0.2, 0.1, 0.4])
surprisal = -np.log2(probs)
contrib = probs * surprisal
H = contrib.sum()

fig, axes = plt.subplots(1, 3, figsize=plot.big_wide_figsize)
colours = ['#4C72B0', '#DD8452', '#55A868']
axes[0].bar(cats, probs, color=colours[0], width=0.6)
axes[0].set_xticks(cats)
axes[0].set_xlabel('category $y$')
axes[0].set_ylabel('$P(y)$')
axes[0].set_title('categorical (four outcomes)')
axes[0].set_ylim(0, 0.5)

axes[1].bar(cats, surprisal, color=colours[1], width=0.6)
axes[1].set_xticks(cats)
axes[1].set_xlabel('category $y$')
axes[1].set_ylabel('$-\\log_2 P(y)$')
axes[1].set_title('surprisal')

axes[2].bar(cats, contrib, color=colours[2], width=0.6)
axes[2].set_xticks(cats)
axes[2].set_xlabel('category $y$')
axes[2].set_ylabel('$-P(y)\\log_2 P(y)$')
axes[2].set_title(f'summands ($H={H:.3f}$ bits)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('multinomial-entropy-panels.svg', directory='./information')
```

## Gaussian Differential Entropy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/gaussian-entropy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/gaussian-entropy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

For a continuous density the same formal expression $$
h(X) = -\int p(y)\log_2 p(y)\,\mathrm{d}y
$$ is called *differential* entropy. It can be negative, and it is not
invariant to rescaling of $y$. For a univariate Gaussian
$\mathcal{N}(\mu,\sigma^2)$, $$
h = \tfrac12\log_2\bigl(2\pi e\sigma^2\bigr)
$$ bits. The panels below use the height example ($\mu=1.7$,
$\sigma^2=0.0225$): density, surprisal $-\log_2 p(y)$, and the integrand
$-p(y)\log_2 p(y)$.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
import mlai
import mlai.plot as plot
```

```
mu, var = 1.7, 0.0225
sigma = np.sqrt(var)
ys = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 400)
# Density in probability per unit length
pdf = norm.pdf(ys, loc=mu, scale=sigma)
# Surprisal in bits; pdf can exceed 1, so -log2(pdf) can be negative
surprisal = -np.log2(pdf)
integrand = pdf * surprisal
# Closed form in bits
h = 0.5 * np.log2(2 * np.pi * np.e * var)

fig, axes = plt.subplots(1, 3, figsize=plot.big_wide_figsize)
colours = ['#4C72B0', '#DD8452', '#55A868']
axes[0].plot(ys, pdf, color=colours[0], lw=2)
axes[0].fill_between(ys, pdf, color=colours[0], alpha=0.25)
axes[0].axvline(mu, color='#666', ls='--', lw=1)
axes[0].set_xlabel('$y$')
axes[0].set_ylabel('$p(y)$')
axes[0].set_title(f'Gaussian($\\mu={mu}, \\sigma^2={var}$)')

axes[1].plot(ys, surprisal, color=colours[1], lw=2)
axes[1].axhline(0, color='#666', ls=':', lw=1)
axes[1].set_xlabel('$y$')
axes[1].set_ylabel('$-\\log_2 p(y)$')
axes[1].set_title('surprisal (can be negative)')

axes[2].plot(ys, integrand, color=colours[2], lw=2)
axes[2].fill_between(ys, integrand, color=colours[2], alpha=0.25)
axes[2].set_xlabel('$y$')
axes[2].set_ylabel('$-p(y)\\log_2 p(y)$')
axes[2].set_title(f'integrand ($h={h:.3f}$ bits)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('gaussian-entropy-panels.svg', directory='./information')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/gaussian-entropy-panels.svg" class="" width="85%" style="vertical-align:middle;">

Figure: <i>Univariate Gaussian (height example). Middle: $-\log_2 p(y)$;
dips below zero where the density exceeds $1$. Right: the integrand
whose integral is differential entropy $h$.</i>

Thermodynamic entropy $S$ and Shannon $H$ have a shared functional form,
in equilibrium we can write $S = k_BH$, where $k_B$ is Boltzmann’s
constant,
$k_B = 1.380649 \times 10^{-23} \tfrac{\text{kg}\text{m}^2 }{\text{s}^2 \text{K}}$.

So there are two parts to a thermodynamic entropy, one is the
uncertainty of a (discrete) distribution. The other translates that
entropy into units of ‘work’
i.e. ($\text{J} = \tfrac{\text{kg}\,\text{m}^2 }{\text{s}^2}$) per
Kelvin ($\tfrac{\text{J}}{\text{K}}$). Boltzmann’s constant makes
entropy physical. When we multiply $S$ by temperature $T$ (in degrees
Kelvin) we recover a term that has units of energy.

But the dependence of entropy on the states of a distribution, or the
probability across those states, also implies that there is a *choice*
in terms of which states we consider. In classical mechanics we have
states like position and velocity (or momentum). But quantum mechanics
also leads to other properties of materials such as the *spin* of an
electron. Choice of which variables you consider when computing e.g. the
entropy of an electron, was referred to as the ‘“anthropomorphic” nature
of entropy’ by Jaynes (1965). He credits this remark originally to
Eugene Wigner. One example that Jaynes gives is thermodynamic steam
tables.

> If we work with a system of degrees of freedom, the experimental
> entropy is a function $S_{0}(X_{1}\ldots X_{n})$ of $n$ independent
> variables. But the system has any number of additional degrees of
> freedom $X_{n + 1},\ X_{n + 2}$, etc. We have to understand that these
> additional degrees of freedom are not to be tampered with during the
> experiments on the $n$ degrees of interest; otherwise one could easily
> produce *apparent violations of the second law.* Jaynes (1965) (my
> emphasis)

He goes on to give an example from entropy of steam (which is very
important in the design of steam engines and power plants).

> For example, the engineers have their “steam tables,” which give
> measured values of the entropy of superheated steam at various
> temperatures and pressures. But the H\$\_2\$0 molecule has a large
> electric dipole moment; and so the entropy of steam depends
> appreciably on the electric field strength present. It must always be
> understood implicitly (because it is never stated explicitly) that
> this extra thermodynamic degree of freedom was not tampered with
> during the experiments on which the steam tables are based; which
> means, in this case, that the electric field was not inadvertently
> varied from one measurement to the next.

This leads him to a conclusion that is worth bearing in mind when we’re
talking about the physical meaning of entropy.

> Recognition that the “entropy of a physical system” is not meaningful
> without further qualifications is important in clarifying many
> questions concerning irreversibility and the second law. For example,
> I have been asked several times whether, in my opinion, a biological
> system, say a cat, which converts inanimate food into a highly
> organized structure and behavior, represents a violation of the second
> law. The answer I always give is that, until we specify the set of
> parameters which define the *thermodynamic state* of the cat, no
> definite question has been asked!

Returning to the mathematical form of entropy, the fact that it is based
on probability means that many of the properties of probability carry
across to entropy. But the use of the logarithm means that they manifest
as additions instead of multiplications.

For example the joint entropy of two variables, $$
H(X, Y) = - \sum_{X, Y} P(X, Y) \log P(X, Y),
$$ can be decomposed as the entropy of the marginal and the entropy of
the conditional, $$
H(X, Y) = H(X \mid Y) + H(Y)
$$ where $$
H(Y \mid X) = - \sum_{X, Y} P(X, Y) \log P(X \mid Y) 
$$ and $$
H(Y) = - \sum_{Y} P(Y) \log P(Y).
$$ This is known as the *chain rule* of entropy. And it follows from the
product rule.

The equivalent of Bayes’ rule is $$
H(Y \mid X) = H(X \mid Y) + H(Y) - H(X).
$$

## Entropy and Probability

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-nogo-probability-prescription.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-nogo-probability-prescription.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

I’ve spent most of my career in machine learning thinking about
uncertainty, and very often that’s been through probability. For
so-called “Bayesian” researchers, one of the most important texts is Ed
Jaynes’s “Probability, the Logic of Science” (Jaynes, 2003), which has
become almost akin to a religious text. However, alongside probability,
and also in Jaynes’s work there’s been a recurring theme of entropy and
information arising.

I started my PhD in 1996 at Aston University, but in 1998 moved to
Cambridge, following my PhD supervisor who’d moved to Microsoft
Research. I was excited because one of the most important researchers
studying uncertainty in neural networks was based in Cambridge in the
Cavendish lab. His name was David MacKay, and because there were no
neural network researchers in the Computer Lab at the time I went to his
group meetings. His group was called the inference group.

MacKay’s work had been inspired by Stephen Gull’s work in the MaxEnt,
and his PhD thesis had introduced an approach to dealing with
uncertainty in neural networks known as the Laplace approximation. I was
excited to learn more. So you can imagine I was surprised and confused
to see that David had become more interested in *error correcting codes*
and information theory. Or at least that’s where his group’s focus had
shifted. And one of the main mathematical objects underpinning these
codes was *entropy*.

As a mechanical engineer, lacking the mathematical background of the
physicists around me, I had to learn somewhat by osmosis. Chemical
osmosis is driven by ionic gradients. Intellectual osmosis is driven by
intriguing explanations.

A mechanical engineer needs to learn thermodynamics. I answered exams
using tables that contained the entropy of superheated steam. I studied
thermodynamics in my undergraduate degree at Southampton from a textbook
by Rogers and Mayhew in the 1990s. It was the same textbook my father
had studied from when he studied mechanical engineering in Sheffield in
the 1960s.

So, I think this got me to thinking about a few things. What was the
entropy of steam to do with the entropy of a communication code (if
anything?). In the nearly three decades since those meetings I find I’ve
been thinking about that for some time. And I notice that while my
understanding of entropy has (I hope!) evolved, Rogers and Mayhew is
still an assigned textbook for some thermodynamics courses
<https://library.buid.ac.ae/meng234>.

Probability and entropy are tightly interlinked. And I’ve been
increasingly thinking if entropy should really be the principal object
we think about when considering uncertainty, not probability. I don’t
know if I’m right, but the idea of this course is to share some of those
ideas.

Why do I think entropy should be the principal object? Well there’s a
funny thing about it that I can’t quite put my finger on, an aspect of
entropy that has teased me for a while. And I’d like to explore that
aspect in this course with you.

Here’s the problem with probability. Every time I specify a probability
distribution, I need to specify all that happens. I don’t get to say
that the distribution looks like a bell curve on the right, but I don’t
know what’s going on on the left. I need my probability to both cover
the things I’m interested in, and the things I might not be interested
in. It feels like with probability I have to specify everything that’s
going to happen.[1] Entropy however is different. With entropy, it feels
like I get to talk about what’s *not* going to happen. In information
theory forms of entropy help us understand “no-go” areas, like the
amount of information a channel is *not* going to carry per second … or
at least an upper bound on the information that the channel will carry.
In physics, the second law of thermodynamics helps us understand that
perpetual motion won’t happen. It doesn’t tell us the mechanism by which
a system will lose energy and eventually slow and stop. It doesn’t tell
us when it will happen (unless we start looking at rates of energy to
entropy conversion) it just tells us that it’s going to happen.

I find that intriguing.

But just because I find something intriguing it doesn’t make it true. So
in some sense this will be a course about opinion. And it needs to be
so. Because opinions about entropy vary. We’ll try and capture that
variety as we go. But to try and bring some narrative coherence to the
course, we will explore the variety through a narrative based on my
opinion. That doesn’t make my opinion right, it just makes it the
easiest way to deliver the course.

But why study entropy in Computer Science? Well, while we’re on the
subject of opinion, and bear in mind I’m really a mechanical engineer,
so don’t take my opinion too seriously. If there is any science in
computer science, then it must be the science of how information
evolves. The time it takes to process it and the transformations we can
apply to it. But information theory is often absent from computer
science courses. Why is that? Well because until now we have had to
focus mainly on the techniques and tools we need to use to process data.
On programming, on hardware implementation. On the mechanics of the
information processing. Not the thermodynamics of the information
processing. We have focussed on the mechanisms through which we
transform information, not the fundamental rules by which it is
transformed.

Today things are shifting rapidly, my community has developed tools that
allow the computer to communicate directly in natural language with
humans and program for them. We have finally created computer systems
that can be driven by anyone through natural language, just as my
mechanical engineering predecessors created cars that could be driven by
anyone without having to have an understanding of how a connecting rod
is connected to a crank shaft.

The field is changing rapidly. But students are being taught
thermodynamics today from books that are the same as those it was taught
from 60 years ago.

It’s my instinct that those books provide important theory that tell us
about the limits of computing machines. Today there are many making
promises about superintelligent AGI machines that do all we can do more
rapidly and more capably. A world where cognitive decision making knows
no bounds. This course is built on the following premise. That premise
is that there will be bounds and that those bounds will be dictated by
information theory. Or more precisely, the information dynamics of the
system.

Just as thermodynamics tells us that there is no perpetual motion, I
believe information dynamics will tell us that there are boundaries to
the notion of superintelligence singularity. We won’t achieve those
boundaries in this course, but we will provide a journey through the
landscape of entropy in a way that may persuade you that these are
interesting tools to seek those boundaries with.[2]

When modelling in practice, we often account for this with cost
functions. There’s a funny thing about entropy too. One idea that will
emerge across the course is the notion that probability tells us what a
system will do. But entropy (somehow) tells us what a system won’t do.
For example, we know that perpetual motion machines can’t exist because,
by the second law of thermodynamics, they cause entropy to be produced.
This tells us that if we see a wheel spinning, and we are not injecting
energy into the wheel, eventually the wheel will stop spinning. It
doesn’t tell us the dynamics by which the spinning will stop. It doesn’t
tell us when it will stop. It just tells us that it won’t keep spinning
for ever. If we had the probability distribution that described the
system, that would tell us when the wheel was likely to stop. So somehow
probability tells us what’s going to happen, whereas entropy tells us
what’s not going to happen. We’ll explore this intuition as we go
through the course, but it’s behind the idea why we might look to
entropy to place limits on the performance of an intelligent system.

The second law of thermodynamics doesn’t tell you how to build an
engine. Shannon’s capacity theorem doesn’t tell you how to construct a
code. They tell you what’s not going to happen. This is common among
entropies: they give you prohibition.

[1] That’s not strictly true because I can marginalise things I’m not
interested in … but gives me a challenge of integrating over those
variables which can lead to mathematical (or computational) headaches.

[2] Fortunately, even if I’m totally wrong, the tools are widely
deployed in a number of domains and your efforts in understanding them
will bear fruit in one domain or another.

## ML to Thermodynamics

Last week we how energy has been used in machine learning and how
entropy is applied to probability distributions. By looking at a
stationary distribution with a given energy we could write down the

p_i = \frac{e^{-\beta E_i}}{Z} \quad Z = \sum_i e^{-\beta E_i}.

relationship between this stationary (and equilibrium) distribution and

and Carnot. Today we’re going deeper into Boltzman’s (and Maxwell, Gibbs
and Einstein) microscopic or *ensemble* interpretation of this equation,
and a derivation that that underpins the famous equation $$
U = W + TS
$$ where $U=\left\langle E\right\rangle$ is total energy (or inner
energy), $F$ is the free engergy, $T$ is temperature and $S$ is physical
entropy. Here what we mean by physical entropy, is the translation of
dimensionless energy, $H$, into physical units of Joules per degree
through Boltzmann’s constant, $k_B$,
i.e. $k_B H = -k_B \sum_i p_i \log p_i$.

## Carnot and Clausius

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/clausius-carnot-second-law.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/clausius-carnot-second-law.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Sadi Carnot (1796–1832) asked, in 1824, what limits the efficiency of a
heat engine. Rudolf Clausius (1822–1888) built on Carnot and Kelvin to
state the second law of thermodynamics in several equivalent forms, and
in 1865 he coined the name *entropy* for the state function that tracks
irreversibility. Boltzmann and Gibbs, later in the same century, gave
the microscopic count behind Clausius’s macroscopic $S$. Shannon and
Jaynes, in the twentieth century, reuse the same functional form with
different operational readings. The historical chain is: engines first,
then entropy as a named quantity, then statistics, then information.

Clausius did not give the Boltzmann distribution. He gave the
macroscopic balance that any prescription must respect. When we write
$p_i \propto e^{-\beta E_i}/Z$ and later derive it from MaxEnt, read it
as the statistical answer to a constraint Clausius already framed: fixed
mean energy, maximum entropy, no perpetual motion.

## The Gibbs–Boltzmann Distribution

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/gibbs-distribution.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/gibbs-distribution.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Name the object and fix the formula. The MaxEnt derivation that *earns*
these weights is a separate step: deriving the occupation from Lagrange
multipliers before Shannon $H$ and Jaynes would put the cart before the
horse.

Write the equilibrium occupation of a discrete system with energies
$\{E_i\}$ as $$
p_i = \frac{e^{-\beta E_i}}{Z(\beta)},
\qquad
Z(\beta)=\sum_j e^{-\beta E_j}.
$$ Physicists call $p_i$ the *Boltzmann distribution* (or Boltzmann
weights) and, for a system exchanging energy with a bath at fixed $T$,
the *Gibbs distribution* or *canonical ensemble*. The three names point
at the same formula. The normalisation $Z(\beta)$ is the *partition
function*. Its logarithm $\log Z(\beta)$ is the cumulant generating
function for the energy under this exponential family: derivatives of
$\log Z$ recover the mean energy, the variance (heat capacity, up to
factors of $\beta$), and higher cumulants. That generating-function
reading is why $Z$ is more than a normalisation constant.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
energies = np.array([0.0, 1.0])
beta = np.linspace(0.1, 3.0, 200)
Z = np.sum(np.exp(-beta[:, None] * energies), axis=1)
p0 = np.exp(-beta * energies[0]) / Z
p1 = np.exp(-beta * energies[1]) / Z
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(beta, p0, linewidth=2, label='$p_0$ (ground)')
ax.plot(beta, p1, linewidth=2, label='$p_1$ (excited)')
ax.set_xlabel(r'coldness $\beta$')
ax.set_ylabel('occupation')
ax.legend()
ax.set_title('Two-state Gibbs--Boltzmann occupations')
mlai.write_figure('two-state-boltzmann.svg', directory='./physics')
```

Figure: <i>Occupation of a two-state system as coldness increases. At
low $\beta$ both states are populated; at high $\beta$ the ground state

The Gibbs–Boltzmann occupation is not only a statement about gases and
magnets. Hopfield networks (Hopfield, 1982) assign an energy to every
binary configuration of a recurrent net and treat recall as a descent
toward low-energy states — equilibrium statistics are again Gibbs.
Ackley, Hinton and Sejnowski (Ackley et al., 1985) made the weights of
that energy *learnable*: a Boltzmann machine is an undirected model
whose distribution over configurations is exactly
$p(s)\propto e^{-E(s)/T}$. The 2024 Nobel Prize in Physics, awarded to
John Hopfield and Geoffrey Hinton, recognised that physical-systems
reading of computation and learning. Name the lineage so the formula
does not feel confined to nineteenth-century heat baths; do not divert
into training algorithms.

## Coldness and Temperature

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/coldness-and-temperature.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/coldness-and-temperature.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The Boltzmann occupation can be written in two ways, $$
p_i = \frac{e^{-E_i/k_B T}}{Z}
= \frac{e^{-\beta E_i}}{Z},
\qquad
\beta = \frac{1}{k_B T}.
$$ The mathematics is the same, but the emphasis is different.

$T$ is the variable of the bath. It is what a thermometer reports, and
it is the intensive parameter in the Helmholtz accounting $F = U - TS$.
In that representation energy comes first and entropy is the correction:
$TS$ is the cut the second law takes from $U$.

$\beta$ is *coldness*. It is the variable conjugate to energy. In that
representation entropy comes first: you maximise $S$ (or write a
generating function in $\beta$) and energy is the constraint.
Temperature is then a derived reading,
$T = 1/k_B\beta = (\partial S/\partial U)^{-1}$.

We will use both. Weeks 1–2 keep $T$ in view so the bath is familiar,
and they already compute in $\beta$ because that is the natural argument
of $Z$. Week 5 makes the $\beta$-first order honest: the Lagrange
multiplier on a mean-energy constraint *is* coldness, and it is the
natural parameter of the exponential family. Weeks 6–7 then treat
$\beta$ as a coordinate on that family. A student who still hears
$\beta$ as \`\`one over the thermometer’’ will miss why the geometry is
written in those coordinates.

## Free Energy Decomposition

<!-- SNIPPET: _physics/includes/free-energy-decomposition.md -->
Helmholtz free energy $F=U-TS=-kT\log Z$ accounts for what remains after
the entropy takes its cut: $U$ is total energy, $TS$ is unavailable, $F$
is what the bath still allows you to do.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
energies = np.array([0.0, 1.0, 3.0])
beta = np.linspace(0.05, 2.5, 200)
Z = np.sum(np.exp(-beta[:, None] * energies), axis=1)
p = np.exp(-beta[:, None] * energies) / Z[:, None]
U = (p * energies).sum(axis=1)
S = -np.sum(p * np.log(p + 1e-300), axis=1)
F = U - S / beta
F_check = -np.log(Z) / beta
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(beta, U, label='$U(\\beta)$')
ax.plot(beta, S, label='$S(\\beta)$ (nats)')
ax.plot(beta, F, label='$F(\\beta)$')
ax.plot(beta, F_check, 'k--', alpha=0.5, label='$-\\ln Z/\\beta$')
ax.set_xlabel(r'$\beta$')
ax.legend()
ax.set_title('Three-state system: energies $E\in\{0,1,3\}$')
mlai.write_figure('three-state-thermo.svg', directory='./ml')
```

## From Loss Landscapes to Hamiltonians

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_ml/includes/from-loss-to-hamiltonian.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_ml/includes/from-loss-to-hamiltonian.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Seeing standard energies from machine learning as simply the logarithm
of the probability leads to a form of static energy. It’s only the
“potential energy” of a system. The aim is to find the minimum potential
energy, and that is what we call learning. So what of “kinetic energy.”

We can think of the movement of the system as being the *learning
dynamics*. The training is the movement of the parameters towards the
minimum. That movement could have mass, and therefore momentum and
kinetic energy.

Sometimes this analogy is proposed, like in momentum variables for
stochastic gradient descent, but the analogy seems loose. One place
where it’s tight is *Hamiltonian Monte-Carlo*.

Imagine a mass on a hill. It is sensitive to the pull of gravity as
transmitted to it through the local slope. This is the potential energy,
$E(\mathbf{ w})$, acting as a force. The force is taken by the gradient
of the potential energy which gives the slope, $\nabla E(\mathbf{ w})$.

In gradient descent we take repeated steps opposite the gradient to
reach a local minimum. We consider a little example below where the loss
surface, the energy, is taken from a small regression problem where the
true line has a gradient of $m=1.4$ and an offset of $c=-3.1$. We sample
some data from that line by adding noise with a standard deviation of
$0.15$. Then we show the energy landscape for this system.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
import mlai.plot as plot
```

```
# Toy line: the error surface is the quadratic energy of the residual.
rng = np.random.default_rng(42)
x = np.linspace(0, 1, 25)
m_true, c_true = 1.4, -3.1
y = m_true * x + c_true + 0.15 * rng.standard_normal(x.shape)

m_vals = np.linspace(m_true - 3, m_true + 3, 80)
c_vals = np.linspace(c_true - 3, c_true + 3, 80)
m_grid, c_grid = np.meshgrid(m_vals, c_vals)
E_grid = np.zeros_like(m_grid)
for i in range(m_grid.shape[0]):
    for j in range(m_grid.shape[1]):
        E_grid[i, j] = ((y - m_grid[i, j] * x - c_grid[i, j]) ** 2).sum()

fig, ax = plt.subplots(figsize=(5, 5))
plot.regression_contour(fig, ax, m_vals, c_vals, E_grid)
mlai.write_figure('loss-energy-contour.svg', directory='./ml')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/ml/loss-energy-contour.svg" class="" width="55%" style="vertical-align:middle;">

Figure: <i>Contours of the least-squares energy
$E(m,c)=\sum_n(y_n-mx_n-c)^2$. Steepest descent walks toward the
bowl.</i>

In physics the parameters would be the position, $q$, instead of the
weights, $w$. And typically (for a Hamiltonian) we denote the potential
energy in physics by $V(\cdot)$, so instead of $E(\mathbf{ w})$ we have
$V(\mathbf{q})$.

So in mechanics, potential energy $V(\mathbf{q})$ depends on position,
$q$. Gravity on a landscape is the everyday example mentioned above. But
there are other forms of potential energy, for example the electrostatic
energy of a charge configuration. Or the displacement of a spring.
Gradient descent treats model parameters as a position $\mathbf{q}$ and
the training objective as $V(\mathbf{q})$. The update
$\mathbf{q}\leftarrow\mathbf{q}-\eta\nabla V$ is a discrete, -dominated
limit of a particle sliding on a potential. There is no inertia, there
is only force from $V(\cdot)$.

Modern neural network models train when the number of data points is
very large. This makes computing the exact gradients very
computationally expensive. The energy is typically decomposed as a sum
over data points, $$
E(\mathbf{ w}, \mathbf{X}) = \sum_i E(\mathbf{ w}, \mathbf{x}_i)
$$ where the full data $\mathbf{X} = \{ \mathbf{x}_i\}_{i=1}^n$.

So when $n$ is huge, the sum becomes prophibitive. Stochastic gradient
descent replaces $\nabla V$ by an unbiased estimate from one example or
a mini-batch. The walk is no longer smooth: the particle feels a
jittering force. That is still dynamics on a *potential*; only the force
estimate is random.

The closely related stochastic gradient with Langevin dynamics (SGLD,
Welling and Teh (2011)) makes this connection very explicit. There the
descent is taken to be a descent in a high friction regime which is
subject to *thermal noise*. Thermal noise is the response of the
descending particle to being bumped by particles of a surrounding fluid,
like in Brownian motion. So SGLD is like Brownian motion in a potential
well.

A classical particle has both potential and kinetic energy. Pure
gradient descent never tracks a momentum conjugate to $\mathbf{q}$; each
step forgets velocity. Momentum SGD and Adam reintroduce inertia
heuristically. Hamiltonian mechanics does it properly: introduce
momentum $\mathbf{p}$ and a kinetic energy $T(\mathbf{p})$, usually
$\frac{1}{2}\mathbf{p}^\top M^{-1}\mathbf{p}$.

The Hamiltonian $H$ is the sum of kinetic and potential energies.
Hamilton’s equations are a mainstay of mechanics. They generate a flow
on the joint $(\mathbf{q},\mathbf{p})$ space that conserves $H$ when the
system is isolated. For sampling, one chooses
$V(\mathbf{q})=-\log p(\mathbf{q})$ (up to a constant) so that the
marginal on $\mathbf{q}$ under the Boltzmann weight $e^{-H}$ recovers
the target density $p(\mathbf{q})$.

Hamiltonian Monte Carlo (originally *hybrid* Monte Carlo) uses
fictitious momentum variables so proposals move far in parameter space
while staying near level sets of $H$. The continuous flow cannot be
integrated in closed form for a general $V$, so we discretise.
Discretisation error is corrected by a Metropolis accept/reject on the
Hamiltonian. The method turns gradient information about $V$ into
efficient exploration of high-dimensional densities. The method was
originally developed for Bayesian sampling of neural network weights.

In these systems time is introduced as a “fictitious variable”
representing the evolution of the algorithm and the momentum variables
are known as “auxiliary variables.” The idea is to show that as time
increases the system converges to the stationary distribution associated
with the energy function, $E(\mathbf{ w})$.

The continuous Hamiltonian flow is ideal: it is reversible, preserves
phase-space volume (Liouville), and conserves $H$ exactly. Any numerical
integrator only approximates that flow. Forward Euler is a poor choice
here: it is neither reversible nor volume-preserving, so the Metropolis
correction based on $\Delta H$ fails and almost every proposal is
rejected. Leapfrog — half-step momentum, full-step position, half-step
momentum — is a second-order symplectic integrator. It is exactly
reversible and (approximately) conserves $H$ over many steps. That is
why HMC uses leapfrog rather than an arbitrary ODE solver: the
discretisation must respect the geometric structure that makes the
accept/reject cheap.

This is the confusion the trajectory figure invites. Hamiltonian
dynamics explore a level set of $H$ to *construct a proposal*. The
sample from the target $p(\mathbf{q})\propto e^{-V(\mathbf{q})}$ is the
position $\mathbf{q}'$ *if* the Metropolis test accepts. Reject, and the
chain stays at $\mathbf{q}$. Either way, the green curve on the contour
plot is scaffolding for one move, not a cloud of draws. Fresh momentum
is drawn for the next proposal; the chain of accepted $\mathbf{q}$
values is what estimates posterior expectations.

Figure: <i>Radford M. Neal (University of Toronto).</i>

Radford Neal is a Canadian computer scientist at the University of
Toronto who (along with David MacKay) introduced ideas form Bayesian
statistics into machine learning and neural networks. He introduced the
hybrid Monte Varlo method for training neural networks with
backpropagation (Neal, 1992). The Hamiltonian dynamics act as a proposal
mechanism for posterior sampling over weights. His 1994 thesis (Neal,
1994) is still worth reading today. Alongside the benefits of the HMC
approach he showed how infinite-width nets with suitable priors become
Gaussian processes. Over time the term Hamiltonian was preferred to
hybrid, and this allows us to really see the connection between “energy
as loss” and “energy as the potential in a physical sampler.” Meaning
that our sampling algorithm is being built by an analogy to a
well-understood physical system.

## Hamiltonian Monte Carlo Example

The reusable component lives in `mlai.hmc`. It has diagonal-mass kinetic
energy, leapfrog *proposals*, and a Metropolis correction on $\Delta H$.
Below we first overlay leapfrog paths on a 2D quadratic potential
(standard normal / least-squares bowl) — read the curves as proposal
trajectories, the markers as accepted samples — then contrast an SGD
point estimate with HMC samples for a small logistic regression — the
same energy language as the lecture.

```
# 2D quadratic potential: V(q)=0.5||q||^2  (standard normal target).
# Leapfrog trajectories on the same contour language as regression_contour.
from mlai import HamiltonianMonteCarlo

potential = lambda q: 0.5 * np.dot(q, q)
grad_potential = lambda q: np.asarray(q, dtype=float)

hmc = HamiltonianMonteCarlo(
    potential=potential,
    grad_potential=grad_potential,
    step_size=0.15,
    n_steps=10,
)
result, trajectories = hmc.sample(
    q0=np.zeros(2), n_samples=40, random_state=0, return_trajectory=True
)

q1 = np.linspace(-3, 3, 80)
q2 = np.linspace(-3, 3, 80)
Q1, Q2 = np.meshgrid(q1, q2)
V_grid = 0.5 * (Q1 ** 2 + Q2 ** 2)

fig, ax = plt.subplots(figsize=(5, 5))
plot.hmc_contour_trajectories(
    ax, q1, q2, V_grid,
    trajectories=trajectories[:12],
    samples=result.samples,
    fontsize=16,
)
ax.set_title(f'HMC leapfrog paths (accept rate {result.accept_rate:.2f})')
mlai.write_figure('hmc-quadratic-trajectories.svg', directory='./ml')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/ml/hmc-quadratic-trajectories.svg" class="" width="55%" style="vertical-align:middle;">

Figure: <i>Leapfrog *trajectories* (curves) are proposals, not samples.
Markers are accepted endpoints only. Target
$V(\mathbf{q})=\tfrac12\|\mathbf{q}\|^2$; Metropolis on $\Delta H$ keeps
the chain on $e^{-V}$.</i>

```
# SGD point estimate vs HMC samples for logistic regression.
# Potential V(w) = -log p(w|D) with a flat prior (negative log-likelihood).
from mlai import LR, Basis, linear, HamiltonianMonteCarlo

rng = np.random.default_rng(0)
X = rng.normal(size=(60, 1))
y = (1.5 * X[:, 0] + 0.25 * rng.normal(size=60) > 0).astype(float).reshape(-1, 1)
model = LR(X, y, Basis(linear, number=2))

def potential(q):
    model.parameters = q
    return -float(model.log_likelihood())

def grad_potential(q):
    model.parameters = q
    return np.asarray(model.gradients, dtype=float)

# Point estimate by steepest descent on V
q_sgd = np.zeros(2)
for _ in range(250):
    q_sgd = q_sgd - 0.05 * grad_potential(q_sgd)

hmc = HamiltonianMonteCarlo(
    potential=potential, grad_potential=grad_potential,
    step_size=0.04, n_steps=10,
)
hmc_result = hmc.sample(q0=q_sgd, n_samples=300, random_state=1)

fig, axes = plt.subplots(3, 1, figsize=(6, 5), sharex=True)
plot.hmc_traces(axes, hmc_result, param_indices=[0, 1], fontsize=12)
axes[0].set_title(
    f'Logistic HMC traces (accept {hmc_result.accept_rate:.2f}); '
    f'SGD point = ({q_sgd[0]:.2f}, {q_sgd[1]:.2f})'
)
mlai.write_figure('hmc-logistic-traces.svg', directory='./ml')
print('SGD:', q_sgd, 'HMC mean:', hmc_result.samples.mean(axis=0))
```

## GAIST on Boltzmann and Free Energy

The opening essay, \`\`Why Stochastic Thermodynamics of Machine
Learning?’’ (Welling et al., 2026), states the free-energy decomposition
we use today: entropy measures the information we are missing; we

Chapter 3 is the physics we need for LO1. Section 3.2 introduces the

$$ The first term is heat ($\rho$ changes at fixed Hamiltonian); the

## Thermodynamic Bath and Schottky’s Anomaly

<!-- SNIPPET: _physics/includes/thermodynamic-bath-schottky.md -->
A thermodynamic bath fixes $T$ and exchanges energy with a small system,
justifying the canonical ensemble. Heat capacity
$C=\partial U/\partial T$ peaks when both states are equally populated —
Schottky’s anomaly.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
epsilon = 1.0
beta = np.linspace(0.05, 5.0, 400)
p1 = 1.0 / (1.0 + np.exp(beta * epsilon))
U = p1 * epsilon
C = (epsilon ** 2) * p1 * (1 - p1) * beta ** 2
T = 1.0 / beta
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(T, U, 'C0', linewidth=2)
axes[0].set_xlabel('$T$ ($k=1$)')
axes[0].set_ylabel('$U$')
axes[0].set_title('Mean energy')
axes[1].plot(T, C, 'C1', linewidth=2)
axes[1].set_xlabel('$T$')
axes[1].set_ylabel('$C$')
axes[1].set_title('Schottky heat-capacity peak')
plt.tight_layout()
mlai.write_figure('schottky-two-state.svg', directory='./ml')
```

## Finite Time Costs More Than $\Delta F$

<!-- SNIPPET: _physics/includes/finite-time-dissipation-intro.md -->

Quasi-static processes achieve reversible work $W=\Delta F$. Finite-time
driving dissipates additional energy beyond the equilibrium bound.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
def two_state_F(beta, eps=1.0):
    Z = 1.0 + np.exp(-beta * eps)
    return -np.log(Z) / beta

beta0, beta1 = 0.5, 2.0
path = np.linspace(beta0, beta1, 80)
F_path = two_state_F(path)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(path, F_path, 'k-', linewidth=2, label='$F(\\beta)$')
ax.annotate('', xy=(beta1, two_state_F(beta1)), xytext=(beta0, two_state_F(beta0)),
            arrowprops=dict(arrowstyle='->', color='green', lw=2))
ax.text(1.0, -0.55, 'quasi-static: $W=\\Delta F$', color='green')
ax.annotate('', xy=(beta1, two_state_F(beta1) + 0.15), xytext=(beta0, two_state_F(beta0) + 0.15),
            arrowprops=dict(arrowstyle='->', color='red', lw=2))
ax.text(1.0, -0.35, 'finite time: $W > \\Delta F$', color='red')
ax.set_xlabel(r'$\beta$')
ax.set_ylabel('$F$')
ax.legend()
mlai.write_figure('finite-time-sketch.svg', directory='./ml')
```
