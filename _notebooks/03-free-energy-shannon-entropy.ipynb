## The Information Game

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Before writing a formula for entropy, play a game. One player thinks of
an object drawn from a known finite list. The other may ask only yes/no
questions. How many questions do you need in the worst case? That count
is already a measure of information: it is how much uncertainty you must
remove to name the object.

With a list of size $N$ and equal prior probability on each item, an
ideal strategy halves the search space at every step. After $k$ answers
you can distinguish at most $2^k$ hypotheses, so you need $$
k \ge \log_2 N
$$ questions in the worst case. Twenty questions reach a million
possibilities: $2^{20} = 1\,048\,576$. A *fair* yes/no answer therefore
carries one *bit* of information. That is Shannon’s unit before
Shannon’s formula: information as the number of balanced binary
distinctions required to resolve uncertainty.

If the prior is not uniform, the best question is no longer a literal
half-split of the *count* of remaining objects. It is a half-split of
their *probability*. The average number of questions under optimal play
becomes the Shannon entropy $H = -\sum_i p_i\log_2 p_i$. We will meet
that formula next; the game already shows what the number is counting.

## MacKay’s Balance and Balls

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

David MacKay opens the compression half of *Information Theory,
Inference, and Learning Algorithms* with a weighing puzzle (MacKay,
2003, p. Exercise~4.1). It is the information game with a ternary
instrument instead of a binary question. We use it here only as a
*counting* argument: how much uncertainty we start with, how much each
use of the balance can remove, and why three weighings are enough in
principle — without exhibiting a particular sequence of measurements.

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/mackay-balance-balls.svg" class="" width="80%" style="vertical-align:middle;">

Figure: <i>A two-pan balance and twelve identical balls. Each use of the
balance returns one of three readings: left heavy, balance, or right
heavy.</i>

You are given twelve balls, identical in appearance. Exactly one differs
in weight from the others; you do not know whether that odd ball is
heavier or lighter. A two-pan balance compares two equal-sized groups
and returns one of three readings: left side heavier, right side
heavier, or equal. Design a strategy that finds the odd ball and the
direction of the defect.

Under a uniform prior over those twenty-four possibilities, the Shannon
uncertainty we must remove is $$
H_{\mathrm{start}} = \log_2 24 \approx 4.58 \text{ bits}.
$$ At the end we need a single named hypothesis: $H_{\mathrm{end}} = 0$.
The *deficit* to close is therefore about $4.58$ bits. That difference —
start minus end — is the information the experiment must supply.

A binary question yields at most one bit. The balance is richer: its
alphabet size is three. If the three readings are equiprobable, one
weighing contributes $$
\log_2 3 \approx 1.585 \text{ bits}.
$$ After $k$ weighings the strategy tree has at most $3^k$ leaves, so it
can name at most $3^k$ hypotheses. Compare budgets: $$
\begin{align*}
3^2 &= 9 < 24,\\
3^3 &= 27 \ge 24.
\end{align*}
$$ Two weighings cannot suffice no matter how clever the groupings.
Three weighings *can*: there is spare capacity of three unused leaves
($27-24=3$). In bits, $$
3\log_2 3 \approx 4.75 > \log_2 24 \approx 4.58.
$$ The gap is small. That is why the puzzle feels tight, and why a first
weighing that wastes an outcome (for example putting six against six,
which can never balance if exactly one ball is odd) fails: you have
thrown away ternary resolution and fallen back toward a binary split.

We stop at the existence argument. Constructing the adaptive tree —
which balls on which pan at each step — is MacKay’s exercise, and
solving it is the best way to feel what \`\`informative experiment’’
means. The only design principle we need today is the one already
present in the information game: choose each measurement so that its
outcomes are as nearly equiprobable as possible under the current
posterior. That maximises expected information gain per weighing and is
what makes the $3^3 \ge 24$ budget achievable in practice.

## Coarse-Graining a Die

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/die-coarse-graining.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/die-coarse-graining.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Take a fair six-sided die, $X\in\{1,\ldots,6\}$ with $p(x)=1/6$, so Ask
questions in two stages: first learn which *group* the face falls in,
then resolve the face inside the group.

$$
H(G_{\mathrm{parity}})=1 \text{ bit},
\qquad
H(X\mid G_{\mathrm{parity}})=\log_2 3 \approx 1.585 \text{ bits}
$$

Let $G_{\mathrm{parity}}=0$ on even faces and $1$ on odd. Then
$w=(1/2,1/2)$, so $H(G_{\mathrm{parity}})=1$. Conditional on the group,
three faces remain, each with probability $1/3$, so
$H(X\mid G_{\mathrm{parity}})=\log_2 3$. The stages add:

$$
w=\Bigl(\tfrac16,\tfrac56\Bigr),
\qquad
H(G_{\mathrm{six}})=h_2\Bigl(\tfrac16\Bigr)\approx 0.65 \text{ bits}
$$

Let $G_{\mathrm{six}}=\mathbf{1}_{\{X=6\}}$. The group weights are
unbalanced, $w=(1/6,5/6)$, so the binary entropy of the first answer is
one bit. After a `yes'' the face is known ($H=0$). After a`no’’ five
faces remain ($H=\log_2 5$). Averaging, Again
$H(G_{\mathrm{six}})+H(X\mid G_{\mathrm{six}})=H(X)$. Note that
\`\`$\ge 4$?’’ on a fair die is *also* a balanced cut ($\{1,2,3\}$ vs
$\{4,5,6\}$), so it *is* one bit — the unbalanced example has to be
chosen carefully.

## Grouping Composition

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-grouping-composition.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-grouping-composition.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

$$
H(X)=H(w)+\sum_i w_i\, H(p_i)
$$

Partition the outcomes of $X$ into groups labelled by $i$, with group
probabilities $w_i$ and within-group distributions $p_i$. Then The die
examples are the case check: $H(w)$ is the entropy of the coarse
question; $\sum_i w_i H(p_i)$ is what remains once the group is known.
If $G=g(X)$ is the group label, this identity is exactly the chain rule
because $G$ is a function of $X$ (so $H(G\mid X)=0$) and
$H(X\mid G)=\sum_i w_i H(p_i)$.

In the reverse direction, observing $X$ fixes $G$ with no leftover
uncertainty. The mutual information $I(X;G)=H(G)-H(G\mid X)=H(G)$ is
exactly the price of the coarse question. After you know the group, the
remaining work is $H(X\mid G)$ — which of the fine outcomes inside that
group. Coarse-graining *keeps* $H(G)$ as your description and *discards*
$H(X\mid G)$:

The die paid $\log_2 6$ whichever cut came first. That invariance under
regrouping is not an accident of base-two logarithms; it is part of what
characterises Shannon entropy among candidate uncertainty measures
(**Khinchin-information57?**). Categorical accounts make the same demand
precise (**Baez-entropy11?**). For this lecture it is enough that staged
questions *add*, so coarse-graining has a well-defined price.

## Coarse-Graining as Abstraction

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/coarse-graining-intelligence.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/coarse-graining-intelligence.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Daniel Bernoulli did not integrate Laplace’s curve for every corpuscle.
He replaced the swarm with a coarse mechanical story that still predicts
pressure — exactly keeping a group-level description and discarding
within-group path detail. Thermodynamic macrostates are the same move
with better bookkeeping: the entropy $S$ is the uncertainty still
unresolved *inside* the macrostate. In intelligent systems the hierarchy
is familiar: raw observations are partitioned into features, features
into concepts. At each level you retain $H(w)$ for the labels you use
and lose $\sum_i w_i H(p_i)$ for the distinctions you refuse to make.
Which partition you choose is Jaynes’ and Wigner’s anthropomorphic
choice of degrees of freedom — coming next in this review.

## Static Bookkeeping Versus a Physical Ensemble

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/grouping-static-vs-ensemble.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/grouping-static-vs-ensemble.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The decomposition $H=H(w)+\sum_i w_i H(p_i)$ is pure bookkeeping on a
single distribution. The die roll has no temporal dynamics because we
used the identity only as accounting. Interpreting $p$ as \`\`the
distribution of the system’s state’’ adds a dynamical hypothesis:
something like stationarity plus ergodicity, so that an ensemble average
equals what one trajectory shows over long time. Equilibrium
thermodynamics makes the same move. The equilibrium density is
stationary under the microscopic dynamics, and thermodynamic entropy is
then a time-independent number of the form when each group has relaxed
to its own conditional stationary law $\pi_i$ and only the coarse
weights remain to carry slow change. Dynamics and groups — local
equilibrium, lumpability, broken ergodicity, and the decay of KL —
return in weeks 3–4 and 8.

## ML to Thermodynamics

Week 1 seeded the occupation $p_i \propto e^{-\beta E_i}$; Week 2
stopped at embodiment. We pick up Gibbs and free energy here before
Shannon. From a stationary distribution with given energy we write the
*Gibbs distribution*, $$
p_i = \frac{e^{-\beta E_i}}{Z} \quad Z = \sum_i e^{-\beta E_i}.
$$ Boltzmann came at energy and entropy from the perspective of
thermodynamics, Clausius and Carnot, but today we will show the
relationship between this stationary (and equilibrium) distribution and
the macroscopic formulae of Mayer, Kelvin, Joule, Helmholtz, Clausius
and Carnot. We go deeper into Boltzmann’s (and Maxwell, Gibbs and
Einstein) microscopic or *ensemble* interpretation of this equation, and
a derivation that that underpins the famous equation $$
U = W + TS
$$ where $U=\left\langle E\right\rangle$ is total energy (or inner
energy), $F$ is the free engergy, $T$ is temperature and $S$ is physical
entropy. Here what we mean by physical entropy, is the translation of
dimensionless energy, $H$, into physical units of Joules per degree
through Boltzmann’s constant, $k_B$,
i.e. $k_B H = -k_B \sum_i p_i \log p_i$.

## Carnot and Clausius

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/clausius-carnot-second-law.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/clausius-carnot-second-law.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Sadi Carnot (1796–1832) asked, in 1824, what limits the efficiency of a
heat engine. Rudolf Clausius (1822–1888) built on Carnot and Kelvin to
state the second law of thermodynamics in several equivalent forms, and
in 1865 he coined the name *entropy* for the state function that tracks
irreversibility. Boltzmann and Gibbs, later in the same century, gave
the microscopic count behind Clausius’s macroscopic $S$. Shannon and
Jaynes, in the twentieth century, reuse the same functional form with
different operational readings. The historical chain is: engines first,
then entropy as a named quantity, then statistics, then information.

Clausius did not give the Boltzmann distribution. He gave the
macroscopic balance that any prescription must respect. When we write
$p_i \propto e^{-\beta E_i}/Z$ and later derive it from MaxEnt, read it
as the statistical answer to a constraint Clausius already framed: fixed
mean energy, maximum entropy, no perpetual motion.

## The Gibbs–Boltzmann Distribution

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/gibbs-distribution.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/gibbs-distribution.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Name the object and fix the formula. The MaxEnt derivation that *earns*
these weights is a separate step: deriving the occupation from Lagrange
multipliers before Shannon $H$ and Jaynes would put the cart before the
horse.

Write the equilibrium occupation of a discrete system with energies
$\{E_i\}$ as $$
p_i = \frac{e^{-\beta E_i}}{Z(\beta)},
\qquad
Z(\beta)=\sum_j e^{-\beta E_j}.
$$ Physicists call $p_i$ the *Boltzmann distribution* (or Boltzmann
weights) and, for a system exchanging energy with a bath at fixed $T$,
the *Gibbs distribution* or *canonical ensemble*. The three names point
at the same formula. The normalisation $Z(\beta)$ is the *partition
function*. Its logarithm $\log Z(\beta)$ is the cumulant generating
function for the energy under this exponential family: derivatives of
$\log Z$ recover the mean energy, the variance (heat capacity, up to
factors of $\beta$), and higher cumulants. That generating-function
reading is why $Z$ is more than a normalisation constant.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
energies = np.array([0.0, 1.0])
beta = np.linspace(0.1, 3.0, 200)
Z = np.sum(np.exp(-beta[:, None] * energies), axis=1)
p0 = np.exp(-beta * energies[0]) / Z
p1 = np.exp(-beta * energies[1]) / Z
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(beta, p0, linewidth=2, label='$p_0$ (ground)')
ax.plot(beta, p1, linewidth=2, label='$p_1$ (excited)')
ax.set_xlabel(r'coldness $\beta$')
ax.set_ylabel('occupation')
ax.legend()
ax.set_title('Two-state Gibbs--Boltzmann occupations')
mlai.write_figure('two-state-boltzmann.svg', directory='./physics')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/physics/two-state-boltzmann.svg" class="" width="75%" style="vertical-align:middle;">

Figure: <i>Occupation of a two-state system as coldness increases. At
low $\beta$ both states are populated; at high $\beta$ the ground state

The Gibbs–Boltzmann occupation is not only a statement about gases and
magnets. Hopfield networks (Hopfield, 1982) assign an energy to every
binary configuration of a recurrent net and treat recall as a descent
toward low-energy states — equilibrium statistics are again Gibbs.
Ackley, Hinton and Sejnowski (Ackley et al., 1985) made the weights of
that energy *learnable*: a Boltzmann machine is an undirected model
whose distribution over configurations is exactly
$p(s)\propto e^{-E(s)/T}$. The 2024 Nobel Prize in Physics, awarded to
John Hopfield and Geoffrey Hinton, recognised that physical-systems
reading of computation and learning. Name the lineage so the formula
does not feel confined to nineteenth-century heat baths; do not divert
into training algorithms.

## Coldness and Temperature

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/coldness-and-temperature.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/coldness-and-temperature.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The Boltzmann occupation can be written in two ways, $$
p_i = \frac{e^{-E_i/k_B T}}{Z}
= \frac{e^{-\beta E_i}}{Z},
\qquad
\beta = \frac{1}{k_B T}.
$$ The mathematics is the same, but the emphasis is different.

$T$ is the variable of the bath. It is what a thermometer reports, and
it is the intensive parameter in the Helmholtz accounting $F = U - TS$.
In that representation energy comes first and entropy is the correction:
$TS$ is the cut the second law takes from $U$.

$\beta$ is *coldness*. It is the variable conjugate to energy. In that
representation entropy comes first: you maximise $S$ (or write a
generating function in $\beta$) and energy is the constraint.
Temperature is then a derived reading,
$T = 1/k_B\beta = (\partial S/\partial U)^{-1}$.

We will use both. Weeks 1–2 keep $T$ in view so the bath is familiar,
and they already compute in $\beta$ because that is the natural argument
of $Z$. Week 5 makes the $\beta$-first order honest: the Lagrange
multiplier on a mean-energy constraint *is* coldness, and it is the
natural parameter of the exponential family. Weeks 6–7 then treat
$\beta$ as a coordinate on that family. A student who still hears
$\beta$ as \`\`one over the thermometer’’ will miss why the geometry is
written in those coordinates.

## Free Energy Decomposition

<!-- SNIPPET: _physics/includes/free-energy-decomposition.md -->
Helmholtz free energy $F=U-TS=-kT\log Z$ accounts for what remains after
the entropy takes its cut: $U$ is total energy, $TS$ is unavailable, $F$
is what the bath still allows you to do.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
energies = np.array([0.0, 1.0, 3.0])
beta = np.linspace(0.05, 2.5, 200)
Z = np.sum(np.exp(-beta[:, None] * energies), axis=1)
p = np.exp(-beta[:, None] * energies) / Z[:, None]
U = (p * energies).sum(axis=1)
S = -np.sum(p * np.log(p + 1e-300), axis=1)
F = U - S / beta
F_check = -np.log(Z) / beta
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(beta, U, label='$U(\\beta)$')
ax.plot(beta, S, label='$S(\\beta)$ (nats)')
ax.plot(beta, F, label='$F(\\beta)$')
ax.plot(beta, F_check, 'k--', alpha=0.5, label='$-\\ln Z/\\beta$')
ax.set_xlabel(r'$\beta$')
ax.legend()
ax.set_title('Three-state system: energies $E\in\{0,1,3\}$')
mlai.write_figure('three-state-thermo.svg', directory='./ml')
```

## From Loss Landscapes to Hamiltonians

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_ml/includes/from-loss-to-hamiltonian.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_ml/includes/from-loss-to-hamiltonian.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Seeing standard energies from machine learning as simply the logarithm
of the probability leads to a form of static energy. It’s only the
“potential energy” of a system. The aim is to find the minimum potential
energy, and that is what we call learning. So what of “kinetic energy”.

We can think of the movement of the system as being the *learning
dynamics*. The training is the movement of the parameters towards the
minimum. That movement could have mass, and therefore momentum and
kinetic energy.

Sometimes this analogy is proposed, like in momentum variables for
stochastic gradient descent, but the analogy seems loose. One place
where it’s tight is *Hamiltonian Monte-Carlo*.

Imagine a mass on a hill. It is sensitive to the pull of gravity as
transmitted to it through the local slope. This is the potential energy,
$E(\mathbf{ w})$, acting as a force. The force is taken by the gradient
of the potential energy which gives the slope, $\nabla E(\mathbf{ w})$.

In gradient descent we take repeated steps opposite the gradient to
reach a local minimum. We consider a little example below where the loss
surface, the energy, is taken from a small regression problem where the
true line has a gradient of $m=1.4$ and an offset of $c=-3.1$. We sample
some data from that line by adding noise with a standard deviation of
$0.15$. Then we show the energy landscape for this system.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
import mlai.plot as plot
```

In physics the parameters would be the position, $q$, instead of the
weights, $w$. And typically (for a Hamiltonian) we denote the potential
energy in physics by $V(\cdot)$, so instead of $E(\mathbf{ w})$ we have
$V(\mathbf{q})$.

So in mechanics, potential energy $V(\mathbf{q})$ depends on position,
$q$. Gravity on a landscape is the everyday example mentioned above. But
there are other forms of potential energy, for example the electrostatic
energy of a charge configuration. Or the displacement of a spring.
Gradient descent treats model parameters as a position $\mathbf{q}$ and
the training objective as $V(\mathbf{q})$. The update
$\mathbf{q}\leftarrow\mathbf{q}-\eta\nabla V$ is a discrete, -dominated
limit of a particle sliding on a potential. There is no inertia, there
is only force from $V(\cdot)$.

Modern neural network models train when the number of data points is
very large. This makes computing the exact gradients very
computationally expensive. The energy is typically decomposed as a sum
over data points, $$
E(\mathbf{ w}, \mathbf{X}) = \sum_i E(\mathbf{ w}, \mathbf{x}_i)
$$ where the full data $\mathbf{X} = \{ \mathbf{x}_i\}_{i=1}^n$.

So when $n$ is huge, the sum becomes prophibitive. Stochastic gradient
descent replaces $\nabla V$ by an unbiased estimate from one example or
a mini-batch. The walk is no longer smooth: the particle feels a
jittering force. That is still dynamics on a *potential*; only the force
estimate is random.

The closely related stochastic gradient with Langevin dynamics (SGLD,
Welling and Teh (2011)) makes this connection very explicit. There the
descent is taken to be a descent in a high friction regime which is
subject to *thermal noise*. Thermal noise is the response of the
descending particle to being bumped by particles of a surrounding fluid,
like in Brownian motion. So SGLD is like Brownian motion in a potential
well.

A classical particle has both potential and kinetic energy. Pure
gradient descent never tracks a momentum conjugate to $\mathbf{q}$; each
step forgets velocity. Momentum SGD and Adam reintroduce inertia
heuristically. Hamiltonian mechanics does it properly: introduce
momentum $\mathbf{p}$ and a kinetic energy $T(\mathbf{p})$, usually
$\frac{1}{2}\mathbf{p}^\top M^{-1}\mathbf{p}$.

The Hamiltonian $H$ is the sum of kinetic and potential energies.
Hamilton’s equations are a mainstay of mechanics. They generate a flow
on the joint $(\mathbf{q},\mathbf{p})$ space that conserves $H$ when the
system is isolated. For sampling, one chooses
$V(\mathbf{q})=-\log p(\mathbf{q})$ (up to a constant) so that the
marginal on $\mathbf{q}$ under the Boltzmann weight $e^{-H}$ recovers
the target density $p(\mathbf{q})$.

Hamiltonian Monte Carlo (originally *hybrid* Monte Carlo) uses
fictitious momentum variables so proposals move far in parameter space
while staying near level sets of $H$. The continuous flow cannot be
integrated in closed form for a general $V$, so we discretise.
Discretisation error is corrected by a Metropolis accept/reject on the
Hamiltonian. The method turns gradient information about $V$ into
efficient exploration of high-dimensional densities. The method was
originally developed for Bayesian sampling of neural network weights.

In these systems time is introduced as a “fictitious variable”
representing the evolution of the algorithm and the momentum variables
are known as “auxiliary variables”. The idea is to show that as time
increases the system converges to the stationary distribution associated
with the energy function, $E(\mathbf{ w})$.

The continuous Hamiltonian flow is ideal: it is reversible, preserves
phase-space volume (Liouville), and conserves $H$ exactly. Any numerical
integrator only approximates that flow. Forward Euler is a poor choice
here: it is neither reversible nor volume-preserving, so the Metropolis
correction based on $\Delta H$ fails and almost every proposal is
rejected. Leapfrog — half-step momentum, full-step position, half-step
momentum — is a second-order symplectic integrator. It is exactly
reversible and (approximately) conserves $H$ over many steps. That is
why HMC uses leapfrog rather than an arbitrary ODE solver: the
discretisation must respect the geometric structure that makes the
accept/reject cheap.

This is the confusion the trajectory figure invites. Hamiltonian
dynamics explore a level set of $H$ to *construct a proposal*. The
sample from the target $p(\mathbf{q})\propto e^{-V(\mathbf{q})}$ is the
position $\mathbf{q}'$ *if* the Metropolis test accepts. Reject, and the
chain stays at $\mathbf{q}$. Either way, the green curve on the contour
plot is scaffolding for one move, not a cloud of draws. Fresh momentum
is drawn for the next proposal; the chain of accepted $\mathbf{q}$
values is what estimates posterior expectations.

Radford Neal is a Canadian computer scientist at the University of
Toronto who (along with David MacKay) introduced ideas form Bayesian
statistics into machine learning and neural networks. He introduced the
hybrid Monte Varlo method for training neural networks with
backpropagation (Neal, 1992). The Hamiltonian dynamics act as a proposal

## Hamiltonian Monte Carlo Example

The reusable component lives in `mlai.hmc`. It has diagonal-mass kinetic
energy, leapfrog *proposals*, and a Metropolis correction on $\Delta H$.
Below we first overlay leapfrog paths on a 2D quadratic potential
(standard normal / least-squares bowl) — read the curves as proposal
trajectories, the markers as accepted samples — then contrast an SGD
point estimate with HMC samples for a small logistic regression — the
same energy language as the lecture.

```
# 2D quadratic potential: V(q)=0.5||q||^2  (standard normal target).
# Leapfrog trajectories on the same contour language as regression_contour.
from mlai import HamiltonianMonteCarlo

potential = lambda q: 0.5 * np.dot(q, q)
grad_potential = lambda q: np.asarray(q, dtype=float)

hmc = HamiltonianMonteCarlo(
    potential=potential,
    grad_potential=grad_potential,
    step_size=0.15,
    n_steps=10,
)
result, trajectories = hmc.sample(
    q0=np.zeros(2), n_samples=40, random_state=0, return_trajectory=True
)

q1 = np.linspace(-3, 3, 80)
q2 = np.linspace(-3, 3, 80)
Q1, Q2 = np.meshgrid(q1, q2)
V_grid = 0.5 * (Q1 ** 2 + Q2 ** 2)

fig, ax = plt.subplots(figsize=(5, 5))
plot.hmc_contour_trajectories(
    ax, q1, q2, V_grid,
    trajectories=trajectories[:12],
    samples=result.samples,
    fontsize=16,
)
ax.set_title(f'HMC leapfrog paths (accept rate {result.accept_rate:.2f})')
mlai.write_figure('hmc-quadratic-trajectories.svg', directory='./ml')
```

Figure: <i>Leapfrog *trajectories* (curves) are proposals, not samples.
Markers are accepted endpoints only. Target
$V(\mathbf{q})=\tfrac12\|\mathbf{q}\|^2$; Metropolis on $\Delta H$ keeps
the chain on $e^{-V}$.</i>

```
# SGD point estimate vs HMC samples for logistic regression.
# Potential V(w) = -log p(w|D) with a flat prior (negative log-likelihood).
from mlai import LR, Basis, linear, HamiltonianMonteCarlo

rng = np.random.default_rng(0)
X = rng.normal(size=(60, 1))
y = (1.5 * X[:, 0] + 0.25 * rng.normal(size=60) > 0).astype(float).reshape(-1, 1)
model = LR(X, y, Basis(linear, number=2))

def potential(q):
    model.parameters = q
    return -float(model.log_likelihood())

def grad_potential(q):
    model.parameters = q
    return np.asarray(model.gradients, dtype=float)

# Point estimate by steepest descent on V
q_sgd = np.zeros(2)
for _ in range(250):
    q_sgd = q_sgd - 0.05 * grad_potential(q_sgd)

hmc = HamiltonianMonteCarlo(
    potential=potential, grad_potential=grad_potential,
    step_size=0.04, n_steps=10,
)
hmc_result = hmc.sample(q0=q_sgd, n_samples=300, random_state=1)

fig, axes = plt.subplots(3, 1, figsize=(6, 5), sharex=True)
plot.hmc_traces(axes, hmc_result, param_indices=[0, 1], fontsize=12)
axes[0].set_title(
    f'Logistic HMC traces (accept {hmc_result.accept_rate:.2f}); '
    f'SGD point = ({q_sgd[0]:.2f}, {q_sgd[1]:.2f})'
)
mlai.write_figure('hmc-logistic-traces.svg', directory='./ml')
print('SGD:', q_sgd, 'HMC mean:', hmc_result.samples.mean(axis=0))
```

Claude Shannon was an American mathematician. He developed his notion of

mind perhaps the most important conceptual contribution from Shannon.
It’s important because it makes information *fungible*. That allows us
to compare information transfer rates across different types of
processing entities.

See Lawrence (2024) p. . See Lawrence (2024) p. .

A typical human, when speaking, shares information at around 2,000 bits
per minute. Two machines will share information at 600 billion bits per
minute. In other words, machines can share information 300 million times
faster than us. This is equivalent to us traveling at walking pace, and
the machine traveling at the speed of light.

From this perspective, machine decision-making belongs to an utterly
different realm to that of humans. Consideration of the relative merits
of the two needs to take these differences into account. This difference
between human and machine underpins the revolution in algorithmic

## Arithmetic Coding

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/arithmetic-coding.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/arithmetic-coding.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

In Chapter 6 of *Information Theory, Inference, and Learning Algorithms*
(MacKay, 2003), MacKay introduces arithmetic coding with a clear
philosophy: *compression of data from a source entails probabilistic
modelling of that source*. The encoding operation is then almost an
afterthought — once you can predict, you can code.

### The guessing game

Section 6.1 motivates the idea with a guessing game. An English speaker
tries to predict the next character of a text; after each correct guess
we record *how many* attempts it took. The sequence of guess-counts is
highly skewed toward 1 and 2, so it compresses easily. Decoding needs an
*identical twin* who makes the same sequence of guesses: stop them after
the recorded number of attempts and they land on the right letter.

The game demonstrates two design principles we will keep: (i) a
*time-varying* predictive mapping is allowed — the predictor may use as
much context as it likes; (ii) encoder and decoder must share an
identical model.

### From guesses to intervals

Section 6.2 replaces the human by a program that, given the string so
far, returns a predictive distribution $\{p_i\}$ over the next symbol.
Arithmetic coding turns that distribution into nested intervals on
$[0,1)$ (Figures 6.1–6.2). A binary transmission itself defines an
interval — the string `01` is the dyadic interval $[0.25, 0.50)$ — so
encoding means: find a bit-string whose interval sits *inside* the
interval that the model assigned to the message.

MacKay’s Algorithm 6.3 is the iterative form. Write $Q_n$ and $R_n$ for
the lower and upper cumulative predictive probabilities. Starting from
$u=0$, $v=1$, $p=v-u$, each symbol updates $$
v \leftarrow u + p\,R_n(x_n\mid x_{<n}),\qquad
u \leftarrow u + p\,Q_n(x_n\mid x_{<n}),\qquad
p \leftarrow v-u.
$$ The Shannon information content of the string is
$h(x\mid H)=\log[1/P(x\mid H)]$; Exercise 6.1 shows the coded length
stays within two bits of that ideal.

First the predictive model. Arithmetic coding does not prescribe *how*
predictions are made — only that encoder and decoder agree. For Dasher
we use a unigram–bigram blend
$P(c \mid c') \propto w\,P_{\mathrm{bi}}(c \mid c') + (1-w)\,P_{\mathrm{uni}}(c)$.

```
DEFAULT_CHARS = "abcdefghijklmnopqrstuvwxyz "
DEFAULT_BIGRAM_WEIGHT = 0.82

def normalise_text(text, chars=DEFAULT_CHARS, map_other_to_space=True):
    """Lower-case onto the alphabet; newlines/tabs → space."""
    alpha = set(chars)
    out = []
    for raw in text:
        ch = raw.lower()
        if ch in alpha:
            out.append(ch)
        elif map_other_to_space and ch in "\n\r\t ":
            if " " in alpha:
                out.append(" ")
    return "".join(out)

class CharModel:
    """Unigram + bigram blend (same predictive form as Dasher)."""

    def __init__(self, chars=DEFAULT_CHARS, bigram_weight=DEFAULT_BIGRAM_WEIGHT,
                 uni=None, bi=None):
        self.chars = chars
        self.bigram_weight = bigram_weight
        self.uni = uni or {}
        self.bi = bi or {}

    def predict(self, context):
        """P(c | last character of context) over self.chars."""
        last = context[-1].lower() if context else ""
        row = self.bi.get(last, {})
        w = self.bigram_weight
        out = {}
        for ch in self.chars:
            bp = row.get(ch, 0.0)
            up = self.uni.get(ch, 1e-4)
            out[ch] = w * bp + (1.0 - w) * up
        z = sum(out.values())
        if z <= 0:
            u = 1.0 / len(self.chars)
            return {ch: u for ch in self.chars}
        return {ch: p / z for ch, p in out.items()}

    def entropy_rate(self, context=""):
        """H(next | context) in bits."""
        probs = self.predict(context)
        return -sum(p * math.log2(p) for p in probs.values() if p > 0)

    def to_dasher_dict(self):
        """JSON object for Dasher.setLanguageModel / dasher-lm.json."""
        return {
            "chars": self.chars,
            "bigramWeight": self.bigram_weight,
            "uni": {ch: self.uni.get(ch, 0.0) for ch in self.chars},
            "bi": {prev: {ch: float(p) for ch, p in row.items()}
                   for prev, row in self.bi.items()},
        }

def train_from_text(text, chars=DEFAULT_CHARS,
                    bigram_weight=DEFAULT_BIGRAM_WEIGHT, add_k=0.0):
    """Count characters → CharModel (MacKay: the predictor is separate)."""
    data = normalise_text(text, chars)
    if not data:
        raise ValueError("training text contains no alphabet characters")
    uni_counts = {ch: add_k for ch in chars}
    bi_counts = defaultdict(lambda: {ch: add_k for ch in chars})
    prev = None
    for ch in data:
        uni_counts[ch] += 1.0
        if prev is not None:
            bi_counts[prev][ch] += 1.0
        prev = ch
    n = sum(uni_counts.values())
    uni = {ch: uni_counts[ch] / n for ch in chars}
    bi = {}
    for prev_ch, row in bi_counts.items():
        total = sum(row.values())
        if total <= 0:
            continue
        sparse = {ch: row[ch] / total for ch in chars if row[ch] > 0}
        if sparse:
            bi[prev_ch] = sparse
    return CharModel(chars=chars, bigram_weight=bigram_weight, uni=uni, bi=bi)

def write_dasher_lm(model, path):
    """Write dasher-lm.json for the visualiser."""
    with open(path, "w", encoding="utf-8") as f:
        json.dump(model.to_dasher_dict(), f, indent=2)
        f.write("\n")
```

```
corpus = """
Alice was beginning to get very tired of sitting by her sister on the bank,
and of having nothing to do: once or twice she had peeped into the book her
sister was reading, but it had no pictures or conversations in it.
"""
model = train_from_text(corpus, bigram_weight=0.82)
print(f"H(next | ∅)    = {model.entropy_rate(''):.3f} bits/char")
print(f"H(next | 't')  = {model.entropy_rate('t'):.3f} bits/char")
print(f"H(next | 'th') = {model.entropy_rate('th'):.3f} bits/char")
```

After `th` the predictive mass concentrates on `e`, so the conditional
entropy drops. That drop *is* the tall `e` box in Dasher.

The helpers below follow MacKay’s Algorithm 6.3 and Figure 6.1: narrow
$[u,v)\subseteq[0,1)$, then pick a dyadic bit-string inside it. The
decoder is the identical twin — same predictions, ask which symbol
interval contains the number defined by the bits.

```
def cumulative_QR(probs, chars):
    """MacKay's Q_n (lower) and R_n (upper) cumulatives."""
    Q, R = {}, {}
    cum = 0.0
    for ch in chars:
        Q[ch] = cum
        cum += probs.get(ch, 0.0)
        R[ch] = cum
    if chars:
        R[chars[-1]] = 1.0
    return Q, R

def encode_interval(message, model):
    """Algorithm 6.3 → (u, v, ideal_bits).  Width v−u = P(message|model)."""
    u, v = 0.0, 1.0
    p = v - u
    ideal = 0.0
    for i, ch in enumerate(message):
        probs = model.predict(message[:i])
        pc = probs[ch]
        if pc <= 0:
            raise ValueError(f"zero probability for {ch!r}")
        ideal -= math.log2(pc)
        Q, R = cumulative_QR(probs, model.chars)
        v = u + p * R[ch]
        u = u + p * Q[ch]
        p = v - u
    return u, v, ideal

def tag_bits(u, v):
    """Bit-string whose dyadic interval lies inside [u, v) (Fig. 6.1)."""
    target = 0.5 * (u + v)
    lo, hi = 0.0, 1.0
    bits = []
    for _ in range(64):
        if lo >= u and hi <= v:
            break
        mid = 0.5 * (lo + hi)
        if target < mid:
            bits.append("0")
            hi = mid
        else:
            bits.append("1")
            lo = mid
    return "".join(bits)

def encode(message, model):
    """Return (bitstring, ideal_bits)."""
    if not message:
        return "", 0.0
    u, v, ideal = encode_interval(message, model)
    return tag_bits(u, v), ideal

def decode(bitstring, length, model):
    """Identical-twin decoder: same model, recover `length` symbols."""
    value = 0.0
    scale = 0.5
    for b in bitstring:
        if b == "1":
            value += scale
        scale *= 0.5
    value += 0.5 * scale  # midpoint of the dyadic interval
    u, v = 0.0, 1.0
    out = []
    for _ in range(length):
        p = v - u
        probs = model.predict("".join(out))
        Q, R = cumulative_QR(probs, model.chars)
        chosen = None
        for ch in model.chars:
            lo = u + p * Q[ch]
            hi = u + p * R[ch]
            if lo <= value < hi or (ch == model.chars[-1] and lo <= value <= hi):
                chosen = ch
                u, v = lo, hi
                break
        if chosen is None:
            raise ValueError("value outside all symbol bins")
        out.append(chosen)
    return "".join(out)

def interval_trace(message, model):
    """Nested [u, v) after each symbol — Fig. 6.2 / Dasher zoom as data."""
    u, v = 0.0, 1.0
    p = v - u
    steps = []
    for i, ch in enumerate(message):
        probs = model.predict(message[:i])
        Q, R = cumulative_QR(probs, model.chars)
        v = u + p * R[ch]
        u = u + p * Q[ch]
        p = v - u
        steps.append({
            "char": ch, "p": probs[ch],
            "bits": -math.log2(probs[ch]),
            "u": u, "v": v, "width": p,
        })
    return steps
```

```
message = normalise_text("the rabbit")
u, v, ideal_bits = encode_interval(message, model)
bitstring, _ = encode(message, model)
recovered = decode(bitstring, len(message), model)
print(f"message:   {message!r}")
print(f"[u, v)  = [{u:.6f}, {v:.6f})  width={v-u:.3e}")
print(f"ideal:     {ideal_bits:.3f} bits  "
      f"({ideal_bits / len(message):.3f} bits/char)")
print(f"bitstring: {bitstring}  ({len(bitstring)} bits)")
print(f"decoded:   {recovered!r}  ok={recovered == message}")
for step in interval_trace(message, model)[:4]:
    print(f"  '{step['char']}': p={step['p']:.3f}  "
          f"bits={step['bits']:.2f}  "
          f"[u,v)=[{step['u']:.4f},{step['v']:.4f})")
```

```
uniform = len(message) * math.log2(len(model.chars))
print(f"uniform: {uniform:.3f} bits  "
      f"({uniform / len(message):.3f} bits/char)")
print(f"saving:  {uniform - ideal_bits:.3f} bits vs uniform")
```

Compare with a uniform code over the same alphabet. A model that knows
English beats uniform — and the gap is exactly what good prediction buys
you.

Dasher (Ward et al., 2000; **MacKay-dasher98?**) *draws* Algorithm 6.3:
each letter is a vertical interval whose height is its predictive

## The Partition Function

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/partition-function-generating.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/partition-function-generating.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The partition function $Z(\beta)=\sum_i e^{-\beta E_i}$ is a generating
function: $U=-\partial_\beta\log Z$, $F=-\beta^{-1}\log Z$,
$S=\beta(U-F)$. The canonical ensemble is an equilibrium construction;
driving in finite time leaves that manifold.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
eps = 1.0
beta = np.linspace(0.1, 3.0, 300)
Z = 1.0 + np.exp(-beta * eps)
U = eps / (1.0 + np.exp(beta * eps))
F = -np.log(Z) / beta
S = beta * (U - F)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(beta, U, label='$U$')
ax.plot(beta, S, label='$S$')
ax.plot(beta, F, label='$F$')
ax.set_xlabel(r'$\beta$')
ax.legend()
ax.set_title('Two-state system')
mlai.write_figure('two-state-partition.svg', directory='./ml')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/ml/two-state-partition.svg" class="" width="75%" style="vertical-align:middle;">

Figure: <i>Thermodynamic quantities from $Z(\beta)$ for the two-state
system used as a running example.</i>

```
import numpy as np

def partition(energies, beta):
    return np.sum(np.exp(-beta * np.asarray(energies)))

def thermo_from_Z(beta, energies):
    e = np.asarray(energies, dtype=float)
    Z = partition(e, beta)
    p = np.exp(-beta * e) / Z
    U = np.sum(p * e)
    F = -np.log(Z) / beta
    S = beta * (U - F)
    return U, S, F
```

## GAIST on Entropy and the Partition Function

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/welling-entropy-partition.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/welling-entropy-partition.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Section 1.2.3 of (Welling et al., 2026) defines Shannon entropy as
expected surprisal, $\mathrm{S}[p] = -\int p\log p$, in nats. It
computes the Gaussian entropy and shows that it depends only on
$\sigma$, not on $\mu$. Section 1.2.4 introduces KL as relative entropy,
the information lost when $q$ stands in for $p$, and records the
chain-rule decomposition of KL. That is useful scaffolding for week 6;
it is not Shannon’s chain rule for $H$.

Section 3.2.4 is the generating-function half of today’s lecture: from
$Z$ to mean energy, entropy, and free energy. Read those two sections.
Do not look here for the Shannon axioms, for $S = kH$, or for channel
capacity. Those are MacKay and Shannon (1948). GAIST treats entropy as a
property of a distribution, not as a no-go on a code.

GAIST §1.2.3 uses $-\int p\log p$ for Gaussians — that is *differential*
entropy. It can be negative and is not bounded like discrete Shannon
$H\in[0,\log n]$. Below we introduce $\text{KL}\left( p\,\|\,q \right)$,
which is always $\ge 0$ discrete and continuous. That is why MaxEnt
projections minimise $\text{KL}\left( \cdot\,\|\,r \right)$, not raw
$H$.

## Discrete Shannon Entropy Is Bounded

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/kl-divergence-discrete-continuous.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/kl-divergence-discrete-continuous.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

KL divergence $\text{KL}\left( \cdot\,\|\,\cdot \right)$ measures
information lost when $q$ is used instead of $p$. It is non-negative in
both discrete and continuous settings (with common support). Unlike
Shannon entropy, it is defined relative to a reference. MaxEnt and
$m$-projections minimise $\text{KL}\left( \cdot\,\|\,r \right)$; that is
why week 4’s die optimisation used $\sum_i p_i\log(p_i/r_i)$.

For $\mathcal{N}(0,\sigma^2)$, differential entropy is
$\frac{1}{2}\log(2\pi e\sigma^2)$ in nats. As $\sigma\to 0$ it diverges
negatively. Discrete Shannon entropy stays in $[0,\log n]$.
Thermodynamic $S=kH$ in week 1 used the discrete sum form on Boltzmann
probabilities. Continuous MaxEnt still works because constraints fix
scale; comparing to a reference uses KL, which remains non-negative.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
p = np.linspace(0.01, 0.99, 200)
H_bin = -p*np.log2(p) - (1-p)*np.log2(1-p)
sigma = np.linspace(0.15, 3.0, 200)
h_diff = 0.5 * np.log(2 * np.pi * np.e * sigma**2)
fig, axes = plt.subplots(1, 2, figsize=(9, 3.5))
axes[0].plot(p, H_bin, 'k-', linewidth=2)
axes[0].axhline(1.0, color='gray', linestyle='--', alpha=0.6, label='max ($n=2$)')
axes[0].set_xlabel('$p$'); axes[0].set_ylabel('$H$ (bits)')
axes[0].set_title('Discrete: $0 \\le H \\le \\log n$')
axes[0].legend(fontsize=8)
axes[1].plot(sigma, h_diff, 'C1', linewidth=2)
axes[1].axhline(0, color='gray', linestyle='--', alpha=0.6)
axes[1].set_xlabel('$\\sigma$'); axes[1].set_ylabel('$h$ (nats)')
axes[1].set_title('Gaussian differential entropy')
fig.tight_layout()
mlai.write_figure('entropy-bounded-vs-differential.svg', directory='./ml')
```

## Maxwell–Boltzmann: Components Versus Speed

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-boltzmann-components-vs-speed.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-boltzmann-components-vs-speed.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Write the joint density on Cartesian velocities as isotropic, Each
marginal $f(v_x)$ is Gaussian. Change to spherical coordinates with
$v=|\mathbf{v}|$ and solid angle $\Omega$. Probability conservation uses
the volume element $d^3v = v^2\,dv\,d\Omega$. Integrating over angles
($\int d\Omega = 4\pi$) gives the Maxwell speed distribution The factor
$v^2$ is the Jacobian of the change of variables — the density of states
of spherical shells in velocity space. Calling \`\`Maxwell–Boltzmann’’
Gaussian is correct for a component and wrong for speed.

$$
f(v_x)=\sqrt{\tfrac{m}{2\pi kT}}\,e^{-mv_x^2/2kT}
\qquad\text{versus}\qquad
f(v)\propto v^2\,e^{-mv^2/2kT}
$$

## How Differential Entropy Transforms

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/differential-entropy-change-of-variables.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/differential-entropy-change-of-variables.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Define differential entropy $h(X)=-\int f_X(x)\ln f_X(x)\,dx$. For a
smooth invertible map $y=g(x)$, probability conservation requires
$f_Y(y)\,|dy|=f_X(x)\,|dx|$, so Substitute into $h(Y)$ and change the
integration variable back to $x$: The shift is the average log-Jacobian.

$$
h(Y)=h(X)+\mathbb{E}\bigl[\ln|g'(X)|\bigr]
$$

For a linear map $y=ax+b$ with $a\neq 0$, the formula collapses to
$h(aX)=h(X)+\ln|a|$. Because a density $f$ has units of inverse length
(or inverse speed), $\ln f$ is the log of a dimensionful quantity. That
is a symptom that differential entropy secretly depends on a reference
scale. Even the claim that a Gaussian maximises entropy at fixed
variance is relative to the variable you chose: fixing
$\mathrm{Var}(v_x)$ and fixing $\mathrm{Var}(v_x^3)$ yield different
maximisers.

With scale $a=\sqrt{kT/m}$, a single Gaussian component has
$h=\ln(a\sqrt{2\pi})+\tfrac12$. The Maxwell speed distribution has a
different value (the expectation of $\ln(v^2)$ against $f(v)$ shifts the
integral). Neither number is \`\`the’’ entropy of the gas; each is
entropy in a chosen coordinate.

## Discrete Entropy from Binning

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/differential-entropy-binning.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/differential-entropy-binning.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Partition the real line into bins of width $\Delta$. Then
$p_i\approx f(x_i)\Delta$ and Shannon entropy of the binned variable is
non-negative and diverges as the bins shrink. Differential entropy is
the $\Delta$-independent remainder.

Now relabel $y=g(x)$. Uniform bins of width $\Delta$ in $y$ correspond
to bins of width $\Delta/|g'|$ in $x$ — a non-uniform partition of the
original line. A *fixed* discrete partition is unchanged by renaming
outcomes, but \`\`uniform bins of fixed width’’ is a *different*
partition in each coordinate. That is exactly where $\mathbb{E}\ln|g'|$
appears. Continuous entropy therefore depends on a choice of reference
measure (here, bin width in the chosen coordinate).

Boltzmann counting needs a phase-space cell of volume $h^{3N}$ (Planck’s
constant to the power of degrees of freedom) before $S=k\ln W$ is
meaningful. Liouville’s theorem helps for Hamiltonian flow: canonical
transformations preserve phase-space volume, so the reference measure is
unchanged. An arbitrary reparametrisation of coordinates does not.

## KL Fixes the Coordinate Dependence

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/relative-entropy-reparametrisation.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/relative-entropy-reparametrisation.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Take two densities $f$ and $g$ on the same space and transform both by
the same invertible map $y=\phi(x)$. Each density picks up the factor
$1/|\phi'|$, which cancels inside the logarithm: KL divergence (and
mutual information) is therefore a property of the pair of
distributions, not of the coordinate chart. It also has a clean discrete
limit: with a common binning,
$\text{KL}\left( p^\Delta\,\|\,q^\Delta \right)\to \text{KL}\left( f\,\|\,g \right)$
as $\Delta\to 0$, with no divergent $-\ln\Delta$ term. Physics often
calls KL relative entropy; we keep
$\text{KL}\left( \cdot\,\|\,\cdot \right)$.

$$
\text{KL}\left( f_Y\,\|\,g_Y \right)=\text{KL}\left( f_X\,\|\,g_X \right)
$$

$$
h_m(f)=-\int f\ln\frac{f}{m}\,dx=-\text{KL}\left( f\,\|\,m \right)
$$

Differential entropy is KL divergence against a reference density $m$
that is usually left unwritten. Taking $m$ uniform (Lebesgue) in the
coordinate $x$ recovers the textbook $h$. Jaynes’ limiting density of
discrete points is the same idea: the binning measure of the previous
section, written as a density. Once $m$ is explicit, the
Maxwell–Boltzmann story becomes: different choices of \`\`equally likely
a priori’’ yield different maximum-entropy distributions under the same
energy constraint.

## Local Equilibrium as Grouping

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/local-equilibrium-grouping.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/local-equilibrium-grouping.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

$$
S(t)=H\bigl(w(t)\bigr)+\sum_i w_i(t)\, H(\pi_i)
$$

Week 2’s grouping identity holds at every instant. Local equilibrium is
the dynamical regime in which each within-group distribution has already
settled to its conditional stationary law $\pi_i$, while the coarse
weights $w_i(t)$ may still change on a slower clock. Then Macroscopic
thermodynamics lives in this picture: the within-group sum is the
entropy attributed to a fixed macrostate, and the slow story is
redistribution of probability among macrostates. Non-equilibrium, at
this first cut, means either the groups have not finished relaxing
($p_i\neq\pi_i$) or an external protocol is driving $w(t)$ (finite-time
thermodynamics; Crooks length in weeks 6–7).

## Scaffolding, Not Outcomes

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/channel-capacity-chain-rule.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/channel-capacity-chain-rule.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The chain rule $H(X,Y)=H(X)+H(Y|X)$ is the algebraic source of
multi-information. Mutual information $I(X;Y)=H(X)-H(X|Y)$ is the
pairwise case; we do not yet treat $n>2$. Channel capacity $C$ is a
no-go on rate; the capacity-achieving input distribution is the
prescription. The data-processing inequality is the no-go on $I$:
processing cannot create information. Cover and Thomas Theorem 2.8.1 is
the reading; the proof waits for week 8, when $I$ is first-class.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
eps = 0.1
p = np.linspace(0.01, 0.99, 200)
H = lambda q: -q*np.log2(q)-(1-q)*np.log2(1-q)
I = H(p) - H(eps)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(p, I, linewidth=2)
ax.scatter([0.5], [H(0.5) - H(eps)], s=80, color='red', zorder=3)
ax.set_xlabel('input bias $p(x=1)$')
ax.set_ylabel('$I(X;Y)$ (bits)')
ax.set_title('Binary symmetric channel')
mlai.write_figure('bsc-capacity.svg', directory='./ml')
```

## Thermodynamic Bath and Schottky’s Anomaly

<!-- SNIPPET: _physics/includes/thermodynamic-bath-schottky.md -->
A thermodynamic bath fixes $T$ and exchanges energy with a small system,
justifying the canonical ensemble. Heat capacity
$C=\partial U/\partial T$ peaks when both states are equally populated —
Schottky’s anomaly.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
epsilon = 1.0
beta = np.linspace(0.05, 5.0, 400)
p1 = 1.0 / (1.0 + np.exp(beta * epsilon))
U = p1 * epsilon
C = (epsilon ** 2) * p1 * (1 - p1) * beta ** 2
T = 1.0 / beta
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(T, U, 'C0', linewidth=2)
axes[0].set_xlabel('$T$ ($k=1$)')
axes[0].set_ylabel('$U$')
axes[0].set_title('Mean energy')
axes[1].plot(T, C, 'C1', linewidth=2)
axes[1].set_xlabel('$T$')
axes[1].set_ylabel('$C$')
axes[1].set_title('Schottky heat-capacity peak')
plt.tight_layout()
mlai.write_figure('schottky-two-state.svg', directory='./ml')
```

## Finite Time Costs More Than $\Delta F$

<!-- SNIPPET: _physics/includes/finite-time-dissipation-intro.md -->

Quasi-static processes achieve reversible work $W=\Delta F$. Finite-time
driving dissipates additional energy beyond the equilibrium bound. In
the grouping language of the die: equilibrium paths keep within-group
distributions at their conditional stationary laws and only the coarse
weights (or bath parameter $\beta$) change; rushing that change is what
costs extra.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
def two_state_F(beta, eps=1.0):
    Z = 1.0 + np.exp(-beta * eps)
    return -np.log(Z) / beta

beta0, beta1 = 0.5, 2.0
path = np.linspace(beta0, beta1, 80)
F_path = two_state_F(path)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(path, F_path, 'k-', linewidth=2, label='$F(\\beta)$')
ax.annotate('', xy=(beta1, two_state_F(beta1)), xytext=(beta0, two_state_F(beta0)),
            arrowprops=dict(arrowstyle='->', color='green', lw=2))
ax.text(1.0, -0.55, 'quasi-static: $W=\\Delta F$', color='green')
ax.annotate('', xy=(beta1, two_state_F(beta1) + 0.15), xytext=(beta0, two_state_F(beta0) + 0.15),
            arrowprops=dict(arrowstyle='->', color='red', lw=2))
ax.text(1.0, -0.35, 'finite time: $W > \\Delta F$', color='red')
ax.set_xlabel(r'$\beta$')
ax.set_ylabel('$F$')
ax.legend()
mlai.write_figure('finite-time-sketch.svg', directory='./ml')
```
