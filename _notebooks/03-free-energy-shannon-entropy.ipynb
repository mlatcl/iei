Week 1 put both figures on the table, and the quiz asked you to keep
them apart. A majority of papers mixed them. Maxwell’s demon is a
sorting intelligence that appears to drop the entropy of a gas.
Laplace’s demon is a predictive intelligence. Laplace’s gremlin (Neil’s

and momenta of every molecule as well as the rules and then doing the
resulting computations. The gremlin is why we *need* a statistical

free-energy to entropy story. The word itself is from aircraft: in the
First World War they blamed gremlins for things going wrong in the
planes. Two demons, one gremlin.

The other expensive item asked which of the Week 1 energy remarks
actually hold for machine learning. Lower energy corresponds to a more
preferred, more probable state. Quadratic loss and cross-entropy can be
written as energy-like scores. What does *not* follow is that every use
of the word energy in an ML paper is a valid thermodynamic internal
energy, or that a heat bath has been specified. We will cover the
relationship today. $\beta$ is inverse temperature, or coldness; Cole
Palmer has a high $\beta$. From an entropic perspective coldness is the
more natural parameter than $T$.

## From Heat to Helmholtz

Week 1 seeded the occupation $p_i \propto e^{-\beta E_i}$. Worksheet 1
told us Clausius did not make entropy a measure of uncertainty. The
lecture is the heat thread that gets us to Helmholtz: Black, Carnot,
Rumford, the air thermometer, Thomson’s 1848 mill-scale, Joule — then,
as taught, the board work on $F=U-TS$ with
$U=\left\langle E\right\rangle$, and Clausius after that.
Maxwell–Boltzmann and Gibbs remain in the file after Clausius. The
entropy game and MacKay’s weighing problem come after that story; they
were not reached on 27 October and now open Week 4 after Quiz 2. Shannon
axioms live in `_lamd/parking.md`.

From a stationary distribution with given energy we will write the
*Gibbs distribution*, $$
p_i = \frac{e^{-\beta E_i}}{Z} \quad Z = \sum_i e^{-\beta E_i}.
$$ The destination of the thread is $$
U = F + TS
$$ where $U=\left\langle E\right\rangle$ is inner energy, $F$ is
Helmholtz free energy, $T$ is temperature and $S$ is physical entropy.
Physical entropy is the dimensionless functional $H=-\sum_i p_i\log p_i$
converted into joules per degree by Boltzmann’s constant, $S=k_B H$.
Helmholtz wrote $A=U-TS$ ($A$ for *Arbeit*); we use $F$ for the same
quantity because week 5 needs $A(\boldsymbol{\theta})$ for the
log-partition.

## Joseph Black and Heat as a Quantity

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/joseph-black-calorimeter.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/joseph-black-calorimeter.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Before heat was understood as a property, it was understood as a
quantity. For Clausius, entropy was not yet a measure of uncertainty,
but to try and understand what he was measuring we need to go back to
heat. For our purposes the intellectual thread starts with Joseph Black
(1728–1799) who is responsible for making heat a quantity you can
measure, mix and even hide. This notion of *heat* as a measurable is
different from the notion of *temperature* as a measurable.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/joseph-black.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>Joseph Black (1728–1799). Black was a physician and chemist
at Glasgow, then Edinburgh. Stipple engraving by James Heath after Henry
Raeburn, 1800 (Wellcome Collection / Wikimedia Commons, public
domain).</i>

Black lectured in medicine and chemistry, in the mid eighteenth century
he was based in Glasgow, where he was a contemporary of James Watt. His
work on heat was done while he was still a physician. It was published
posthumously by John Robison as *Lectures on the Elements of Chemistry*
(Edinburgh, 1803) (Black, 1803).

Thermometers had existed since about the 17th century. This leads to a
mystery: ice melts, but the temperature doesn’t change. The same thing
happens at the other end: water boils, steam is produced, and the
thermometer sits at the boiling point. If you heat an equal mass of
water equally as a mass of mercury, they do not change temperature
equally. The important step that Joseph Black made was to realise that
temperature and heat are separate quantities. That is quite a big leap.
Most children, and most of us at first, think heat and temperature are
the same. Temperature is a reading, but heat is a quantity that can
enter a body without changing the reading.

Black’s notion is that heat is a quantity you can add, mix, and hide.
Take known masses, mix them, and they come up to a final temperature:
the heat has mixed. Capacity is when something absorbs heat at a
different rate. Mercury, he says, has less *capacity for the matter of
heat* than water: the same heat produces a larger change of temperature.
That is specific heat. The ice can start below zero, but once you get it
to zero it stops absorbing heat — it is swallowing heat, latent heat.
The heat that turns water into steam does not raise the thermometer
either. Recovered when the change of state is reversed.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/ice-calorimeter.jpg" style="width:45%; height:auto; display:block;">

Figure: <i>The calorimeter of Lavoisier and Laplace. Heat delivered is
read as the mass of ice melted; the conversion factor is Black’s latent
heat of fusion. Plate XI from *Encyclopaedia Londinensis* (1801–02),
after H. Lascelles (Wikimedia Commons / Science History Institute,
public domain).</i>

Black did not build this box. His own apparatus was an open fire and a
pan. He tried to keep the fire steady — bellows, not a closed system —
and timed how long it takes to melt ice or to boil a pan dry. The ice
calorimeter is Lavoisier and Laplace in the winter of 1782–83: a sample
in the inner chamber, a jacket of ice, an outer ice shield against the
room. Heat that leaves the sample melts ice in the middle jacket; you
weigh the meltwater. The conversion factor is Black’s latent heat of
fusion: the heat that would raise an equal mass of water by about
$143$–$144$ Fahrenheit *degrees*. That is a quantity of heat, not a
temperature on the thermometer. It is close to the modern figure of
$80^\circ\mathrm{C}$, or $334\,\mathrm{kJ\,kg}^{-1}$. The thermometer is
no longer the meter. The meltwater is.

One of the striking things in this period, even though Britain and
France are at war half the time, is how much scientific traffic there is
between them, including people travelling. Laplace is famous in England.
Lavoisier, with Laplace, forms an early table of chemical elements;
Lavoisier is killed in the Revolution; Laplace survives it and later
becomes a count under the Bourbon restoration. Their calorimeter is
inspired by Black.

Black’s unit of heat is not a temperature. It is the heat that raises an
equal mass of water by one degree on Fahrenheit’s thermometer. On that
scale the interval from melting ice to boiling water is $180$ units. The
extra heat to turn the boiling water into steam he estimates at about
$810$ of those units: the pan sits on the fire four and a half times
longer boiling dry than it did coming up to the boil — say fifteen
minutes to the boil, then about an hour and a quarter to evaporate. He
knows the timing method is rough because the fire is not a closed
system. His contemporary, James Watt, condensing steam and measuring the
water it warms, gets nearer $950$, within a few percent of the modern
figure. The modern figure is about $967^\circ\mathrm{F}$ on the same
scale, $540\,\mathrm{cal\,g}^{-1}$, or $2256\,\mathrm{kJ\,kg}^{-1}$: a
little more than five times the heat that takes water from $0$ to
$100^\circ\mathrm{C}$.

The number is already important industrially. Until Watt, a Newcomen
engine condensed steam by spraying cold water onto the cylinder. That
makes a vacuum and the piston comes down; then you heat the cylinder
again and it goes up. The condensation on the cylinder is latent heat
the engine has to repay on every stroke. Watt’s question is: how much
energy is that spray? His answer is the separate condenser — take the
steam out of the cylinder and cool it somewhere that is always cold.
That is why the engine becomes more efficient. Manufacturing it takes
him twenty-five to thirty years. Glasgow does not yet have the
machinery; he has to come to Boulton in Birmingham. You do not see these
engines in industrial use until about twenty years after the 1760s work.
The Lap engine in the Science Museum in London is late 1770s, early
1780s.

## Sadi Carnot and the Fall of Heat

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/sadi-carnot-caloric.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/sadi-carnot-caloric.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Black made heat a quantity you can add, mix and hide. From our
perspective, the next step comes from Sadi Carnot (1796–1832). Carnot
has a practical question, he’s not focussed on what heat is made of, he
wants to understand the limits of a steam engine.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/sadi-carnot.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>Sadi Carnot (1796–1832), aged 17, in the uniform of the École
Polytechnique. Portrait by Louis-Léopold Boilly, 1813 (Wikimedia
Commons, public domain).</i>

Nicolas Léonard Sadi Carnot was a military engineer, son of Lazare
Carnot. Lazare who was a mathematician and the “Organizer of Victory” of
the Revolutionary armies. After Waterloo Lazare he was driven into
exile, and Sadi’s career in army under the Bourbon restoration stalled.
The portrait is a decade before the book. By 1824 he is on reduced pay
in Paris, attending lectures at the Conservatoire, and paying himself to
print six hundred copies of a short essay: *Réflexions sur la puissance
motrice du feu* (Carnot, 1824).

After his time in Glasgow, James Watt had moved to Birmingham and
together with Matthew Boulton, commercialised a more efficient steam
engine. By the 1820s the steam engine was widely used in mines, mills,
looms and forges. Carnot believes that one day it will serve as a
universal motor, but he notes that it doesn’t have a theory of
mechanics. France is watching Britain do steam better. Watt’s engines
are a few percent. One reason they stay that low is that he refused
high-pressure steam: he thought it was dangerous, and his patent covered
engines that were not high-pressure. He was probably right about the
boilers they could manufacture then. Later, corrugated boilers, high
temperature and high pressure, and a modern steam turbine is about sixty
percent. The so-called “practical men” are improving the machine, but
Carnot wants to know whether any of that can be unbounded, or whether
there is a ceiling independent of the machine and of the substance that
runs through it. That is the same kind of question we would like to be
able to ask of intelligence. We are in almost the same position he was
in: a load of our quantities are not defined.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-reflexions-thurston.jpg" style="width:45%; height:auto; display:block;">

Figure: <i>Title page of R.\~H. Thurston’s 1890 English translation of
Carnot’s 1824 *Réflexions sur la puissance motrice du feu*. The original
was printed in Paris by Bachelier, at Carnot’s expense, in six hundred
copies.</i>

The 1824 original is a short book, 118 pages, almost unread in his
lifetime. Thurston’s translation is how most English readers met it. The
argument does not depend on calculus. It depends on a picture of heat
that Black had made plausible: caloric, a conserved fluid, falling from
hot to cold.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-1824-fig1.png" style="width:32%; height:auto; display:block;">

Figure: <i>Carnot’s own figure of the mill (1824). The cylinder $ab$
holds the working body; $cd$ is the piston. Body $A$ is the furnace,
body $B$ the refrigerator — two reservoirs of caloric at different
temperatures. Caloric taken from $A$ is given up to $B$. Vectorised from
the plate in *Réflexions sur la puissance motrice du feu* (Wikimedia
Commons, public domain).</i>

Carnot’s own comparison: the motive power of heat can be compared to
that of a waterfall. A waterfall has a maximum you cannot exceed,
whatever machine you put in the stream. The power depends on the
quantity of water and the height of the fall. Heat, likewise, depends on
the quantity of caloric and on what he calls the height of its fall —
the difference of temperature between the bodies exchanging it. In
English, *mill* until relatively recently also meant a factory; the
picture here is a water mill. In a steam engine the boiler is the upper
reservoir, the condenser the lower. Carnot’s own plate is that mill:
body $A$ the furnace, body $B$ the refrigerator, the cylinder the wheel.
The caloric developed in the firebox is taken up by the steam, passes
through the cylinder, and is given up to the cold water. The mill turns
because the caloric *moves*. The same quantity that left the hot
reservoir arrives at the cold one: heat moving like water does in a
waterfall. Conceptually he is thinking of the fall of water driving a
wheel, rather than a direct conversion of some of the heat into the
work. He is honest that he does not fully understand how the work comes
from the fall of caloric. Watt already knows, from Black, that steam
carries latent heat around and gives it up when it condenses. What is
surprising, looking back, is that this conservation theory of heat
exists *before* conservation of energy.

He pays for six hundred copies. In the summer of 1832 he falls ill; he
dies of cholera on 24 August, at Ivry, aged 36. The book, and the work,
almost vanish with him. Two years later Émile Clapeyron, another
polytechnicien, recasts the argument in 1834 in the language of
indicator diagrams, and people form what becomes known as the caloric
theory of heat — the dominant theory, a conservation theory before
conservation of energy.

## Count Rumford’s Cannon

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/count-rumford-cannon.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/count-rumford-cannon.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Carnot treated heat as a conserved fluid because that is what a mill
requires. The difficulty with that picture was already in print. In
1798, twenty-six years before the *Réflexions*, Benjamin Thompson, who
styled himself Count Rumford, published a paper in the *Philosophical
Transactions* on heat excited by friction (**Rumford-friction98?**).

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/count-rumford.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>Benjamin Thompson, Count Rumford (1753–1814). American-born,
then in the service of the Elector of Bavaria. Portrait by Moritz
Kellerhoven (Wikimedia Commons, public domain).</i>

Rumford was born in the United States and becomes superintendent for the
boring of brass cannons in the military arsenal at Munich, in the
service of Bavaria. An American working for a German army, submitting to
the Royal Society. He is struck by how hot the gun becomes, and by how
much hotter the metallic chips are (he checks and they are hotter than
boiling water). If heat were conserved, perhaps the shavings have lost
caloric and the cannon has gained it. He measures the specific heat of
the shavings and of the cannon and finds they are the same. His paper to
the Royal Society asks where the heat is coming from.

He asks whether the chips have changed their capacity for heat, as
Black’s latent-heat doctrine would require if caloric were being
squeezed out of the metal, but he finds they haven’t. He builds an
experiment with a short hollow cylinder is turned against a blunt steel
borer by machinery carried round by horses. A wooden box filled with
water surrounds the cylinder. At one hour the water is at
$107^\circ\mathrm{F}$; at two hours $178^\circ\mathrm{F}$; at two hours
thirty minutes it boils.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/rumford-cannon-boring.jpg" style="width:55%; height:auto; display:block;">

Figure: <i>Rumford’s 1798 apparatus, as reprinted in *The Complete Works
of Count Rumford* (1870). Fig.\~2: the cannon in the boring machine,
turned by the bar $w$. Fig.\~3: the same cylinder in the water box that
is brought to the boil.</i>

Rumford remarks that the source of the heat generated by friction
appeared evidently to be *inexhaustible*. But anything which an
insulated body can continue to furnish without limitation cannot
possibly be a material substance. That is 1798. Caloric is in the table
of the elements. Lavoisier, with Laplace, is the person who names
oxygen, previously phlogiston; he was a smart person, and caloric still
sat in the table. Rumford’s conclusion is that heat is a *mode of
motion* — a kinetic theory of heat. For what we are looking at here,
that is the form of heat: the air in this room is hot because it is in
motion. Heat at some point also becomes radiation; ignore radiation for
the moment. It is a qualitative blow to caloric, not a mechanical
equivalent. He does not weigh a known quantity of work against a known
quantity of heat. He does not say where the heat is coming from. He sets
it up as an example against the theories of conserved heat.

## Mercury, Ice, Steam, and Air

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/fahrenheit-celsius-air-thermometer.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/fahrenheit-celsius-air-thermometer.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Black already treats temperature as a *reading* and heat as a
*quantity*. The reading, in the 1840s, comes from one of three expansion
thermometers. Fahrenheit and Celsius use mercury. Accurate laboratory
work uses air. All three pin two fixed points — ice and steam, or a
close cousin — and then divide the interval so that equal expansions of
the chosen substance are equal degrees. Thomson will inherit that
practice, and then refuse it as a definition of a degree.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/fahrenheit-thermometer.jpg" style="width:42%; height:auto; display:block;">

Figure: <i>Daniel Gabriel Fahrenheit (1686–1736) with a mercury
thermometer. Anonymous engraving, 18th century (Wikimedia Commons,
public domain). The instrument is mercury in glass; the scale is equal
expansion of that mercury between two pins.</i>

Fahrenheit’s mercury thermometer is the household instrument of the
eighteenth century. He marks a mixture of ice, water and sal-ammoniac as
$0$, and (originally) blood heat as $96$. The scale that survives is a
recalibration of the same tube: ice at $32$, steam at $212$, so there
are $180$ equal expansions of mercury between the two pins. Celsius, in
1742, uses the same pins on a centesimal division. His own numbering ran
the other way — $100$ at ice, $0$ at steam — and was inverted within a
few years. By Thomson’s time \`\`Celsius’’ means ice $0$, steam $100$.
The physics is the same as Fahrenheit: you have chosen a substance, and
you have called equal expansions of it equal degrees. Black’s whole
point is that this reading is *not* the quantity of heat. The same rise
of the same mercury takes different heat in water, in oil, in ice that
is melting.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/air-thermometer-scheme.png" style="width:72%; height:auto; display:block;">

Figure: <i>(a) Constant *pressure*: a sliding mercury index, open to the
room; the volume of the trapped air is the reading. (b) Constant
*volume*: a mercury manometer; the height $h$ is the reading. Scheme by
P. Peterlin (Wikimedia Commons, CC BY-SA 3.0). Thomson’s 1848 comparison
is with (a).</i>

The air thermometer is the 1840s laboratory standard, the scale to which
mercury and alcohol are reduced (Regnault). Take a glass bulb of dry air
and confine it with mercury. The mercury is open to the room: either a
sliding thread in a capillary, as in (a), or a cistern you raise and
lower. You move the mercury until the two surfaces sit at the same
height, so the trapped air is at atmospheric pressure. The *volume* of
that air is then the reading. Ice water and steam still fix two points;
the interval is divided so that equal increases of volume are equal
degrees. That is Gay-Lussac made into an instrument: far from
condensation, the volume of air rises almost linearly with this
temperature. Push the same line far enough and the volume would vanish
at about $-273^\circ$ of the air scale. That number lives in the
instrument, not in Carnot. Panel (b) is the other common form, a
constant-*volume* gas thermometer, where you read pressure from the
manometer height. Thomson in 1848 is arguing with (a). Air is a better
expansion-substance than mercury — different gases agree closely unless
they are about to liquefy — but it is still a substance. A degree that
means \`\`equal expansion of air’’ is not yet independent of the
thermometer.

## Thomson’s Scale from the Mill

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/kelvin-absolute-scale.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/kelvin-absolute-scale.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

William Thomson, who later became Lord Kelvin, takes Carnot seriously as
theory. In 1848 he is a twenty-four year old professor of natural
philosophy at Glasgow, and he has read Clapeyron’s recasting of the
*Réflexions*. Fahrenheit, Celsius and the air thermometer already exist:
ice and steam already mark $0$ and $100$ on the scale he will compare
against. Those pins are a century old. What does not exist is a meaning
of *degree* that is independent of mercury, or of air. If the motive
power of heat depends only on the quantity of caloric and the height of
its fall, then that height should be usable as a thermometer
(**Thomson-absolute48?**). That is what he adds at twenty-four: a
definition of the interval, not a new pair of fixed points, and not a
bottom of the fall. The 1848 answer still assumes that heat is
conserved. It is not yet the scale that later bears his name.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/william-thomson-kelvin.jpg" style="width:38%; height:auto; display:block;">

Figure: <i>William Thomson (1824–1907), later Lord Kelvin. Photograph
from *The World’s Work*, 1902 (Wikimedia Commons, public domain) — half
a century after the 1848 scale.</i>

Thomson studied as an undergraduate at Cambridge and spent his career as
professor of natural philosophy at Glasgow. Maxwell is a younger
contemporary in the same group: people who combined the empirical work
going on in places like Glasgow, where they really cared about the
engineering, with a deep mathematical understanding. In 1848 Thomson is
looking for a thermometric scale that is not an accident of mercury or
air. He believes that Carnot’s theory provides such a scale. He already
has a quantity of heat: Black’s unit, the heat that raises a kilogramme
of water by one degree of the air thermometer. What he still lacks is a
meaning for an *interval* of temperature that does not belong to one
substance. When he is twenty-four he tries to interpret the degree as a
measurement related to specific heat: how much work could you extract
from one unit of Black’s heat falling one degree. Carnot’s mill supplies
it: the motive power of a given quantity of caloric falling through one
degree. That is mill-work from a *fall*, as water on a wheel.

The air thermometer is the scale Thomson is arguing with. Household
thermometers use mercury because it is convenient; in accurate work you
compare the mercury against air and quote the air reading (Regnault).
Air is a better expansion-substance than mercury, but it is still a
substance. A degree that means \`\`equal expansion of air’’ is not
independent of the thermometer. Thomson will not call that absolute.

What he adds is a meaning for the interval. Take a unit of heat — Black
already told you how to measure that — and let it fall through Carnot’s
mill. Call one degree whatever drop of temperature always yields the
same motive power from that unit, whether you are near steam or near
ice. The degree now belongs to the mill, not to mercury, and not to air.
Independence of substance here means independence of the thermometer’s
fluid and of the engine’s working fluid. The fixed points stay where
Fahrenheit and Celsius left them. That is why the paper matters in 1848,
before anyone has broken conservation of heat.

Thomson’s own statement of the characteristic property: all degrees have
the same value; a unit of heat descending from a body $A$ at temperature
$T^\circ$ to a body $B$ at $(T-1)^\circ$ gives out the same mechanical
effect, whatever the number $T$. That is the mill, turned into a ruler.
The working fluid is still caloric. The fall is still a fall, not a
conversion. He then *compares* the new ruler with the air thermometer,
using Clapeyron’s formulae and Regnault’s steam data. The two scales are
forced to agree at ice and at steam. Everything else is a table of how
unequal an air-thermometer degree is, in mill-value, as you go up the
scale. Work is how we make the degrees equal. It is not how we make the
heat, and it is not a new zero.

The $-273^\circ$ that looks like absolute zero is already sitting in the
air thermometer — the same bulb of air, volume as the reading. If that
volume is linear in temperature, pushing the graduation far enough
reduces the air to nothing at $-100/0.366$, which Thomson writes as
$-273^\circ$. He uses that fact only as a reason why Clapeyron’s table
was to be expected: an air-thermometer degree is *worth more* when it is
cold, because a finite number of those degrees is trying to reach
infinite cold. It is not the zero of his scale, and he does not compute
a table down to it. On the mill scale a unit of heat can always be let
down one more equal-value degree, and still arrive. Infinite cold is at
$-\infty$. 1848 has done something narrower: it has made a degree a
Carnot object. There is not yet a bottom, because caloric is still
conserved.

Thomson writes that in the present state of science no operation is
known by which heat can be absorbed without either raising a temperature
or becoming latent. Conversion of heat, or caloric, into mechanical
effect is probably impossible, certainly undiscovered. Actual engines
must therefore look for their power not in absorption and conversion,
but in the transmission of heat. That is Carnot, stated as a fact of
1848, and it is the assumption his mill-degree rests on.

## Joule and the Mechanical Equivalent

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/joule-mechanical-equivalent.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/joule-mechanical-equivalent.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

James Prescott Joule (1818–1889) gives the answer to Rumford’s riddle.
He theorises that a measured quantity of mechanical work could *create*
a measured quantity of heat. If it can then heat is not a conserved
fluid. This would imply that Carnot’s mill becomes a special case, not
the general law. Joule is a Manchester brewer who tries this experiment
with a paddle wheel (**Joule-equivalent50?**).

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/james-prescott-joule.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>James Prescott Joule (1818–1889). Seated portrait, later in
life (Wikimedia Commons, public domain).</i>

The family brewery is in Manchester where Joule has a private
laboratory. From 1840 he is working to relate heat to electricity and to
chemical affinity. In 1843 he argues that the heat from
magneto-electricity is proportional to the work absorbed. Mayer, in
1842, has already claimed that agitating water warms it, without a
measured force. Joule wants the equivalent.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/joule-apparatus-harpers.png" style="width:48%; height:auto; display:block;">

Figure: <i>Joule’s apparatus for the mechanical equivalent of heat. A
falling weight turns a paddle in water; the work is known, the rise in
temperature is measured. Engraving from *Harper’s New Monthly Magazine*,
August 1869 (Wikimedia Commons, public domain).</i>

In Joule’s apparatus weights descend and turn a paddle in a can of
water. The work associated with dropping a mass through a height is
$mgh$. It is coiled around a bearing; there is a water paddle in a tank.
To do the experiment you try to get rid of friction in the spindle. He
did not drop it once: he had to drop it multiple times to show how much
work converted into heat. It is a very similar experiment to Rumford’s,
but he has a better thermometer (mercury or other liquid, not air, by
the look of the apparatus) and he is measuring the specific heat
capacity of water. Mechanical power is being *spent*, and heat appears.
Finally people realise that heat energy and work can transform into one
another. That is the 1840s. Not very long ago.

Cambridge here is the British Association meeting of June 1845, not the
Cambridge Philosophical Society — that society is Thomson’s 1848 scale.
Joule exhibits the brass paddle to the Chemical Section; the published
letter opens by saying so (**Joule-equivalent45?**). That run gives
about $890$ foot-pounds per degree per pound of water. He also uses
other methods such as magneto-electricity, rarefaction of air, passing
water through narrow tubes to give neighbouring figures. His mean
estimate is near $817$. By 1850 the number is settled as $772.692$
foot-pounds (**Joule-equivalent50?**). The scientific world, he notes,
preoccupied with the hypothesis that heat is a substance, had almost
unanimously denied that fluid friction could generate heat.

You will hear people say that when they built steam engines they had no
theory of thermodynamics. That is not true. They had theories of
thermodynamics. They did not have *complete* theories. We may not have a
complete theory of thermodynamics today. This is where we are with
machine learning: we do not have good theories of what is going on in
the models, and we do not even have good definitions of intelligence.
People can go around saying AGI or superintelligence, and it does not
mean very much. Caloric is better defined. At least caloric is a notion
you can look at and say: it is not conserved. It is telling you
something specific you should expect. It is falsifiable. If you do not
define your terms, you cannot do falsifiability.

Once you realise that Joule converts work to heat, you also get the
sense that heat converts to work. You can convert all your work into
heat. You cannot get all the heat back as work. Carnot is assuming that
the work comes from the fall, as in a mill. Here the heat itself
converts. You cannot keep that theory of heat engines and Joule’s
demonstration as general laws. Clausius in 1850, and Thomson in 1851,
will keep Carnot’s *limit* and drop conservation of heat. We take
Clausius first: two laws, and $\mathrm{d}Q/T$. Thomson then returns to
rewrite the scale so that it has a bottom.

## Clausius and $\mathrm{d}Q/T$

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/clausius-carnot-second-law.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/clausius-carnot-second-law.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Heat is not conserved. Carnot’s limit remains. That is the problem
Clausius inherits from Joule and Thomson. If the caloric mill is gone,
what still closes when a reversible engine runs a cycle? The answer is
not a quantity of heat. It is heat divided by temperature. A reversible
cycle produces no entropy.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/rudolf-clausius.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>Rudolf Clausius (1822–1888). Photograph from the AIP Emilio
Segrè Visual Archives, Lande Collection (Wikimedia Commons, CC0).</i>

Clausius is a mathematical physicist, first in Berlin, later Zurich,
Würzburg and Bonn. It is quite close between Kelvin and Clausius to get
to the formula. In 1850, independently of Thomson’s 1851 dynamical
theory, he writes the two laws (**Clausius-motive50?**). He holds on to
Joule’s equivalent between work and energy, keeps Carnot’s limit, and
drops the notion that heat is conserved. He is doing all this without a
notion of conservation of energy, which is gobsmacking. Bernoulli too:
$PV=NRT$ is later; he was doing it with momentum. The first law is
Joule’s equivalent made universal: heat is a form of work. The second is
what remains of the mill: heat does not pass from cold to hot by itself,
and no cyclic engine converts the heat of a single bath entirely into
work.

The first law is why $\oint \mathrm{d}Q$ is not zero for an engine. Some
of the heat taken from the hot bath does not arrive at the cold bath; it
has become work. That is Joule, inside Carnot’s cycle. The second law is
why you still cannot take *all* of it. There is a remainder that must be
rejected to the cold bath, and the size of that remainder, for a
reversible engine, depends only on the two temperatures — not on whether
the working substance is steam, air, or anything else. That was Carnot’s
other claim, and it survives.

Thomson’s 1851 scale is this statement: the absolute temperatures stand
in the same ratio as the two heats. What Clausius notices is that in a
reversible engine the heat energy at the hot temperature, divided by
that temperature, equals the heat energy at the cold temperature,
divided by that temperature, as you go around. If the cycle is
reversible, that quantity is conserved: the integral of the small
changes in heat, over the whole loop, is zero. That is what Clausius
told us. The second law then says it can also be positive: a
non-reversible process. Clausius’s 1854 paper calls the quantity the
equivalence-value of the transformation (**Clausius-equivalence54?**).

If every reversible cycle gives zero, the integral of $\mathrm{d}Q/T$
between two states cannot depend on the reversible path you take. That
is the definition of a state function. Clausius writes $$
S - S_0 = \int \frac{\mathrm{d}Q_{\mathrm{rev}}}{T}.
$$ Heat $\mathrm{d}Q$ is not a state function: the heat you put in
depends on the path, which is why $\oint \mathrm{d}Q$ is the work of the
engine. Dividing by $T$ along a *reversible* path removes that
path-dependence. Reversibility is the condition. It means the engine can
be run backwards and the baths restored: no finite drops of temperature,
no friction, no paddle wheel hidden in the mechanism.

This is the mill, rewritten after Joule. The working fluid comes home.
Its entropy comes home with it. The hot bath has lost $Q_h/T_h$; the
cold bath has gained $Q_c/T_c$; those two are equal. No transformation
has been left over. Reversible does not mean \`\`nothing happened.’’
Work was done. It means the transformation can be undone, and when you
close the loop there is no residue.

An irreversible cycle is still a cycle: the working substance returns,
so its $S$ is unchanged. What has changed is the baths. You took $Q_h$
from the hot bath and, because some of the fall was wasted, you
delivered *more* than the reversible $Q_c$ to the cold bath. The
integral $\oint \mathrm{d}Q/T$, with $\mathrm{d}Q$ the heat absorbed by
the working substance, is then strictly negative
(**Clausius-equivalence54?**). Equivalently, the entropy lost by the hot
bath is less than the entropy gained by the cold bath. The difference
was produced. The second law says this integral can also be positive:
that is a non-reversible process.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/clausius-coins-entropy.jpg" style="width:62%; height:auto; display:block;">

Figure: <i>Clausius names entropy, 1865. English translation of the
memoir: $\int \mathrm{d}Q/T = S-S_0$, from Greek τροπή,
*transformation*. He forms the word to sit beside *energy*.</i>

In 1865 he gives the state function a name (**Clausius-entropy65?**).
Clausius writes that we might call $S$ the transformational content of
the body, just as we term the magnitude $U$ its thermal and ergonal
content, the inner energy; but that it is better to borrow terms for
important magnitudes from the ancient languages, so that they may be
adopted unchanged in all modern languages. He proposes to call $S$ the
*entropy* of the body, from the Greek word τροπή, *transformation*. He
has put an $n$ on the front. He says explicitly that he has formed the
word to be as similar as possible to *energy*, because the two
magnitudes are so nearly allied in their physical meanings that a
certain similarity of designation appears desirable. They are not allied
because of the division by $T$. There is no Boltzmann’s constant here
yet: this is a macroscopic property, something he can derive from
measuring what was lost, or noticing it is not lost in a reversible
cycle. He has no notion underneath it that this is anything to do with
uncertainty. He just says: I need this piece to fit the puzzle. The
assumption, coming at it afterwards, is that they were looking at energy
conserved and therefore this piece is missing. It is not quite like
that.

Clausius did not give the Boltzmann distribution, and he did not make
entropy a measure of uncertainty. He gave the macroscopic balance that
any later prescription must respect. $S$ is a macroscopic measurement
that is *extensive*: the entropy of a system increases as we increase
the size of the system. Temperature is *intensive*: it does not. When we
write $p_i \propto e^{-\beta E_i}/Z$ and later derive it from MaxEnt,
read it as a statistical answer to a constraint he already framed: fixed
mean energy, maximum entropy, no perpetual motion. The historical chain
is engines, then the equivalent, then this integral, then statistics,
then information.

Later, Maxwell, then Boltzmann and Gibbs, come to the same sort of
conclusions. This century, Shannon and Jaynes reuse the functional form
for what might be said to be a different job. Shannon is always
discrete. Wiener is continuous. Wiener never drops the mechanism behind
the entropy: he always thinks about what drives the distribution. In
Shannon you just say it is a probability. A lot of the theory does not
have to specify what the probability *is*; you use that it normalises to
one and is positive, and you derive a lot from that. That is
mathematics. The other bit, where you have to talk about mechanism, is
physics.

## Thomson Returns: The Scale After Joule

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/kelvin-dynamical-theory.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/kelvin-dynamical-theory.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Clausius has kept Carnot’s limit and dropped conserved heat. Thomson
does the same, independently, a year later. In 1851 he writes the
dynamical theory of heat (**Thomson-dynamical51?**). The mill is still
there. The ruler is not. Heat taken in and heat rejected now stand in a
*ratio* of the two temperatures, and for the first time the fall has a
bottom.

The 1848 paper had already heard of Joule, in a footnote, and had
refused him: conversion of heat into work was probably impossible,
certainly undiscovered. The paddle wheel is the experiment Thomson said
had not been shown. By 1851 he accepts it. Heat can be generated by
work, and some of the heat taken from the hot bath does not arrive at
the cold bath — it has become work. Conservation of caloric cannot be
the general law. Clausius had already said so. Thomson’s contribution,
for us, is what happens to the *scale*.

In 1848 a degree was a height of fall of a conserved unit of heat: the
same mill-work for every rung, whatever $T$ you started from. After
Joule the unit does not arrive. What *is* equal, for a reversible
engine, is $Q_h/T_h$ and $Q_c/T_c$. That is Clausius’s entropy leaving
the hot bath and arriving at the cold bath. Thomson takes it as the
definition of the scale. Ice is no longer the zero. The numbers $T_h$
and $T_c$ are the absolute temperatures of the two baths; their ratio is
the ratio of the heat taken in to the heat rejected.

This is the genuine floor that 1848 did not have. If
$Q_c/Q_h = T_c/T_h$, then $T_c=0$ is the temperature at which a
reversible engine, taking $Q_h$ from the hot bath, rejects nothing. The
$-273^\circ$ of Gay-Lussac’s expansion was an accident of air. The zero
of the 1851 scale is an accident of *engines*: it is the temperature you
cannot fall below, because there is no heat left to reject. We still use
this scale. We name it after the title Thomson later took.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/james-clerk-maxwell.png" style="width:100%; height:auto; display:block;">

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/boltzmann2.jpg" style="width:100%; height:auto; display:block;">

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/j-w-gibbs.jpg" style="width:100%; height:auto; display:block;">

Figure: <i>James Clerk Maxwell (1831–1879), Ludwig Boltzmann
(1844–1906), Josiah Willard Gibbs (1839–1903). After Clausius the
account becomes statistical: components of velocity, then occupation of
energy, then free energy.</i>

Clausius gave a macroscopic balance. Maxwell, Boltzmann and Gibbs give a
microscopic count of the same $S$. Maxwell’s first move is the velocity
distribution: each Cartesian component is Gaussian, because the kinetic
energy is quadratic. That quadratic is the energy whose expectation we
will split as $U=F+TS$.

## Maxwell–Boltzmann: Components Versus Speed

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-boltzmann-components-vs-speed.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-boltzmann-components-vs-speed.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Write the joint density on Cartesian velocities as isotropic, Each
marginal $f(v_x)$ is Gaussian. Change to spherical coordinates with
$v=|\mathbf{v}|$ and solid angle $\Omega$. Probability conservation uses
the volume element $d^3v = v^2\,dv\,d\Omega$. Integrating over angles
($\int d\Omega = 4\pi$) gives the Maxwell speed distribution The factor
$v^2$ is the Jacobian of the change of variables — the density of states
of spherical shells in velocity space. Calling \`\`Maxwell–Boltzmann’’
Gaussian is correct for a component and wrong for speed.

$$
f(v_x)=\sqrt{\tfrac{m}{2\pi kT}}\,e^{-mv_x^2/2kT}
\qquad\text{versus}\qquad
f(v)\propto v^2\,e^{-mv^2/2kT}
$$

## The Gibbs–Boltzmann Distribution

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/gibbs-distribution.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/gibbs-distribution.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Name the object and fix the formula. The MaxEnt derivation that *earns*
these weights is a separate step: deriving the occupation from Lagrange
multipliers before Shannon $H$ and Jaynes would put the cart before the
horse.

Write the equilibrium occupation of a discrete system with energies
$\{E_i\}$ as $$
p_i = \frac{e^{-\beta E_i}}{Z(\beta)},
\qquad
Z(\beta)=\sum_j e^{-\beta E_j}.
$$ Physicists call $p_i$ the *Boltzmann distribution* (or Boltzmann
weights) and, for a system exchanging energy with a bath at fixed $T$,
the *Gibbs distribution* or *canonical ensemble*. The three names point
at the same formula. The normalisation $Z(\beta)$ is the *partition
function*. Its logarithm $\log Z(\beta)$ is the cumulant generating
function for the energy under this exponential family: derivatives of
$\log Z$ recover the mean energy, the variance (heat capacity, up to
factors of $\beta$), and higher cumulants. That generating-function
reading is why $Z$ is more than a normalisation constant.

The Gibbs–Boltzmann occupation is not only a statement about gases and
magnets. Hopfield networks (Hopfield, 1982) assign an energy to every
binary configuration of a recurrent net and treat recall as a descent
toward low-energy states — equilibrium statistics are again Gibbs.
Ackley, Hinton and Sejnowski (Ackley et al., 1985) made the weights of
that energy *learnable*: a Boltzmann machine is an undirected model
whose distribution over configurations is exactly
$p(s)\propto e^{-E(s)/T}$. The 2024 Nobel Prize in Physics, awarded to
John Hopfield and Geoffrey Hinton, recognised that physical-systems
reading of computation and learning. Name the lineage so the formula
does not feel confined to nineteenth-century heat baths; do not divert
into training algorithms.

## Coldness and Temperature

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/coldness-and-temperature.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/coldness-and-temperature.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The Boltzmann occupation can be written in two ways, $$
p_i = \frac{e^{-E_i/k_B T}}{Z}
= \frac{e^{-\beta E_i}}{Z},
\qquad
\beta = \frac{1}{k_B T}.
$$ The mathematics is the same, but the emphasis is different.

$T$ is the variable of the bath. It is what a thermometer reports, and
it is the intensive parameter in the Helmholtz accounting $F = U - TS$.
In that representation energy comes first and entropy is the correction:
$TS$ is the cut the second law takes from $U$.

$\beta$ is *coldness*. It is the variable conjugate to energy. In that
representation entropy comes first: you maximise $S$ (or write a
generating function in $\beta$) and energy is the constraint.
Temperature is then a derived reading,
$T = 1/k_B\beta = (\partial S/\partial U)^{-1}$.

We will use both. Weeks 1–2 keep $T$ in view so the bath is familiar,
and they already compute in $\beta$ because that is the natural argument
of $Z$. Week 5 makes the $\beta$-first order honest: the Lagrange
multiplier on a mean-energy constraint *is* coldness, and it is the
natural parameter of the exponential family. Weeks 6–7 then treat
$\beta$ as a coordinate on that family. A student who still hears
$\beta$ as \`\`one over the thermometer’’ will miss why the geometry is
written in those coordinates.

## Free Energy Decomposition

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/free-energy-decomposition.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/free-energy-decomposition.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Once the Gibbs occupation $$
p(x)=\frac{e^{-\beta E(x)}}{Z},\qquad Z=\sum_x e^{-\beta E(x)}
$$ is given, we can derive the macroscopic formulae that characterise
the gas. They come to us from Helmholtz. We can compute them through the
expectation of that occupation of the given energies, taken at
equilibrium. The relationship between energy and probability is
$p_i\propto e^{-\beta E_i}$. Here we will identify the “inner energy” or
“total energy” with $\left\langle E\right\rangle$ and read off $F=U-TS$.
$Z$ is German: it comes from Boltzmann, the sum of states, Summe der
Zustände. That is why it is $Z$, because it is a sum. If the system were
continuous, a Gaussian, $Z$ would be an integral.

Start from the occupation and take the logarithm, $$
\log p(x)=-\beta E(x)-\log Z.
$$ Note that nothing statistical has happened here, this is an identity
for every probability that is a member of the exponential family, which
in physics we think of as the Gibbs distribution. It’s valid for every
configuration $x$ that the Gibbs distribution supports. If we solve for
the energy of that configuration we can restore temperature through
recognising that $\beta=1/(k_BT)$, $$
E(x)=-k_BT\log p(x)-k_BT\log Z.
$$ Note that the first term on the right (the surprisal multiplied by
the temperature and Boltzmann’s constant) still depends on $x$. The
second does not: $-k_BT\log Z$ is a property of the whole ensemble, not
of a particular microstate.

At equilibrium the macroscopic inner energy can be related to the
expectation of the microscopic score under the same distribution, $$
U\equiv\left\langle E\right\rangle=\sum_x p(x)\,E(x).
$$ This is where the *ensemble* is entering in physics. The ensemble
that comes to us from Boltzmann, Maxwell, Gibbs, and Einstein. The idea
is that the system is not in one configuration. When we coarse-grain
this room we are not going to look at the actual energy of every
particle. There are of order $10^{23}$ of them; at that scale the
variance around the expectation does not show. That expected energy
becomes, at this scale, the total energy of the gas in the room. The
index $i$ is the *space* — discrete energy levels, which is one of the
challenges Boltzmann received: what *is* this discrete space? That is
where the phase-volume arguments come from. We will just assume it is
discrete; an expectation is a sum or an integral. Substitute the
identity for $E(x)$, The second term comes out of the sum because
$\log Z$ does not depend on $x$, and $\sum_x p(x)=1$.

The two remaining pieces already have names. The
configuration-independent term is the *Helmholtz free energy*, $$
F=-k_BT\log Z,
$$ it is often known as the “free energy” or “available energy.” It is
the energy that we can use for work. The average of $-\log p$ is the
dimensionless Shannon functional $H=-\sum_x p(x)\log p(x)$. Multiplying
by Boltzmann’s constant produces physical entropy, $S=k_B H$, with units
of energy per degree. Then $$
U=F+TS\qquad\text{or}\qquad F=U-TS.
$$ Helmholtz originally wrote $A=U-TS$, with $A$ for *Arbeit*, the work
still available. This module uses $F$ for the same quantity, because
week 5 needs $A(\boldsymbol{\theta})$ for the log-partition of an
exponential family. But Helmholtz’s $A$ and our $F$ are the same object;
$U$ is the inner energy $\left\langle E\right\rangle$, not a free
energy. The fact that they are different is down to energy that is
stored as entropy. It’s in this term that we can find Black’s latent
heat.

From the physics and thermodynamics perspective $U$ is the energy in the
system, $TS$ is the “cut” that entropy takes from that total energy. It
is tied up in the uncertainty of the occupation, unavailable as work.
Then $F$ is what the thermodynamic bath still allows you to do. Given
the Gibbs occupation, the macroscopic split $F=U-TS$ is an identity.

The point of the derivation, boarded, is that inner energy is composed
of the entropy plus the available energy. Entropy is now your
bookkeeping device for why you cannot get all the work back out of the
heat. You can convert all your work into heat and produce the inner
energy with your motion. Then the motion is going in different
directions. The energy is still there at the particle level. Maxwell’s
demon is an attempt to reverse that by seeing where the particles are.
The scale we have just written assumes an ensemble. Dive into the
ensemble and play with it, and you cross levels of coarse graining. That
is easy to do in modelling, and it is what goes on in a lot of
Maxwell’s-demon arguments. It is not always clear at which point you are
allowed to cross. Week 4 is the treatment; here it is the warning.

Watch the signs on the board. Boltzmann’s constant is of order
$10^{-23}\,\mathrm{J\,K}^{-1}$. There is a $10^{-23}$ scaling on these
numbers, and then of order $10^{23}$ particles, so you get back to
joules. The units of $k_B$ are joules per kelvin — joules per degree.
$S=k_B H$ takes a dimensionless functional of a probability and converts
it into the units we speak about. If you think of $H$ as the more
fundamental object, coldness $\beta$ is the more natural parameter than
$T$. The reason we usually write it the other way around is that it was
discovered in Clausius’s way: energy first, then entropy times
temperature.

The $T$-first reading puts energy first and treats entropy as a
correction. The $\beta$-first reading treats $\log Z$ as a generating
function: $U=-\partial_\beta\log Z$ and $S=\beta(U-F)$, which we compute
from $Z$ later. Both are the identity just derived. Later we will derive
the occupation from MaxEnt and identify $\beta$ as the Lagrange
multiplier on the mean-energy constraint; until then the occupation is
given, and we are reading it.

If you have a process whose entropy has not changed, inner energy and
available energy are the same. That is a reversible system: you have not
taken work and converted it into inner energy. Under this definition
$\beta\,\mathrm{d}S = \mathrm{d}Q$: the thing they were measuring the
flow of, when they talked about conservation, was entropy. That is the
caloric they thought was conserved. If the entropy is conserved you are
looking at a reversible system, which is what Carnot was trying to look
at, and why his engine is the most efficient you can be.

A short history-of-science aside, boarded before the algebra. A lot of
the *practice* of science in this period — what we might call
engineering — is in Birmingham, Glasgow, Aberdeen, driven by
application. Late nineteenth century, a rising capability in Germany.
After the Great Exhibition of 1851 there was a royal commission to ask
why Oxford and Cambridge had contributed so little. The result of that
inquiry was the Cavendish Laboratory (1870) and the Natural Sciences
Tripos. Maxwell is the scientist this thread is heading towards; we skip
him for a moment.

```
energies = np.array([0.0, 1.0, 3.0])
beta = np.linspace(0.05, 2.5, 200)
Z = np.sum(np.exp(-beta[:, None] * energies), axis=1)
p = np.exp(-beta[:, None] * energies) / Z[:, None]
U = (p * energies).sum(axis=1)
S = -np.sum(p * np.log(p + 1e-300), axis=1)
F = U - S / beta
F_check = -np.log(Z) / beta
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(beta, U, label=r'$U(\beta)$')
ax.plot(beta, S, label=r'$S(\beta)$ (nats, $k_B=1$)')
ax.plot(beta, F, label=r'$F(\beta)$')
ax.plot(beta, F_check, 'k--', alpha=0.5, label=r'$-\ln Z/\beta$')
ax.set_xlabel(r'$\beta$')
ax.legend()
ax.set_title(r'Three-state system: energies $E\in\{0,1,3\}$')
mlai.write_figure('three-state-thermo.svg', directory='./ml')
```

The Worksheet 1 three-state system, $E\in\{0,1,3\}$, is a numerical
check of the same algebra with $k_B=1$. Compute $p_i=e^{-\beta E_i}/Z$,
then $U=\sum_i p_i E_i$ and $S=-\sum_i p_i\log p_i$. The two expressions
for free energy, $F=U-S/\beta$ and $F=-\log Z/\beta$, overlie. If they
do not, the signs in the derivation are wrong.

```
import numpy as np
```

```
def gibbs_thermo(energies, beta):
    """U, S, F from a discrete Gibbs occupation ($k_B=1$)."""
    e = np.asarray(energies, dtype=float)
    weights = np.exp(-beta * e)
    Z = weights.sum()
    p = weights / Z
    U = np.sum(p * e)
    S = -np.sum(p * np.log(p))
    F = -np.log(Z) / beta
    return U, S, F, U - S / beta

U, S, F, F_from_US = gibbs_thermo([0.0, 1.0, 3.0], beta=1.0)
U, S, F, F_from_US
```

## The Carnot Cycle on $T$-$S$

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/carnot-ts-cycle.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/carnot-ts-cycle.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

A reversible engine is the most efficient you can be, such an engine is
what Carnot studied. Under the Helmholtz split,
$\beta\,\mathrm{d}S=\mathrm{d}Q$. When they talked about heat flow and
conservation, what they were actually measuring the flow of was entropy.
This was the source of the caloric that they believed to be conserved.
The ideal Carnot cycle makes that visible. It can be drawn on a
temperature-entropy diagram as a square. That square is our picture, not
his. Before drawing it, it is worth hearing what he actually asked.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-1824-fig1.png" style="width:28%; height:auto; display:block;">

Figure: <i>Carnot’s own plate, 1824. Cylinder $ab$, piston $cd$, body
$A$ the furnace, body $B$ the refrigerator. This is the mill as he drew
it: not a waterwheel, a piston and two reservoirs. Vectorised from
*Réflexions sur la puissance motrice du feu* (Wikimedia Commons, public
domain).</i>

We have already seen the mill: caloric falls from a hot body to a cold
one, and the same caloric arrives at the bottom. His own plate is not a
waterwheel. It is a cylinder and a movable piston, and two bodies, $A$
and $B$, held at different temperatures. The waterfall lives in the
text, not in the figure (Carnot, 1824).

The sentence is careful. The motive power of a waterfall depends on its
height and on the quantity of the liquid. The motive power of heat
depends on the quantity of caloric and on what he will call the *height
of its fall* — the difference of temperature between the bodies
exchanging it. In a waterfall the work is exactly proportional to the
difference of level. In the fall of caloric, he says, the motive power
undoubtedly increases with the difference of temperature — but he does
not know whether it is proportional. Would the same quantity of caloric
falling from $100^\circ$ to $50^\circ$ furnish more or less motive power
than the same caloric falling from $50^\circ$ to zero? That is his
question. Sit with it before the square appears. The square is our
answer, not his. He identified the fall, and the reversible ceiling. He
did not write $W=\Delta T\times\Delta S$.

<img src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-mill-square.svg" class="" width="85%" style="vertical-align:middle;">

Figure: <i>The same cycle twice. Left: a water mill on height against
mass of water. Fill the bucket at the top, let it descend, empty it,
return the empty bucket. The water is not consumed; the work is
$g\,\Delta h\,\Delta m$. Right: a reversible heat engine on $T$ against
$S$. The conserved quantity in the reversible model is what we now call
entropy. The vertical brace is Carnot’s fall — the difference between
the reservoirs — not the downward stroke of the working body.</i>

If you take the mill seriously as a cycle, it is already a rectangle.
Fill the bucket at the top: mass comes in at height $h_h$. The loaded
bucket descends, the shaft turns. Empty at the bottom. Return the empty
bucket. The water is not consumed. The work is $g\,\Delta h\,\Delta m$.

Replace height by temperature and mass of water by entropy, and you have
the $T$–$S$ square. The conserved quantity in the reversible model is
what we now call entropy. That is an interpretation of his caloric, not
a claim that he had already discovered it without naming it. Falk reads
Clausius’s entropy as a resurrection of caloric; Meyn draws the mill as
this rectangle. Historians are not unanimous, and they should not be:
the identification is a reading of the reversible model, not a
reconstruction of what was in his head. What we can say is narrower.
Once the transferred quantity is entropy and the height is thermodynamic
temperature, the product is exactly the area, and proportionality is no
longer a question. For a computer scientist this is a beautiful example
of how choosing the right state variable turns a complicated process
into simple geometry.

```
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch, Polygon, Rectangle
import mlai
```

```
diagrams = './physics'
Sc, Sh = 1.0, 2.4
Tc, Th = 0.8, 2.2
A = (Sc, Th)
B = (Sh, Th)
C = (Sh, Tc)
D = (Sc, Tc)
hot_c = '#C44E52'
cold_c = '#4C72B0'
adiab_c = '#333333'
work_c = '#DD8452'
ink = '#222222'
legs = [
    dict(p0=A, p1=B, color=hot_c,
         label='expansion', sub=r'isothermal, $T_h$'),
    dict(p0=B, p1=C, color=adiab_c,
         label='isolated cool', sub='adiabatic'),
    dict(p0=C, p1=D, color=cold_c,
         label='contraction', sub=r'isothermal, $T_c$'),
    dict(p0=D, p1=A, color=adiab_c,
         label='isolated heat', sub='adiabatic'),
]
titles = [
    r'The $T$–$S$ plane',
    'Start at some temperature',
    'Temperature constant, entropy increases',
    'The system cools rapidly',
    'Entropy comes back down',
    'Heat rapidly: the square',
    'Four legs of the mill',
    r'Work extracted: $\Delta T\times\Delta S$',
]

def _axes(title):
    fig, ax = plt.subplots(figsize=(7.2, 5.6))
    ax.set_xlim(0.15, 3.05)
    ax.set_ylim(0.25, 2.85)
    ax.set_aspect('equal')
    ax.set_xlabel(r'entropy $S$')
    ax.set_ylabel(r'temperature $T$')
    ax.set_xticks([Sc, Sh])
    ax.set_xticklabels([r'$S_c$', r'$S_h$'])
    ax.set_yticks([Tc, Th])
    ax.set_yticklabels([r'$T_c$', r'$T_h$'])
    ax.axhline(Th, color=hot_c, ls='--', lw=1.0, alpha=0.45, zorder=0)
    ax.axhline(Tc, color=cold_c, ls='--', lw=1.0, alpha=0.45, zorder=0)
    ax.set_title(title)
    for spine in ax.spines.values():
        spine.set_color('#444444')
    return fig, ax

def _save(frame):
    mlai.write_figure(f'carnot-ts{frame:03d}.svg', directory=diagrams)

def _dot(ax, p, color=ink):
    ax.plot(p[0], p[1], 'o', color=color, ms=8, zorder=5)

def _draw_leg(ax, p0, p1, color, arrow=False, lw=2.7):
    if arrow:
        ax.add_patch(FancyArrowPatch(
            p0, p1, arrowstyle='-|>', mutation_scale=16,
            lw=lw, color=color, shrinkA=0, shrinkB=2.5,
            clip_on=False, zorder=3))
    else:
        ax.plot([p0[0], p1[0]], [p0[1], p1[1]], color=color, lw=lw,
                solid_capstyle='round', zorder=3)

def _label_leg(ax, p0, p1, text, sub, color):
    mx = 0.5 * (p0[0] + p1[0])
    my = 0.5 * (p0[1] + p1[1])
    if abs(p1[0] - p0[0]) > abs(p1[1] - p0[1]):
        ha = 'center'
        if my > 1.4:
            va, offx, offy = 'bottom', 0.0, 0.08
        else:
            va, offx, offy = 'top', 0.0, -0.08
    else:
        va = 'center'
        if mx > 1.6:
            ha, offx, offy = 'left', 0.10, 0.0
        else:
            ha, offx, offy = 'right', -0.10, 0.0
    ax.text(mx + offx, my + offy, text + '\n' + sub, color=color,
            ha=ha, va=va, fontsize=11, linespacing=1.25)

# 000: axes
fig, ax = _axes(titles[0])
_save(0)

# 001: start
fig, ax = _axes(titles[1])
_dot(ax, A, hot_c)
_save(1)

# 002--005: add one leg at a time
for n_done in range(1, 5):
    fig, ax = _axes(titles[1 + n_done])
    _dot(ax, A, hot_c)
    for i in range(n_done):
        lg = legs[i]
        _draw_leg(ax, lg['p0'], lg['p1'], lg['color'], arrow=True)
        _dot(ax, lg['p1'], lg['color'])
    _save(1 + n_done)

# 006: labels on the closed square
fig, ax = _axes(titles[6])
_dot(ax, A, hot_c)
for lg in legs:
    _draw_leg(ax, lg['p0'], lg['p1'], lg['color'], arrow=True)
    _dot(ax, lg['p1'], lg['color'])
    _label_leg(ax, lg['p0'], lg['p1'], lg['label'], lg['sub'], lg['color'])
_save(6)

# 007: work is the area
fig, ax = _axes(titles[7])
ax.add_patch(Rectangle(
    (Sc, Tc), Sh - Sc, Th - Tc,
    facecolor=work_c, alpha=0.28, edgecolor='none', zorder=1))
_dot(ax, A, hot_c)
for lg in legs:
    _draw_leg(ax, lg['p0'], lg['p1'], lg['color'], arrow=True)
    _dot(ax, lg['p1'], lg['color'])
ax.annotate(
    '', xy=(Sc - 0.22, Th), xytext=(Sc - 0.22, Tc),
    arrowprops=dict(arrowstyle='<->', color=ink, lw=1.2), zorder=4)
ax.text(Sc - 0.30, 0.5 * (Th + Tc), r'$\Delta T$', ha='right', va='center',
        color=ink, fontsize=14)
ax.annotate(
    '', xy=(Sh, Tc - 0.20), xytext=(Sc, Tc - 0.20),
    arrowprops=dict(arrowstyle='<->', color=ink, lw=1.2), zorder=4)
ax.text(0.5 * (Sc + Sh), Tc - 0.32, r'$\Delta S$', ha='center', va='top',
        color=ink, fontsize=14)
ax.text(0.5 * (Sc + Sh), 0.5 * (Tc + Th),
        r'$W=\Delta T\times\Delta S$', ha='center', va='center',
        color=ink, fontsize=15)
_save(7)

# 008: real engine cuts the other two corners; same ΔT, same ΔS
fig, ax = _axes('In practice: cut two corners, not a smaller square')
ax.add_patch(Rectangle(
    (Sc, Tc), Sh - Sc, Th - Tc,
    facecolor=work_c, alpha=0.12, edgecolor='none', zorder=1))
for lg in legs:
    _draw_leg(ax, lg['p0'], lg['p1'], lg['color'], arrow=False, lw=1.6)
T2 = Tc + (2.0 / 3.0) * (Th - Tc)
T4 = Th + Tc - T2
inner = Polygon(
    [(Sc, Tc), (Sc, T2), (Sh, Th), (Sh, T4)],
    closed=True, facecolor='#55A868', alpha=0.35,
    edgecolor='#2E7D4F', lw=2.2, zorder=2)
ax.add_patch(inner)
_dot(ax, (Sh, Th), hot_c)
_dot(ax, (Sc, Tc), cold_c)
ax.text(0.5 * (Sc + Sh), 0.5 * (Tc + Th),
        r'$\eta\sim$ area ratio', ha='center', va='center',
        color=ink, fontsize=14)
_save(8)

# p–V: same four legs, chosen so the loop is readable (not the T–S numbers)
gamma = 1.4
nR = 1.0
T_hot, T_cold = 1.4, 1.0
Va, Vb = 1.0, 1.5
Vc = Vb * (T_hot / T_cold) ** (1.0 / (gamma - 1.0))
Vd = Va * (T_hot / T_cold) ** (1.0 / (gamma - 1.0))

def p_of(T, V):
    return nR * T / V

def adiabat_T(T0, V0, V):
    return T0 * (V0 / V) ** (gamma - 1.0)

V_ab = np.linspace(Va, Vb, 120)
V_bc = np.linspace(Vb, Vc, 120)
V_cd = np.linspace(Vc, Vd, 120)
V_da = np.linspace(Vd, Va, 120)
p_ab = p_of(T_hot, V_ab)
p_bc = p_of(adiabat_T(T_hot, Vb, V_bc), V_bc)
p_cd = p_of(T_cold, V_cd)
p_da = p_of(adiabat_T(T_cold, Vd, V_da), V_da)
V_cycle = np.concatenate([V_ab, V_bc, V_cd, V_da])
p_cycle = np.concatenate([p_ab, p_bc, p_cd, p_da])

V_bg = np.linspace(0.7, Vc + 0.4, 200)
fig, ax = plt.subplots(figsize=(7.2, 5.6))
ax.plot(V_bg, p_of(T_hot, V_bg), color=hot_c, ls='--', lw=1.0, alpha=0.45,
        label=r'isotherm $T_h$')
ax.plot(V_bg, p_of(T_cold, V_bg), color=cold_c, ls='--', lw=1.0, alpha=0.45,
        label=r'isotherm $T_c$')
ax.fill(V_cycle, p_cycle, color=work_c, alpha=0.28, zorder=1)
ax.plot(V_ab, p_ab, color=hot_c, lw=2.7)
ax.plot(V_bc, p_bc, color=adiab_c, lw=2.7)
ax.plot(V_cd, p_cd, color=cold_c, lw=2.7)
ax.plot(V_da, p_da, color=adiab_c, lw=2.7)
ax.add_patch(FancyArrowPatch(
    (V_ab[50], p_ab[50]), (V_ab[70], p_ab[70]),
    arrowstyle='-|>', mutation_scale=14, lw=0, color=hot_c, zorder=4))
ax.add_patch(FancyArrowPatch(
    (V_bc[50], p_bc[50]), (V_bc[70], p_bc[70]),
    arrowstyle='-|>', mutation_scale=14, lw=0, color=adiab_c, zorder=4))
ax.add_patch(FancyArrowPatch(
    (V_cd[50], p_cd[50]), (V_cd[70], p_cd[70]),
    arrowstyle='-|>', mutation_scale=14, lw=0, color=cold_c, zorder=4))
ax.add_patch(FancyArrowPatch(
    (V_da[50], p_da[50]), (V_da[70], p_da[70]),
    arrowstyle='-|>', mutation_scale=14, lw=0, color=adiab_c, zorder=4))
ax.plot(Va, p_of(T_hot, Va), 'o', color=hot_c, ms=7)
ax.plot(Vb, p_of(T_hot, Vb), 'o', color=hot_c, ms=7)
ax.plot(Vc, p_of(T_cold, Vc), 'o', color=cold_c, ms=7)
ax.plot(Vd, p_of(T_cold, Vd), 'o', color=cold_c, ms=7)
ax.text(Va - 0.06, p_of(T_hot, Va) + 0.04, '1', ha='right', color=ink)
ax.text(Vb + 0.06, p_of(T_hot, Vb) + 0.04, '2', ha='left', color=ink)
ax.text(Vc + 0.06, p_of(T_cold, Vc) - 0.06, '3', ha='left', color=ink)
ax.text(Vd - 0.06, p_of(T_cold, Vd) - 0.06, '4', ha='right', color=ink)
ax.set_xlim(0.6, Vc + 0.4)
ax.set_ylim(0.0, p_of(T_hot, Va) * 1.18)
ax.set_xlabel(r'volume $V$')
ax.set_ylabel(r'pressure $p$')
ax.set_title(r'Same cycle on $p$–$V$: not a square')
ax.legend(loc='upper right', frameon=False)
mlai.write_figure('carnot-pv.svg', directory=diagrams)

# Two-panel: mill on (m, h) beside engine on (S, T)
mh, ml = Sc, Sh
hh, hc = Th, Tc
fig, axes = plt.subplots(1, 2, figsize=(11.4, 5.5))
panels = [
    dict(ax=axes[0],
         xlab=r'mass of water $m$', ylab=r'height $h$',
         xt=[mh, ml], xtl=[r'$m$', r'$m+\Delta m$'],
         yt=[hc, hh], ytl=[r'$h_c$', r'$h_h$'],
         top='fill bucket', bot='empty bucket',
         right='loaded bucket\ndescends',
         work=r'$W=g\Delta h\Delta m$',
         fall='height of the fall',
         title='Water mill'),
    dict(ax=axes[1],
         xlab=r'entropy $S$', ylab=r'temperature $T$',
         xt=[Sc, Sh], xtl=[r'$S_c$', r'$S_h$'],
         yt=[Tc, Th], ytl=[r'$T_c$', r'$T_h$'],
         top=r'take in entropy at $T_h$', bot=r'reject entropy at $T_c$',
         right='working body cools;\nno heat transfer',
         work=r'$W=\Delta T\times\Delta S$',
         fall="Carnot's fall of caloric",
         title='Heat engine'),
]
for pan in panels:
    ax = pan['ax']
    x0, x1 = pan['xt']
    y0, y1 = pan['yt']
    ax.set_xlim(0.05, 3.55)
    ax.set_ylim(0.15, 2.95)
    ax.set_aspect('equal')
    ax.set_xlabel(pan['xlab'])
    ax.set_ylabel(pan['ylab'])
    ax.set_xticks(pan['xt'])
    ax.set_xticklabels(pan['xtl'])
    ax.set_yticks(pan['yt'])
    ax.set_yticklabels(pan['ytl'])
    ax.axhline(y1, color=hot_c, ls='--', lw=1.0, alpha=0.45, zorder=0)
    ax.axhline(y0, color=cold_c, ls='--', lw=1.0, alpha=0.45, zorder=0)
    ax.add_patch(Rectangle(
        (x0, y0), x1 - x0, y1 - y0,
        facecolor=work_c, alpha=0.22, edgecolor='none', zorder=1))
    _draw_leg(ax, (x0, y1), (x1, y1), hot_c, arrow=True)
    _draw_leg(ax, (x1, y1), (x1, y0), adiab_c, arrow=True)
    _draw_leg(ax, (x1, y0), (x0, y0), cold_c, arrow=True)
    _draw_leg(ax, (x0, y0), (x0, y1), adiab_c, arrow=True)
    _dot(ax, (x0, y1), hot_c)
    ax.text(0.5 * (x0 + x1), y1 + 0.10, pan['top'], ha='center', va='bottom',
            color=hot_c, fontsize=9)
    ax.text(0.5 * (x0 + x1), y0 - 0.10, pan['bot'], ha='center', va='top',
            color=cold_c, fontsize=9)
    ax.text(x1 + 0.16, 0.5 * (y0 + y1), pan['right'], ha='left', va='center',
            color=adiab_c, fontsize=8, linespacing=1.2)
    ax.text(0.5 * (x0 + x1), 0.5 * (y0 + y1), pan['work'],
            ha='center', va='center', color=ink, fontsize=13)
    ax.annotate(
        '', xy=(0.42, y1), xytext=(0.42, y0),
        arrowprops=dict(arrowstyle='<->', color=ink, lw=1.2), zorder=4)
    ax.text(0.28, 0.5 * (y0 + y1), pan['fall'], ha='center', va='center',
            color=ink, fontsize=8, rotation=90)
    ax.set_title(pan['title'])
    for spine in ax.spines.values():
        spine.set_color('#444444')
fig.tight_layout()
mlai.write_figure('carnot-mill-square.svg', directory=diagrams)

# Fall versus adiabatic cool, on the T-S square already drawn
fig, ax = _axes("The fall is the reservoirs, not the downward stroke")
ax.set_xlim(0.05, 3.45)
ax.add_patch(Rectangle(
    (Sc, Tc), Sh - Sc, Th - Tc,
    facecolor=work_c, alpha=0.22, edgecolor='none', zorder=1))
_dot(ax, A, hot_c)
for lg in legs:
    _draw_leg(ax, lg['p0'], lg['p1'], lg['color'], arrow=True)
    _dot(ax, lg['p1'], lg['color'])
ax.annotate(
    '', xy=(Sc - 0.22, Th), xytext=(Sc - 0.22, Tc),
    arrowprops=dict(arrowstyle='<->', color=ink, lw=1.2), zorder=4)
ax.text(Sc - 0.34, 0.5 * (Th + Tc), "Carnot's fall of caloric",
        ha='right', va='center', color=ink, fontsize=11, rotation=90)
ax.text(Sh + 0.10, 0.5 * (Th + Tc),
        'working body cools\nby expansion;\nno heat transfer',
        ha='left', va='center', color=adiab_c, fontsize=10)
ax.text(0.5 * (Sc + Sh), 0.5 * (Tc + Th),
        r'$W=\Delta T\times\Delta S$', ha='center', va='center',
        color=ink, fontsize=15)
mlai.write_figure('carnot-ts-fall.svg', directory=diagrams)
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-ts007.svg" class="" width="70%" style="vertical-align:middle;">

Figure: <i>The ideal Carnot cycle on a temperature-entropy ($T$-$S$)
diagram is a square. Heat is taken in along the top ($T$ constant, $S$
increasing) and rejected along the bottom. The two vertical legs are
adiabatic: the working body changes temperature without exchanging
entropy. The work extracted is the area, $\Delta T\times\Delta S$.</i>

You start at some temperature. In the Carnot cycle the temperature stays
constant while the entropy increases: that is the expansion of the
system, the increasing volume is the opening of the piston.[1] Then the
system cools — an isolated expansion, entropy holding. Once it has
cooled, the entropy comes slowly back down, temperature staying the
same: the contraction, heat rejected to the cold bath. That cold
rejection is what Watt’s condenser is doing: it is the bottom of the
mill, not the cooling stroke itself. Then you compress in isolation, the
steam heats, and you have a square. This is the ideal engine. It is not
how Carnot described it, but it is what he was describing.

On the visualiser I drew the two temperature changes as rapid, almost
instantaneous. In the ideal cycle they need not be. They are adiabatic:
thermally isolated, reversible expansions and compressions. No heat
crosses the boundary, so entropy of the working body holds, and the
temperature changes because the volume does. The entropy changes, by
contrast, have to happen in contact with a bath. Do them too fast and
you are no longer isothermal: compression would heat the system and
expansion would cool it, and you have left the square.

$\Delta T$ is the difference between the two temperatures and $\Delta S$
the difference between the two entropies. The energy extracted on the
cycle is $\Delta T\times\Delta S$.

<img src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-ts-fall.svg" class="" width="70%" style="vertical-align:middle;">

Figure: <i>Two annotations that are easy to conflate. The vertical brace
is Carnot’s fall of caloric: entropy is taken in at $T_h$ and given up
at $T_c$. The right-hand leg is not that fall. It is an isolated
expansion of the working body; no heat crosses, entropy holds,
temperature drops.</i>

It is easy to look at the square and take the downward stroke for the
fall. That leg is where the working substance cools by expansion, with
no heat transfer. Carnot’s fall is the difference between the
temperatures at which entropy enters and leaves, across the whole cycle.
The mill turns because caloric moves from $A$ to $B$, not because the
piston is dropping.

He gives a concrete steam-engine example. A boiler at about
$160^\circ\mathrm{C}$, a condenser at about $40^\circ\mathrm{C}$: a fall
of caloric of $120^\circ$. Combustion could furnish a much larger fall,
a thousand degrees or more, and ordinary engines used only a modest part
of it. That was his engineering point: look after the height. The modern
reading, which he did not write, puts those on an absolute scale,
$T_h = 433\,\mathrm{K}$, $T_c = 313\,\mathrm{K}$, and the ceiling is
$1-T_c/T_h$, a little under twenty-eight percent. The analogy suggested
quantity of caloric times temperature fall. The $T$–$S$ square tells us
how that product arises, once the transferred quantity is entropy and
temperature is thermodynamic temperature.

<img src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-ts008.svg" class="" width="70%" style="vertical-align:middle;">

Figure: <i>A real engine between the same reservoirs, with the same
entropy swing, does not shrink the square. It still visits the top-right
and the bottom-left. Heat is not taken in along the whole top, nor
rejected along the whole bottom, so those two corners are cut off. The
remaining area over the square is the efficiency.</i>

In a motor-car engine the same loop is a four-stroke: ignition, piston
down, exhaust, compression. This is known as the Otto cycle. But this is
the mechanism behind the theory, or one possible mechanism. What the
theory says is that the maximum energy you can get out is the area of
the square.

The inner loop is easy to draw wrongly. You do not scale the square
down. The fall is still $T_h$ to $T_c$, and the entropy change of the
working body is still $S_c$ to $S_h$: those are known. So the top-right
corner stays, and the bottom-left stays. What you lose is the other two.
Heat does not go in isothermally along the whole top, nor come out along
the whole bottom, and the area that remains, over the square, is the
efficiency of the engine.

As an undergraduate mechanical engineer one learns these diagrams
without anyone saying that the energy is stored in probability, or that
this is the form of what entropy is. When you realise that the engine is
at the heart of the industrial revolution and that it’s really
converting entropy into work, then there is an argument that at its
heart even the Industrial Revolution is an information revolution. That
is, of course, if you accept the interpretation of entropy as
information.

<img src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-pv.svg" class="" width="70%" style="vertical-align:middle;">

Figure: <i>The same Carnot cycle on a $p$-$V$ diagram. The hot and cold
legs follow isotherms $pV=\mathrm{const}$; the isolated cool and
isolated heat follow adiabats, which are steeper. The enclosed area is
still the work, $\oint p\,\mathrm{d}V$, but you cannot read
$\Delta T\times\Delta S$ off it by eye.</i>

The other way these things are drawn is as pressure-volume diagrams. You
can draw the same stages as the $T$-$S$ diagram, but they no longer look
like straight lines. That’s because isothermal on this chart is a
constraint on $pV$ due to the relationship between pressure and volume
($pV = nRT$ for an ideal gas). This makes it not so easy to read off
what energy has been extracted because it’s the area inside the shape
still, but the area is no longer a square.

But as a reminder, we are imposing a modern perspective on how Carnot
thought about it. Recall his thinking was that it was more like a mill:
a waterfall from which we are extracting energy. He did not talk about
entropy. He was interested in change in temperature as being equivalent
to height change. The $p$–$V$ chart is Clapeyron’s language; the square
is ours. The mill is his.

[1] Focussing on volume increase, normally we would expect increasing
volume to cause cooling, but in a heat engine if we increase the volume
slowly enough, heat can flow from a heat bath to the cylinder and it
stays at constant temperature. The energy is flowing in from the heat
bath.

## Gaussian: Expected Quadratic Energy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/gaussian-quadratic-free-energy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/gaussian-quadratic-free-energy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

We’ve already introduced the univariate Gaussian as an energy-based
model with quadratic energy. We also looked at animations of elastic
collisions and argued that the velocities were given by the
Maxwell-Boltzmann distribution that has the form of a Gibbs
distribution. We can apply the Helmholtz split to this system in closed
form. Take $$
E(x)=\frac{x^2}{2}
$$ and a bath at inverse temperature $\beta$ (here $k_B=1$, so
$T=1/\beta$). The Gibbs density is $$
p(x)=\frac{e^{-\beta x^2/2}}{Z}=\sqrt{\frac{\beta}{2\pi}}\,\exp\Bigl(-\beta\frac{x^2}{2}\Bigr),
$$ which is $\mathcal{N}(0,T)$. The partition function is the Gaussian
integral $$
Z=\int_{-\infty}^{\infty}e^{-\beta x^2/2}\,\mathrm{d}x=\sqrt{\frac{2\pi}{\beta}}.
$$

The inner energy is the expectation of the quadratic, $$
U=\left\langle E\right\rangle=\int p(x)\,\frac{x^2}{2}\,\mathrm{d}x=\frac{1}{2}\left\langle x^2\right\rangle.
$$ Under $p=\mathcal{N}(0,T)$ we have $\left\langle x^2\right\rangle=T$,
so $U=T/2$. That is the equipartition theorem for a single quadratic
degree of freedom: each such term in the energy contributes $k_BT/2$ at
equilibrium. Maxwell’s three Cartesian components — a few slides back —
are three independent copies of this quadratic. Kinetic energy
$\tfrac32 k_BT$ is $d=3$. The Jacobian that produces *speed* is why the
speed density is not itself Gaussian.

The free energy is the generating-function term from the derivation, $$
F=-T\log Z=-T\log\sqrt{2\pi T}=-\frac{T}{2}\log(2\pi T).
$$ Entropy follows from $S=(U-F)/T$, $$
S=\frac{1}{2}+\frac{1}{2}\log(2\pi T).
$$ That is the differential entropy of $\mathcal{N}(0,T)$, in nats, with
$k_B=1$. Here $S$ is simply the name we gave to
$-k_B\left\langle\log p\right\rangle$ when we took the expectation of
$E(x)$. The continuous functional can be negative and is not invariant
to a change of units. The algebra we need is already visible: $$
F+TS=-\frac{T}{2}\log(2\pi T)+T\Bigl(\frac{1}{2}+\frac{1}{2}\log(2\pi T)\Bigr)=\frac{T}{2}=U.
$$ The $\tfrac12$ in $U$ and the $\tfrac12$ sitting in $T\cdot\tfrac12$
— the $e$ inside $\tfrac12\log(2\pi e T)$ — cancel, leaving
$F=-T\log Z$. Spreading the Gaussian (raising $T$) does not change the
*shape* of the well $E(x)=x^2/2$; it changes how far the occupation
climbs the walls, and therefore how much of $U$ is locked in $TS$.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
T = np.linspace(0.15, 3.0, 400)
U = T / 2.0
S = 0.5 + 0.5 * np.log(2.0 * np.pi * T)
TS = T * S
F = -0.5 * T * np.log(2.0 * np.pi * T)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(T, U, label=r'$U=T/2$')
ax.plot(T, TS, label=r'$TS$')
ax.plot(T, F, label=r'$F=-T\log Z$')
ax.axhline(0.0, color='#666', ls=':', lw=1)
ax.set_xlabel(r'$T$ ($k_B=1$)')
ax.legend()
ax.set_title(r'Quadratic $E(x)=x^2/2$: $U=F+TS$')
mlai.write_figure('gaussian-quadratic-thermo.svg', directory='./ml')

xs = np.linspace(-4.0, 4.0, 400)
Ex = 0.5 * xs ** 2
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(xs, Ex, 'k', lw=2, label=r'$E(x)=x^2/2$')
for Tval, colour in [(0.3, '#4C72B0'), (1.0, '#DD8452'), (2.5, '#55A868')]:
    ax.axhline(Tval / 2.0, color=colour, ls='--', lw=1.5,
               label=rf'$\langle E\rangle=T/2$ at $T={Tval:.1f}$')
ax.set_xlabel(r'$x$')
ax.set_ylabel(r'$E(x)$')
ax.set_ylim(0.0, 4.0)
ax.legend()
ax.set_title(r'Occupation climbs the well as $T$ rises')
mlai.write_figure('gaussian-quadratic-well.svg', directory='./ml')
```

Figure: <i>Closed form for $E(x)=x^2/2$ with $k_B=1$. Inner energy

```
import numpy as np
```

```
def gaussian_quadratic_thermo(T):
    """U, S, F for $E(x)=x^2/2$ at temperature $T$ ($k_B=1$)."""
    U = T / 2.0
    Z = np.sqrt(2.0 * np.pi * T)
    F = -T * np.log(Z)
    S = (U - F) / T
    return U, S, F

T = 1.0
U, S, F = gaussian_quadratic_thermo(T)
U, S, F, np.isclose(U, F + T * S)
```

If the energy is a sum of independent quadratics,
$E(\mathbf{x})=\tfrac12\sum_{i=1}^d x_i^2$, the partition function
factors and $$
U=\frac{d\,T}{2},\qquad
Z=(2\pi T)^{d/2},\qquad
F=-\frac{d T}{2}\log(2\pi T).
$$ Each coordinate contributes its own $T/2$ to the inner energy and its
own $\tfrac12\log(2\pi e T)$ to the entropy. That is why an ideal gas of
$N$ particles in three dimensions has kinetic energy $\tfrac32 N k_BT$:
$3N$ quadratic terms in the Hamiltonian, each holding $k_BT/2$. The
derivation did not change. We took the expectation of a bigger $E$.

The air thermometer is a bulb of gas whose *volume*, at constant
pressure, is the reading. Thomson already had Gay-Lussac’s wall: if that
volume is linear in temperature it vanishes at about $-273^\circ$. The
quadratic well we just split has no box, so it cannot be that
thermometer. Maxwell’s three Cartesian components *are* three copies of
$E=v_i^2/2$: the kinetic half of an ideal-gas particle is already on the
board. What is missing is the configuration integral. A free particle in
a container of volume $V$ has, up to constants of mass and $h$ that we
do not need, $$
Z_1 = V\,(2\pi T)^{3/2}.
$$ The Gaussian integrals we just evaluated are the momentum half; $V$
is the position half. For $N$ particles, $Z \propto V^N T^{3N/2}$, so
$F=-T\log Z$ depends on $V$. Pressure is a Helmholtz derivative, not a
new potential: $$
P=-\Bigl(\frac{\partial F}{\partial V}\Bigr)_{T,N}=\frac{NT}{V}.
$$ At constant $P$, $V\propto T$. Gay-Lussac’s wall is $T=0$ on the same
scale whose variance we just read off Maxwell’s Gaussians. Thomson in
1848 treated $-273^\circ$ as an accident of air. After Joule it is the
floor of the engine scale; after Maxwell it is also the temperature at
which those Gaussians have collapsed to a delta, and a finite pressure
would require zero volume.

We do not need the Gibbs free energy $G=F+PV$ for this. $G$ is the
potential when *pressure* is the control variable. Charles’s law is
already in $PV=NT$. Coupled multivariate Gaussians are not required
either: independent coordinates suffice, and that is what Maxwell wrote.
The one extra Helmholtz fact is the derivative with respect to $V$. We
name it. We do not compute Stirling, $h^3$, or chemical potential today.

## Two Sides of the Same Coin

We have just derived $F=U-TS$ from the Gibbs occupation, with inner
energy $U=\left\langle E\right\rangle$. Helmholtz free energy is what
remains after entropy’s cut. That identity is the whole of the LO1
payload. It is also why two research programmes that sound opposed are
not. Welling, Lu and Holdijk’s *Generative AI and Stochastic
Thermodynamics* was published this year. Its subtitle is *A Tale of Free
Energies* (Welling et al., 2026). If that book had a slogan, it would be
close to *free energy is all you need*. If this course had a slogan, it
would be *entropy is all you need*. They are two sides of the same coin,
and the coin is inner energy.

GAIST explores the shared mathematics of generative models and
nonequilibrium thermodynamics. The mathematics is broadly the same; the
philosophy is not. Their Chapter 3 is the Helmholtz split we have just
done. Their Chapter 5 will identify variational free energy with the
ELBO: inference as heat, learning as work. We will meet that
identification in week 5 as a *correspondence*, not as a collapse of
Shannon, Boltzmann and Jaynes into one theory. Read GAIST by chapter,
not cover to cover. Callen remains the undergraduate thermodynamics. The
motivations of this course — perpetual motion, the human bandwidth
constraint — are not in that book.

Karl Friston’s free-energy principle treats the same variational free
energy as an account of brains and of action: perception as inference
that lowers a bound on surprise (**Friston-free10?**). Under a trial
distribution $q$ the functional is again $U[q]-TS[q]$, which is
Helmholtz’s split with the occupation replaced by $q$. That is why it
belongs in this lecture, and why it does not get a derivation today. We
stay with entropy. Entropy is the cut that forbids perpetual motion.
Free energy is what you can still do once that cut is taken. Inner
energy is what you started with. Three names, one identity.
