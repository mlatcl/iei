energy, or that a heat bath has been specified. We will cover the
relationship today. $\beta$ is inverse temperature, or coldness; Cole
Palmer has a high $\beta$. From an entropic perspective coldness is the
more natural parameter than $T$.

## From Heat to Helmholtz

Week 1 seeded the occupation $p_i \propto e^{-\beta E_i}$. Worksheet 1
told us Clausius did not make entropy a measure of uncertainty. The
lecture is the heat thread that gets us to Helmholtz: Black, Carnot,
Rumford, the air thermometer, Thomson’s 1848 mill-scale, Joule — then,
as taught, the board work on $F=U-TS$ with
$U=\left\langle E\right\rangle$, and Clausius after that.
Maxwell–Boltzmann and Gibbs remain in the file after Clausius. The
entropy game and MacKay’s weighing problem come after that story; they
were not reached on 27 October. Shannon axioms are parked in Week 4.

From a stationary distribution with given energy we will write the
*Gibbs distribution*, $$
p_i = \frac{e^{-\beta E_i}}{Z} \quad Z = \sum_i e^{-\beta E_i}.
$$ The destination of the thread is $$
U = F + TS
$$ where $U=\left\langle E\right\rangle$ is inner energy, $F$ is
Helmholtz free energy, $T$ is temperature and $S$ is physical entropy.
Physical entropy is the dimensionless functional $H=-\sum_i p_i\log p_i$
converted into joules per degree by Boltzmann’s constant, $S=k_B H$.
Helmholtz wrote $A=U-TS$ ($A$ for *Arbeit*); we use $F$ for the same
quantity because week 5 needs $A(\boldsymbol{\theta})$ for the
log-partition.

## Joseph Black and Heat as a Quantity

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/joseph-black-calorimeter.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/joseph-black-calorimeter.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Before heat was understood as a property, it was understood as a
quantity. For Clausius, entropy was not yet a measure of uncertainty,
but to try and understand what he was measuring we need to go back to
heat. For our purposes the intellectual thread starts with Joseph Black
(1728–1799) who is responsible for making heat a quantity you can
measure, mix and even hide. This notion of *heat* as a measurable is
different from the notion of *temperature* as a measurable.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/joseph-black.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>Joseph Black (1728–1799). Black was a physician and chemist
at Glasgow, then Edinburgh. Stipple engraving by James Heath after Henry
Raeburn, 1800 (Wellcome Collection / Wikimedia Commons, public
domain).</i>

Black lectured in medicine and chemistry, in the mid eighteenth century
he was based in Glasgow, where he was a contemporary of James Watt. His
work on heat was done while he was still a physician. It was published
posthumously by John Robison as *Lectures on the Elements of Chemistry*
(Edinburgh, 1803) (Black, 1803).

Thermometers had existed since about the 17th century. This leads to a
mystery, ice melts, but the temperature doesn’t change. The same thing
happens at the other end: water boils, steam is produced, and the
thermometer sits at the boiling point. If you heat an equal mass of
water equally as a mass of mercury, they do not change temperture
equally. The important step that Joseph Black made was to realise that
temperature and heat are separate quantities. Temperature is a reading,
but heat is a ‘quantity’ that can enter that body without changing the
reading.

Black’s mixing calorimetry is the mechanism. You account for heat as you
would account for a fluid: what leaves one body arrives in the other.
The maths is an additive ledger, not a kinetic theory. Mercury, he says,
has less *capacity for the matter of heat* than water: the same heat
produces a larger change of temperature. That is specific heat, named as
a capacity. The heat that melts ice, or that turns water into steam,
does not raise the temperature at all. Black calls it *latent*: combined
with the particles, hidden from the thermometer, recovered when the
change of state is reversed.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/ice-calorimeter.jpg" style="width:45%; height:auto; display:block;">

Figure: <i>The calorimeter of Lavoisier and Laplace. Heat delivered is
read as the mass of ice melted; the conversion factor is Black’s latent
heat of fusion. Plate XI from *Encyclopaedia Londinensis* (1801–02),
after H. Lascelles (Wikimedia Commons / Science History Institute,
public domain).</i>

Black did not build this box. He measured by mixing, and by timing how
long a steady fire takes to melt ice or to boil a pan dry. The ice
calorimeter is Lavoisier and Laplace in the winter of 1782–83: a sample
in the inner chamber, a jacket of ice, an outer ice shield against the
room. Heat that leaves the sample melts ice in the middle jacket; you
weigh the meltwater. The conversion factor is Black’s latent heat of
fusion: the heat that would raise an equal mass of water by about
$143$–$144$ Fahrenheit *degrees*. That is a quantity of heat, not a
temperature on the thermometer. It is close to the modern figure of
$80^\circ\mathrm{C}$, or $334\,\mathrm{kJ\,kg}^{-1}$. The instrument
makes his accounting automatic: the thermometer is no longer the meter.
The meltwater is.

Black’s unit of heat is not a temperature. It is the heat that raises an
equal mass of water by one degree on Fahrenheit’s thermometer. On that
scale the interval from melting ice to boiling water is $180$ units. The
extra heat to turn the boiling water into steam he estimates at about
$810$ of those units: the pan sits on the fire four and a half times
longer boiling dry than it did coming up to the boil. He knows the
timing method is rough because the fire is not a closed system. His
contemporary, James Watt, condensing steam and measuring the water it
warms, gets nearer $950$. The modern figure is about
$967^\circ\mathrm{F}$ on the same scale, $540\,\mathrm{cal\,g}^{-1}$, or
$2256\,\mathrm{kJ\,kg}^{-1}$: a little more than five times the heat
that takes water from $0$ to $100^\circ\mathrm{C}$. The number is
already important industrially. James Watt is famous for improving the
efficiency of the steam engine. Until his time, a Newcomen engine would
condense steam by spraying cold water onto the engine cylinder. That
condensation occuring on the cylinder becomes latent heat that the
engine has to re-pay on every stroke. Watt’s invention of the separate
condenser can be motivated by Black’s latent heat.

## Sadi Carnot and the Fall of Heat

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/sadi-carnot-caloric.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/sadi-carnot-caloric.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Black made heat a quantity you can measure, mix and hide. Once it is a
quantity, it looks like a fluid you can account for. From our
perspective, the next step comes from Sadi Carnot (1796–1832). Carnot
has a practical question, he’s not focussed on what heat is made of, he
wants to understand the limits of a steam engine.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/sadi-carnot.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>Sadi Carnot (1796–1832), aged 17, in the uniform of the École
Polytechnique. Portrait by Louis-Léopold Boilly, 1813 (Wikimedia
Commons, public domain).</i>

Nicolas Léonard Sadi Carnot was a military engineer, son of Lazare
Carnot. Lazare who was a mathematician and the “Organizer of Victory” of
the Revolutionary armies. After Waterloo Lazare he was driven into
exile, and Sadi’s career in army under the Bourbon restoration stalled.
The portrait is a decade before the book. By 1824 he is on reduced pay
in Paris, attending lectures at the Conservatoire, and paying himself to
print six hundred copies of a short essay: *Réflexions sur la puissance
motrice du feu* (Carnot, 1824).

After his time in Glasgow, James Watt had moved to Birmingham and
together with Matthew Boulton, commercialised a more efficient steam
engine. By the 1820s the steam engine was widely used in mines, mills,
looms and forges. Carnot beleives that one day it will serve as a
universal motor, but he notes that it doesn’t have a theory of
mechanics. France is watching Britain do steam better. The so-called
“practical men” are improving the capabilities of the machine, but
Carnot wants to know whether any of that can be unbounded, or whether
there is a ceiling independent of the machine and of the substance that
runs through it.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-reflexions-thurston.jpg" style="width:45%; height:auto; display:block;">

Figure: <i>Title page of R.~H. Thurston’s 1890 English translation of
Carnot’s 1824 *Réflexions sur la puissance motrice du feu*. The original
was printed in Paris by Bachelier, at Carnot’s expense, in six hundred
copies.</i>

The 1824 original is a short book, 118 pages, almost unread in his
lifetime. Thurston’s translation is how most English readers met it. The
argument does not depend on calculus. It depends on a picture of heat
that Black had made plausible: caloric, a conserved fluid, falling from
hot to cold.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/carnot-1824-fig1.png" style="width:32%; height:auto; display:block;">

Figure: <i>Carnot’s own figure of the mill (1824). The cylinder $ab$
holds the working body; $cd$ is the piston. Body $A$ is the furnace,
body $B$ the refrigerator — two reservoirs of caloric at different
temperatures. Caloric taken from $A$ is given up to $B$. Vectorised from
the plate in *Réflexions sur la puissance motrice du feu* (Wikimedia
Commons, public domain).</i>

Carnot’s own comparison: the motive power of heat can be compared to
that of a waterfall. A waterfall has a maximum you cannot exceed,
whatever machine you put in the stream. The power depends on the
quantity of water and the height of the fall. Heat, likewise, depends on
the quantity of caloric and on what he calls the height of its fall —
the difference of temperature between the bodies exchanging it. In a
steam engine the boiler is the upper reservoir, the condenser the lower.
Carnot’s own plate is that mill: body $A$ the furnace, body $B$ the
refrigerator, the cylinder the wheel. The caloric developed in the
firebox is taken up by the steam, passes through the cylinder, and is
given up to the cold water. The mill turns because the caloric *moves*.
The same quantity that left the hot reservoir arrives at the cold one.
That is Black’s ledger, used as the mechanics of an engine: a conserved
fluid falling between two baths.

In the summer of 1832 Carnot falls ill; he dies of cholera on 24 August,
at Ivry, aged 36. The *Réflexions* have made no impression. Two years
later Émile Clapeyron, another polytechnicien, writes the argument in
the language of indicator diagrams, and the book becomes findable.

## Count Rumford’s Cannon

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/count-rumford-cannon.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/count-rumford-cannon.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Carnot treated heat as a conserved fluid because that is what a mill
requires. The difficulty with that picture was already in print. In
1798, twenty-six years before the *Réflexions*, Benjamin Thompson, who
styled himself Count Rumford, published a paper in the *Philosophical
Transactions* on heat excited by friction (Thompson, 1798).

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/count-rumford.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>Benjamin Thompson, Count Rumford (1753–1814). American-born,
then in the service of the Elector of Bavaria. Portrait by Moritz
Kellerhoven (Wikimedia Commons, public domain).</i>

Rumford was the superintendent for the boring of brass cannons in the
military arsenal at Munich. He is struck by how hot the gun becomes, and
by how much hotter the metallic chips are (he checks and they are hotter
than boiling water). His paper to the royal society asks where does the
heat come from?

He asks whether the chips have changed their capacity for heat, as
Black’s latent-heat doctrine would require if caloric were being
squeezed out of the metal, but he finds they haven’t. He builds an
experiment with a short hollow cylinder is turned against a blunt steel
borer by machinery carried round by horses. A wooden box filled with
water surrounds the cylinder. At one hour the water is at
$107^\circ\mathrm{F}$; at two hours $178^\circ\mathrm{F}$; at two hours
thirty minutes it boils.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/rumford-cannon-boring.jpg" style="width:55%; height:auto; display:block;">

Figure: <i>Rumford’s 1798 apparatus, as reprinted in *The Complete Works
of Count Rumford* (1870). Fig.~2: the cannon in the boring machine,
turned by the bar $w$. Fig.~3: the same cylinder in the water box that
is brought to the boil.</i>

Rumford remarks that the the source of the heat generated by friction
appeared evidently to be *inexhaustible*. But anything which an
insulated body can continue to furnish without limitation cannot
possibly be a material substance. That is 1798. It is a qualitative blow
to caloric, not a mechanical equivalent. He does not weigh a known
quantity of work against a known quantity of heat. He has shown
generation. He has shown that the thermodynamic ledger doesn’t close.

## Mercury, Ice, Steam, and Air

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/fahrenheit-celsius-air-thermometer.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/fahrenheit-celsius-air-thermometer.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Black already treats temperature as a *reading* and heat as a
*quantity*. The reading, in the 1840s, comes from one of three expansion
thermometers. Fahrenheit and Celsius use mercury. Accurate laboratory
work uses air. All three pin two fixed points — ice and steam, or a
close cousin — and then divide the interval so that equal expansions of
the chosen substance are equal degrees. Thomson will inherit that
practice, and then refuse it as a definition of a degree.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/fahrenheit-thermometer.jpg" style="width:42%; height:auto; display:block;">

Figure: <i>Daniel Gabriel Fahrenheit (1686–1736) with a mercury
thermometer. Anonymous engraving, 18th century (Wikimedia Commons,
public domain). The instrument is mercury in glass; the scale is equal
expansion of that mercury between two pins.</i>

Fahrenheit’s mercury thermometer is the household instrument of the
eighteenth century. He marks a mixture of ice, water and sal-ammoniac as
$0$, and (originally) blood heat as $96$. The scale that survives is a
recalibration of the same tube: ice at $32$, steam at $212$, so there
are $180$ equal expansions of mercury between the two pins. Celsius, in
1742, uses the same pins on a centesimal division. His own numbering ran
the other way — $100$ at ice, $0$ at steam — and was inverted within a
few years. By Thomson’s time \`\`Celsius’’ means ice $0$, steam $100$.
The physics is the same as Fahrenheit: you have chosen a substance, and
you have called equal expansions of it equal degrees. Black’s whole
point is that this reading is *not* the quantity of heat. The same rise
of the same mercury takes different heat in water, in oil, in ice that
is melting.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/air-thermometer-scheme.png" style="width:72%; height:auto; display:block;">

Figure: <i>(a) Constant *pressure*: a sliding mercury index, open to the
room; the volume of the trapped air is the reading. (b) Constant
*volume*: a mercury manometer; the height $h$ is the reading. Scheme by
P. Peterlin (Wikimedia Commons, CC BY-SA 3.0). Thomson’s 1848 comparison
is with (a).</i>

The air thermometer is the 1840s laboratory standard, the scale to which
mercury and alcohol are reduced (Regnault). Take a glass bulb of dry air
and confine it with mercury. The mercury is open to the room: either a
sliding thread in a capillary, as in (a), or a cistern you raise and
lower. You move the mercury until the two surfaces sit at the same
height, so the trapped air is at atmospheric pressure. The *volume* of
that air is then the reading. Ice water and steam still fix two points;
the interval is divided so that equal increases of volume are equal
degrees. That is Gay-Lussac made into an instrument: far from
condensation, the volume of air rises almost linearly with this
temperature. Push the same line far enough and the volume would vanish
at about $-273^\circ$ of the air scale. That number lives in the
instrument, not in Carnot. Panel (b) is the other common form, a
constant-*volume* gas thermometer, where you read pressure from the
manometer height. Thomson in 1848 is arguing with (a). Air is a better
expansion-substance than mercury — different gases agree closely unless
they are about to liquefy — but it is still a substance. A degree that
means \`\`equal expansion of air’’ is not yet independent of the
thermometer.

## Thomson’s Scale from the Mill

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/kelvin-absolute-scale.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/kelvin-absolute-scale.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

William Thomson, who later became Lord Kelvin, takes Carnot seriously as
theory. In 1848 he is a twenty-four year old professor of natural
philosophy at Glasgow, and he has read Clapeyron’s recasting of the
*Réflexions*. Fahrenheit, Celsius and the air thermometer already exist:
ice and steam already mark $0$ and $100$ on the scale he will compare
against. Those pins are a century old. What does not exist is a meaning
of *degree* that is independent of mercury, or of air. If the motive
power of heat depends only on the quantity of caloric and the height of
its fall, then that height should be usable as a thermometer (Thomson,
1848). That is what he adds at twenty-four: a definition of the
interval, not a new pair of fixed points, and not a bottom of the fall.
The 1848 answer still assumes that heat is conserved. It is not yet the
scale that later bears his name.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/william-thomson-kelvin.jpg" style="width:38%; height:auto; display:block;">

Figure: <i>William Thomson (1824–1907), later Lord Kelvin. Photograph
from *The World’s Work*, 1902 (Wikimedia Commons, public domain) — half
a century after the 1848 scale.</i>

In 1848 Thomson is looking for a thermometric scale that is not an
accident of mercury or air. He believes that Carnot’s theory provides
such a scale. He already has a quantity of heat: Black’s unit, the heat
that raises a kilogramme of water by one degree of the air thermometer.
What he still lacks is a meaning for an *interval* of temperature that
does not belong to one substance. Carnot’s mill supplies it: the motive
power of a given quantity of caloric falling through one degree. That is
mill-work from a *fall*, as water on a wheel.

The air thermometer is the scale Thomson is arguing with. Household
thermometers use mercury because it is convenient; in accurate work you
compare the mercury against air and quote the air reading (Regnault).
Air is a better expansion-substance than mercury, but it is still a
substance. A degree that means \`\`equal expansion of air’’ is not
independent of the thermometer. Thomson will not call that absolute.

What he adds is a meaning for the interval. Take a unit of heat — Black
already told you how to measure that — and let it fall through Carnot’s
mill. Call one degree whatever drop of temperature always yields the
same motive power from that unit, whether you are near steam or near
ice. The degree now belongs to the mill, not to mercury, and not to air.
Independence of substance here means independence of the thermometer’s
fluid and of the engine’s working fluid. The fixed points stay where
Fahrenheit and Celsius left them. That is why the paper matters in 1848,
before anyone has broken conservation of heat.

Thomson’s own statement of the characteristic property: all degrees have
the same value; a unit of heat descending from a body $A$ at temperature
$T^\circ$ to a body $B$ at $(T-1)^\circ$ gives out the same mechanical
effect, whatever the number $T$. That is the mill, turned into a ruler.
The working fluid is still caloric. The fall is still a fall, not a
conversion. He then *compares* the new ruler with the air thermometer,
using Clapeyron’s formulae and Regnault’s steam data. The two scales are
forced to agree at ice and at steam. Everything else is a table of how
unequal an air-thermometer degree is, in mill-value, as you go up the
scale. Work is how we make the degrees equal. It is not how we make the
heat, and it is not a new zero.

The $-273^\circ$ that looks like absolute zero is already sitting in the
air thermometer — the same bulb of air, volume as the reading. If that
volume is linear in temperature, pushing the graduation far enough
reduces the air to nothing at $-100/0.366$, which Thomson writes as
$-273^\circ$. He uses that fact only as a reason why Clapeyron’s table
was to be expected: an air-thermometer degree is *worth more* when it is
cold, because a finite number of those degrees is trying to reach
infinite cold. It is not the zero of his scale, and he does not compute
a table down to it. On the mill scale a unit of heat can always be let
down one more equal-value degree, and still arrive. Infinite cold is at
$-\infty$. 1848 has done something narrower: it has made a degree a
Carnot object. There is not yet a bottom, because caloric is still
conserved.

Thomson writes that in the present state of science no operation is
known by which heat can be absorbed without either raising a temperature
or becoming latent. Conversion of heat, or caloric, into mechanical
effect is probably impossible, certainly undiscovered. Actual engines
must therefore look for their power not in absorption and conversion,
but in the transmission of heat. That is Carnot, stated as a fact of
1848, and it is the assumption his mill-degree rests on.

## Joule and the Mechanical Equivalent

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/joule-mechanical-equivalent.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/joule-mechanical-equivalent.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

James Prescott Joule (1818–1889) gives the answer to Rumford’s riddle.
He theorises that a measured quantity of mechanical work could *create*
a measured quantity of heat. If it can then heat is not a conserved
fluid. This would imply that Carnot’s mill becomes a special case, not
the general law. Joule is a Manchester brewer whoe tries this experiment
with a paddle wheel (Joule, 1850).

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/james-prescott-joule.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>James Prescott Joule (1818–1889). Seated portrait, later in
life (Wikimedia Commons, public domain).</i>

The family brewery is in Manchester where Joule has a private
laboratory. From 1840 he is working to relate heat to electricity and to
chemical affinity. In 1843 he argues that the heat from
magneto-electricity is proportional to the work absorbed. Mayer, in
1842, has already claimed that agitating water warms it, without a
measured force. Joule wants the equivalent.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/joule-apparatus-harpers.png" style="width:48%; height:auto; display:block;">

Figure: <i>Joule’s apparatus for the mechanical equivalent of heat. A
falling weight turns a paddle in water; the work is known, the rise in
temperature is measured. Engraving from *Harper’s New Monthly Magazine*,
August 1869 (Wikimedia Commons, public domain).</i>

In Joules’ apparatus weights descend and turn a paddle in a can of
water. The paddle moves against the water’s resistance. When the weights
have run down they are wound up and the motion is repeated. The
thermometer is read after multiple descents. The work is easily
calculated as the weight ($\text{mg}) multiplied by the height ($)\$.
The heat is the rise in temperature times the capacity of the water.
Mechanical power is being *spent*, and heat appears.

Cambridge here is the British Association meeting of June 1845, not the
Cambridge Philosophical Society — that society is Thomson’s 1848 scale.
Joule exhibits the brass paddle to the Chemical Section; the published
letter opens by saying so (Joule, 1845). That run gives about $890$
foot-pounds per degree per pound of water. He also uses other methods
such as magneto-electricity, rarefaction of air, passing water through
narrow tubes to give neighbouring figures. His mean estimate is near
$817$. By 1850 the number is settled as $772.692$ foot-pounds (Joule,
1850). The scientific world, he notes, preoccupied with the hypothesis
that heat is a substance, had almost unanimously denied that fluid
friction could generate heat.

The contradiction is the cue, not the resolution. Clausius in 1850, and
Thomson in 1851, will keep Carnot’s *limit* and drop conservation of
heat. We take Clausius first: two laws, and $\mathrm{d}Q/T$. Thomson
then returns to rewrite the scale so that it has a bottom.

## Clausius and $\mathrm{d}Q/T$

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/clausius-carnot-second-law.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/clausius-carnot-second-law.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Heat is not conserved. Carnot’s limit remains. That is the problem
Clausius inherits from Joule and Thomson. If the caloric mill is gone,
what still closes when a reversible engine runs a cycle? The answer is
not a quantity of heat. It is heat divided by temperature. A reversible
cycle produces no entropy.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/rudolf-clausius.jpg" style="width:40%; height:auto; display:block;">

Figure: <i>Rudolf Clausius (1822–1888). Photograph from the AIP Emilio
Segrè Visual Archives, Lande Collection (Wikimedia Commons, CC0).</i>

Clausius is a mathematical physicist, first in Berlin, later Zurich,
Würzburg and Bonn. In 1850, independently of Thomson’s 1851 dynamical
theory, he writes the two laws that keep Joule and Carnot in the same
ledger (Clausius, 1850). The first law is Joule’s equivalent made
universal: heat is a form of work. The second is what remains of the
mill: heat does not pass from cold to hot by itself, and no cyclic
engine converts the heat of a single bath entirely into work.

The first law is why $\oint \mathrm{d}Q$ is not zero for an engine. Some
of the heat taken from the hot bath does not arrive at the cold bath; it
has become work. That is Joule, inside Carnot’s cycle. The second law is
why you still cannot take *all* of it. There is a remainder that must be
rejected to the cold bath, and the size of that remainder, for a
reversible engine, depends only on the two temperatures — not on whether
the working substance is steam, air, or anything else. That was Carnot’s
other claim, and it survives.

Thomson’s 1851 scale is this statement: the absolute temperatures stand
in the same ratio as the two heats. Write it the other way up and the
books close. The heat that leaves the hot bath, divided by $T_h$, equals
the heat that arrives at the cold bath, divided by $T_c$. The
*difference* of the heats is the work. The *quotient* of each heat by
its temperature is the same number. That number is the transformation.
Clausius’s 1854 paper calls it the equivalence-value of the
transformation (Clausius, 1854). We are looking at a two-bath cycle; the
same accounting will apply to a loop assembled from many small
exchanges.

If every reversible cycle gives zero, the integral of $\mathrm{d}Q/T$
between two states cannot depend on the reversible path you take. That
is the definition of a state function. Clausius writes $$
S - S_0 = \int \frac{\mathrm{d}Q_{\mathrm{rev}}}{T}.
$$ Heat $\mathrm{d}Q$ is not a state function: the heat you put in
depends on the path, which is why $\oint \mathrm{d}Q$ is the work of the
engine. Dividing by $T$ along a *reversible* path removes that
path-dependence. Reversibility is the condition. It means the engine can
be run backwards and the baths restored: no finite drops of temperature,
no friction, no paddle wheel hidden in the mechanism.

This is the mill, rewritten after Joule. The working fluid comes home.
Its entropy comes home with it. The hot bath has lost $Q_h/T_h$; the
cold bath has gained $Q_c/T_c$; those two are equal. No transformation
has been left over. Reversible does not mean \`\`nothing happened.’’
Work was done. It means the transformation can be undone, and when you
close the loop there is no residue.

An irreversible cycle is still a cycle: the working substance returns,
so its $S$ is unchanged. What has changed is the baths. You took $Q_h$
from the hot bath and, because some of the fall was wasted, you
delivered *more* than the reversible $Q_c$ to the cold bath. The
integral $\oint \mathrm{d}Q/T$, with $\mathrm{d}Q$ the heat absorbed by
the working substance, is then strictly negative (Clausius, 1854).
Equivalently, the entropy lost by the hot bath is less than the entropy
gained by the cold bath. The difference was produced. That is the second
law as an accounting inequality, not as a prohibition in words.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/clausius-coins-entropy.jpg" style="width:62%; height:auto; display:block;">

Figure: <i>Clausius names entropy, 1865. English translation of the
memoir: $\int \mathrm{d}Q/T = S-S_0$, from Greek τροπή,
*transformation*. He forms the word to sit beside *energy*.</i>

In 1865 he gives the state function a name (Clausius, 1865).
*Verwandlungsinhalt*, the transformation content, would have done; he
prefers a Greek root, τροπή, so that the word can pass unchanged between
languages, and so that it sits next to *energy*. The two quantities, he
says, are so nearly allied in their physical meanings that a similarity
in designation appears desirable. Energy is the first-law account.
Entropy is the second-law account. His summary of the two laws: the
energy of the universe is constant; the entropy of the universe tends to
a maximum.

Clausius did not give the Boltzmann distribution, and he did not make
entropy a measure of uncertainty. He gave the macroscopic balance that
any later prescription must respect. When we write
$p_i \propto e^{-\beta E_i}/Z$ and later derive it from MaxEnt, read it
as a statistical answer to a constraint he already framed: fixed mean
energy, maximum entropy, no perpetual motion. The historical chain is
engines, then the equivalent, then this integral, then statistics, then
information.

## Thomson Returns: The Scale After Joule

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/kelvin-dynamical-theory.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/kelvin-dynamical-theory.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Clausius has kept Carnot’s limit and dropped conserved heat. Thomson
does the same, independently, a year later. In 1851 he writes the
dynamical theory of heat (Thomson, 1851). The mill is still there. The
ruler is not. Heat taken in and heat rejected now stand in a *ratio* of
the two temperatures, and for the first time the fall has a bottom.

The 1848 paper had already heard of Joule, in a footnote, and had
refused him: conversion of heat into work was probably impossible,
certainly undiscovered. The paddle wheel is the experiment Thomson said
had not been shown. By 1851 he accepts it. Heat can be generated by
work, and some of the heat taken from the hot bath does not arrive at
the cold bath — it has become work. Conservation of caloric cannot be
the general law. Clausius had already said so. Thomson’s contribution,
for us, is what happens to the *scale*.

In 1848 a degree was a height of fall of a conserved unit of heat: the
same mill-work for every rung, whatever $T$ you started from. After
Joule the unit does not arrive. What *is* equal, for a reversible
engine, is $Q_h/T_h$ and $Q_c/T_c$. That is Clausius’s entropy leaving
the hot bath and arriving at the cold bath. Thomson takes it as the
definition of the scale. Ice is no longer the zero. The numbers $T_h$
and $T_c$ are the absolute temperatures of the two baths; their ratio is
the ratio of the heat taken in to the heat rejected.

This is the genuine floor that 1848 did not have. If
$Q_c/Q_h = T_c/T_h$, then $T_c=0$ is the temperature at which a
reversible engine, taking $Q_h$ from the hot bath, rejects nothing. The
$-273^\circ$ of Gay-Lussac’s expansion was an accident of air. The zero
of the 1851 scale is an accident of *engines*: it is the temperature you
cannot fall below, because there is no heat left to reject. We still use
this scale. We name it after the title Thomson later took.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/boltzmann2.jpg" style="width:100%; height:auto; display:block;">

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/j-w-gibbs.jpg" style="width:100%; height:auto; display:block;">

Figure: <i>James Clerk Maxwell (1831–1879), Ludwig Boltzmann
(1844–1906), Josiah Willard Gibbs (1839–1903). After Clausius the
account becomes statistical: components of velocity, then occupation of
energy, then free energy.</i>

Clausius gave a macroscopic balance. Maxwell, Boltzmann and Gibbs give a
microscopic count of the same $S$. Maxwell’s first move is the velocity
distribution: each Cartesian component is Gaussian, because the kinetic
energy is quadratic. That quadratic is the energy whose expectation we
will split as $U=F+TS$.

## Maxwell–Boltzmann: Components Versus Speed

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-boltzmann-components-vs-speed.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/maxwell-boltzmann-components-vs-speed.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Write the joint density on Cartesian velocities as isotropic, Each
marginal $f(v_x)$ is Gaussian. Change to spherical coordinates with
$v=|\mathbf{v}|$ and solid angle $\Omega$. Probability conservation uses
the volume element $d^3v = v^2\,dv\,d\Omega$. Integrating over angles
($\int d\Omega = 4\pi$) gives the Maxwell speed distribution The factor
$v^2$ is the Jacobian of the change of variables — the density of states
of spherical shells in velocity space. Calling \`\`Maxwell–Boltzmann’’
Gaussian is correct for a component and wrong for speed.

$$
f(v_x)=\sqrt{\tfrac{m}{2\pi kT}}\,e^{-mv_x^2/2kT}
\qquad\text{versus}\qquad
f(v)\propto v^2\,e^{-mv^2/2kT}
$$

## The Gibbs–Boltzmann Distribution

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/gibbs-distribution.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/gibbs-distribution.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Name the object and fix the formula. The MaxEnt derivation that *earns*
these weights is a separate step: deriving the occupation from Lagrange
multipliers before Shannon $H$ and Jaynes would put the cart before the
horse.

Write the equilibrium occupation of a discrete system with energies
$\{E_i\}$ as $$
p_i = \frac{e^{-\beta E_i}}{Z(\beta)},
\qquad
Z(\beta)=\sum_j e^{-\beta E_j}.
$$ Physicists call $p_i$ the *Boltzmann distribution* (or Boltzmann
weights) and, for a system exchanging energy with a bath at fixed $T$,
the *Gibbs distribution* or *canonical ensemble*. The three names point
at the same formula. The normalisation $Z(\beta)$ is the *partition
function*. Its logarithm $\log Z(\beta)$ is the cumulant generating
function for the energy under this exponential family: derivatives of
$\log Z$ recover the mean energy, the variance (heat capacity, up to
factors of $\beta$), and higher cumulants. That generating-function
reading is why $Z$ is more than a normalisation constant.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
energies = np.array([0.0, 1.0])
beta = np.linspace(0.1, 3.0, 200)
Z = np.sum(np.exp(-beta[:, None] * energies), axis=1)
p0 = np.exp(-beta * energies[0]) / Z
p1 = np.exp(-beta * energies[1]) / Z
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(beta, p0, linewidth=2, label='$p_0$ (ground)')
ax.plot(beta, p1, linewidth=2, label='$p_1$ (excited)')
ax.set_xlabel(r'coldness $\beta$')
ax.set_ylabel('occupation')
ax.legend()
ax.set_title('Two-state Gibbs--Boltzmann occupations')
mlai.write_figure('two-state-boltzmann.svg', directory='./physics')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/physics/two-state-boltzmann.svg" class="" width="75%" style="vertical-align:middle;">

Figure: <i>Occupation of a two-state system as coldness increases. At
low $\beta$ both states are populated; at high $\beta$ the ground state
dominates.</i>

The Gibbs–Boltzmann occupation is not only a statement about gases and
magnets. Hopfield networks (Hopfield, 1982) assign an energy to every
binary configuration of a recurrent net and treat recall as a descent
toward low-energy states — equilibrium statistics are again Gibbs.
Ackley, Hinton and Sejnowski (Ackley et al., 1985) made the weights of
that energy *learnable*: a Boltzmann machine is an undirected model
whose distribution over configurations is exactly
$p(s)\propto e^{-E(s)/T}$. The 2024 Nobel Prize in Physics, awarded to
John Hopfield and Geoffrey Hinton, recognised that physical-systems
reading of computation and learning. Name the lineage so the formula
does not feel confined to nineteenth-century heat baths; do not divert
into training algorithms.

## Coldness and Temperature

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/coldness-and-temperature.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/coldness-and-temperature.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The Boltzmann occupation can be written in two ways, $$
p_i = \frac{e^{-E_i/k_B T}}{Z}
= \frac{e^{-\beta E_i}}{Z},
\qquad
\beta = \frac{1}{k_B T}.
$$ The mathematics is the same, but the emphasis is different.

$T$ is the variable of the bath. It is what a thermometer reports, and
it is the intensive parameter in the Helmholtz accounting $F = U - TS$.
In that representation energy comes first and entropy is the correction:
$TS$ is the cut the second law takes from $U$.

$\beta$ is *coldness*. It is the variable conjugate to energy. In that
representation entropy comes first: you maximise $S$ (or write a
generating function in $\beta$) and energy is the constraint.
Temperature is then a derived reading,
$T = 1/k_B\beta = (\partial S/\partial U)^{-1}$.

We will use both. Weeks 1–2 keep $T$ in view so the bath is familiar,
and they already compute in $\beta$ because that is the natural argument
of $Z$. Week 5 makes the $\beta$-first order honest: the Lagrange
multiplier on a mean-energy constraint *is* coldness, and it is the
natural parameter of the exponential family. Weeks 6–7 then treat
$\beta$ as a coordinate on that family. A student who still hears
$\beta$ as \`\`one over the thermometer’’ will miss why the geometry is
written in those coordinates.

## Free Energy Decomposition

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/free-energy-decomposition.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/free-energy-decomposition.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Once the Gibbs occupation $$
p(x)=\frac{e^{-\beta E(x)}}{Z},\qquad Z=\sum_x e^{-\beta E(x)}
$$ is given, we can derive the macroscopic formulae that characterise
the gas. They come to us from Helmholtz. We can compute them through the
expectation of that occupation of the given engergies, taken at
equilibrium. The relationship between energy and probability is
$p_i\propto e^{-\beta E_i}$. Here we will identify the “inner energy” or
“total energy” with $\left\langle E\right\rangle$ and read off $F=U-TS$.

Start from the occupation and take the logarithm, $$
\log p(x)=-\beta E(x)-\log Z.
$$ Note that nothing statistical has happened here, this is an identity
for every probability that is a member of the exponential family, which
in physics we think of as the Gibbs distribution. It’s valid for every
configuration $x$ that the Gibbs distribution supports. If we solve for
the energy of that configuration we can restore temperature through
recognising that$\beta=1/(k_BT)$, $$
E(x)=-k_BT\log p(x)-k_BT\log Z.
$$ Note that the first term on the right (the surprisal multiplied by
the temperature and Boltzmann’s constant) still depends on $x$. The
second does not: $-k_BT\log Z$ is a property of the whole ensemble, not
of a particular microstate.

At equilibrium the macroscopic inner energy can be related to the
expectation of the microscopic score under the same distribution, $$
U\equiv\left\langle E\right\rangle=\sum_x p(x)\,E(x).
$$ This is where the *ensemble* is entering in physics. The ensemble
that comes to us from Boltzmann, Maxwell, Gibbs, and Einstein made. The
idea is that the system is not in one configuration. The energy we write
in the “thermodynamic ledger” is an average over the different
occupations. Substitute the identity for $E(x)$, The second term comes
out of the sum because $\log Z$ does not depend on $x$, and
$\sum_x p(x)=1$.

The two remaining pieces already have names. The
configuration-independent term is the *Helmholtz free energy*, $$
F=-k_BT\log Z,
$$ it is often known as the “free energy” or “available energy”. It is
the energy that we can use for work. The average of $-\log p$ is the
dimensionless Shannon functional $H=-\sum_x p(x)\log p(x)$. Multiplying
by Boltzmann’s constant produces physical entropy, $S=k_B H$, with units
of energy per degree. Then $$
U=F+TS\qquad\text{or}\qquad F=U-TS.
$$ Helmholtz originally wrote $A=U-TS$, with $A$ for *Arbeit*, the work
still available. This module uses $F$ for the same quantity, because
week 5 needs $A(\boldsymbol{\theta})$ for the log-partition of an
exponential family. But Helmholtz’s $A$ and our $F$ are the same object;
$U$ is the inner energy $\left\langle E\right\rangle$, not a free
energy. The fact that they are different is down to energy that is
stored as entropy. It’s in this term that we can find Black’s latent
heat.

From the physics and thermodynamics perspective $U$ is the energy in the
system, $TS$ is the “cut” that entropy takes from that total energy. It
is the tied up in the uncertainty of the occupation, unavailable as
work. Then $F$ is what the thermodynamic bath still allows you to do.
Ggiven the Gibbs occupation, the macroscopic split $F=U-TS$ is an
identity.

The $T$-first reading puts energy first and treats entropy as a
correction. The $\beta$-first reading treats $\log Z$ as a generating
function: $U=-\partial_\beta\log Z$ and $S=\beta(U-F)$, which we compute
from $Z$ later. Both are the identity just derived. Later we will derive
the occupation from MaxEnt and identify $\beta$ as the Lagrange
multiplier on the mean-energy constraint; until then the occupation is
given and the accounting follows.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
energies = np.array([0.0, 1.0, 3.0])
beta = np.linspace(0.05, 2.5, 200)
Z = np.sum(np.exp(-beta[:, None] * energies), axis=1)
p = np.exp(-beta[:, None] * energies) / Z[:, None]
U = (p * energies).sum(axis=1)
S = -np.sum(p * np.log(p + 1e-300), axis=1)
F = U - S / beta
F_check = -np.log(Z) / beta
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(beta, U, label=r'$U(\beta)$')
ax.plot(beta, S, label=r'$S(\beta)$ (nats, $k_B=1$)')
ax.plot(beta, F, label=r'$F(\beta)$')
ax.plot(beta, F_check, 'k--', alpha=0.5, label=r'$-\ln Z/\beta$')
ax.set_xlabel(r'$\beta$')
ax.legend()
ax.set_title(r'Three-state system: energies $E\in\{0,1,3\}$')
mlai.write_figure('three-state-thermo.svg', directory='./ml')
```

Figure: <i>$U$, $S$, and $F$ for the Worksheet 1 three-state system
($k_B=1$). The dashed curve is $-\log Z/\beta$; it sits on $F$, which is
the identity just derived.</i>

The Worksheet 1 three-state system, $E\in\{0,1,3\}$, is a numerical
check of the same algebra with $k_B=1$. Compute $p_i=e^{-\beta E_i}/Z$,
then $U=\sum_i p_i E_i$ and $S=-\sum_i p_i\log p_i$. The two expressions
for free energy, $F=U-S/\beta$ and $F=-\log Z/\beta$, overlie. If they
do not, the signs in the derivation are wrong.

```
import numpy as np
```

```
def gibbs_thermo(energies, beta):
    """U, S, F from a discrete Gibbs occupation ($k_B=1$)."""
    e = np.asarray(energies, dtype=float)
    weights = np.exp(-beta * e)
    Z = weights.sum()
    p = weights / Z
    U = np.sum(p * e)
    S = -np.sum(p * np.log(p))
    F = -np.log(Z) / beta
    return U, S, F, U - S / beta

U, S, F, F_from_US = gibbs_thermo([0.0, 1.0, 3.0], beta=1.0)
U, S, F, F_from_US
```

## Gaussian: Expected Quadratic Energy

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/gaussian-quadratic-free-energy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/gaussian-quadratic-free-energy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

We’ve already introduced the univariate Gaussian as an energy-based
model with quadratic energy. We also looked at animations of elastic
collisions and argued that the velocities were given by the
Maxwell-Boltzmann distribution that has the form of a Gibbs
distribution. We can apply the Helmholtz split to this system in closed
form. Take $$
E(x)=\frac{x^2}{2}
$$ and a bath at inverse temperature $\beta$ (here $k_B=1$, so
$T=1/\beta$). The Gibbs density is $$
p(x)=\frac{e^{-\beta x^2/2}}{Z}=\sqrt{\frac{\beta}{2\pi}}\,\exp\Bigl(-\beta\frac{x^2}{2}\Bigr),
$$ which is $\mathcal{N}(0,T)$. The partition function is the Gaussian
integral $$
Z=\int_{-\infty}^{\infty}e^{-\beta x^2/2}\,\mathrm{d}x=\sqrt{\frac{2\pi}{\beta}}.
$$

The inner energy is the expectation of the quadratic, $$
U=\left\langle E\right\rangle=\int p(x)\,\frac{x^2}{2}\,\mathrm{d}x=\frac{1}{2}\left\langle x^2\right\rangle.
$$ Under $p=\mathcal{N}(0,T)$ we have $\left\langle x^2\right\rangle=T$,
so $U=T/2$. That is the equipartition theorem for a single quadratic
degree of freedom: each such term in the energy contributes $k_BT/2$ at
equilibrium. Maxwell’s three Cartesian components — a few slides back —
are three independent copies of this quadratic. Kinetic energy
$\tfrac32 k_BT$ is $d=3$. The Jacobian that produces *speed* is why the
speed density is not itself Gaussian.

The free energy is the generating-function term from the derivation, $$
F=-T\log Z=-T\log\sqrt{2\pi T}=-\frac{T}{2}\log(2\pi T).
$$ Entropy follows from $S=(U-F)/T$, $$
S=\frac{1}{2}+\frac{1}{2}\log(2\pi T).
$$ That is the differential entropy of $\mathcal{N}(0,T)$, in nats, with
$k_B=1$. Here $S$ is simply the name we gave to
$-k_B\left\langle\log p\right\rangle$ when we took the expectation of
$E(x)$. The continuous functional can be negative and is not invariant
to a change of units. The algebra we need is already visible: $$
F+TS=-\frac{T}{2}\log(2\pi T)+T\Bigl(\frac{1}{2}+\frac{1}{2}\log(2\pi T)\Bigr)=\frac{T}{2}=U.
$$ The $\tfrac12$ in $U$ and the $\tfrac12$ sitting in $T\cdot\tfrac12$
— the $e$ inside $\tfrac12\log(2\pi e T)$ — cancel, leaving
$F=-T\log Z$. Spreading the Gaussian (raising $T$) does not change the
*shape* of the well $E(x)=x^2/2$; it changes how far the occupation
climbs the walls, and therefore how much of $U$ is locked in $TS$.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
T = np.linspace(0.15, 3.0, 400)
U = T / 2.0
S = 0.5 + 0.5 * np.log(2.0 * np.pi * T)
TS = T * S
F = -0.5 * T * np.log(2.0 * np.pi * T)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(T, U, label=r'$U=T/2$')
ax.plot(T, TS, label=r'$TS$')
ax.plot(T, F, label=r'$F=-T\log Z$')
ax.axhline(0.0, color='#666', ls=':', lw=1)
ax.set_xlabel(r'$T$ ($k_B=1$)')
ax.legend()
ax.set_title(r'Quadratic $E(x)=x^2/2$: $U=F+TS$')
mlai.write_figure('gaussian-quadratic-thermo.svg', directory='./ml')

xs = np.linspace(-4.0, 4.0, 400)
Ex = 0.5 * xs ** 2
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(xs, Ex, 'k', lw=2, label=r'$E(x)=x^2/2$')
for Tval, colour in [(0.3, '#4C72B0'), (1.0, '#DD8452'), (2.5, '#55A868')]:
    ax.axhline(Tval / 2.0, color=colour, ls='--', lw=1.5,
               label=rf'$\langle E\rangle=T/2$ at $T={Tval:.1f}$')
ax.set_xlabel(r'$x$')
ax.set_ylabel(r'$E(x)$')
ax.set_ylim(0.0, 4.0)
ax.legend()
ax.set_title(r'Occupation climbs the well as $T$ rises')
mlai.write_figure('gaussian-quadratic-well.svg', directory='./ml')
```

Figure: <i>The energy is the same parabola at every temperature. Raising
$T$ spreads $p=\mathcal{N}(0,T)$ and the occupation climbs the walls:

<img src="https://mlatcl.github.io/iei/slides/diagrams/ml/gaussian-quadratic-thermo.svg" class="" width="75%" style="vertical-align:middle;">

Figure: <i>Closed form for $E(x)=x^2/2$ with $k_B=1$. Inner energy
$U=T/2$ is the straight line; $TS$ takes a growing cut; $F=U-TS$ is

```
import numpy as np
```

```
def gaussian_quadratic_thermo(T):
    """U, S, F for $E(x)=x^2/2$ at temperature $T$ ($k_B=1$)."""
    U = T / 2.0
    Z = np.sqrt(2.0 * np.pi * T)
    F = -T * np.log(Z)
    S = (U - F) / T
    return U, S, F

T = 1.0
U, S, F = gaussian_quadratic_thermo(T)
U, S, F, np.isclose(U, F + T * S)
```

If the energy is a sum of independent quadratics,
$E(\mathbf{x})=\tfrac12\sum_{i=1}^d x_i^2$, the partition function
factors and $$
U=\frac{d\,T}{2},\qquad
Z=(2\pi T)^{d/2},\qquad
F=-\frac{d T}{2}\log(2\pi T).
$$ Each coordinate contributes its own $T/2$ to the inner energy and its
own $\tfrac12\log(2\pi e T)$ to the entropy. That is why an ideal gas of
$N$ particles in three dimensions has kinetic energy $\tfrac32 N k_BT$:
$3N$ quadratic terms in the Hamiltonian, each holding $k_BT/2$. The
derivation did not change. We took the expectation of a bigger $E$.

The air thermometer is a bulb of gas whose *volume*, at constant
pressure, is the reading. Thomson already had Gay-Lussac’s wall: if that
volume is linear in temperature it vanishes at about $-273^\circ$. The
quadratic well we just split has no box, so it cannot be that
thermometer. Maxwell’s three Cartesian components *are* three copies of
$E=v_i^2/2$: the kinetic half of an ideal-gas particle is already on the
board. What is missing is the configuration integral. A free particle in
a container of volume $V$ has, up to constants of mass and $h$ that we
do not need, $$
Z_1 = V\,(2\pi T)^{3/2}.
$$ The Gaussian integrals we just evaluated are the momentum half; $V$
is the position half. For $N$ particles, $Z \propto V^N T^{3N/2}$, so
$F=-T\log Z$ depends on $V$. Pressure is a Helmholtz derivative, not a
new potential: $$
P=-\Bigl(\frac{\partial F}{\partial V}\Bigr)_{T,N}=\frac{NT}{V}.
$$ At constant $P$, $V\propto T$. Gay-Lussac’s wall is $T=0$ on the same
scale whose variance we just read off Maxwell’s Gaussians. Thomson in
1848 treated $-273^\circ$ as an accident of air. After Joule it is the
floor of the engine scale; after Maxwell it is also the temperature at
which those Gaussians have collapsed to a delta, and a finite pressure
would require zero volume.

We do not need the Gibbs free energy $G=F+PV$ for this. $G$ is the
potential when *pressure* is the control variable. Charles’s law is
already in $PV=NT$. Coupled multivariate Gaussians are not required
either: independent coordinates suffice, and that is what Maxwell wrote.
The one extra Helmholtz fact is the derivative with respect to $V$. We
name it. We do not compute Stirling, $h^3$, or chemical potential today.

## The Information Game

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Before writing a formula for entropy, play a game. One player thinks of
an object drawn from a known finite list. The other may ask only yes/no
questions. How many questions do you need in the worst case? That count
is already a measure of information: it is how much uncertainty you must
remove to name the object.

With a list of size $N$ and equal prior probability on each item, an
ideal strategy halves the search space at every step. After $k$ answers
you can distinguish at most $2^k$ hypotheses, so you need $$
k \ge \log_2 N
$$ questions in the worst case. Twenty questions reach a million
possibilities: $2^{20} = 1\,048\,576$. A *fair* yes/no answer therefore
carries one *bit* of information. That is Shannon’s unit before
Shannon’s formula: information as the number of balanced binary
distinctions required to resolve uncertainty.

If the prior is not uniform, the best question is no longer a literal
half-split of the *count* of remaining objects. It is a half-split of
their *probability*. The average number of questions under optimal play
becomes the Shannon entropy $H = -\sum_i p_i\log_2 p_i$. We will meet
that formula next; the game already shows what the number is counting.

## MacKay’s Balance and Balls

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

David MacKay opens the compression half of *Information Theory,
Inference, and Learning Algorithms* with a weighing puzzle (MacKay,
2003, p. Exercise~4.1). State the puzzle first. Do not yet say that the
instrument is ternary. The information-game habit is to ask how we
should proceed — what we must name — and then to try the instrument as
if it had two readings, like a yes/no question. Keep the *mechanism*
(what we put on the pans) on a different slide from the *maths* (how
many bits that use can return). Only after that sequence do we notice
that the balance has a third reading.

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/mackay-balance-balls.svg" class="" width="80%" style="vertical-align:middle;">

Figure: <i>A two-pan balance and twelve identical balls.</i>

You are given twelve balls, identical in appearance. Exactly one differs
in weight from the others; you do not know whether that odd ball is
heavier or lighter. A two-pan balance compares two equal-sized groups.
Design a strategy that finds the odd ball and the direction of the
defect.

Resist jumping to a particular grouping. Count first: how many states of
the world. Then try the instrument as a two-outcome device, because that
is the game we already know. Keep what we put on the pans separate from
the information budget those pans can return.

Under a uniform prior over those twenty-four possibilities, the Shannon
uncertainty we must remove is $$
H_{\mathrm{start}} = \log_2 24 \approx 4.58 \text{ bits}.
$$ At the end we need a single named hypothesis: $H_{\mathrm{end}} = 0$.
The *deficit* to close is therefore about $4.58$ bits. That difference —
start minus end — is the information the experiment must supply. We have
not yet said what the balance returns.

## Try Two Outcomes

A two-outcome instrument, used optimally, returns at most $\log_2 2 = 1$
bit per measurement. The information budget then says $$
2^4 = 16 < 24 \le 32 = 2^5,
$$ so five uses can suffice on paper and four cannot. Whether a physical
balance *achieves* that one bit is a different question: it depends on
what we put on the pans.

$$
\log_2 24 \approx 4.58,\qquad 2^4 = 16 < 24 \le 32 = 2^5
$$

Write two columns if you can: left, the information budget; right, the
pans. The left column does not know about equal group sizes, leftover
balls, or borrowing a ball of known mass. Those constraints live on the
right. The rest of this block is the right-hand column, scored
afterwards by the left.

The move 6 vs 6 is the binary-search instinct. If we already knew the
odd ball was heavier, it would be a perfect split of twelve balls. We do
not know that. A down-left reading means \`\`one of the left six is
heavy, *or* one of the right six is light’’: twelve of the original
twenty-four hypotheses remain. One bit toward the 24, and *no* ball has
been eliminated. Mechanism and maths have already come apart: the pans
moved, the pile of balls did not shrink.

The information gain can be perfect for the two-outcome model — an even
split of 24 — while the mechanism fails to name a set of six suspect
*balls*. What we carry forward is six maybe-heavy and six maybe-light.
If we now *pretend* it was heavier and keep only the down pan, we have
thrown away half of the remaining hypotheses by assumption, not by
measurement.

This is the sequence people sketch on the board: 6 vs 6, then 3 vs 3. It
is the right mechanism for a different puzzle — twelve balls, one
*heavier*. Hold that sequence, because the next move is where the pans
refuse to cooperate.

With three suspects we have two mechanical choices. Weigh 1 vs 1 and
leave one off: the left-off ball is the odd one *if the pans balance*.
That reading is not in the two-outcome story we are trying to tell. Or
weigh 2 vs 2 by padding with a ball already known to be normal: A,B
against C plus a normal. The pans are now full, so in the heavier-only
story they must tip, and we have not used a third symbol. We have
changed the split.

$$
\left(\tfrac{2}{3},\tfrac{1}{3}\right)
\quad\Rightarrow\quad
H = h_2\!\left(\tfrac{2}{3}\right) \approx 0.92 < 1 \text{ bit}
$$

If A,B,C are the three remaining heavier-suspects and N is known normal,
the reading A,B down leaves two hypotheses and has probability $2/3$;
C,N down names C and has probability $1/3$. Binary entropy
$h_2(2/3)\approx 0.92$ bits. The deficit $1-0.92$ is exactly the price
of the mechanical constraint \`\`equal numbers on the pans, and no third
reading’’. We did not lose those bits because the mathematics of two
outcomes changed. We lost them because the mechanism could not realise
an equiprobable two-way split of three objects.

1 vs 1 of two remaining heavier-suspects returns the missing bit: the
down pan is the odd ball, if the heavier-side story is true. The
sequence of *acts* was 6 vs 6, 3 vs 3, 2 vs 2 with a known, then 1 vs 1.
The sequence of *gains* was 1, 1, $0.92$, 1 — already not four bits. And
this named a ball only inside the heavier-side story.

To finish the original task we must name the direction as well as the
ball. The obvious mechanism is candidate versus known normal. If the
two-outcome story were true, this use would return the last bit: heavy
or light. But if we followed the down pan all the way and the odd ball
was actually light, the candidate may be normal — and then the pans
*balance*. The third reading has been sitting in the mechanism whenever
a known ball appears. We cannot keep it out without lying about what the
device does. That is the cue to abandon the two-outcome model, not to
patch it again.

The moral is not that the binary budget is wrong. It is that an
information budget is silent about pans. Equal group sizes, leftover
balls, and borrowed normals are constraints on the *channel we actually
have*. Score every weighing twice: what the device did, then how many
bits that split was worth.

## Three Readings

A two-pan balance has a ternary alphabet. The reading \`\`level’’ is not
an annoyance to be designed out; it is a symbol that can carry
information. The first move of the two-outcome story, 6 vs 6, sets the
probability of that symbol to zero: with exactly one odd ball among the
twelve, the pans cannot balance. We spent a mechanism choice to turn a
three-outcome device into a two-outcome device, and then paid further
mechanism taxes (the known ball) to keep the third symbol from coming
back.

$$
\log_2 3 \approx 1.58,\qquad 3^2 = 9 < 24 \le 27 = 3^3
$$

A binary question yields at most one bit. If the three readings are
equiprobable, one weighing contributes $$
\log_2 3 \approx 1.585 \text{ bits}.
$$ After $k$ weighings the strategy tree has at most $3^k$ leaves, so it
can name at most $3^k$ hypotheses. Compare budgets: $$
\begin{align*}
3^2 &= 9 < 24,\\
3^3 &= 27 \ge 24.
\end{align*}
$$ Two weighings cannot suffice no matter how clever the groupings.
Three weighings *can*: there is spare capacity of three unused leaves
($27-24=3$). In bits, $$
3\log_2 3 \approx 4.75 > \log_2 24 \approx 4.58.
$$ The gap is small. That is why the puzzle feels tight. It is the same
comparison we already made with $\log_2 2$, with a larger per-use
budget.

The first weighing that realises the ternary budget puts four against
four and leaves four off. Level then means the odd ball is in the four
left off; a tip couples a group of four with a direction, as 6 vs 6 did,
but now the third symbol is alive. We do not give the rest of the tree.
The design principle is the one already present in the information game:
choose each measurement so that its outcomes are as nearly equiprobable
as possible under the current posterior. That maximises expected
information gain per weighing and is what makes the $3^3 \ge 24$ budget
achievable in practice. Adding a known-normal ball, or arranging a
weighing that can never balance, is the same kind of mechanism tax we
already computed for $h_2(2/3)$.

We stop at the existence argument. Constructing the adaptive tree —
which balls on which pan at each step — is MacKay’s exercise, and
solving it is the best way to feel what \`\`informative experiment’’
means. What the lecture owns is the split between ledgers: $\log_2 24$
versus $\log_2 2$ or $\log_2 3$ is the information account; 6 vs 6, 4 vs
4, and padding with a known ball are the mechanical account. The two are
not the same, and the known-ball tax is the exhibit.

## Coarse-Graining a Die

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/die-coarse-graining.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/die-coarse-graining.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Take a fair six-sided die, $X\in\{1,\ldots,6\}$ with $p(x)=1/6$, so Ask
questions in two stages: first learn which *group* the face falls in,
then resolve the face inside the group.

$$
H(G_{\mathrm{parity}})=1 \text{ bit},
\qquad
H(X\mid G_{\mathrm{parity}})=\log_2 3 \approx 1.585 \text{ bits}
$$

Let $G_{\mathrm{parity}}=0$ on even faces and $1$ on odd. Then
$w=(1/2,1/2)$, so $H(G_{\mathrm{parity}})=1$. Conditional on the group,
three faces remain, each with probability $1/3$, so
$H(X\mid G_{\mathrm{parity}})=\log_2 3$. The stages add:

$$
w=\Bigl(\tfrac16,\tfrac56\Bigr),
\qquad
H(G_{\mathrm{six}})=h_2\Bigl(\tfrac16\Bigr)\approx 0.65 \text{ bits}
$$

Let $G_{\mathrm{six}}=\mathbf{1}_{\{X=6\}}$. The group weights are
unbalanced, $w=(1/6,5/6)$, so the binary entropy of the first answer is
one bit. After a `yes'' the face is known ($H=0$). After a`no’’ five
faces remain ($H=\log_2 5$). Averaging, Again
$H(G_{\mathrm{six}})+H(X\mid G_{\mathrm{six}})=H(X)$. Note that
\`\`$\ge 4$?’’ on a fair die is *also* a balanced cut ($\{1,2,3\}$ vs
$\{4,5,6\}$), so it *is* one bit — the unbalanced example has to be
chosen carefully.

## Grouping Composition

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-grouping-composition.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/entropy-grouping-composition.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

$$
H(X)=H(w)+\sum_i w_i\, H(p_i)
$$

Partition the outcomes of $X$ into groups labelled by $i$, with group
probabilities $w_i$ and within-group distributions $p_i$. Then The die
examples are the case check: $H(w)$ is the entropy of the coarse
question; $\sum_i w_i H(p_i)$ is what remains once the group is known.
If $G=g(X)$ is the group label, this identity is exactly the chain rule
because $G$ is a function of $X$ (so $H(G\mid X)=0$) and
$H(X\mid G)=\sum_i w_i H(p_i)$.

In the reverse direction, observing $X$ fixes $G$ with no leftover
uncertainty. The mutual information $I(X;G)=H(G)-H(G\mid X)=H(G)$ is
exactly the price of the coarse question. After you know the group, the
remaining work is $H(X\mid G)$ — which of the fine outcomes inside that
group. Coarse-graining *keeps* $H(G)$ as your description and *discards*
$H(X\mid G)$:

The die paid $\log_2 6$ whichever cut came first. That invariance under
regrouping is not an accident of base-two logarithms; it is part of what
characterises Shannon entropy among candidate uncertainty measures
(**Khinchin-information57?**). Categorical accounts make the same demand
precise (**Baez-entropy11?**). For this lecture it is enough that staged
questions *add*, so coarse-graining has a well-defined price.

## Coarse-Graining as Abstraction

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/coarse-graining-intelligence.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/coarse-graining-intelligence.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Daniel Bernoulli did not integrate Laplace’s curve for every corpuscle.
He replaced the swarm with a coarse mechanical story that still predicts
pressure — exactly keeping a group-level description and discarding
within-group path detail. Thermodynamic macrostates are the same move
with better bookkeeping: the entropy $S$ is the uncertainty still
unresolved *inside* the macrostate. In intelligent systems the hierarchy
is familiar: raw observations are partitioned into features, features
into concepts. At each level you retain $H(w)$ for the labels you use
and lose $\sum_i w_i H(p_i)$ for the distinctions you refuse to make.
Which partition you choose is Jaynes’ and Wigner’s anthropomorphic
choice of degrees of freedom — coming next in this review.

## Static Bookkeeping Versus a Physical Ensemble

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/grouping-static-vs-ensemble.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/grouping-static-vs-ensemble.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The decomposition $H=H(w)+\sum_i w_i H(p_i)$ is pure bookkeeping on a
single distribution. The die roll has no temporal dynamics because we
used the identity only as accounting. Interpreting $p$ as \`\`the
distribution of the system’s state’’ adds a dynamical hypothesis:
something like stationarity plus ergodicity, so that an ensemble average
equals what one trajectory shows over long time. Equilibrium
thermodynamics makes the same move. The equilibrium density is
stationary under the microscopic dynamics, and thermodynamic entropy is
then a time-independent number of the form when each group has relaxed
to its own conditional stationary law $\pi_i$ and only the coarse
weights remain to carry slow change. Dynamics and groups — local
equilibrium, lumpability, broken ergodicity, and the decay of KL —
return in weeks 3–4 and 8.
