## Liouville, Detailed Balance, and the Trapdoor

Many of the Worksheet 3 dialogues usded Liouville’s theorem, the
detailed balance we suggested, or Smoluchowski’s trapdoor. Today we’ll
review a little about what those different ideas say and how they relate
to our coarse graining perspective.

```
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import mlai.plot as plot
import mlai
```

The first question is about the rule itself. Maxwell’s being lets fast
molecules through one way and slow ones the other. Is that a rule that
Newton’s laws, on their own, could ever carry out? This is the question
that Liouville’s theorem can address. The second question is about
equilibrium. If the whole box, gate included, is at one temperature,
what must be true of the traffic between any two states? That’s detailed
balance. The third question considers a mechanism, suppose we try to
build the gate out of real parts, specifically a little hinged flap
sitting in the same gas. Why doesn’t it work? That’s Smoluchowski’s
trapdoor.

```
def leapfrog(q, p, dt, steps):
    # Pendulum, H = p^2/2 - cos q. Leapfrog is symplectic: it preserves area exactly.
    for _ in range(steps):
        p = p - 0.5*dt*np.sin(q)
        q = q + dt*p
        p = p - 0.5*dt*np.sin(q)
    return q, p

def coarse_entropy(q, p, edges_q, edges_p):
    counts, _, _ = np.histogram2d(q, p, bins=[edges_q, edges_p])
    w = counts.ravel()/counts.sum()
    w = w[w > 0]
    return -np.sum(w*np.log(w)), counts

def enclosed_area(q, p):
    # Shoelace formula on the (unwrapped) boundary of the cloud.
    return 0.5*abs(np.dot(q, np.roll(p, -1)) - np.dot(p, np.roll(q, -1)))

rng = np.random.default_rng(1)
n = 40000
r = 0.35*np.sqrt(rng.uniform(size=n))
theta = rng.uniform(0, 2*np.pi, size=n)
q0, p0 = -1.6 + r*np.cos(theta), 0.9 + r*np.sin(theta)
# Follow the edge of the disc too, so we can measure the cloud's true area.
phi = np.linspace(0, 2*np.pi, 100000, endpoint=False)
qb0, pb0 = -1.6 + 0.35*np.cos(phi), 0.9 + 0.35*np.sin(phi)

edges_q = np.linspace(-np.pi, np.pi, 17)
edges_p = np.linspace(-2.5, 2.5, 13)
qq, pp = np.meshgrid(np.linspace(-np.pi, np.pi, 300), np.linspace(-2.5, 2.5, 300))
energy = 0.5*pp**2 - np.cos(qq)

times = [0, 6, 40]
fig, axes = plt.subplots(1, 3, figsize=(12, 4.6), sharey=True)
q, p, t_prev = q0.copy(), p0.copy(), 0
qb, pb = qb0.copy(), pb0.copy()
for ax, t in zip(axes, times):
    steps = int(round((t - t_prev)/0.01))
    q, p = leapfrog(q, p, 0.01, steps)
    qb, pb = leapfrog(qb, pb, 0.01, steps)
    t_prev = t
    qw = (q + np.pi) % (2*np.pi) - np.pi
    H_cg, counts = coarse_entropy(qw, p, edges_q, edges_p)
    area = enclosed_area(qb, pb)
    ax.pcolormesh(edges_q, edges_p, (counts.T > 0), cmap='Greys', vmin=0, vmax=4, alpha=0.6)
    ax.contour(qq, pp, energy, levels=12, colors='#bbbbbb', linewidths=0.6, linestyles='solid')
    ax.plot(qw, p, ',', color='#cc4444', alpha=0.4, rasterized=True)
    for e in edges_q: ax.axvline(e, color='white', lw=0.5)
    for e in edges_p: ax.axhline(e, color='white', lw=0.5)
    ax.set_xlim(-np.pi, np.pi); ax.set_ylim(-2.5, 2.5)
    ax.set_xlabel('angle $q$ (radians)')
    ax.set_title(f't = {t}: true area {area:.3f}' + chr(10) + f'{int((counts>0).sum())} cells, coarse entropy {H_cg:.2f} nats', fontsize=11)
# Direction of motion: p > 0 means q increasing, so the flow runs clockwise.
axes[0].annotate('', xy=(0.6, 1.25), xytext=(-0.6, 1.25),
                 arrowprops=dict(arrowstyle='->', color='#555555', lw=1.5))
axes[0].annotate('', xy=(-0.6, -1.25), xytext=(0.6, -1.25),
                 arrowprops=dict(arrowstyle='->', color='#555555', lw=1.5))
axes[0].set_ylabel('momentum $p$')
fig.tight_layout()
mlai.write_figure('liouville-pendulum-cloud.svg', directory='./physics')
```

Liouville’s theorem is a statement about what mechanics does to
*uncertainty*. Take a state of a mechanical system — positions
$\mathbf{q}$ and momenta $\mathbf{p}$ for every particle — and suppose
we don’t know it exactly. We know it lies somewhere in a region of phase
space. Let each point in that region evolve under Newton’s laws (or
Hamilton’s equations, which are the same laws written more tidily). The
region moves, it stretches, it folds back on itself. But its volume
doesn’t change.

The figure shows this for a single pendulum. Phase space is the space of
complete states. For a pendulum the state is two numbers: the angle $q$
it hangs at and its momentum $p$ (mass times angular velocity). Knowing
both fixes the whole future, so each point in the plane is one possible
pendulum, and the motion of the pendulum is the motion of that point.
Hamilton’s equations say $\dot q = p$ and $\dot p = -\sin q$ (in units
where mass, length and gravity are 1). So when $p$ is positive the point
moves right, and gravity pulls $p$ back towards zero: the flow goes
round clockwise.

Energy $E=\tfrac{1}{2}p^2-\cos q$ is conserved, so each point stays on
its own grey curve. The closed ovals round the centre are ordinary
swings: the bottom of the swing ($q=0$) is where $|p|$ is largest, and
the turning points are where the oval crosses $p=0$. The wavy lines
along the top and bottom are a pendulum with enough energy to go right
over the top and keep rotating. The angle wraps round, so the left and
right edges of each panel are the same place ($q=\pm\pi$, pointing
straight up), and the boundary between the two kinds of motion is the
separatrix through those points.

The red cloud is not one pendulum but 40,000 possible starting states of
it. We can read it as our uncertainty: we know the pendulum is somewhere
in the disc, but not where. The cloud starts as a disc on the left of
the picture: the pendulum is swinging right, about 90 degrees from the
bottom. Each point then moves along its own oval. But a pendulum’s
period grows with the size of its swing, so points on the outer ovals
take longer to go round than points on the inner ones. The outer edge of
the cloud falls behind the inner edge, the disc is sheared into a
crescent ($t=6$), and then into a long thin thread wound round and round
($t=40$). Its area is exactly what it was at the start. The panel titles
give that area, measured from the edge of the cloud, and it agrees to
three decimal places at all three times.

Why, then, does the red look bigger at $t=40$? Because the thread
becomes thinner than a pixel, and a pixel can only be red or not.
Counting red pixels gives an area about twice the true one by $t=40$.
The picture is itself a coarse graining, and it shows the same effect as
the shaded grid: the region we can *see* grows while the region the
states actually occupy does not.

That has a direct consequence for entropy. If our uncertainty is spread
uniformly over the cloud, its Shannon entropy is the log of the cloud’s
volume. Same volume, same entropy. More generally, for any density
$\rho_t$ carried along by Hamiltonian flow, the fine-grained entropy
$-\int \rho_t \log \rho_t$ is constant in time.

So where does the second law come from? From the grid. We can’t follow
the thread; we only see which cells it is in. The entropy computed from
the cell occupancies rises as the thread spreads across more cells. This
is the Gibbs and Ehrenfest account of the second law: what increases is
a *coarse-grained* entropy, the one that forgets the fine detail. That
fits the grouping picture we have been using since the die in Week 2:
the macrostate is the group, and the detail within each group is what we
have chosen not to track.

The reason volume can’t shrink is that mechanics is reversible. If I
hand you the state of the system now, you can flip all the momenta, run
the same laws forward, and recover where it came from. So two different
starting points can never arrive at the same end point: if they did, the
backward run would have to choose between two pasts. Any rule that sent
two distinct states to one would squeeze phase-space volume, and
mechanics never does that.

This gives us a test we can apply to the demon’s gate. Don’t ask whether
it is clever, or what it knows. Just ask: does it ever send two
different states to the same one?

```
vs, vmax, xw = 1.0, 2.0, 1.0   # threshold, velocity range, half-width of each side
col = {'A': '#cc4444', 'C': '#ee9944', 'B': '#4488cc', 'D': '#44aa99'}
# arriving at the gate: (x0, x1, v0, v1)
arrive = {'A': (-xw, 0, vs, vmax),      # on the left, moving right, fast
          'B': (-xw, 0, 0, vs),         # on the left, moving right, slow
          'C': (0, xw, -vmax, -vs),     # on the right, moving left, fast
          'D': (0, xw, -vs, 0)}         # on the right, moving left, slow
# leaving the gate
TR_hi, TR_lo = (0, xw, vs, vmax), (0, xw, 0, vs)
BL_lo, BL_hi = (-xw, 0, -vmax, -vs), (-xw, 0, -vs, 0)
cases = {
    'Open hole': {'A': TR_hi, 'B': TR_lo, 'C': BL_lo, 'D': BL_hi},
    'Solid wall': {'A': BL_lo, 'B': BL_hi, 'C': TR_hi, 'D': TR_lo},
    "Maxwell's gate": {'A': TR_hi, 'B': BL_hi, 'C': TR_hi, 'D': BL_hi},
}

def box(ax, r, **kw):
    x0, x1, v0, v1 = r
    ax.add_patch(mpatches.Rectangle((x0, v0), x1 - x0, v1 - v0, **kw))

fig, axes = plt.subplots(1, 3, figsize=(12, 4.4), sharey=True)
for ax, (name, rule) in zip(axes, cases.items()):
    for k, r in arrive.items():
        box(ax, r, facecolor=col[k], edgecolor='white', lw=1.5, alpha=0.85)
        ax.text(0.5*(r[0]+r[1]), 0.5*(r[2]+r[3]), k, ha='center', va='center',
                color='white', fontsize=13, fontweight='bold')
    landed = {}
    for k, r in rule.items():
        landed.setdefault(r, []).append(k)
    for r in [TR_hi, TR_lo, BL_lo, BL_hi]:
        ks = landed.get(r, [])
        if not ks:
            box(ax, r, facecolor='none', edgecolor='#aaaaaa', lw=1, ls='--')
            ax.text(0.5*(r[0]+r[1]), 0.5*(r[2]+r[3]), 'empty', ha='center',
                    va='center', color='#888888', fontsize=10)
        elif len(ks) == 1:
            box(ax, r, facecolor=col[ks[0]], edgecolor='white', lw=1.5, alpha=0.35)
            ax.text(0.5*(r[0]+r[1]), 0.5*(r[2]+r[3]), ks[0], ha='center',
                    va='center', color=col[ks[0]], fontsize=13, fontweight='bold')
        else:
            box(ax, r, facecolor=col[ks[0]], edgecolor=col[ks[1]], lw=0,
                alpha=0.5, hatch='///')
            box(ax, r, facecolor='none', edgecolor='#222222', lw=2)
            ax.text(0.5*(r[0]+r[1]), 0.5*(r[2]+r[3]), '+'.join(ks), ha='center',
                    va='center', color='#222222', fontsize=13, fontweight='bold')
    ax.axvline(0, color='#2ca02c', lw=4)
    ax.axhline(0, color='#333333', lw=0.8)
    for v in (vs, -vs):
        ax.axhline(v, color='#888888', lw=0.6, ls=':')
    ax.set_xlim(-xw, xw); ax.set_ylim(-vmax, vmax)
    ax.set_xticks([-0.5*xw, 0.5*xw]); ax.set_xticklabels(['left side', 'right side'])
    ax.set_title(name, fontsize=12)
axes[0].set_yticks([-vs, 0, vs]); axes[0].set_yticklabels([r'$-v_\star$', '0', r'$v_\star$'])
axes[0].set_ylabel('velocity across the gate $v$')
fig.text(0.5, 0.005, 'Strong colour: about to reach the gate.   Pale colour: just left it.   Dashed: nothing arrives here.',
         ha='center', fontsize=10, color='#444444')
fig.tight_layout(rect=(0, 0.04, 1, 1))
mlai.write_figure('maxwell-gate-phase-space-map.svg', directory='./physics')
```

Figure: <i>What happens to a molecule at the gate, drawn in (side,
velocity) space. Molecules about to arrive are in the strong-coloured
quadrants: A and B on the left heading right (fast and slow), C and D on
the right heading left. An open hole lets everything through; a solid
wall bounces everything back. Both send each arriving region to its own
leaving region. Maxwell’s gate sends A and C to the same place (fast,
leaving to the right) and B and D to the same place (slow, leaving to
the left). Two quadrants end up empty. The gate halves the phase-space
volume.</i>

Here is the gate drawn in phase space. Look only at the direction across
the gate: which side a molecule is on, and its velocity $v$ across the
gate (positive means heading right). Molecules about to hit the gate are
on the left moving right, or on the right moving left. I’ve split each
of those by speed, giving four regions: A (left, fast), B (left, slow),
C (right, fast), D (right, slow).

With an open hole every molecule carries on through. A ends up on the
right moving right, C ends up on the left moving left, and so on. Four
regions in, four distinct regions out. With a solid wall every molecule
bounces: its velocity flips and it stays on its own side. Again four in,
four distinct out. Both are perfectly good mechanics.

Now apply Maxwell’s rule. A fast molecule from the left (A) is let
through, so it leaves to the right, moving right, fast. A fast molecule
from the right (C) is bounced, so it also leaves on the right, moving
right, fast. A and C land in the same place. In the same way B (slow
from the left, bounced) and D (slow from the right, let through) both
end up on the left moving left, slow. Four regions in, two out. Half of
phase space is never visited after an encounter with the gate.

Look at one of the fast molecules leaving to the right. Did it come
through the gate, or bounce off it? The state of the gas no longer tells
you. That is exactly the many-to-one map that Liouville forbids. So the
gate’s rule is not something mechanics, on its own, can do. The rule is
fine as a thought experiment, and it does sort the gas: the simulation
in lecture 4 shows that, and so does the flux calculation. What
Liouville tells us is that the rule cannot be the whole of the physics.
Something else has to enter the story.

Figure: <i>Left: on its own the gate sends two different pasts to one

If the gas alone can’t hold the history, maybe something else can.
Suppose the gate keeps a record each time it acts: “let through” or
“bounced”. Now the joint state of the gas and the record is one-to-one
again. A fast molecule leaving to the right with the record “bounced”
came from C; with “let through” it came from A. No volume has been lost.
It has been moved from the gas into the record.

This is why Liouville sits so close to Landauer’s principle. A finite
memory fills up. To keep the demon running in a cycle we have to reset
the record to blank, and that reset is itself a two-to-one map: “let
through” and “bounced” both go to “blank”. The merge is back, this time
in the memory. The only way a reset can happen in a world governed by
mechanics is if the lost volume is pushed into something else, usually a
heat bath. Landauer’s figure of at least $k_BT\ln 2$ of heat per erased
bit is the size of that push (Landauer, 1961). Bennett used this to put
the demon’s cost at erasure rather than at measurement (Bennett, 1982).

Earman and Norton argue that this resolution assumes some of what it
sets out to prove (Earman and Norton, 1999, 1998). That debate is real,
and I don’t want to present the erasure story as settled. But the
phase-space picture is not in dispute: the gate’s rule, on its own,
throws away volume, and the volume has to go somewhere.

Figure: <i>Two three-state systems with the same stationary distribution
(one third each). On the left every flow is cancelled by its own
reverse: detailed balance. On the right the probabilities still don’t
change in time, because each state gains as much as it loses, but the
clockwise flows are twice the anticlockwise ones and a net current goes
round the loop. That is a steady state, not an equilibrium.</i>

Detailed balance is a statement about transitions between states. Write
$\pi_i$ for the equilibrium probability of state $i$ and $k_{ij}$ for
the rate of jumping from $i$ to $j$. The probability flowing from $i$ to
$j$ per unit time is $\pi_i k_{ij}$. Detailed balance says that for
*every* pair, $$
\pi_i k_{ij} = \pi_j k_{ji}.
$$ Note that this is a balance of probability flows, not of rates. If
$j$ is much less likely than $i$, the rate back from $j$ has to be
correspondingly higher.

It is easy to confuse detailed balance with stationarity, and the figure
shows why they differ. Stationarity only asks that each state’s total
inflow equals its total outflow, so the probabilities stop changing. On
the right of the figure that holds, but probability is circulating
clockwise round the loop. You can have a steady current with unchanging
probabilities. Detailed balance rules that out. It is what equilibrium
means at the level of individual transitions, and it follows from
reversible microscopic laws together with a thermal stationary
distribution. (When velocities are involved the reverse of a transition
also flips the velocities: the reverse of “fast molecule moves left to
right” is “fast molecule moves right to left”.)

Now look at the gate again. A fast molecule crossing left to right is
allowed. Its reverse, a fast molecule crossing right to left, is
bounced. A slow molecule crossing right to left is allowed; its reverse
is bounced. The gate is built from transitions whose reverses it
forbids. That is the right-hand panel: a steady state with a current
running through it, not an equilibrium. If the gas and the gate are
really at one temperature, detailed balance has to hold, so once we
include the gate’s own workings the reverse moves must come back. If
they don’t, something outside the box is driving it.

In lecture 4 we called the condition “as many molecules cross left to
right as right to left” flux balance. That is stationarity across the
membrane. It is not detailed balance, and the gate can satisfy one while
violating the other.

Let me be careful about how this connects to lecture 4. There we treated
each side as a Maxwellian at its own temperature and asked whether the
flows across the gate could balance. They can’t, for both number and
energy at once: a fast molecule carries more energy than a slow one, so
when equal numbers cross each way more energy goes right than comes
back. That told us the steady state is not two clean Maxwellians. The
tails near the gate are continually being eaten away and refilled by
collisions.

That is a statement about which description of the gas is right. It is
not, by itself, a proof that the demon fails. The gas really does sort
under the gate’s rule: the simulation shows it, and the entropy of the
gas really does fall. What Liouville and detailed balance add is the
reason the sorting can’t be free. The gate’s rule merges states, so the
merge has to be paid for somewhere. And at a single temperature the
reverse transitions it forbids have to reappear once the gate is made of
real physics. Which brings us to the trapdoor.

Figure: <i>Smoluchowski’s trapdoor. A light flap on a soft spring is
meant to open only when a fast molecule hits it from the left. But a
flap light enough to be pushed by one molecule is light enough to be
shaken by the gas around it. It has its own thermal energy, rattles open

Smoluchowski asked the engineering question in 1912. Replace the demon
with a mechanism: a small flap over the hole, held shut by a weak
spring, that a molecule coming from the left can push open but one from
the right can’t. No intelligence, no measurement, no memory. Surely that
sorts the gas?

It doesn’t, and the reason has nothing to do with information. The flap
is itself a physical object in contact with the gas. For one molecule to
push it open it has to be very light and the spring very soft. But then
the flap is shaken by the gas just like any other light object.
Equipartition gives it $\tfrac{1}{2}k_BT$ of kinetic energy in its swing
and another $\tfrac{1}{2}k_BT$ stored in the spring, on average, just as
every molecule has $\tfrac{1}{2}k_BT$ per direction of motion. It
rattles. Sometimes it is open when a molecule arrives from the right,
and lets it through the wrong way. Sometimes it slams shut and knocks a
molecule back. Once we include the flap in the description, detailed
balance holds for gas plus flap together, and the net sorting averages
to zero. Feynman’s ratchet and pawl, in the *Feynman Lectures* (Volume
I, chapter 46), is the same argument for extracting work rather than
sorting.

There is a tempting fix: keep the flap cold so that it doesn’t rattle.
That does work for a while. But a cold flap in a warm gas is a second
heat bath at a different temperature, and what we have built is a small
heat engine running between the flap and the gas. It stops when the flap
has warmed up. We haven’t beaten the second law; we’ve brought in a
second reservoir.

So the trapdoor is not a Liouville-and-memory story. Each time the flap
shuts, the two-to-one merge from the gate picture is being absorbed by
the flap’s own jiggling, which is passed straight back to the gas as
heat. There is no record to fill or wipe. If we want to coarse-grain
this system we can’t just add a “door open / door closed” label to our
compartments. We have to add the flap as a real degree of freedom, with
its own position, momentum, and temperature, and check detailed balance
for the enlarged system.

## Shannon Capacity and DPI

The Liouville block ended with Maxwell’s gate merging four regions of
phase space into two. That picture is already a channel, and it carries
us straight into Shannon’s capacity and the data-processing inequality.
Live today: the gate as a channel, what capacity is, and the statement
and moral of the DPI. The DPI proof, its form for distributions, and the
KL-decay account of the second law are in the notes overflow below.

The phase-space map from the Liouville block is already a channel. Its
input is the region a molecule arrives from, A, B, C or D. Its output is
the region it leaves into. Suppose the four arrivals are equally likely,
so $H(\text{in})=2$ bits. The open hole and the solid wall are
one-to-one: knowing where a molecule left tells you exactly where it
came from, so $I(\text{in};\text{out})=H(\text{in})=2$ bits. Maxwell’s
gate sends A and C to one output and B and D to another. Now
$I(\text{in};\text{out})=1$ bit, and one bit is left over,
$H(\text{in}\mid\text{out})=1$: given a fast molecule leaving to the
right, did it come through or bounce? That is the merge we saw in the
picture, measured in bits. A deterministic channel has no noise, so its
capacity is just $\log_2$ of the number of distinct outputs it can
produce: 2 bits for the hole or the wall, 1 bit for the gate.

Once the gate has acted, anything we later compute from the state of the
gas is a function of the output. So
$\text{in}\to\text{out}\to f(\text{out})$ is a Markov chain, and the
data-processing inequality says no such computation can recover more
than the 1 bit that survived. The only way to keep the missing bit is to
store it somewhere else as the gate acts: a record $M$ of “let through”
or “bounced”. With the record, $I(\text{in};\text{out},M)=2$ bits and
nothing has been lost; this is the middle panel of the memory figure.
Wiping the record maps both of its states to “blank”, another two-to-one
channel, and the same inequality says the bit is now gone from the
record as well. Information theory says the bit is lost from the system
we are describing. Landauer’s principle is the physical claim about
where it goes: into the surroundings as heat.

This is the bridge between the physics and the information theory. The
data-processing inequality says information can only be lost along a
chain, and it is lost exactly when a step merges states that carried
different information. Liouville’s theorem says Hamiltonian mechanics
never merges states: it is a one-to-one map that keeps volume. So
mechanics is the equality case of the data-processing inequality. The
fine-grained entropy of the pendulum cloud stays fixed for the same
reason $I(\text{in};\text{out})$ stays at 2 bits through the open hole.

Strict loss needs a merge. The gate merges. Erasure merges. And a coarse
graining merges every state inside a cell. The rise in the pendulum’s
cell entropy is data processing applied to the cells, and the KL-decay
notes below make that precise: with equal-volume cells, Liouville makes
the cell-to-cell transitions doubly stochastic, and forgetting the
detail inside the cells at each step makes the cell entropy rise.

Two cautions. First, the bits here are bookkeeping in units of $\log 2$.
Turning them into heat needs Boltzmann’s constant and a temperature, and
that step is Landauer’s, not Shannon’s. Second, phase space is
continuous, so mutual information between exact states is infinite. The
statements are about cells, or regions like A to D, not about exact
points.

## Channels and Capacity

<span class="editsection-bracket"
style="">\[</span><span class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/channel-capacity-chain-rule.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/channel-capacity-chain-rule.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The chain rule $H(X,Y)=H(X)+H(Y\mid X)$ says the uncertainty in a pair
is the uncertainty in the first plus what is left in the second once the
first is known. Mutual information is the overlap: how much knowing $Y$
reduces uncertainty about $X$, $I(X;Y)=H(X)-H(X\mid Y)$. Substituting
the chain rule gives the symmetric form $H(X)+H(Y)-H(X,Y)$. It is the KL
divergence between the joint and the product of the marginals, so it is
never negative and is zero only when $X$ and $Y$ are independent.

Shannon’s setting (Shannon, 1948) has three parts. A channel takes an
input symbol $x$ and returns an output $y$ with probability
$p(y\mid x)$, independently at each use. A code assigns each of $M$
messages a block of $n$ input symbols. The rate is how many bits of
message each channel use carries, $R=\tfrac{1}{n}\log_2 M$. A rate is
called achievable if there are codes at that rate whose probability of
decoding error goes to zero as the blocks get longer. Note what the
definition does not ask for: no single use of the channel need be
reliable. Reliability is bought with block length.

The noisy-channel coding theorem (Cover and Thomas, 1991, ch. 7;
Shannon, 1948) has two halves. The converse is a no-go: at any rate
above $C$ the probability of error stays bounded away from zero,
whatever code you use. The achievability half says every rate below $C$
can be reached. Shannon’s proof draws the codewords at random from the
capacity-achieving input distribution $p^*(x)$ and shows that the
average code does well, so some code must. It is an existence proof.
Building practical codes that come close to $C$ took another fifty
years.

Two special cases help. A *noiseless deterministic* channel, where each
input gives one fixed output, has $H(Y\mid X)=0$, so $I(X;Y)=H(Y)$. Its
capacity is $\log_2$ of the number of distinct outputs it can produce.
If two inputs share an output, the channel can no longer tell them
apart, and capacity falls. A channel whose output is independent of its
input has $I(X;Y)=0$ for every $p(x)$ and capacity zero.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
def Hb(q):
    q = np.clip(q, 1e-12, 1 - 1e-12)
    return -q*np.log2(q) - (1 - q)*np.log2(1 - q)

p = np.linspace(0, 1, 401)
fig, ax = plt.subplots(figsize=(7, 4))
for eps in [0.0, 0.05, 0.1, 0.25, 0.5]:
    py1 = p*(1 - eps) + (1 - p)*eps          # P(Y=1)
    I = Hb(py1) - Hb(eps)                    # I(X;Y) = H(Y) - H(Y|X)
    line, = ax.plot(p, I, linewidth=2, label=f'$\\epsilon={eps}$')
    ax.scatter([0.5], [1 - Hb(eps)], s=50, color=line.get_color(), zorder=3)
ax.set_xlabel('input bias $p(x=1)$')
ax.set_ylabel('$I(X;Y)$ (bits)')
ax.set_ylim(-0.02, 1.05)
ax.legend(title='flip probability')
ax.set_title('Binary symmetric channel: $C = 1 - H_2(\\epsilon)$')
plt.tight_layout()
mlai.write_figure('bsc-capacity.svg', directory='./ml')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/ml/bsc-capacity.svg" class="" width="70%" style="vertical-align:middle;">

Figure: <i>Mutual information of a binary symmetric channel against the
input bias, for several flip probabilities $\epsilon$. Each curve peaks
at a uniform input (dots), giving capacity $C=1-H_2(\epsilon)$. At
$\epsilon=1/2$ the output is independent of the input and $C=0$.</i>

The binary symmetric channel flips each bit with probability $\epsilon$.
With $P(X=1)=p$ the output has $P(Y=1)=p(1-\epsilon)+(1-p)\epsilon$, and
$H(Y\mid X)=H_2(\epsilon)$ whatever the input, so $$
I(X;Y)=H_2\big(p(1-\epsilon)+(1-p)\epsilon\big)-H_2(\epsilon).
$$ The first term is largest when the output is uniform, which happens
at $p=1/2$, so $C=1-H_2(\epsilon)$. A channel that flips one bit in ten
still carries just over half a bit per use, reliably, if you code over
long enough blocks.

The data-processing inequality is the no-go on mutual information. If
$Z$ is computed from $Y$ alone, possibly with extra randomness, then $Z$
knows no more about $X$ than $Y$ did: $I(X;Z)\le I(X;Y)$. Equality holds
when $Z$ keeps everything in $Y$ that bears on $X$, in which case $Z$ is
a sufficient statistic. We use it again in a moment: in error-controlled
regulation, $D\to S\to A$ is a chain, so $I(A;D)\le I(S;D)$. The proof
(Cover and Thomas Theorem 2.8.1) is in the notes overflow. Quiz 4
examines the statement and the moral, not the proof.

Overview only. Full Conant–Ashby development — concavity argument,
existence of a deterministic optimum, and the two caveats in depth — is
the Week 8 review. Today’s job is the setup, the notation, and the
slogan and its two caveats. Diagrams follow Joery’s oneshot figure in
the good-regulator notes: disturbance $D$, system state $S$, action $A$,
reguland $Z$.

Conant and Ashby (Conant and Ashby, 1970) often label the regulator $R$.
We write $A$ for the action so the policy $\pi(a\mid s)$ is first-class.
The reguland — the variable whose entropy successful regulation
minimises — is $Z$. The system state is $S$; the disturbance is $D$. The
cumulant generating function earlier today was
$\psi(\boldsymbol{\theta})=\log Z(-\boldsymbol{\theta})$; that $Z$ was
the partition function, not the reguland — different objects, same
letter in the physics tradition.

Figure: <i>Cause-controlled regulation (left) acts on the disturbance
$D$; error-controlled regulation (right) acts on the state $S$ that $D$

A disturbance $D$ causes events in a system $S$; the regulator counters
with an action $A$; the pair produces a reguland $Z=\phi(S,A)$ (Conant
and Ashby, 1970). Cause-controlled regulation acts on $D$ (or on an
inferred model of $D$). Error-controlled regulation acts on the state
$S$ caused by $D$, so by the data-processing inequality
$I(A;D)\le I(S;D)$. Both diagrams are shown above; the existence proof
that an $H(Z)$-optimal policy may be taken deterministic waits for Week
8.

Successful regulation, for Conant and Ashby, minimises the entropy of
the reguland: $\min_\pi H(Z)$. Among optima there exists a deterministic
policy $A=h(S)$, equivalently $H(A\mid S)=0$. That is the sense in which
the regulator is a “model”. This is deliberately weak. Do not collapse
the slogan into “$A$ contains a rich model of $S$”: a constant action
also has $H(A\mid S)=0$. Do not confuse low $H(Z)$ with a desirable
outcome. Requisite variety in Shannon form will use the same letters:
residual uncertainty satisfies $H(Z)\ge H(D)-I(D;A)$ when $A$ is the
regulatory response. Week 8 reviews the theorem properly and applies it
to capability claims.

Ashby’s law of requisite variety in Shannon form (Ashby, 1956), written
in today’s letters: residual uncertainty in the reguland is at least the
disturbance entropy minus the information the action carries about the
disturbance. Requisite variety asks how much relevant $I$ is required;
the Good Regulator Theorem asks how that $I$ may be organised.

The Good Regulator Theorem, in entropy form, is an existence result:
among policies that minimise $H(Z)$ there is a deterministic one,
$A=h(S)$, equivalently $H(A\mid S)=0$ (Conant and Ashby, 1970). The full
concavity / extreme-point argument is in the notes include; today the
diagrams and the slogan-with-fences are enough for assessment.

## Goals as Radon–Nikodym

Conant and Ashby minimise the entropy of the reguland. We might ask the
question, how does that relate to the more usual approach in
reinforcement learning where we maximise a reward function. In ongoing
work (with Joery de Vries) we suggest that the long-run state–action
occupancy is pushed through a fixed channel $K(z\mid s,a)$ to an outcome
law $p_\pi$ on $Z$. Minimising $H(Z)$ is concave over the occupancy
polytope, so a deterministic policy is optimal.

If the “good” outcomes are not the whole alphabet, then we introduce a
goal measure $\nu$, a “measure of success” as it were. This must be
absolutely continuous with respect to the base measure $\lambda$. The
change of base is $$
H_\nu(Z)=H_\lambda(Z)+\mathbb{E}\!\left[\log\tfrac{\mathrm{d}\nu}{\mathrm{d}\lambda}(Z)\right].
$$ The second term is the goal: writing
$r=-\log(\mathrm{d}\nu/\mathrm{d}\lambda)$ makes it $-\mathbb{E}[r(Z)]$.
So ordinary reward maximisation is entropy minimisation under a
reference (the measure of success). This is the same change-of-measure
we reviewed when considering speed and velocity of a Maxwellian gas. Our
interpretation is that Conant and Ashby’s good set $G$ is the special
case that puts smaller $\nu$-volume on $G$ than on $G^c$, so the goal
term pays for landing in $G$.

## The Measure of Success as a Coarse Graining

This block extends “Goals as Radon–Nikodym”. The identity
$H_\nu(Z)=H(Z)-\mathbb{E}[r(Z)]$, with
$r=-\log(\mathrm{d}\nu/\mathrm{d}\lambda)$, was stated as a change of
measure. Here we read it as the grouping formula from Week 4. Two
lessons follow. First, coarser success is easier: the more outcomes you
are willing to lump together, the less entropy is left to remove and the
less requisite variety the regulator needs. Second, any reward function
can be read backwards as a coarse graining. It says how many finer
outcomes each observed outcome stands for, and the regulator is assumed
to have no grip inside each one.

Conant and Ashby minimise the entropy of the reguland, but the reguland
is a description, not a fact of nature. Suppose the world has a fine
description $X$ and we measure success on a summary $Z=T(X)$: a
temperature band rather than a temperature, “patient discharged” rather
than a full physiological state. Because $T$ is deterministic,
$H(T(X))\le H(X)$ for every policy. This is the same DPI-style fact we
met with $h_0$ in Week 6. So the coarser the measure of success, the
lower the entropy the regulator has to remove. At the extreme, a single
cell (“anything is fine”) gives $H(Z)=0$ under every policy. Every
regulator is then optimal, including the constant one. That is caveat 1
seen from the other side: the constant action $A=a_0$ is a good
regulator exactly when the measure of success is coarse enough not to
care.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
n = 8
widths = [1, 2, 4, 8]
variety = [1, 2, 4, 8]
fig, ax = plt.subplots(figsize=(7, 4))
for m in variety:
    floor = [max(0.0, np.log2(n/w) - np.log2(m)) for w in widths]
    ax.plot(widths, floor, marker='o', label=f'$m={m}$ actions')
ax.set_xscale('log', base=2)
ax.set_xticks(widths)
ax.set_xticklabels([str(w) for w in widths])
ax.set_xlabel('cell width $w$ (states per success class)')
ax.set_ylabel('minimum achievable $H(C)$ (bits)')
ax.legend()
ax.set_title('Coarser success needs less requisite variety')
plt.tight_layout()
mlai.write_figure('coarse-success-requisite-variety.svg', directory='./information')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/coarse-success-requisite-variety.svg" class="" width="75%" style="vertical-align:middle;">

Figure: <i>Minimum achievable entropy of the success class against cell
width, for regulators with $m$ evenly spaced actions that observe $D$.
Each doubling of tolerance saves one bit, exactly as each doubling of
action variety does.</i>

A toy thermostat. The disturbance $D$ is uniform on $\{0,\dots,7\}$, so
$H(D)=3$ bits. The regulator sees $D$ and picks one of $m$ evenly spaced
corrective shifts, and the outcome is $X=(D+A)\bmod 8$. We call the run
a success at the level of bands of width $w$, $C=\lfloor X/w\rfloor$.
Brute force over all deterministic policies gives $$
\min_h H(C)=\max\left(0,\ \log_2\tfrac{8}{w}-\log_2 m\right).
$$ Doubling the tolerance $w$ saves exactly one bit, and so does
doubling the action variety $m$: the two trade bit for bit. This is
requisite variety in Shannon form, $H(Z)\ge H(D)-I(D;A)$, applied to the
coarse variable. A coarse measure of success shrinks the effective
$H(D)$, because disturbances that land in the same cell no longer need
different responses. It also bounds how fine the “model” $h:S\to A$ has
to be. The regulator only needs to tell apart states whose correct
responses differ at the level of $C$. The measure of success sets the
resolution of the model the theorem forces.

Week 4 gave the grouping formula $H=H(w)+\sum_i w_i H(p_i)$. Write it
with the fine description $X$ grouped by the outcome $Z$: $$
H(X)=H(Z)+\sum_z P(Z=z)\,H(X\mid Z=z).
$$ Now suppose each outcome $z$ is a cell containing $\nu(z)$ fine
states, and the regulator has no control inside a cell. The MaxEnt
description of “no control” is uniform within the cell, so
$H(X\mid Z=z)=\log\nu(z)$. Define $r(z)=-\log\nu(z)$. Then $$
H(X)=H(Z)-\mathbb{E}[r(Z)]=H_\nu(Z),
$$ which is exactly the goal-measure identity from the previous block,
with $\nu$ as the cell-volume measure and $\lambda$ as counting on $Z$.
So minimising entropy relative to the measure of success is minimising
the entropy of the fine world, on the assumption that the regulator only
acts on which cell it lands in. The reward of an outcome is minus the
entropy left inside it. High reward means a small cell: landing there
certifies a precise fine state. Low reward means a big cell: landing
there leaves the world mostly unconstrained. This is Week 6’s density of
states again: $\nu$ plays the role of $h_0$, Boltzmann’s $W$, and
$\mathbb{E}[\log h_0]$ was “the entropy lost by recording only $k$”.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
def shannon(p):
    p = np.asarray(p, dtype=float)
    p = p[p > 0]
    return -np.sum(p * np.log2(p))

q = np.linspace(0, 1, 201)
nu_good, nu_bad = 1, 9
H_Z = np.array([shannon([qi, 1-qi]) for qi in q])
H_nu = H_Z + q*np.log2(nu_good) + (1-q)*np.log2(nu_bad)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(q, H_Z, label='$H(Z)$ (counting measure)')
ax.plot(q, H_nu, label='$H_\\nu(Z)=H(Z)-\\mathbb{E}[r(Z)]$')
ax.set_xlabel('$q = P(Z\\in G)$')
ax.set_ylabel('bits')
ax.legend()
ax.set_title('One good state in ten: $\\nu(G)=1$, $\\nu(G^c)=9$')
plt.tight_layout()
mlai.write_figure('goal-measure-one-in-ten.svg', directory='./information')
```

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/goal-measure-one-in-ten.svg" class="" width="75%" style="vertical-align:middle;">

Figure: <i>Plain $H(Z)$ cannot tell always-succeed from always-fail.
Entropy relative to the measure of success can: its minimum is at $q=1$
and its maximum is at $q=0.1$, the uniform distribution over the ten
fine states.</i>

Ten fine states, one of them good. Coarse-grain into $G$ (one state) and
$G^c$ (nine states), so $r(G)=0$ and $r(G^c)=-\log_2 9\approx-3.17$
bits. With $q=P(Z\in G)$:

| $q$ | $H(Z)$ | $H_\nu(Z)$ |
|:----|:-------|:-----------|
| 0   | 0      | 3.17       |
| 0.5 | 1.00   | 2.58       |
| 0.9 | 0.47   | 0.79       |
| 1   | 0      | 0          |

Under the counting measure, “always fail” ($q=0$) and “always succeed”
($q=1$) are equally good regulators, because both are perfectly
predictable. That is caveat 2. The measure of success breaks the tie:
always failing leaves $\log_2 9$ bits in the fine world, always
succeeding leaves none. The maximum of $H_\nu$ is at $q=0.1$, where the
fine state is uniform on all ten: $\log_2 10\approx3.32$ bits. The
reward gap, $\log_2 9$ bits, is just the log of the ratio of the cell
sizes.

The construction runs both ways. Given a coarse graining with cell
volumes $\nu$, the reward is $r=-\log\nu$. Given any reward $r$, the
measure $\nu(z)=e^{-r(z)}$ can be read as the volume of a cell of hidden
fine states that outcome $z$ stands for. Several familiar facts about
rewards become facts about coarse graining.

-   *Shift invariance.* Adding a constant to $r$ multiplies every volume
    by the same factor. That is a change of unit, so the optimal policy
    does not move. This is why reward shifts do not matter in RL.
-   *The good set.* Conant and Ashby’s $G$ is the two-cell case. The
    size of the reward for success is the log of how much bigger failure
    is than success.
-   *Reward scale.* Multiplying $r$ by $\beta$ raises the volume ratios
    to the power $\beta$. A sparse reward of 1 versus 0 (in nats) says
    success is a cell $e\approx2.7$ times smaller than failure. That is
    a mild preference; $\beta=10$ makes it about 22,000 times smaller.
-   *Additive rewards over time.* If $Z=(Z_1,\dots,Z_T)$ and
    $r(z)=\sum_t r(z_t)$, then $\nu(z)=\prod_t\nu(z_t)$. The cell of a
    trajectory is the product of the per-step cells, as if each step had
    its own independent fine degrees of freedom.
-   *Constant reward.* Every cell has the same volume, so $H_\nu$ and
    $H$ differ by a constant. We are back to plain Conant and Ashby:
    predictability with no preference.

Not every reward comes from a literal partition of a finite set, because
$e^{-r}$ need not be an integer. That is why the course states the goal
as a measure $\nu\ll\lambda$ and the reward as a Radon–Nikodym
derivative. The volumes are like the Poisson’s $1/k!$ in Week 6: a
volume, not a count.

The between-cell term $H(Z)$ is bounded by $\log|\mathcal{Z}|$. The
within-cell term $-\mathbb{E}[r(Z)]$ is not. So if one policy’s expected
reward exceeds another’s by more than $\log|\mathcal{Z}|$, it has lower
$H_\nu$ whatever the outcome entropies are: $$
H_\nu(\pi)-H_\nu(\pi')=\big[H(Z_\pi)-H(Z_{\pi'})\big]-\big[\mathbb{E}_\pi r-\mathbb{E}_{\pi'}r\big]<\log|\mathcal{Z}|-\log|\mathcal{Z}|=0.
$$ This is the “volume threshold” on the RL recovery slide. When cell
volumes differ by much more than the number of cells, the regulator
gains far more by landing in a small cell than by being predictable
about which cell it lands in. Entropy minimisation then becomes reward
maximisation. When reward gaps are small compared with
$\log|\mathcal{Z}|$, the entropy term matters: the regulator will trade
some reward for predictability. In both regimes $H_\nu$ is concave in
the occupancy, because $H$ is concave and $\mathbb{E}[r]$ is linear. So
the minimum is at a vertex of the occupancy polytope, a deterministic
policy. That is the same structure that gives deterministic optimal
policies in MDPs.

The identity $H(X)=H_\nu(Z)$ needs the fine state to be uniform inside
each cell. If the regulator does have some grip inside a cell,
$H(X\mid Z=z)\le\log\nu(z)$, and minimising $H_\nu$ minimises an upper
bound on the fine-grained entropy. This is Jaynes’ reading of MaxEnt
from Week 5: uniform within cells is the honest description of what we
have declared we will not track. So a reward function carries three
choices, usually hidden: what counts as an outcome (the partition), how
big each outcome is ($\nu$), and what the regulator is assumed unable to
control (MaxEnt inside each cell). Changing any one of them changes what
“good regulation” means.

Taken together: Conant and Ashby’s criterion, $\min H(Z)$, is only as
meaningful as the description $Z$. Coarse-graining the description makes
regulation easier, in bits you can count. A reward function is a compact
way of specifying a coarse graining with unequal cells, and its value at
each outcome is minus the entropy that outcome leaves unresolved. Reward
maximisation is the large-gap limit of entropy minimisation relative to
that coarse graining.
