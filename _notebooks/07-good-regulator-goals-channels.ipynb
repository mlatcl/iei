<span class="editsection-bracket" style="">\[</span><span

Overview only. Full Conant–Ashby development — concavity argument,
existence of a deterministic optimum, and the two caveats in depth — is
the Week 8 review. Today’s job is the setup, the notation, and the
slogan and its two caveats. Diagrams follow Joery’s oneshot figure in
the good-regulator notes: disturbance $D$, system state $S$, action $A$,
reguland $Z$.

Conant and Ashby (Conant and Ashby, 1970) often label the regulator $R$.
We write $A$ for the action so the policy $\pi(a\mid s)$ is first-class.
The reguland — the variable whose entropy successful regulation
minimises — is $Z$. The system state is $S$; the disturbance is $D$. The
cumulant generating function earlier today was
$\psi(\boldsymbol{\theta})=\log Z(-\boldsymbol{\theta})$; that $Z$ was
the partition function, not the reguland — different objects, same
letter in the physics tradition.

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/good-regulator-cause-controlled.svg" class="" width="100%" style="vertical-align:middle;">

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/good-regulator-error-controlled.svg" class="" width="100%" style="vertical-align:middle;">

Figure: <i>Cause-controlled regulation (left) acts on the disturbance
$D$; error-controlled regulation (right) acts on the state $S$ that $D$

A disturbance $D$ causes events in a system $S$; the regulator counters
with an action $A$; the pair produces a reguland $Z=\phi(S,A)$ (Conant
and Ashby, 1970). Cause-controlled regulation acts on $D$ (or on an
inferred model of $D$). Error-controlled regulation acts on the state
$S$ caused by $D$, so by the data-processing inequality
$I(A;D)\le I(S;D)$. Both diagrams are shown above; the existence proof
that an $H(Z)$-optimal policy may be taken deterministic waits for Week
8.

Successful regulation, for Conant and Ashby, minimises the entropy of
the reguland: $\min_\pi H(Z)$. Among optima there exists a deterministic
policy $A=h(S)$, equivalently $H(A\mid S)=0$. That is the sense in which
the regulator is a
`model'' --- deliberately weak. Do not collapse the slogan into`$A$
contains a rich model of $S$’’: a constant action also has
$H(A\mid S)=0$. Do not confuse low $H(Z)$ with a desirable outcome.
Requisite variety in Shannon form will use the same letters: residual
uncertainty satisfies $H(Z)\ge H(D)-I(D;A)$ when $A$ is the regulatory
response. Week 8 reviews the theorem properly and applies it to
capability claims.

Ashby’s law of requisite variety in Shannon form (Ashby, 1956), written
in today’s letters: residual uncertainty in the reguland is at least the
disturbance entropy minus the information the action carries about the
disturbance. Requisite variety asks how much relevant $I$ is required;
the Good Regulator Theorem asks how that $I$ may be organised. Full
development of both, with the concavity argument, is Week 8.

The Good Regulator Theorem, in entropy form, is an existence result:
among policies that minimise $H(Z)$ there is a deterministic one,
$A=h(S)$, equivalently $H(A\mid S)=0$ (Conant and Ashby, 1970). The full
concavity / extreme-point argument is in the notes include; today the
diagrams and the slogan-with-fences are enough for assessment. Requisite
variety remains the quantity bound $H(Z)\ge H(D)-I(D;A)$, not the
structure theorem.

## Goals as Radon–Nikodym

*Class test timer (Goal measure): 11:00*

Conant and Ashby minimise the entropy of the reguland. In the
reinforcement-learning reading developed with Joery, the long-run
state–action occupancy is pushed through a fixed channel $K(z\mid s,a)$
to an outcome law $p_\pi$ on $Z$. Minimising $H(Z)$ is concave over the
occupancy polytope, so a deterministic policy is optimal — the same
structural reason as an ordinary MDP return, with a different objective.

When the \`\`good’’ outcomes are not the whole alphabet, introduce a
goal measure $\nu$ absolutely continuous with respect to a base
$\lambda$. Entropy relative to $\nu$ differs from entropy relative to
$\lambda$ by the goal term $$
\mathbb{E}\bigl[\ln\tfrac{\mathrm{d}\nu}{\mathrm{d}\lambda}(Z)\bigr]=-\mathbb{E}[r(Z)],\qquad r=-\ln\tfrac{\mathrm{d}\nu}{\mathrm{d}\lambda}.
$$ So ordinary reward maximisation is entropy minimisation under a
tilted reference — the same change-of-measure grammar as Week 5’s KL /
Radon–Nikodym block. Conant and Ashby’s good set $G$ is the special case
that puts smaller $\nu$-volume on $G$ than on $G^c$, so the goal term
pays for landing in $G$.

This is a course-level vocabulary map, not the Joery paper. Assessed:
letters $D,S,A,Z$; $\min H(Z)$; $H(A\mid S)=0$ with both caveats;
$r=-\ln(\mathrm{d}\nu/\mathrm{d}\lambda)$ as the goal/reward bridge to
Week 5. Not assessed: majorisation–minimisation, empowerment as a
summand, DeepSea thresholds.

```
import mlai.plot as plot
import mlai
import numpy as np
import os
```

```
def plot_correlated_gaussian(mu_x = 0, 
                             var_x = 1, mu_y = 0, var_y = 1, 
                             xlabel = "$v_x$", ylabel = "$v_y$", correlation = 0.5, ax=None, diagrams="./ml"):

    sd_x = np.sqrt(var_x)
    sd_y = np.sqrt(var_y)
    tau = 2*np.pi

    x = np.linspace(mu_x-3*sd_x, mu_x+3*sd_x, 100)[:, np.newaxis]
    y = np.linspace(mu_y-3*sd_y, mu_y+3*sd_y, 100)[:, np.newaxis]

    p_x = 1/np.sqrt(tau*var_x)*np.exp(-1/(2*var_x)*(x - mu_x)**2)
    p_y = 1/np.sqrt(tau*var_y)*np.exp(-1/(2*var_y)*(y - mu_y)**2)

    covMat = np.asarray([[1, correlation], [correlation, 1]])
    fact = np.asarray([[sd_x, 0], [0, sd_y]])
    covMat = np.dot(np.dot(fact,covMat), fact)
    v, R = np.linalg.eig(covMat)
    if ax is None:
        fig, ax = plt.subplots(1, 1, figsize=plot.big_figsize)

    ax.plot(mu_x, mu_y, 'x', color=[1., 0., 1.], markersize=5, linewidth=3)
    theta = np.linspace(0, tau, 100)
    xel = np.sin(theta)*np.sqrt(v[0])
    yel = np.cos(theta)*np.sqrt(v[1])
    vals = np.dot(R,np.vstack([xel, yel]))
    ax.plot(vals[0, :]+mu_x, vals[1, :]+mu_y, '-', color=[1., 0., 1.], linewidth=3)
    ax.set_xlim([np.min(x), np.max(x)])
    ax.set_ylim([np.min(y), np.max(y)])
    ax.set_xticks([mu_x-3*sd_x, mu_x, mu_x+3*sd_x])
    ax.set_yticks([mu_y-3*sd_y, mu_y, mu_y+3*sd_y])
    ax.set_xlabel(xlabel, fontsize=20)
    ax.set_ylabel(ylabel, fontsize=20)
```

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
```

```
def shannon(p):
    p = np.asarray(p, dtype=float)
    p = p[p > 0]
    return -np.sum(p * np.log2(p))

def correlated_pair(rho):
    return np.array([[0.25*(1+rho), 0.25*(1-rho)],
                     [0.25*(1-rho), 0.25*(1+rho)]])

rhos = np.linspace(-1, 1, 200)
I_vals, H_vals = [], []
for rho in rhos:
    joint = correlated_pair(rho)
    hx = shannon(joint.sum(axis=1))
    hy = shannon(joint.sum(axis=0))
    H = shannon(joint.ravel())
    I_vals.append(hx + hy - H)
    H_vals.append(H)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(rhos, I_vals, label='$I(\\rho)$')
ax.plot(rhos, H_vals, label='$H(\\rho)$')
ax.set_xlabel('$\\rho$')
ax.set_ylabel('bits')
ax.legend()
ax.set_title('Correlated-pair family: $I+H=2$')
mlai.write_figure('multi-information-rho.svg', directory='./ml')
```

## Data-Processing Inequality

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/data-processing-inequality.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/data-processing-inequality.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Mutual information is the pairwise case of multi-information:
$I(X;Y)=H(X)-H(X|Y)=H(Y)-H(Y|X)=H(X)+H(Y)-H(X,Y)$. Week 2 defined it
from the chain rule and stated the data-processing inequality. We prove
it now, because this week’s object is $I$.

The data-processing inequality (Cover and Thomas, 1991, Thm 2.8.1) is
the no-go on $I$. If $X\to Y\to Z$ is Markov, then $I(X;Z)\le I(X;Y)$.
The first expansion of $I(X;Y,Z)$ uses the chain rule and drops
$I(X;Z\mid Y)$ by the Markov property. The second expansion is at least
$I(X;Z)$ because conditional mutual information is non-negative.
Equality holds when $I(X;Y\mid Z)=0$ — $Y$ is a sufficient statistic of
$X$ for $Z$, or $Z$ already captures everything $Y$ knew about $X$.

This is the same grammar as $I+H=C$. Conservation says you cannot have
both high stored correlation and high free uncertainty without bound.
Data processing says you cannot *increase* stored correlation by further
processing. A claim that an intelligent system will “just process more”
and thereby recover information that a Markov stage has already
destroyed is a DPI violation.

## KL Decay and Grouping

<span class="editsection-bracket" style="">\[</span><span

$$
\text{KL}\left( p\,\|\,\pi \right)=\text{KL}\left( w\,\|\,w_\pi \right)+\sum_i w_i\, \text{KL}\left( p_i\,\|\,\pi_i \right)
$$

For a Markov process with stationary distribution $\pi$,
$\text{KL}\left( p_t\,\|\,\pi \right)$ is nonincreasing — the
data-processing inequality applied to the evolving law. (Physics
literature often calls the same functional relative entropy.) Decompose
$p$ into group weights $w$ and conditionals $p_i$, and likewise $\pi$
into $w_\pi$ and $\pi_i$. Then The first term is how far the macro
weights sit from equilibrium weights; the second is residual
non-equilibrium inside groups. Local equilibrium is the regime
$\text{KL}\left( p_i\,\|\,\pi_i \right)\approx 0$ with
$\text{KL}\left( w\,\|\,w_\pi \right)$ still free to fall. Under
deterministic Hamiltonian flow the fine-grained Shannon entropy of the
phase-space density is conserved (Liouville); the Gibbs–Ehrenfest
account of the second law is that only a entropy increases. Week 2’s die
made the static split; this is the same algebra along a trajectory
toward equilibrium.

Information engines and finite-time protocols act primarily on the
between-group term — changing which macrostate or which bit is occupied
— while assuming or enforcing fast decay of the within-group
divergences. That is why week 2’s stationary examples and week 6’s paths
on the equilibrium manifold belong to one story.

## Information Bottleneck

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/information-bottleneck.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/information-bottleneck.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The data-processing inequality forbids creating information by
processing. It does not say which information to keep. The information
bottleneck (Tishby et al., 1999) is the prescription: given an
observation $X$ and a relevance variable $Y$, choose a representation
$T$ of $X$ that keeps $I(T;Y)$ and discards the rest of $I(X;T)$.

The information bottleneck of Tishby, Pereira and Bialek is a
constrained optimisation on mutual information. The encoder $p(t\mid x)$
is Markov given $X$, so $Y \leftrightarrow X \leftrightarrow T$. Data
processing then supplies the bound $I(T;Y)\le I(X;Y)$: no representation
of $X$ can be more informative about $Y$ than $X$ itself. The Lagrange
multiplier $\beta$ interpolates. Small $\beta$ spends almost no $I$ on
the representation. Large $\beta$ recovers $T=X$. The self-consistent
encoder is an exponential family, $$
p(t\mid x) \propto p(t)\,\exp\bigl(-\beta\,\mathrm{KL}\bigl(p(y\mid x)\,\|\,p(y\mid t)\bigr)\bigr),
$$ the same Boltzmann form as week 4, now with a KL sufficient
statistic. We do not run Blahut–Arimoto. The point is the pair: DPI is
the no-go; the bottleneck is how you spend $I$.

Week 4 asked for the least-committal $p$ given moments. Here the
constraint is a mutual information. Same Lagrange move, different
constraint. Later in this session the bottleneck evaluates intelligence
claims: a system that keeps only what is relevant for action is making
an IB statement. The Good Regulator existence claim $H(A\mid S)=0$ comes
from concavity of entropy; only then may one offer the control
interpretation $\min I(S;A)$ subject to $H(Z)\le\epsilon$ as an
IB-shaped reading — an interpretation after the theorem, not the theorem
itself. Requisite variety and the Good Regulator are tools we use in
Week 8; the information bottleneck itself is the relevant-compression
theorem.

course interpretation, not Conant and Ashby’s equation. Optional LO9

## The Inaccessible Game

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information-game/includes/inaccessible-game-introduction.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information-game/includes/inaccessible-game-introduction.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

We call our framework the “inaccessible game” because the system is
isolated from external observation: an outside observer cannot extract
or inject information, making the game’s internal state inaccessible.

Like other zero-player games, such as Conway’s Game of Life (Gardner,
1970), the system evolves according to internal rules without external
interference. But unlike cellular automata where rules are chosen by
design, in our game the rules emerge from an information-theoretic
principle.

## Why “Inaccessible?”

The game is inaccessible because our fourth axiom—information
isolation—ensures that an external observer learns nothing. Recall from
Baez-Fritz-Leinster that information gained from observing a process
$f: p \rightarrow q$ is measured by entropy change: $$
F(f) = H(p) - H(q).
$$

But if marginal entropies are conserved ($\sum h_i = C$), then an
external observer sees:

-   Before observation: $\sum h_i = C$
-   After observation: $\sum h_i = C$  
-   Information gained: $\Delta(\sum h_i) = 0$

The observer learns nothing! The system is informationally isolated from
the outside world.

## What Makes It a Game?

The “game” has the following characteristics:

**Players:** None (zero-player game)

**State:** A probability distribution over variables, parametrized by
natural parameters $\boldsymbol{\theta}$

**Rules:** Evolve to maximize entropy production subject to the
constraint $\sum h_i = C$

**Dynamics:** Emerge from information geometry (Fisher information) and
the constraint structure

The game starts in a maximally correlated state (high multi-information
$I$, low joint entropy $H$) and evolves toward states of higher entropy.
Since $I + H = C$, this means the system breaks down correlations ($I$
decreases) as entropy increases.

## Connection to Physical Reality

Why should we care about this abstract game? Because its dynamics
exhibit structure that connects to fundamental physics:

1.  **GENERIC structure:** The dynamics decompose into reversible
    (conservative) and irreversible (dissipative) components
2.  **Energy-entropy equivalence:** In the thermodynamic limit, our
    marginal entropy constraint becomes equivalent to energy
    conservation
3.  **Landauer’s principle:** Information erasure necessarily involves
    the dissipative part of the dynamics

The inaccessible game provides an information-theoretic foundation for
thermodynamic principles. It suggests that physical laws might emerge
from information-theoretic constraints, rather than information theory
being derived from physics.
