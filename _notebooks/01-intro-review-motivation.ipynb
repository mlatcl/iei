## Perpetual Motion and Superintelligence

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/perpetual-motion-superintelligence-analogy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/perpetual-motion-superintelligence-analogy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Imagine in 1925 a world where the automobile is already transforming
society, but big promises are being made for things to come. The stock
market is soaring, the 1918 pandemic is forgotten. And every major
automobile manufacturer is investing heavily on the promise they will
each be the first to produce a car that needs no fuel. A perpetual
motion machine.

Well, of course that didn’t happen. But I sometimes wonder if what we’re
seeing today 100 years later is the modern equivalent of that. In 2026
billions are being invested in promises of superintelligence and
artificial general intelligence that will transform everything.

We know why perpetual motion is impossible, but the two slogans often
mixed together belong to different laws. Perpetual motion of the *first
kind* would extract energy from nothing: that is forbidden by the first
law. Perpetual motion of the *second kind* would extract useful work
from a single heat bath, with no colder reservoir to dump heat into:
that is the Kelvin–Clausius form of the second law. Frictionless motion
produces no entropy. Organised motion that you can sell as an engine,
indefinitely, from one bath, does. For an isolated system the entropy
does not decrease — almost always, because the law is statistical.

How might we make an equivalent statement for the bizarre claims around
superintelligence? Some inspiration comes from Maxwell’s demon, an
“intelligent” entity which operates against the laws of thermodynamics.
The inspiration comes because the demon suggests that for the second law
to hold there must be a relationship between the demon’s decisions and
thermodynamic entropy.

One of the resolutions comes from *Landauer’s principle*, the notion
that erasure of information requires heat dissipation. This suggests
there are fundamental information-theoretic constraints on intelligent
systems, just as there are thermodynamic constraints on engines.

I’ve no doubt that AI technologies will transform our world just as much
as the automobile has. But I also have no doubt that the promise of
unconstrained superintelligence is just as silly as the promise of
perpetual motion.

## Preview: Maxwell’s Demon

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/james-clerk-maxwell.png" style="width:35%; height:auto; display:block;">

Figure: <i>James Clerk Maxwell (1831–1879). The demon is an 1860s
thought experiment, not a machine we will build today.</i>

Boltzmann’s $H$ is not entropy. In 1872 he wrote $E=\int f\log f$; he
switched to $H$ in the 1890s after Burbury’s usage. The $H$-theorem is
$\mathrm{d}H/\mathrm{d}t\le 0$. Entropy is (up to constants) the
*negative* of this $H$. Shannon later named his $H$ after the
$H$-theorem, which is why the sign is part of the story Worksheet 1 asks
about. The claim that Boltzmann chose capital eta, which looks like a
Latin $H$, is a long-standing conjecture (Chapman, Brush, Hjalmars), not
an established fact. Isolated entropy does not decrease — almost always.
The USPTO does examine perpetual-motion applications; it can require a
working model and typically rejects them for lack of operability or
utility.

Why should any of that matter for intelligence? Maxwell’s demon is the
1860s thought experiment that ties a *decision* to a thermodynamic
ledger. An entity sits in a box of ideal gas, at a partition, with a
trap door it can open without friction. If a fast molecule approaches
from one side it lets it through; if a slow one approaches from the
other, likewise. Temperature sets the variance of each Cartesian
component of velocity, $\sigma^2=k_BT/m$, not the variance of *speed*.
Speed has a nonzero mean; its variance is $(3-8/\pi)k_BT/m$. For
nitrogen at $300\,\mathrm{K}$ the component standard deviation is about
$300\,\mathrm{m\,s}^{-1}$, the rms speed about
$517\,\mathrm{m\,s}^{-1}$, and the standard deviation of speed about
$200\,\mathrm{m\,s}^{-1}$. After enough such decisions one side is
hotter than the other. The entropy of the gas has fallen. The second law
sits with the speed of light in the physicist’s mind: if you think you
have transgressed it, the error is in the thinking.

The demon is also the place where nineteenth-century physics starts to
conflate information, entropy, and energy. If information processing has
a physical cost, we can begin to ask questions about intelligence that
are not just polemic. I am happy to rank *an action* under a tightly
specified goal and a time bound — get these tables out of the room,
stacked, in twenty-five minutes — and then compare the energy two
methods spent. Thermodynamics already knows the tension: wait for
equilibrium and the engine is more efficient; demand the work in a hurry
and you pay in dissipation. Curzon and Ahlborn’s efficiency at maximum
power is $1-\sqrt{T_c/T_h}$ (1975). What I will not do is rank *an
entity* across every criterion by which a human might be judged. That is
Bostrom’s superintelligence (Bostrom, 2014). That last judgement — that
this is not a scientific ranking — is mine, not a result.

## Preview: Landauer’s Principle

One resolution offered for the demon is Landauer’s principle: deleting
information has a minimum energy cost, $k_B T\ln 2$ per bit, at the
temperature of the bath (Landauer, 1961). Boltzmann’s constant is of
order $10^{-23}$ joules per kelvin, so at $300\,\mathrm{K}$ the bound is
about $2.9\times 10^{-21}\,\mathrm{J}$ — not a constraint on any machine
we have built. It has been measured, not merely conjectured: Bérut,
Arakelyan, Petrosyan, Ciliberto, Dillenschneider and Lutz (2012) did so
with a colloidal particle. The engineering point is that today’s
dissipation sits far above the bound; the physical point is that the
bound is real.

The usual further claim is that *computation* need not cost energy,
because it can be made reversible. The computers we actually run are not
reversible, which is why data centres draw power. After Feynman’s death,
Tony Hey collected the lectures as *Feynman Lectures on Computation*;
Chapter 5 is the place he talks about this. Some of what he says remains
contested, including the claim that measurement and computation are the
same operation. Early arguments against the demon blamed the
*measurement*. Bennett’s position, and the majority view, is that
measurement need not produce entropy if computation is reversible;
Earman and Norton dispute that reading. I am not sure I believe the
reversible ideal. We do not have *practical* general-purpose reversible
computers. Adiabatic and reversible-logic prototypes have been built,
and quantum gates are logically reversible.

A student question from the first delivery is the right one, and we will
spend Week 4 on it: why is Maxwell’s demon a story about *erasure*? The
sketch — not a proof — is that the demon needs a memory. If that memory
scales with the number of particles that pass, either the tape is
unbounded (a Turing machine that will not fit in the box) or at some
point the demon must erase. Arguments then split. Physicists almost
always declare a *mechanism*: heat baths, frictionless doors, pistons.
Counter-arguments then say that mechanism cannot exist — thermal noise,
no frictionless door. The second law, like Shannon’s capacity, is not a
mechanism argument. It is a no-go. Mixing the two levels is why the
demon literature is still noisy. Week 4 is the treatment; today is the
hook.

## Preview: No-Go versus Mechanism

Shannon’s contribution, for this course, is less the proofs than the
refusal to talk about mechanism. He modelled sources as ergodic Markov
processes, which is a mechanism of a kind. The sharper point is that the
capacity theorem’s proof by random coding is *non-constructive*: it says
a code exists without building one. That is the no-go-versus-mechanism
contrast. The second law has the same shape: entropy must be produced
somewhere. It does not tell you when organised motion will stop.
Dynamics would. Entropy will not. That no-go/mechanism reading, and the
claim that probability is not enough, are the lecturer’s framing of the
course, not a settled result.

That is also why I have come to doubt that probability is enough, even
though most of my working life has been probabilistic modelling.
Probability insists that we specify what *will* happen, including the
parts of the support we do not care about. Entropy is the language in
which we say what will *not* happen. Week 2 takes that contrast
properly, including MacKay, no-go versus prescription, and the common
distributions as entropies. Today we only need the distinction, because
it is why the demon debate talks past itself: one side is arguing
mechanism, the other is arguing a no-go.

## Laplace’s Determinism

<span class="editsection-bracket" style="">\[</span><span
class="editsection"

Famously, Laplace considered the idea of a deterministic Universe, one
in which the model is *known*, or as the below translation refers to it,
“an intelligence which could comprehend all the forces by which nature
is animated.” He speculates on an “intelligence” that can submit this
vast data to analysis and proposes that such an entity would be able to
predict the future.

> Given for one instant an intelligence which could comprehend all the
> forces by which nature is animated and the respective situation of the
> beings who compose it—an intelligence sufficiently vast to submit
> these data to analysis—it would embrace in the same formula the
> movements of the greatest bodies of the universe and those of the
> lightest atom; for it, nothing would be uncertain and the future, as
> the past, would be present in its eyes. — Pierre-Simon Laplace, *A
> Philosophical Essay on Probabilities*, trans. F. W. Truscott and
> F. L. Emory (1902).

This notion is known as *Laplace’s demon* or *Laplace’s superman*.

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/laplacesDeterminismEnglish.png" style="width:60%; height:auto; display:block;">

Figure: <i>Laplace’s determinism in English translation.</i>

Unfortunately, most analyses of his ideas stop at that point, whereas
his real point is that such a notion is unreachable. Not so much
*superman* as *strawman*. A few pages later in the same Truscott and
Emory translation, Laplace goes on to observe:

> The curve described by a simple molecule of air or vapor is regulated
> in a manner just as certain as the planetary orbits; the only
> difference between them is that which comes from our ignorance.
>
> Probability is relative, in part to this ignorance, in part to our
> knowledge.}

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/philosophicaless00lapliala.png" style="width:60%; height:auto; display:block;">

In other words, we can never make use of the idealistic deterministic
Universe due to our ignorance about the world, Laplace’s suggestion, and
focus in this essay is that we turn to probability to deal with this
uncertainty. This is also our inspiration for using probability in
machine learning. This is the true message of Laplace’s essay, not
determinism, but the gremlin of uncertainty that emerges from our
ignorance.

The “forces by which nature is animated” is our *model*, the “situation
of beings that compose it” is our *data* and the “intelligence
sufficiently vast enough to submit these data to analysis” is our
compute. The fly in the ointment is our *ignorance* about these aspects.
And *probability* is the tool we use to incorporate this ignorance
leading to uncertainty or *doubt* in our predictions.

## Laplace’s Gremlin

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/laplaces-gremlin.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/laplaces-gremlin.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

> The curve described by a simple molecule of air or vapor is regulated
> in a manner just as certain as the planetary orbits; the only
> difference between them is that which comes from our ignorance.
> Probability is relative, in part to this ignorance, in part to our
> knowledge. We know that of three or greater number of events a single
> one ought to occur; but nothing induces us to believe that one of them
> will occur rather than the others. In this state of indecision it is
> impossible for us to announce their occurrence with certainty. It is,
> however, probable that one of these events, chosen at will, will not
> occur because we see several cases equally possible which exclude its
> occurrence, while only a single one favors it.
>
> — Pierre-Simon Laplace, *A Philosophical Essay on Probabilities*,
> trans. F. W. Truscott and F. L. Emory (1902)

The representation of ignorance through probability is the true message
of Laplace. I refer to this message as “Laplace’s gremlin,” because it
is the gremlin of uncertainty that interferes with the demon of
determinism to mean that our predictions are not deterministic.

Our separation of the uncertainty into the data, the model and the
computation gives three domains in which doubt can enter a prediction.
Bayes’ rule, a rearrangement of the product rule, is the foundation of
how we update beliefs in the presence of new information. We review it
later in this lecture.

The real point of Laplace’s essay was that we do not have access to all
the data, we do not have access to a complete physical understanding,
and even if we did we still would not have the compute needed for
deterministic predictions. There is uncertainty in the system which
means we cannot make precise predictions.

Gremlins are imaginary creatures used as an explanation of failure in
aircraft, causing crashes. In that sense the Gremlin represents the
uncertainty that a pilot felt about what might go wrong in a plane which
might be “theoretically sound” but in practice is poorly maintained or
exposed to conditions that take it beyond its design criteria. Laplace’s
gremlin is all the things that your model, data and ability to compute
don’t account for bringing about failures in your ability to predict.
Laplace’s gremlin is the uncertainty in the system.

## Preview: From Insufficient Reason to MaxEnt

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/e-t-jaynes.jpg" style="width:30%; height:auto; display:block;">

Figure: <i>E. T. Jaynes. Maximum entropy is the modern writing of
Laplace’s principle of insufficient reason.</i>

Laplace’s gremlin is uncertainty from ignorance — of data, of law, of
compute. He had a rule for turning that ignorance into a distribution:
the principle of insufficient reason. If $n$ outcomes are mutually
exclusive and nothing distinguishes them, take $p_i = 1/n$. He never
wrote that rule in the language of entropy. Jaynes, who read Laplace
hard, did: the principle of maximum entropy is the modern form of
insufficient reason (Jaynes, 2003). Laplace stayed with probability. One
of the arguments of this course is that entropy lets us make
declarations about what will not happen that probability, used alone,
does not. Week 5 is where we derive the occupation. Do not do Lagrange
multipliers today.

A question from the first delivery: what about uncertainty about the
specification itself — the risks you cannot even name? Dennis Lindley
argued that you integrate that level out and it propagates. I believed
that for a long time. I am less sure it is that simple. Anders Hansen,
in mathematics here, has written about the need for an “I do not know”
that current LLMs lack. Those are pointers, not this week’s work.

## Probability Review

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_ml/includes/probability-review-compact.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_ml/includes/probability-review-compact.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Notation: we often write $P(x,y)$ for $P(X=x,Y=y)$. Unlike a generic
bivariate function, $P(x,y)=P(y,x)$.

The product rule relates joint and conditional. The sum rule recovers a
marginal by summing out the variable you do not care about. Continuous
analogues replace sums by integrals. Expectation of a discrete random
variable is $\mathbb{E}[f(Y)]=\sum_y f(y)P(y)$. The four-outcome energy
table later in the lecture is that average with $f=-\log P$.

Bayes is not a third axiom; it is the product rule rearranged. The
barrels example below is a small discrete inversion of the conditioning.

<img src="https://mlatcl.github.io/iei/slides/diagrams/ml/bayes-two-barrels.svg" class="" width="80%" style="vertical-align:middle;">

Figure: <i>Two barrels: Barrel One holds 20 apples and 4 oranges; Barrel
Two holds 4 apples and 8 oranges. A barrel is chosen uniformly at random
and a single fruit is drawn.</i>

Work the numbers on the board if time allows. $P(\text{apple})=7/12$,

## Common Distributions

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_ml/includes/common-distributions-review.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_ml/includes/common-distributions-review.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

These are prerequisites restated, not new theory. The Gaussian (and the
categorical) reappear later as maximum-entropy distributions under
ordinary moment constraints. The Poisson is maximum-entropy only
relative to the base measure $1/k!$, and the binomial only relative to
$\binom{n}{k}$. Among distributions on the non-negative integers with a
fixed mean, the maximum-entropy one is the geometric. Week 4’s point
that \`\`the discretisation is the measure’’ — Jaynes’s $m$ — is the
reason those caveats exist.

A fair coin is Bernoulli($1/2$). A two-state thermal system with fixed
energies is Bernoulli once occupation probabilities are written down.
Binomial$(n,p)$ is $n$ independent Bernoulli trials,
$P(K=k)=\binom{n}{k}p^k(1-p)^{n-k}$, with mean $np$.

Poisson$(\lambda)$ is a count model with mean equal to variance.
Multinomial$(n,\mathbf{p})$ is $n$ trials into $K$ exclusive categories;
the binomial is the $K=2$ case.

The new idea we’re testing “Socratic method” can be deployed by

So in other words we are asking you to take on the role of Socrates in
these conversations.

The general background is an idea that in order to develop your
understanding of a subject through interaction with an LLM you need two
components to your enquiry: *curiosity* and *skepticism*. The curiosity
allows you to generate the prompt and the LLM to regurgitate some of its
knowledge (or perform searches that it summarises). But the skepticism
engages with that summary through challenging the conclusions that the
LLM has. In the Socratic *elenchus* that challenge is through pointing
out a logical inconsistency that arises (Vlastos (1993)), perhaps
through a side implication. For our purpose that challenge may not take
exactly that form. But it should push back on the narrative the LLM
provides. Generating such push back also requires you to engage with the
material the LLM has provided.

For Socrates these are curated dialogues (written by Plato; the modest
“I do not think I know what I do not know” is *Apology* 21d, alongside
the *Euthyphro* Plato (1914)). So it is normally the case that his
challenges hit home. In your case, that will not normally be the case.
And we don’t expect you to curate your dialogue. What we’d like instead
is a period of inquiry that is then summarised by a single dialogue that
is played out with one LLM in a short session of 10 prompts and
responses.

Classical Socratic practice (*elenchus*) tests consistency by
questioning, not by lecturing. Contemporary seminar pedagogy keeps the
same habit: the questioner holds the inquiry. Current LLMs default to
exposition and agreement; they rarely sustain adversarial follow-ups
without being steered. Assigning the student the Socrates role forces
engagement with the subject matter rather than passive acceptance of a
fluent summary.

## Energy in Machine Learning

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_ml/includes/energy-in-machine-learning.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_ml/includes/energy-in-machine-learning.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

An *energy* in machine learning is a scalar score on a configuration:
parameters, predictions, or binary states. Often an energy is equivalent
to an objective function or an error function. In this respect lower is
preferred. Energy-based models turn that score into a distribution by
the same exponential map used in statistical physics, \[ p(x)e^{-E(x)}.
\] Training normally then amounts to shaping $E$ so that observed
configurations sit in low-energy valleys.

We can reinterpret many machine learning algorithms in this way by
simply writing familiar loss functions in the energy language. Note
these are not new algorithms, just the energy interpretation.

## Energies from Distributions You Already Know

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_ml/includes/energy-from-distributions-examples.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_ml/includes/energy-from-distributions-examples.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The map $p\propto e^{-E}$ runs both ways. Given a distribution we
already introduced, set $$
E_i = -\log p_i
$$ (up to an additive constant that cancels in the normalisation). High
probability is low energy. For each classic family we plot $P$ (or $p$)
beside $E=-\log P$, then overlay the two on twin axes.

Exponentiating recovers the distribution (natural units):
$p(y)=e^{-E(y)}$ already normalises here because we took $E=-\log p$
exactly. In general one writes $p_i = e^{-E_i}/Z$ with
$Z=\sum_j e^{-E_j}$. Shannon entropy of this table is therefore also an
average energy.

A Bernoulli($p$) on $\{0,1\}$ has energies $E(x)=-\log P(x)$ (nats).
High probability is low energy. The panels below use a biased coin
$p=0.3$: first $P$ and $E$ side by side, then the same two curves
overlaid on twin axes so the inverse relationship is visible in one
frame.

````
import numpy as np
import matplotlib.pyplot as plt
import mlai
import mlai.plot as plot
````

````
p_success = 0.3
xs = np.array([0, 1])
probs = np.array([1 - p_success, p_success])
energy = -np.log(probs)
colours = ['#4C72B0', '#DD8452']

fig, axes = plt.subplots(1, 2, figsize=plot.big_wide_figsize)
axes[0].bar(xs, probs, color=colours[0], width=0.5)
axes[0].set_xticks(xs)
axes[0].set_xlabel('$x$')
axes[0].set_ylabel('$P(x)$')
axes[0].set_title(f'Bernoulli($p={p_success}$)')
axes[0].set_ylim(0, 1)

axes[1].bar(xs, energy, color=colours[1], width=0.5)
axes[1].set_xticks(xs)
axes[1].set_xlabel('$x$')
axes[1].set_ylabel('$E(x)=-\\log P(x)$')
axes[1].set_title('energy (nats)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('bernoulli-energy-panels.svg', directory='./ml')

fig, ax = plt.subplots(figsize=plot.big_figsize)
bars = ax.bar(xs, probs, color=colours[0], width=0.45, alpha=0.85, label='$P(x)$')
ax.set_xticks(xs)
ax.set_xlabel('$x$')
ax.set_ylabel('$P(x)$', color=colours[0])
ax.tick_params(axis='y', labelcolor=colours[0])
ax.set_ylim(0, 1)
ax2 = ax.twinx()
ax2.plot(xs, energy, 'o-', color=colours[1], lw=2, markersize=10, label='$E(x)$')
ax2.set_ylabel('$E(x)=-\\log P(x)$', color=colours[1])
ax2.tick_params(axis='y', labelcolor=colours[1])
ax.set_title(f'Bernoulli($p={p_success}$): $P$ and $E$ overlaid')
ax.spines['top'].set_visible(False)
handles = [bars, ax2.lines[0]]
ax.legend(handles, ['$P(x)$', '$E(x)$'], loc='upper center')
fig.tight_layout()
mlai.write_figure('bernoulli-energy-overlay.svg', directory='./ml')
````

Reading the Gaussian density as $p\propto e^{-E}$ makes the link to
regression immediate: minimising
$\sum_n(y_n-\mathbf{w}^\top\mathbf{x}_n)^2$ is minimising a sum of
quadratic energies, which is maximum likelihood under independent
Gaussian noise. The continuous normaliser plays the role of $Z$.

Under $y_n=\mathbf{w}^\top\mathbf{x}_n+\varepsilon_n$ with
$\varepsilon_n\sim\mathscr{N}\left(0,\sigma^2\right)$, the negative
log-likelihood is $\frac{1}{2\sigma^2}E(\mathbf{w})$ plus constants. So
ordinary least squares *is* maximum likelihood for a Gaussian or
equivalently, minimising a quadratic energy.

When $y$ is a one-hot label and $\hat{y}$ a categorical prediction,
cross-entropy is the energy of that pair. Softmax turns a vector of
logits into a distribution. Logits are *negative* energies: if $E_c$ is
the class energy then $\hat{y}=\mathrm{softmax}(-E)$. The functional
$-\log\hat{y}_c$ is the same one that appears in Shannon entropy
$\mathbb{E}[-\log P(y)]$, applied to a *model* distribution rather than
the data distribution.

Ackley, Hinton and Sejnowski (Ackley et al., 1985) made the couplings
$W_{ij}$ learnable; Hopfield (Hopfield, 1982) used a similar pairwise
energy for associative memory with fixed weights. When a Boltzmann or
Gibbs occupation is written $p\propto e^{-\beta E}$, the symbol $E$ is
the same kind of object minimised in regression and classification — a
score over configurations, exponentiated into a distribution.

## Boltzmann Seed

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/mlatcl/iei/edit/gh-pages/_lamd/intro-review-motivation.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/mlatcl/iei/edit/gh-pages/_lamd/intro-review-motivation.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

In machine learning an energy is often just a negative log probability,
measured in nats, which is a peculiar unit for something that in physics
is measured in joules. Energy-based models are, among other things, a
way of not writing the partition function $Z$. In physics $Z$ is the
whole story. We return to it with free energy; do not compute it today.

````
import numpy as np
````

````
def boltzmann(energies, beta):
    """Boltzmann probabilities $p_i \\propto e^{-\\beta E_i}$."""
    log_w = -beta * np.asarray(energies, dtype=float)
    log_w -= log_w.max()
    w = np.exp(log_w)
    return w / w.sum()

# Live check: boltzmann([0, 1], 1.0) -> about (0.731, 0.269)
````
