## Perpetual Motion and Superintelligence

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/perpetual-motion-superintelligence-analogy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/perpetual-motion-superintelligence-analogy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Imagine in 1925 a world where the automobile is already transforming
society, but big promises are being made for things to come. The stock
market is soaring, the 1918 pandemic is forgotten. And every major
automobile manufacturer is investing heavily on the promise they will
each be the first to produce a car that needs no fuel. A perpetual
motion machine.

Well, of course that didn’t happen. But I sometimes wonder if what we’re
seeing today 100 years later is the modern equivalent of that. In 2026
billions are being invested in promises of superintelligence and
artificial general intelligence that will transform everything.

We know why perpetual motion is impossible: the second law of
thermodynamics tells us that entropy always increases. So we can’t have
motion without entropy production. No matter how clever the design, you
cannot extract energy from nothing, and you cannot create a closed
system that does useful work indefinitely without an external energy
source.

How might we make an equivalent statement for the bizarre claims around
superintelligence? Some inspiration comes from Maxwell’s demon, an
“intelligent” entity which operates against the laws of thermodynamics.
The inspiration comes because the demon suggests that for the second law
to hold there must be a relationship between the demon’s decisions and
thermodynamic entropy.

One of the resolutions comes from *Landauer’s principle*, the notion
that erasure of information requires heat dissipation. This suggests
there are fundamental information-theoretic constraints on intelligent
systems, just as there are thermodynamic constraints on engines.

I’ve no doubt that AI technologies will transform our world just as much
as the automobile has. But I also have no doubt that the promise of
unconstrained superintelligence is just as silly as the promise of
perpetual motion.

## Laplace’s Gremlin

Unfortunately, most analyses of his ideas stop at that point, whereas
his real point is that such a notion is unreachable. Not so much
*superman* as *strawman*. Just three pages later in the “Philosophical
Essay on Probabilities” (Laplace, 1814), Laplace goes on to observe:

> The curve described by a simple molecule of air or vapor is regulated
> in a manner just as certain as the planetary orbits; the only
> difference between them is that which comes from our ignorance.
>
> Probability is relative, in part to this ignorance, in part to our
> knowledge.

```
import notutils as nu
nu.display_google_book(id='1YQPAAAAQAAJ', page='PR17-IA4')
```

Figure: <i>Laplace’s gremlin in the original French.</i>

<img class="" src="https://mlatcl.github.io/iei/slides/diagrams/physics/philosophicaless00lapliala.png" style="width:60%; height:auto; display:block;">

Figure: <i>To Laplace, determinism is a strawman. Ignorance of mechanism
and data leads to uncertainty which should be dealt with through

In other words, we can never make use of the idealistic deterministic
Universe due to our ignorance about the world, Laplace’s suggestion, and
focus in this essay is that we turn to probability to deal with this
uncertainty. This is also our inspiration for using probability in
machine learning. This is the true message of Laplace’s essay, not
determinism, but the gremlin of uncertainty that emerges from our
ignorance.

The “forces by which nature is animated” is our *model*, the “situation
of beings that compose it” is our *data* and the “intelligence
sufficiently vast enough to submit these data to analysis” is our
compute. The fly in the ointment is our *ignorance* about these aspects.
And *probability* is the tool we use to incorporate this ignorance
leading to uncertainty or *doubt* in our predictions.

## Laplace’s Gremlin

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_physics/includes/laplaces-gremlin.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_physics/includes/laplaces-gremlin.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

> The curve described by a simple molecule of air or vapor is regulated
> in a manner just as certain as the planetary orbits; the only
> difference between them is that which comes from our ignorance.
> Probability is relative, in part to this ignorance, in part to our
> knowledge. We know that of three or greater number of events a single
> one ought to occur; but nothing induces us to believe that one of them
> will occur rather than the others. In this state of indecision it is
> impossible for us to announce their occurrence with certainty. It is,
> however, probable that one of these events, chosen at will, will not
> occur because we see several cases equally possible which exclude its
> occurrence, while only a single one favors it.
>
> — Pierre-Simon Laplace (Laplace, 1814), pg 5

The representation of ignorance through probability is the true message
of Laplace, I refer to this message as “Laplace’s gremlin,” because it
is the gremlin of uncertainty that interferes with the demon of
determinism to mean that our predictions are not deterministic.

Our separation of the uncertainty into the data, the model and the
computation give us three domains in which our doubts can creep into our
ability to predict. Over the last three lectures we’ve introduced some
of the basic tools we can use to unpick this uncertainty. You’ve been
introduced to, (or have yow reviewed) *Bayes’ rule*. The rule, which is
a simple consequence of the product rule of probability, is the
foundation of how we update our beliefs in the presence of new
information.

The real point of Laplace’s essay was that we don’t have access to all
the data, we don’t have access to a complete physical understanding, and
as the example of the Game of Life shows, even if we did have access to
both (as we do for “Conway’s universe”) we still don’t have access to
all the compute that we need to make deterministic predictions. There is
uncertainty in the system which means we can’t make precise predictions.

Gremlins are imaginary creatures used as an explanation of failure in
aircraft, causing crashes. In that sense the Gremlin represents the
uncertainty that a pilot felt about what might go wrong in a plane which
might be “theoretically sound” but in practice is poorly maintained or
exposed to conditions that take it beyond its design criteria. Laplace’s
gremlin is all the things that your model, data and ability to compute
don’t account for bringing about failures in your ability to predict.
Laplace’s gremlin is the uncertainty in the system.

## Doubt

What Laplace’s gremlin implies in practice is that we can never be
certain about outcome, so doubt is an inevitable consequence of our lack
of knowledge. It is what drives us to develop quantitative notions for
uncertainty. We have already introduced probability. Next we meet
*energy* as used in machine learning, then *entropy* — the concept that
unifies information and uncertainty, and therefore we argue (perhaps)
intelligence.

We will look to apply these ideas to how we approach our learning as
well as what we learn. Bringing doubt into your own reasoning is a way
in which you can develop your understanding. Learning to express that
doubt through challenging a statement. The model we will use finds its
origins in Socrates.

## Worksheets and LLMs: Can you be Socrates?

The general approach we’d like to take to this course is that of a
“community of inquiry” (see Chapter 4, Lipman (2012)). The unusual
modern twist on this notion is that the LLMs themselves become part of
that community.

Each of us will have different perspectives on what an LLM does and does
not provide. You are welcome to bring those perspectives into your work.
In particular, for each worksheet, you will be asked to reflect on the
LLM responses and the process. Part of that reflection should be
specific to the exercise and what you learnt about the subject. But I
would like part of that reflection to be general about your
understanding of the LLM and what it does and doesn’t provide. For
worksheets 2, 3, and 4 a portion of that reflection will be on how you
feel your understanding of LLMs as a tool of inquiry has evolved (if it
evolved!).

The assessment model is based on is 1. a form of questioning enquiry
generally called “the Socratic method” is an informative way of
exploring a topic 2. an impression that the current generation of LL
models is weak when asked to sustain Socratic dialogue. Answers tend to
be expansive and when read carefully can lead to ‘diaclectic vertigo’ 3.
The new idea we’re testing “Socratic method” can be deployed by
reversing the role of Socrates and the student.

So in other words we are asking you to take on the role of Socrates in
these conversations.

The general background is an idea that in order to develop your
understanding of a subject through interaction with an LLM you need two
components to your enquiry: *curiousity* and *skepticism*. The
curiousity allows you to generate the prompt and the LLM to regurgitate
some of its knowled (or perform searches that it summarises). But the
skepticism engages with that summary through challenging the conclusions
that the LLM has. In the Socratic *elenchus* that challenge is through
pointing out a logical inconsistency that arises (Vlastos (1993)),
perhaps through a side implication. For our purpose that challenge may
not take exactly that form. But it should push back on the narrative the
LLM provides. Generating such push back also requires you to engage with
the material the LLM has provided.

For Socrates these are curated dialogues (written by Plato, e.g. Plato
(1914)). So its normally the case that his challenges hit home. In your
case, that won’t normally be the case. And we don’t expect you to curate
your dialogue. What we’d like instead is a period of inquiry that is
then summarised by a single dialogue that is played out with one LLM in
a short session of 10 prompts and responses.

Classical Socratic practice (*elenchus*) tests consistency by
questioning, not by lecturing. Contemporary seminar pedagogy keeps the
same habit: the questioner holds the inquiry. Current LLMs default to
exposition and agreement; they rarely sustain adversarial follow-ups
without being steered. Assigning the student the Socrates role forces
engagement with the subject matter rather than passive acceptance of a
fluent summary.

## Energy in Machine Learning

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_ml/includes/energy-in-machine-learning.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_ml/includes/energy-in-machine-learning.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

An *energy* in machine learning is a scalar score on a configuration:
parameters, predictions, or binary states. Often an energy is equivalent
to an objective function or an error function. In this respect lower is
preferred. Energy-based models turn that score into a distribution by
the same exponential map used in statistical physics, \[ p(x)e^{-E(x)}.
\] Training normally then amounts to shaping $E$ so that observed
configurations sit in low-energy valleys.

We can reinterpret many machine learning algorithms in this way by
simply writing familiar loss functions in the energy language. Note
these are not new algorithms, just the energy interpretation.

## Energies from Distributions You Already Know

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_ml/includes/energy-from-distributions-examples.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_ml/includes/energy-from-distributions-examples.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

The map $p\propto e^{-E}$ runs both ways. Given a distribution we
already introduced, set $$
E_i = -\log p_i
$$ (up to an additive constant that cancels in the normalisation). High
probability is low energy. The numbers below reuse examples from earlier
in the lecture. For each classic family we plot $P$ (or $p$) beside
$E=-\log P$, then overlay the two on twin axes.

Exponentiating recovers the distribution (natural units):
$p(y)=e^{-E(y)}$ already normalises here because we took $E=-\log p$
exactly. In general one writes $p_i = e^{-E_i}/Z$ with
$Z=\sum_j e^{-E_j}$. The entropy calculation from the review is
therefore also an average energy.

## Poisson: Distribution and Energy

A Poisson($\lambda$) on the non-negative integers has $$
P(k)=\frac{\lambda^k e^{-\lambda}}{k!},\qquad E(k)=-\log P(k)
$$ (nats). The bulk of the mass sits in a shallow energy valley; the
tails climb as $k!$ in the denominator of $P$ becomes the dominant cost.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
import mlai
import mlai.plot as plot
```

```
lam = 4.0
ks = np.arange(0, 16)
probs = poisson.pmf(ks, lam)
energy = -np.log(np.clip(probs, 1e-300, None))
colours = ['#4C72B0', '#DD8452']

fig, axes = plt.subplots(1, 2, figsize=plot.big_wide_figsize)
axes[0].bar(ks, probs, color=colours[0], width=0.8)
axes[0].set_xlabel('$k$')
axes[0].set_ylabel('$P(k)$')
axes[0].set_title(f'Poisson($\\lambda={lam:g}$)')

axes[1].bar(ks, energy, color=colours[1], width=0.8)
axes[1].set_xlabel('$k$')
axes[1].set_ylabel('$E(k)=-\\log P(k)$')
axes[1].set_title('energy (nats)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('poisson-energy-panels.svg', directory='./ml')

fig, ax = plt.subplots(figsize=plot.big_figsize)
bars = ax.bar(ks, probs, color=colours[0], width=0.8, alpha=0.85, label='$P(k)$')
ax.set_xlabel('$k$')
ax.set_ylabel('$P(k)$', color=colours[0])
ax.tick_params(axis='y', labelcolor=colours[0])
ax2 = ax.twinx()
ax2.plot(ks, energy, 'o-', color=colours[1], lw=2, markersize=5, label='$E(k)$')
ax2.set_ylabel('$E(k)=-\\log P(k)$', color=colours[1])
ax2.tick_params(axis='y', labelcolor=colours[1])
ax.set_title(f'Poisson($\\lambda={lam:g}$): $P$ and $E$ overlaid')
ax.spines['top'].set_visible(False)
handles = [bars, ax2.lines[0]]
ax.legend(handles, ['$P(k)$', '$E(k)$'], loc='upper right')
fig.tight_layout()
mlai.write_figure('poisson-energy-overlay.svg', directory='./ml')
```

The univariate Gaussian already written as $$
p(y)\propto\exp\!\left(-\frac{(y-\mu)^2}{2\sigma^2}\right)
$$ is an energy-based model with quadratic energy. Taking
$E(y)=-\log p(y)$ (nats) adds only the normaliser
$\tfrac12\log(2\pi\sigma^2)$, so the shape remains a parabola about
$\mu$. The height example $\mu=1.7$, $\sigma^2=0.0225$ is reused below.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
import mlai
import mlai.plot as plot
```

```
mu, var = 1.7, 0.0225
sigma = np.sqrt(var)
ys = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 400)
pdf = norm.pdf(ys, loc=mu, scale=sigma)
energy = -np.log(pdf)
colours = ['#4C72B0', '#DD8452']

fig, axes = plt.subplots(1, 2, figsize=plot.big_wide_figsize)
axes[0].plot(ys, pdf, color=colours[0], lw=2)
axes[0].fill_between(ys, pdf, color=colours[0], alpha=0.25)
axes[0].axvline(mu, color='#666', ls='--', lw=1)
axes[0].set_xlabel('$y$')
axes[0].set_ylabel('$p(y)$')
axes[0].set_title(f'Gaussian($\\mu={mu}, \\sigma^2={var}$)')

axes[1].plot(ys, energy, color=colours[1], lw=2)
axes[1].axvline(mu, color='#666', ls='--', lw=1)
axes[1].set_xlabel('$y$')
axes[1].set_ylabel('$E(y)=-\\log p(y)$')
axes[1].set_title('energy (nats; quadratic)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('gaussian-energy-panels.svg', directory='./ml')

fig, ax = plt.subplots(figsize=plot.big_figsize)
line_p, = ax.plot(ys, pdf, color=colours[0], lw=2, label='$p(y)$')
ax.fill_between(ys, pdf, color=colours[0], alpha=0.2)
ax.axvline(mu, color='#666', ls='--', lw=1)
ax.set_xlabel('$y$')
ax.set_ylabel('$p(y)$', color=colours[0])
ax.tick_params(axis='y', labelcolor=colours[0])
ax2 = ax.twinx()
line_e, = ax2.plot(ys, energy, color=colours[1], lw=2, label='$E(y)$')
ax2.set_ylabel('$E(y)=-\\log p(y)$', color=colours[1])
ax2.tick_params(axis='y', labelcolor=colours[1])
ax.set_title('Gaussian: density and quadratic energy overlaid')
ax.spines['top'].set_visible(False)
ax.legend([line_p, line_e], ['$p(y)$', '$E(y)$'], loc='upper right')
fig.tight_layout()
mlai.write_figure('gaussian-energy-overlay.svg', directory='./ml')
```

Reading the Gaussian density as $p\propto e^{-E}$ makes the link to
regression immediate: minimising
$\sum_n(y_n-\mathbf{w}^\top\mathbf{x}_n)^2$ is minimising a sum of
quadratic energies, which is maximum likelihood under independent
Gaussian noise. The continuous normaliser plays the role of $Z$.

Under $y_n=\mathbf{w}^\top\mathbf{x}_n+\varepsilon_n$ with
$\varepsilon_n\sim\mathscr{N}\left(0,\sigma^2\right)$, the negative
log-likelihood is $\frac{1}{2\sigma^2}E(\mathbf{w})$ plus constants. So
ordinary least squares *is* maximum likelihood for a Gaussian or
equivalently, minimising a quadratic energy.

When $y$ is a one-hot label and $\hat{y}$ a categorical prediction,
cross-entropy is the energy of that pair. Softmax turns a vector of
class energies (or logits) into a distribution. The functional
$-\log\hat{y}_c$ is the same one that appears in Shannon entropy
$\mathbb{E}[-\log P(y)]$, applied to a *model* distribution rather than
the data distribution.

Ackley, Hinton and Sejnowski (Ackley et al., 1985) made the couplings
$W_{ij}$ learnable; Hopfield (Hopfield, 1982) used a similar pairwise
energy for associative memory with fixed weights. When a Boltzmann or
Gibbs occupation is written $p\propto e^{-\beta E}$, the symbol $E$ is
the same kind of object minimised in regression and classification — a
score over configurations, exponentiated into a distribution.

## Boltzmann Seed

```
import numpy as np
```

```
def boltzmann(energies, beta):
    """Boltzmann probabilities $p_i \\propto e^{-\\beta E_i}$."""
    log_w = -beta * np.asarray(energies, dtype=float)
    log_w -= log_w.max()
    w = np.exp(log_w)
    return w / w.sum()

# Live check: boltzmann([0, 1], 1.0) -> about (0.731, 0.269)
```

## The Information Game

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/information-game-guessing.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

Before writing a formula for entropy, play a game. One player thinks of
an object drawn from a known finite list. The other may ask only yes/no
questions. How many questions do you need in the worst case? That count
is already a measure of information: it is how much uncertainty you must
remove to name the object.

With a list of size $N$ and equal prior probability on each item, an
ideal strategy halves the search space at every step. After $k$ answers
you can distinguish at most $2^k$ hypotheses, so you need $$
k \ge \log_2 N
$$ questions in the worst case. Twenty questions reach a million
possibilities: $2^{20} = 1\,048\,576$. A *fair* yes/no answer therefore
carries one *bit* of information. That is Shannon’s unit before
Shannon’s formula: information as the number of balanced binary
distinctions required to resolve uncertainty.

If the prior is not uniform, the best question is no longer a literal
half-split of the *count* of remaining objects. It is a half-split of
their *probability*. The average number of questions under optimal play
becomes the Shannon entropy $H = -\sum_i p_i\log_2 p_i$. We will meet
that formula next; the game already shows what the number is counting.

## MacKay’s Balance and Balls

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/mackay-weighing-problem.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

David MacKay opens the compression half of *Information Theory,
Inference, and Learning Algorithms* with a weighing puzzle (MacKay,
2003, p. Exercise\~4.1). It is the information game with a ternary
instrument instead of a binary question. We use it here only as a
*counting* argument: how much uncertainty we start with, how much each
use of the balance can remove, and why three weighings are enough in
principle — without exhibiting a particular sequence of measurements.

<img src="https://mlatcl.github.io/iei/slides/diagrams/information/mackay-balance-balls.svg" class="" width="80%" style="vertical-align:middle;">

Figure: <i>A two-pan balance and twelve identical balls. Each use of the
balance returns one of three readings: left heavy, balance, or right
heavy.</i>

You are given twelve balls, identical in appearance. Exactly one differs
in weight from the others; you do not know whether that odd ball is
heavier or lighter. A two-pan balance compares two equal-sized groups
and returns one of three readings: left side heavier, right side
heavier, or equal. Design a strategy that finds the odd ball and the
direction of the defect.

Under a uniform prior over those twenty-four possibilities, the Shannon
uncertainty we must remove is $$
H_{\mathrm{start}} = \log_2 24 \approx 4.58 \text{ bits}.
$$ At the end we need a single named hypothesis: $H_{\mathrm{end}} = 0$.
The *deficit* to close is therefore about $4.58$ bits. That difference —
start minus end — is the information the experiment must supply.

A binary question yields at most one bit. The balance is richer: its
alphabet size is three. If the three readings are equiprobable, one
weighing contributes $$
\log_2 3 \approx 1.585 \text{ bits}.
$$ After $k$ weighings the strategy tree has at most $3^k$ leaves, so it
can name at most $3^k$ hypotheses. Compare budgets: $$
\begin{align*}
3^2 &= 9 < 24,\\
3^3 &= 27 \ge 24.
\end{align*}
$$ Two weighings cannot suffice no matter how clever the groupings.
Three weighings *can*: there is spare capacity of three unused leaves
($27-24=3$). In bits, $$
3\log_2 3 \approx 4.75 > \log_2 24 \approx 4.58.
$$ The gap is small. That is why the puzzle feels tight, and why a first
weighing that wastes an outcome (for example putting six against six,
which can never balance if exactly one ball is odd) fails: you have
thrown away ternary resolution and fallen back toward a binary split.

We stop at the existence argument. Constructing the adaptive tree —
which balls on which pan at each step — is MacKay’s exercise, and
solving it is the best way to feel what \`\`informative experiment’’
means. The only design principle we need today is the one already
present in the information game: choose each measurement so that its
outcomes are as nearly equiprobable as possible under the current
posterior. That maximises expected information gain per weighing and is
what makes the $3^3 \ge 24$ budget achievable in practice.

That last number is Shannon entropy in nats:
$H(p)=\mathbb{E}_{p}[-\log p(y)]=-\sum_y p(y)\log p(y)$. Same arithmetic
under another name.

Under Shannon’s definition, entropy is dimensionless. It is the
expectation of the negative log surprisal,
$-\left\langle\log P(y)\right\rangle$. The information game and MacKay’s
balance give the operational reading: $H$ is how much uncertainty you
must remove; the log alphabet size of your instrument bounds how much
each query can buy.

## Bernoulli Entropy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/bernoulli-entropy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/bernoulli-entropy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

A Bernoulli($p$) random variable takes values $\{0,1\}$ with
probabilities $1-p$ and $p$. Its Shannon entropy is the binary entropy
$$
H(X) = -p\log_2 p - (1-p)\log_2(1-p)
$$ bits. The panels below show, for a fixed $p$, the distribution, the
pointwise surprisal $-\log_2 P(x)$, and the summands $-P(x)\log_2 P(x)$
whose total is $H$. The same $p=0.3$ coin appeared in the energy
overlay: middle panel is that energy in bits; the right panel averages
it to $H$.

```
import numpy as np
import matplotlib.pyplot as plt
import mlai
import mlai.plot as plot
```

```
p_success = 0.3
xs = np.array([0, 1])
probs = np.array([1 - p_success, p_success])
surprisal = -np.log2(probs)
contrib = probs * surprisal
H = contrib.sum()

fig, axes = plt.subplots(1, 3, figsize=plot.big_wide_figsize)
colours = ['#4C72B0', '#DD8452', '#55A868']
axes[0].bar(xs, probs, color=colours[0], width=0.5)
axes[0].set_xticks(xs)
axes[0].set_xlabel('$x$')
axes[0].set_ylabel('$P(x)$')
axes[0].set_title(f'Bernoulli($p={p_success}$)')
axes[0].set_ylim(0, 1)

axes[1].bar(xs, surprisal, color=colours[1], width=0.5)
axes[1].set_xticks(xs)
axes[1].set_xlabel('$x$')
axes[1].set_ylabel('$-\\log_2 P(x)$')
axes[1].set_title('surprisal')

axes[2].bar(xs, contrib, color=colours[2], width=0.5)
axes[2].set_xticks(xs)
axes[2].set_xlabel('$x$')
axes[2].set_ylabel('$-P(x)\\log_2 P(x)$')
axes[2].set_title(f'summands ($H={H:.3f}$ bits)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('bernoulli-entropy-panels.svg', directory='./information')
```

## Binomial Entropy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/binomial-entropy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/binomial-entropy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

A Binomial($n,p$) count is the sum of $n$ i.i.d. Bernoulli trials. There
is no elementary closed form for $H$, but the same three-panel reading
applies: plot $P(k)$, plot $-\log_2 P(k)$, and sum $-P(k)\log_2 P(k)$.
For large $n$ a Gaussian approximation gives
$H\approx\tfrac12\log_2\bigl(2\pi e\, np(1-p)\bigr)$ bits.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
import mlai
import mlai.plot as plot
```

```
n, p = 12, 0.4
ks = np.arange(0, n + 1)
probs = binom.pmf(ks, n, p)
# Guard zeros for log (binom pmf is positive on 0..n)
surprisal = -np.log2(np.clip(probs, 1e-300, None))
contrib = probs * surprisal
H = contrib.sum()

fig, axes = plt.subplots(1, 3, figsize=plot.big_wide_figsize)
colours = ['#4C72B0', '#DD8452', '#55A868']
axes[0].bar(ks, probs, color=colours[0], width=0.8)
axes[0].set_xlabel('$k$')
axes[0].set_ylabel('$P(k)$')
axes[0].set_title(f'Binomial($n={n}, p={p}$)')

axes[1].bar(ks, surprisal, color=colours[1], width=0.8)
axes[1].set_xlabel('$k$')
axes[1].set_ylabel('$-\\log_2 P(k)$')
axes[1].set_title('surprisal')

axes[2].bar(ks, contrib, color=colours[2], width=0.8)
axes[2].set_xlabel('$k$')
axes[2].set_ylabel('$-P(k)\\log_2 P(k)$')
axes[2].set_title(f'summands ($H={H:.3f}$ bits)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('binomial-entropy-panels.svg', directory='./information')
```

## Poisson Entropy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/poisson-entropy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/poisson-entropy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

A Poisson($\lambda$) distribution on the non-negative integers has $$
P(k)=\frac{\lambda^k e^{-\lambda}}{k!}.
$$ Entropy again has no elementary closed form; numerically
$H=\sum_k -P(k)\log_2 P(k)$. For large $\lambda$,
$H\approx\tfrac12\log_2(2\pi e\lambda)$ bits. The $\lambda=4$ panels
reuse the count distribution from the energy overlay: surprisal is that
energy in bits; the right panel is the average that defines $H$.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
import mlai
import mlai.plot as plot
```

```
lam = 4.0
ks = np.arange(0, 16)
probs = poisson.pmf(ks, lam)
surprisal = -np.log2(np.clip(probs, 1e-300, None))
contrib = probs * surprisal
H = contrib.sum()

fig, axes = plt.subplots(1, 3, figsize=plot.big_wide_figsize)
colours = ['#4C72B0', '#DD8452', '#55A868']
axes[0].bar(ks, probs, color=colours[0], width=0.8)
axes[0].set_xlabel('$k$')
axes[0].set_ylabel('$P(k)$')
axes[0].set_title(f'Poisson($\\lambda={lam:g}$)')

axes[1].bar(ks, surprisal, color=colours[1], width=0.8)
axes[1].set_xlabel('$k$')
axes[1].set_ylabel('$-\\log_2 P(k)$')
axes[1].set_title('surprisal')

axes[2].bar(ks, contrib, color=colours[2], width=0.8)
axes[2].set_xlabel('$k$')
axes[2].set_ylabel('$-P(k)\\log_2 P(k)$')
axes[2].set_title(f'summands ($H={H:.3f}$ bits)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('poisson-entropy-panels.svg', directory='./information')
```

## Gaussian Differential Entropy

<span class="editsection-bracket" style="">\[</span><span
class="editsection"
style=""><a href="https://github.com/lawrennd/snippets/edit/main/_information/includes/gaussian-entropy.md" target="_blank" onclick="ga('send', 'event', 'Edit Page', 'Edit', 'https://github.com/lawrennd/snippets/edit/main/_information/includes/gaussian-entropy.md', 13);">edit</a></span><span class="editsection-bracket" style="">\]</span>

For a continuous density the same formal expression $$
h(X) = -\int p(y)\log_2 p(y)\,\mathrm{d}y
$$ is called *differential* entropy. It can be negative, and it is not
invariant to rescaling of $y$. For a univariate Gaussian
$\mathcal{N}(\mu,\sigma^2)$, $$
h = \tfrac12\log_2\bigl(2\pi e\sigma^2\bigr)
$$ bits. The panels below use the height example ($\mu=1.7$,
$\sigma^2=0.0225$): density, surprisal $-\log_2 p(y)$, and the integrand
$-p(y)\log_2 p(y)$. That is the same height Gaussian as the
quadratic-energy overlay: the middle curve is $E=-\log p$ in bits (still
a parabola); the right panel integrates $p\,E$ to differential entropy.

```
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
import mlai
import mlai.plot as plot
```

```
mu, var = 1.7, 0.0225
sigma = np.sqrt(var)
ys = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 400)
# Density in probability per unit length
pdf = norm.pdf(ys, loc=mu, scale=sigma)
# Surprisal in bits; pdf can exceed 1, so -log2(pdf) can be negative
surprisal = -np.log2(pdf)
integrand = pdf * surprisal
# Closed form in bits
h = 0.5 * np.log2(2 * np.pi * np.e * var)

fig, axes = plt.subplots(1, 3, figsize=plot.big_wide_figsize)
colours = ['#4C72B0', '#DD8452', '#55A868']
axes[0].plot(ys, pdf, color=colours[0], lw=2)
axes[0].fill_between(ys, pdf, color=colours[0], alpha=0.25)
axes[0].axvline(mu, color='#666', ls='--', lw=1)
axes[0].set_xlabel('$y$')
axes[0].set_ylabel('$p(y)$')
axes[0].set_title(f'Gaussian($\\mu={mu}, \\sigma^2={var}$)')

axes[1].plot(ys, surprisal, color=colours[1], lw=2)
axes[1].axhline(0, color='#666', ls=':', lw=1)
axes[1].set_xlabel('$y$')
axes[1].set_ylabel('$-\\log_2 p(y)$')
axes[1].set_title('surprisal (can be negative)')

axes[2].plot(ys, integrand, color=colours[2], lw=2)
axes[2].fill_between(ys, integrand, color=colours[2], alpha=0.25)
axes[2].set_xlabel('$y$')
axes[2].set_ylabel('$-p(y)\\log_2 p(y)$')
axes[2].set_title(f'integrand ($h={h:.3f}$ bits)')
for ax in axes:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
fig.tight_layout()
mlai.write_figure('gaussian-entropy-panels.svg', directory='./information')
```

Figure: <i>Univariate Gaussian (height example). Middle: $-\log_2 p(y)$
— quadratic energy in bits; dips below zero where the density exceeds

Thermodynamic entropy $S$ and Shannon $H$ have a shared functional form,
in equilibrium we can write $S = k_BH$, where $k_B$ is Boltzmann’s
constant,
$k_B = 1.380649 \times 10^{-23} \tfrac{\text{kg}\text{m}^2 }{\text{s}^2 \text{K}}$.

So there are two parts to a thermodynamic entropy, one is the
uncertainty of a (discrete) distribution. The other translates that
entropy into units of ‘work’
i.e. ($\text{J} = \tfrac{\text{kg}\,\text{m}^2 }{\text{s}^2}$) per
Kelvin ($\tfrac{\text{J}}{\text{K}}$). Boltzmann’s constant makes
entropy physical. When we multiply $S$ by temperature $T$ (in degrees
Kelvin) we recover a term that has units of energy.

But the dependence of entropy on the states of a distribution, or the
probability across those states, also implies that there is a *choice*
in terms of which states we consider. In classical mechanics we have
states like position and velocity (or momentum). But quantum mechanics
also leads to other properties of materials such as the *spin* of an
electron. Choice of which variables you consider when computing e.g. the
entropy of an electron, was referred to as the ‘“anthropomorphic” nature
of entropy’ by Jaynes (1965). He credits this remark originally to
Eugene Wigner. One example that Jaynes gives is thermodynamic steam
tables.

> If we work with a system of degrees of freedom, the experimental
> entropy is a function $S_{0}(X_{1}\ldots X_{n})$ of $n$ independent
> variables. But the system has any number of additional degrees of
> freedom $X_{n + 1},\ X_{n + 2}$, etc. We have to understand that these
> additional degrees of freedom are not to be tampered with during the
> experiments on the $n$ degrees of interest; otherwise one could easily
> produce *apparent violations of the second law.* Jaynes (1965) (my
> emphasis)

He goes on to give an example from entropy of steam (which is very
important in the design of steam engines and power plants).

> For example, the engineers have their “steam tables,” which give
> measured values of the entropy of superheated steam at various
> temperatures and pressures. But the H\$\_2\$0 molecule has a large
> electric dipole moment; and so the entropy of steam depends
> appreciably on the electric field strength present. It must always be
> understood implicitly (because it is never stated explicitly) that
> this extra thermodynamic degree of freedom was not tampered with
> during the experiments on which the steam tables are based; which
> means, in this case, that the electric field was not inadvertently
> varied from one measurement to the next.

This leads him to a conclusion that is worth bearing in mind when we’re
talking about the physical meaning of entropy.

> Recognition that the “entropy of a physical system” is not meaningful
> without further qualifications is important in clarifying many
> questions concerning irreversibility and the second law. For example,
> I have been asked several times whether, in my opinion, a biological
> system, say a cat, which converts inanimate food into a highly
> organized structure and behavior, represents a violation of the second
> law. The answer I always give is that, until we specify the set of
> parameters which define the *thermodynamic state* of the cat, no
> definite question has been asked!

Returning to the mathematical form of entropy, the fact that it is based
on probability means that many of the properties of probability carry
across to entropy. But the use of the logarithm means that they manifest
as additions instead of multiplications.

For example the joint entropy of two variables, $$
H(X, Y) = - \sum_{X, Y} P(X, Y) \log P(X, Y),
$$ can be decomposed as the entropy of the marginal and the entropy of
the conditional, $$
H(X, Y) = H(X \mid Y) + H(Y)
$$ where $$
H(Y \mid X) = - \sum_{X, Y} P(X, Y) \log P(X \mid Y) 
$$ and $$
H(Y) = - \sum_{Y} P(Y) \log P(Y).
$$ This is known as the *chain rule* of entropy. And it follows from the
product rule.

The equivalent of Bayes’ rule is $$
H(Y \mid X) = H(X \mid Y) + H(Y) - H(X).
$$

## Entropy and Probability

<span class="editsection-bracket" style="">\[</span><span

I’ve spent most of my career in machine learning thinking about
uncertainty, and very often that’s been through probability. For
so-called “Bayesian” researchers, one of the most important texts is Ed

Figure: <i>Embodiment factors are the ratio between our ability to
compute and our ability to communicate. Relative to the machine we are
also locked in. In the table we represent embodiment as the length of
time it would take to communicate one second’s worth of computation. For

Shannon measured information in bits: one bit is the result of a fair
coin toss. He estimated $\sim 12$ bits per English word on average
(Shannon, 1948), which with typical speaking rates gives $\sim 10$–$60$
bits per second for human communication Lawrence (2024). Machines
communicate orders of magnitude faster — the embodiment factor is the
ratio between compute and that narrow channel.

Shannon measured information in bits. Human communication is slow
relative to machines — the embodiment factor. Lecture 3 derives $H$;
today we only need the bit as a unit of uncertainty and of bandwidth.

Shannon gave the unit used for bandwidth and embodiment factors. The
derivation of $H$ and the statement $S=kH$ are LO2 in lecture 3. The
bandwidth gap is a bottleneck on sharing thought, not a second no-go
paired with Boltzmann. Lecture 4 applies the same bit accounting to
locked-in communication.
